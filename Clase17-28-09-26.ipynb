## V.a. uniforme continua

Sea

$$
X\sim U[a,b].
$$

$X$ representa un resultado equiprobable en el intervalo $[a,b]$.

Su función de densidad es

$$
f_X(x)
=
\frac{1}{b-a}\mathbf{1}_{[a,b]}(x)
=
\frac{1}{b-a}\chi_{[a,b]}(x).
$$

Ahora calculamos la función de distribución acumulada:

$$
F_X(z)
=
\mathbb{P}(X\leq z)
=
\mathbb{P}\bigl(X\in(-\infty,z]\bigr).
$$

Por definición,

$$
F_X(z)
=
\int_{-\infty}^{z}f_X(t)\,dt
=
\int_{-\infty}^{z}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt.
$$

### Si $z<a$

Como el intervalo $(-\infty,z]$ no intersecta al soporte $[a,b]$,

$$
\int_{-\infty}^{z}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt
=
0.
$$

Por lo tanto,

$$
F_X(z)=0.
$$

### Si $a<z<b$

Tenemos

$$
\int_{-\infty}^{z}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt
=
\int_{-\infty}^{a}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt
+
\int_{a}^{z}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt.
$$

La primera integral es cero, por lo que

$$
F_X(z)
=
\int_{a}^{z}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt.
$$

Como $\chi_{[a,b]}(t)=1$ para $t\in[a,z]$,

$$
F_X(z)
=
\frac{1}{b-a}\int_a^z dt.
$$

Así,

$$
F_X(z)
=
\frac{1}{b-a}\left.t\right|_a^z
=
\frac{z-a}{b-a}.
$$

### Si ahora $z>b$

Continuamos calculando

$$
F_X(z)
=
\int_{-\infty}^{z}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt.
$$

### Caso $z>b$

Separamos la integral como

$$
F_X(z)
=
\int_{-\infty}^{a}
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt
+
\int_a^b
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt
+
\int_b^z
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt.
$$

La primera y la tercera integral son cero, de modo que

$$
F_X(z)
=
\int_a^b
\frac{1}{b-a}\chi_{[a,b]}(t)\,dt.
$$

Como $\chi_{[a,b]}(t)=1$ para $t\in[a,b]$,

$$
F_X(z)
=
\frac{1}{b-a}\int_a^b dt
=
\frac{b-a}{b-a}
=
1.
$$

De donde obtenemos la función de distribución:

$$
F_X(z)
=
\begin{cases}
0,
& z\leq a,\\[4pt]
\dfrac{z-a}{b-a},
& a<z<b,\\[8pt]
1,
& z\geq b.
\end{cases}
$$

> **Nota.** Si nos dan una función de densidad, podemos obtener la función de distribución integrando:
>
> $$
> F_X(z)=\int_{-\infty}^{z}f_X(t)\,dt.
> $$


# 2. V.a. exponencial

Sea

$$
X\sim \operatorname{Exp}(\lambda),
\qquad \lambda>0.
$$

$X$ representa el tiempo que tenemos que esperar hasta que suceda un evento que ocurre, en promedio, cada

$$
\frac{1}{\lambda}
$$

unidades de tiempo.

Su función de densidad es

$$
f_X(x)
=
\lambda e^{-\lambda x}
\mathbf{1}_{[0,\infty)}(x).
$$

Ahora, para calcular $F_X(z)$:

### Si $z<0$

Tenemos

$$
F_X(z)
=
\int_{-\infty}^{z}
\lambda e^{-\lambda t}
\mathbf{1}_{[0,\infty)}(t)\,dt.
$$

Como la densidad es cero para $t<0$,

$$
F_X(z)=0.
$$

### Si $z>0$

Entonces

### Caso $z>0$

Tenemos

$$
F_X(z)
=
\int_{-\infty}^{z}
\lambda e^{-\lambda t}
\mathbf{1}_{[0,\infty)}(t)\,dt.
$$

Separamos la integral en

$$
F_X(z)
=
\int_{-\infty}^{0}
\lambda e^{-\lambda t}
\mathbf{1}_{[0,\infty)}(t)\,dt
+
\int_0^z
\lambda e^{-\lambda t}
\mathbf{1}_{[0,\infty)}(t)\,dt.
$$

La primera integral es cero, por lo que

$$
F_X(z)
=
\int_0^z \lambda e^{-\lambda t}\,dt.
$$

Calculando,

$$
F_X(z)
=
\lambda
\left(
-\frac{1}{\lambda}e^{-\lambda t}
\right)\Bigg|_0^z.
$$

Por lo tanto,

$$
F_X(z)
=
-e^{-\lambda t}\Big|_0^z
=
1-e^{-\lambda z}.
$$

Así, la función de distribución de una variable aleatoria exponencial es

$$
F_X(z)
=
\begin{cases}
0,
& z<0,\\[4pt]
1-e^{-\lambda z},
& z\geq 0.
\end{cases}
$$

Además, su función de supervivencia es

$$
S_X(z)
=
\mathbb{P}(X>z).
$$

Para $z\geq0$,

$$
S_X(z)
=
1-F_X(z)
=
e^{-\lambda z}.
$$


## Relación entre la distribución geométrica y la exponencial

La v.a. exponencial es el análogo continuo de la v.a. geométrica.

Sea

$$
Y_n\sim \operatorname{Geo}\left(\frac{\lambda}{n}\right).
$$

La función de supervivencia de una variable geométrica es

$$
\mathbb{P}(Y_n>y)
=
(1-p)^y.
$$

Tomando

$$
p=\frac{\lambda}{n},
$$

obtenemos

$$
\mathbb{P}(Y_n>y)
=
\left(1-\frac{\lambda}{n}\right)^y.
$$

Ahora consideremos la variable reescalada

$$
\frac{Y_n}{n}.
$$

Entonces

$$
\mathbb{P}\left(\frac{Y_n}{n}>x\right)
=
\mathbb{P}(Y_n>nx).
$$

Para tomar en cuenta que $Y_n$ toma valores enteros, podemos escribir

$$
\mathbb{P}\left(\frac{Y_n}{n}>x\right)
=
\left(
1-\frac{\lambda}{n}
\right)^{\lfloor nx\rfloor}.
$$

Por lo tanto,

$$
\lim_{n\to\infty}
\mathbb{P}\left(\frac{Y_n}{n}>x\right)
=
\lim_{n\to\infty}
\left(
1-\frac{\lambda}{n}
\right)^{\lfloor nx\rfloor}.
$$

Usando el límite

$$
\lim_{n\to\infty}
\left(1-\frac{\lambda}{n}\right)^{nx}
=
e^{-\lambda x},
$$

obtenemos

$$
\lim_{n\to\infty}
\mathbb{P}\left(\frac{Y_n}{n}>x\right)
=
e^{-\lambda x}.
$$

Pero

$$
e^{-\lambda x}
=
\mathbb{P}(X>x),
$$

donde

$$
X\sim\operatorname{Exp}(\lambda).
$$

Por lo tanto, la distribución exponencial puede verse como un límite continuo de la distribución geométrica.


## Propiedad de falta de memoria

La distribución exponencial tiene la propiedad de falta de memoria.

Queremos calcular

$$
\mathbb{P}(X>s+t\mid X>s).
$$

Por definición de probabilidad condicional,

$$
\mathbb{P}(X>s+t\mid X>s)
=
\frac{
\mathbb{P}\bigl(\{X>s+t\}\cap\{X>s\}\bigr)
}{
\mathbb{P}(X>s)
}.
$$

Como

$$
\{X>s+t\}\subseteq\{X>s\},
$$

tenemos

$$
\{X>s+t\}\cap\{X>s\}
=
\{X>s+t\}.
$$

Por tanto,

$$
\mathbb{P}(X>s+t\mid X>s)
=
\frac{\mathbb{P}(X>s+t)}
{\mathbb{P}(X>s)}.
$$

Usando la función de supervivencia de la exponencial,

$$
\mathbb{P}(X>u)=e^{-\lambda u},
$$

obtenemos

$$
\mathbb{P}(X>s+t\mid X>s)
=
\frac{e^{-\lambda(s+t)}}{e^{-\lambda s}}.
$$

Simplificando,

$$
\mathbb{P}(X>s+t\mid X>s)
=
e^{-\lambda t}.
$$

Pero

$$
e^{-\lambda t}
=
\mathbb{P}(X>t).
$$

Por lo tanto,

$$
\boxed{
\mathbb{P}(X>s+t\mid X>s)
=
\mathbb{P}(X>t)
}
$$

Esta es la **propiedad de falta de memoria** de la distribución exponencial.
