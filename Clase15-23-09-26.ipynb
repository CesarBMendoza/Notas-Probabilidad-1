## Ejemplo de variable aleatoria binomial

Podemos pensar una variable aleatoria binomial como la suma de $n$ variables aleatorias Bernoulli independientes.

Por ejemplo,

$$
X
=
0+1+1+0+1+\cdots,
$$

donde cada término representa el resultado de un intento:

- $1$: éxito,
- $0$: fracaso.

Por tanto, si

$$
X_1,\ldots,X_n
$$

son variables aleatorias Bernoulli independientes con parámetro $p$, entonces

$$
X=X_1+\cdots+X_n
$$

tiene distribución binomial:

$$
X\sim\operatorname{Binomial}(n,p).
$$

Nos gustaría verificar que

$$
\sum_{j\in R_X}p_X(j)=1.
$$

Como

$$
R_X=\{0,1,\ldots,n\},
$$

tenemos

$$
\sum_{j\in R_X}p_X(j)
=
\sum_{j=0}^{n}p_X(j).
$$

Recordando que

$$
p_X(j)
=
\binom{n}{j}p^j(1-p)^{n-j},
$$

obtenemos

$$
\sum_{j=0}^{n}p_X(j)
=
\sum_{j=0}^{n}
\binom{n}{j}
p^j(1-p)^{n-j}.
$$

Por el teorema del binomio,

$$
\sum_{j=0}^{n}
\binom{n}{j}
p^j(1-p)^{n-j}
=
\bigl(p+(1-p)\bigr)^n.
$$

Por tanto,

$$
\sum_{j\in R_X}p_X(j)
=
1^n
=
1.
$$

---

### Ejemplo

Tenemos una urna con

- $3$ bolas negras,
- $3$ bolas blancas.

Sacamos $4$ bolas **con reemplazo**.

Sea

$$
X=\text{número de bolas negras obtenidas}.
$$

En cada extracción, la probabilidad de obtener una bola negra es

$$
p
=
\frac{3}{6}
=
\frac12.
$$

Además, como las extracciones son con reemplazo, los intentos son independientes.

Por tanto,

$$
\boxed{
X\sim\operatorname{Binomial}\left(4,\frac12\right)
}.
$$

## Variable aleatoria de Poisson

Sea

$$
N\sim\operatorname{Poisson}(\lambda),
\qquad \lambda>0.
$$

La variable aleatoria $N$ representa el número de ocurrencias de cierto evento en un intervalo de tiempo fijo, donde $\lambda$ representa el número promedio de ocurrencias en dicho intervalo.

El rango de $N$ es

$$
R_N
=
\{0,1,2,\ldots\}
=
\mathbb{N}_0.
$$

Su función de probabilidad puntual está dada por

$$
p_N(k)
=
\mathbb{P}[N=k]
=
e^{-\lambda}
\frac{\lambda^k}{k!},
\qquad
k=0,1,2,\ldots
$$

Verifiquemos que las probabilidades suman $1$:

$$
\sum_{j\in R_N}p_N(j)
=
\sum_{j=0}^{\infty}p_N(j).
$$

Sustituyendo la función de probabilidad,

$$
\sum_{j=0}^{\infty}
e^{-\lambda}
\frac{\lambda^j}{j!}.
$$

Sacando $e^{-\lambda}$ de la suma,

$$
=
e^{-\lambda}
\sum_{j=0}^{\infty}
\frac{\lambda^j}{j!}.
$$

Recordemos que la expansión en serie de la exponencial es

$$
e^\lambda
=
\sum_{j=0}^{\infty}
\frac{\lambda^j}{j!}.
$$

Por tanto,

$$
\sum_{j\in R_N}p_N(j)
=
e^{-\lambda}e^\lambda
=
1.
$$

Así,

$$
\boxed{
\sum_{j\in R_N}p_N(j)=1
}.
$$

---

## Aproximación de Poisson a la binomial

Gráficamente, y también analíticamente, la distribución de Poisson puede aparecer como un límite de distribuciones binomiales.

### Teorema

Sea

$$
X_n\sim\operatorname{Binomial}(n,p_n),
$$

donde

$$
p_n=\frac{\lambda}{n},
\qquad
\lambda>0.
$$

Sea además

$$
N\sim\operatorname{Poisson}(\lambda).
$$

Entonces, para cada $k\in\mathbb{N}_0$ fijo,

$$
\boxed{
\lim_{n\to\infty}p_{X_n}(k)
=
p_N(k)
}.
$$

Es decir,

$$
\lim_{n\to\infty}
\mathbb{P}[X_n=k]
=
e^{-\lambda}
\frac{\lambda^k}{k!}.
$$

## Demostración de la aproximación de Poisson

Sea

$$
X_n\sim\operatorname{Binomial}
\left(
n,\frac{\lambda}{n}
\right).
$$

Para $k$ fijo,

$$
p_{X_n}(k)
=
\binom{n}{k}
\left(\frac{\lambda}{n}\right)^k
\left(
1-\frac{\lambda}{n}
\right)^{n-k}.
$$

Utilizando

$$
\binom{n}{k}
=
\frac{n!}{k!(n-k)!},
$$

obtenemos

$$
p_{X_n}(k)
=
\frac{n!}{k!(n-k)!}
\frac{\lambda^k}{n^k}
\left(
1-\frac{\lambda}{n}
\right)^n
\left(
1-\frac{\lambda}{n}
\right)^{-k}.
$$

Por tanto,

$$
\lim_{n\to\infty}p_{X_n}(k)
=
\frac{\lambda^k}{k!}
\lim_{n\to\infty}
\left[
\frac{n!}{(n-k)!n^k}
\left(
1-\frac{\lambda}{n}
\right)^n
\left(
1-\frac{\lambda}{n}
\right)^{-k}
\right].
$$

Veamos cada término por separado.

---

### Primer término

Tenemos

$$
\frac{n!}{(n-k)!}
=
n(n-1)\cdots(n-k+1).
$$

Por tanto,

$$
\frac{n!}{(n-k)!n^k}
=
\frac{n}{n}
\frac{n-1}{n}
\cdots
\frac{n-k+1}{n}.
$$

Hay $k$ factores.

Para cada factor,

$$
\frac{n-r}{n}
=
1-\frac{r}{n}
\longrightarrow1
$$

cuando $n\to\infty$.

Así,

$$
\lim_{n\to\infty}
\frac{n!}{(n-k)!n^k}
=
1.
$$

---

### Segundo término

Utilizando el límite exponencial conocido,

$$
\lim_{n\to\infty}
\left(
1-\frac{\lambda}{n}
\right)^n
=
e^{-\lambda}.
$$

---

### Tercer término

Como $k$ es fijo,

$$
\lim_{n\to\infty}
\left(
1-\frac{\lambda}{n}
\right)^{-k}
=
1.
$$

En efecto,

$$
\left(
1-\frac{\lambda}{n}
\right)^{-k}
=
\underbrace{
\left(
1-\frac{\lambda}{n}
\right)^{-1}
\cdots
\left(
1-\frac{\lambda}{n}
\right)^{-1}
}_{k\text{ factores}},
$$

y cada factor converge a $1$.

---

Por lo tanto,

$$
\lim_{n\to\infty}p_{X_n}(k)
=
\frac{\lambda^k}{k!}
(1)
e^{-\lambda}
(1).
$$

Así,

$$
\boxed{
\lim_{n\to\infty}p_{X_n}(k)
=
e^{-\lambda}
\frac{\lambda^k}{k!}
}.
$$

Pero

$$
e^{-\lambda}
\frac{\lambda^k}{k!}
=
p_N(k),
$$

donde

$$
N\sim\operatorname{Poisson}(\lambda).
$$

Por tanto,

$$
\boxed{
\lim_{n\to\infty}p_{X_n}(k)
=
p_N(k)
}.
$$

## Variable aleatoria binomial negativa

Sea

$$
Y\sim\operatorname{BN}(k,p).
$$

Existen dos convenciones frecuentes para definir una variable aleatoria binomial negativa.

---

## Definición 1: número de intentos necesarios para obtener $k$ éxitos

Definimos

$$
Y
=
\text{número de intentos necesarios para obtener $k$ éxitos},
$$

donde cada intento tiene probabilidad de éxito

$$
p.
$$

Esta variable aleatoria es una generalización de la variable geométrica.

En efecto, para

$$
k=1,
$$

tenemos

$$
Y\sim\operatorname{Geom}(p).
$$

El rango de $Y$ es

$$
R_Y
=
\{k,k+1,k+2,\ldots\}.
$$

Queremos calcular

$$
p_Y(j)
=
\mathbb{P}[Y=j].
$$

Si pensamos en una realización de $j$ intentos, para que

$$
Y=j,
$$

debemos tener exactamente $k-1$ éxitos entre los primeros $j-1$ intentos y el último intento necesariamente debe ser un éxito.

Por ejemplo, una realización podría verse como

$$
E,\ F,\ F,\ E,\ F,\ E,\ F,\ E,
$$

donde el último lugar contiene el $k$-ésimo éxito.

Entre los primeros $j-1$ lugares debemos escoger las posiciones de los $k-1$ éxitos.

Esto puede hacerse de

$$
\binom{j-1}{k-1}
$$

formas.

Cada configuración tiene

- $k$ éxitos,
- $j-k$ fracasos.

Por independencia, su probabilidad es

$$
p^k(1-p)^{j-k}.
$$

Por tanto,

$$
\boxed{
p_Y(j)
=
\binom{j-1}{k-1}
p^k(1-p)^{j-k}
}
$$

para

$$
j=k,k+1,k+2,\ldots
$$

---

## Definición 2: número de fracasos antes de obtener $k$ éxitos

Otra convención consiste en definir

$$
Y
=
\text{número de fracasos que ocurren antes de obtener $k$ éxitos}.
$$

En este caso,

$$
R_Y
=
\{0,1,2,\ldots\}
=
\mathbb{N}_0.
$$

Si

$$
Y=j,
$$

entonces ocurrieron

- $j$ fracasos,
- $k$ éxitos,

y el último intento necesariamente fue un éxito.

Antes de ese último éxito tenemos

$$
j+k-1
$$

intentos, en los cuales debemos colocar los $j$ fracasos.

Esto puede hacerse de

$$
\binom{j+k-1}{j}
$$

formas.

Por tanto,

$$
\boxed{
p_Y(j)
=
\binom{j+k-1}{j}
p^k(1-p)^j
}
$$

para

$$
j=0,1,2,\ldots
$$

Equivalentemente,

$$
p_Y(j)
=
\binom{j+k-1}{k-1}
p^k(1-p)^j.
$$
