# Ejemplo 2. Cumpleaños en todos los días del año

¿Cuál es la probabilidad de que, entre $k$ amigos, haya **al menos un amigo que cumpla años en cada uno de los días del año**?

Tomaremos nuevamente

$n=365$.

> **Suposición.** Suponemos que cada uno de los $n=365$ días es igualmente probable como cumpleaños y que los cumpleaños de los amigos son independientes.

---

## Caso 1: $k<365$

Si $k<365$, entonces es imposible que haya al menos una persona cumpliendo años en cada uno de los $365$ días.

Por lo tanto,

$P(A)=0$.

> **¿Por qué?** Tenemos menos de $365$ personas y queremos cubrir $365$ días diferentes. Aun si todos los amigos cumplen años en días distintos, quedarían días sin ningún cumpleaños.

---

## Caso 2: $k=365$

Si $k=365$, para que todos los días del año aparezcan como cumpleaños es necesario que **exactamente una persona cumpla años en cada día**.

Existen

$365!$

formas de asignar los $365$ días distintos a los $365$ amigos.

Por otro lado, existen

$365^{365}$

posibles asignaciones de cumpleaños sin ninguna restricción.

Por lo tanto,

$P(A)=\frac{365!}{365^{365}}$.

En general, si $k=n$,

$P(A)=\frac{n!}{n^n}$.

Para $n=365$,

$P(A)=\frac{365!}{365^{365}}\approx1.455\times10^{-157}$.

> **¿De dónde sale $365!$?** El primer amigo puede ocupar cualquiera de los $365$ días, el segundo debe ocupar uno de los $364$ días restantes, el tercero uno de los $363$ restantes, y así sucesivamente. Por la regla de multiplicación obtenemos $365\cdot364\cdot363\cdots1=365!$.

---

## Caso 3: $k>365$

Ahora sí es posible que haya varias personas cumpliendo años el mismo día, pero queremos que **ninguno de los $365$ días quede vacío**.

Definamos

$A=\{\text{hay al menos un amigo que cumple años en cada día del año}\}$.

Entonces,

$A^c=\{\text{hay al menos un día del año en el que nadie cumple años}\}$.

Para estudiar $A^c$, definimos los eventos

$B_i=\{\text{no hay ningún cumpleaños en el día }i\}$,

para $i=1,\ldots,n$.

Entonces,

$A^c=\bigcup_{i=1}^{n}B_i$.

Por lo tanto,

$P(A)=1-P(A^c)=1-P\left(\bigcup_{i=1}^{n}B_i\right)$.

Para calcular esta unión utilizaremos el **principio de inclusión-exclusión**.

---

## Probabilidad de $B_i$

Fijemos un día $i$.

Queremos que ninguno de los $k$ amigos cumpla años ese día.

Cada amigo puede cumplir años en cualquiera de los otros $n-1$ días.

Por la regla de multiplicación, existen

$(n-1)^k$

asignaciones de cumpleaños en las que nadie cumple años en el día $i$.

Como el espacio muestral tiene $n^k$ elementos,

$P(B_i)=\frac{(n-1)^k}{n^k}$.

Por lo tanto,

$P(B_i)=\left(\frac{n-1}{n}\right)^k$.

Equivalentemente,

$P(B_i)=\left(1-\frac{1}{n}\right)^k$.

> **¿De dónde sale esta potencia?** Cada amigo tiene probabilidad $\frac{n-1}{n}$ de evitar el día $i$. Como los $k$ cumpleaños son independientes, multiplicamos esa probabilidad $k$ veces.

Como existen $n$ posibles días que podrían quedar vacíos, hay

$\binom{n}{1}=n$

eventos de este tipo.

---

## Intersección de dos eventos

Ahora consideremos dos días distintos $i$ y $j$.

El evento

$B_i\cap B_j$

significa que nadie cumple años ni en el día $i$ ni en el día $j$.

Por lo tanto, cada amigo tiene solamente $n-2$ días disponibles.

Así,

$P(B_i\cap B_j)=\frac{(n-2)^k}{n^k}$.

Equivalentemente,

$P(B_i\cap B_j)=\left(1-\frac{2}{n}\right)^k$.

Como debemos escoger qué dos días quedan vacíos, existen

$\binom{n}{2}$

intersecciones de este tipo.

> **¿De dónde sale $\binom{n}{2}$?** De los $n$ días elegimos los dos días que queremos dejar sin cumpleaños.

---

## Intersección de tres eventos

Si elegimos tres días distintos,

$B_i\cap B_j\cap B_\ell$

significa que ninguno de los $k$ amigos cumple años en esos tres días.

Cada amigo dispone entonces de $n-3$ posibles días.

Por lo tanto,

$P(B_i\cap B_j\cap B_\ell)=\frac{(n-3)^k}{n^k}$.

Equivalentemente,

$P(B_i\cap B_j\cap B_\ell)=\left(1-\frac{3}{n}\right)^k$.

Existen

$\binom{n}{3}$

formas de escoger esos tres días.

---

## Intersección de $j$ eventos

En general, si escogemos $j$ días que deben permanecer sin cumpleaños, cada amigo puede cumplir años solamente en alguno de los $n-j$ días restantes.

Por lo tanto,

$P(B_{i_1}\cap B_{i_2}\cap\cdots\cap B_{i_j})=\frac{(n-j)^k}{n^k}$.

Equivalentemente,

$P(B_{i_1}\cap B_{i_2}\cap\cdots\cap B_{i_j})=\left(1-\frac{j}{n}\right)^k$.

Como podemos escoger los $j$ días vacíos de

$\binom{n}{j}$

maneras distintas, el término correspondiente en inclusión-exclusión es

$\binom{n}{j}\left(1-\frac{j}{n}\right)^k$.

Finalmente,

$P(B_1\cap B_2\cap\cdots\cap B_n)=0$,

pues no puede ocurrir que los $k$ amigos eviten simultáneamente todos los días del año.

---

## Aplicando inclusión-exclusión

Por el principio de inclusión-exclusión,

$P(A^c)=\binom{n}{1}\left(1-\frac{1}{n}\right)^k-\binom{n}{2}\left(1-\frac{2}{n}\right)^k+\binom{n}{3}\left(1-\frac{3}{n}\right)^k-\cdots$.

Por lo tanto,

$P(A)=1-P(A^c)$.

Así,

$P(A)=1-\binom{n}{1}\left(1-\frac{1}{n}\right)^k+\binom{n}{2}\left(1-\frac{2}{n}\right)^k-\binom{n}{3}\left(1-\frac{3}{n}\right)^k+\cdots$.

Podemos incorporar el término inicial $1$ escribiéndolo como

$1=\binom{n}{0}\left(1-\frac{0}{n}\right)^k$.

Por lo tanto,

$P(A)=\sum_{j=0}^{n}(-1)^j\binom{n}{j}\left(1-\frac{j}{n}\right)^k$.

Como para $k>0$ el término correspondiente a $j=n$ es cero, también podemos escribir

$P(A)=\sum_{j=0}^{n-1}(-1)^j\binom{n}{j}\left(1-\frac{j}{n}\right)^k$.

---

## Interpretación de la fórmula

Cada término tiene una interpretación combinatoria:

- $\binom{n}{j}$ representa las formas de escoger los $j$ días que queremos dejar sin cumpleaños.
- $\left(1-\frac{j}{n}\right)^k$ es la probabilidad de que los $k$ amigos eviten esos $j$ días.
- El factor $(-1)^j$ aparece por el principio de inclusión-exclusión, que alterna restas y sumas para corregir el sobreconteo.

Por tanto, la fórmula final es

$P(A)=\sum_{j=0}^{n}(-1)^j\binom{n}{j}\left(1-\frac{j}{n}\right)^k$.

Para nuestro problema,

$n=365$.
