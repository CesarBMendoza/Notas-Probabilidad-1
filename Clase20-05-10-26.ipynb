# Transformaciones de variables aleatorias

Una variable aleatoria puede transformarse mediante una función y producir una nueva distribución.

Dependiendo de la distribución original de $X$ y de la función $g$, la variable aleatoria $Y=g(X)$ puede tener una distribución discreta, continua o mixta.

**Observación:** Si $X$ es discreta, entonces $Y=g(X)$ necesariamente es discreta. En cambio, una variable aleatoria continua puede transformarse en una discreta, continua o mixta.

Recordemos que, si $X$ es una variable aleatoria y

$$
g:\mathbb{R}\longrightarrow\mathbb{R}
$$

es una función medible, entonces

$$
Y=g(X)
$$

es una variable aleatoria.

Ahora nos gustaría determinar su distribución.

---

## Transformaciones de variables aleatorias discretas

Supongamos que $X$ es una variable aleatoria discreta con función de masa

$$
p_X(x)=P(X=x).
$$

Definimos una nueva variable aleatoria

$$
Y=g(X).
$$

Queremos encontrar su función de masa.

Por definición,

$$
p_Y(y)=P(Y=y).
$$

Como $Y=g(X)$, tenemos que

$$
\begin{aligned}
p_Y(y)
&=P(Y=y)\\
&=P(g(X)=y).
\end{aligned}
$$

Observemos que el evento $\{g(X)=y\}$ ocurre cuando $X$ toma cualquiera de los valores $x$ tales que $g(x)=y$.

Por lo tanto,

$$
P(g(X)=y)
=
\sum_{\substack{x\in R_X\\g(x)=y}}P(X=x).
$$

Así,

$$
\boxed{
p_Y(y)=
\sum_{\substack{x\in R_X\\g(x)=y}}p_X(x).
}
$$

Esta fórmula nos permite obtener la función de masa de $Y$ a partir de la función de masa de $X$.

---

## Ejemplo 1: Transformación de una variable Poisson

Supongamos que

$$
X\sim\operatorname{Poisson}(\lambda),
$$

donde $\lambda>0$.

Sabemos que

$$
R_X=\mathbb{N}_0=\{0,1,2,3,\ldots\}
$$

y que su función de masa está dada por

$$
p_X(k)=e^{-\lambda}\frac{\lambda^k}{k!},
\qquad k=0,1,2,\ldots
$$

Definimos

$$
Y=X^2.
$$

Queremos encontrar la función de masa de $Y$.

Observemos que

$$
R_Y=\{0,1,4,9,16,\ldots\}.
$$

Es decir, $Y$ solamente puede tomar valores que sean cuadrados perfectos de enteros no negativos.

Por definición,

$$
\begin{aligned}
p_Y(k)
&=P(Y=k)\\
&=P(X^2=k).
\end{aligned}
$$

Como $X\geq 0$, tenemos que

$$
X^2=k
\iff
X=\sqrt{k}.
$$

Por lo tanto,

$$
\begin{aligned}
p_Y(k)
&=P(X=\sqrt{k})\\
&=p_X(\sqrt{k})\\
&=e^{-\lambda}
\frac{\lambda^{\sqrt{k}}}{(\sqrt{k})!}.
\end{aligned}
$$

Esta expresión es válida cuando $\sqrt{k}$ es un entero no negativo.

Continuando con el ejemplo anterior, tenemos que

$$
X\sim\operatorname{Poisson}(\lambda)
$$

y

$$
Y=X^2.
$$

Por lo tanto,

$$
p_Y(k)=
e^{-\lambda}
\frac{\lambda^{\sqrt{k}}}{(\sqrt{k})!},
\qquad k\in R_Y.
$$

Es decir, la función de masa de $Y$ está dada por

$$
\boxed{
p_Y(k)=
\begin{cases}
e^{-\lambda}\dfrac{\lambda^{\sqrt{k}}}{(\sqrt{k})!},
& k=n^2,\quad n\in\mathbb{N}_0,\\[8pt]
0, & \text{en otro caso}.
\end{cases}
}
$$

## Ejemplo 2: Transformación de una variable aleatoria discreta

Sea $X$ una variable aleatoria con función de masa dada por

$$
p_X(k)=P(X=k)=
\begin{cases}
\dfrac{1}{5}, & k=-2,\\[6pt]
\dfrac{1}{6}, & k=-1,\\[6pt]
\dfrac{1}{5}, & k=0,\\[6pt]
\dfrac{1}{15}, & k=1,\\[6pt]
\dfrac{11}{30}, & k=2,\\[6pt]
0, & \text{en otro caso}.
\end{cases}
$$

Por lo tanto,

$$
R_X=\{-2,-1,0,1,2\}.
$$

Veamos primero que efectivamente se trata de una función de masa.

Para ello, debemos verificar que

$$
\sum_{k\in R_X}p_X(k)=1.
$$

En efecto,

$$
\begin{aligned}
\sum_{k\in R_X}p_X(k)
&=\frac15+\frac16+\frac15+\frac1{15}+\frac{11}{30}\\
&=\frac{6+5+6+2+11}{30}\\
&=\frac{30}{30}\\
&=1.
\end{aligned}
$$

Además, todas las probabilidades son no negativas.

Por lo tanto, $p_X$ es una función de masa de probabilidad.

---

Ahora consideremos la transformación

$$
Y=X^2+3.
$$

Es decir,

$$
g(x)=x^2+3.
$$

Queremos determinar los valores que puede tomar $Y$ y encontrar su función de masa.

Para ello, evaluamos $g$ en cada uno de los valores de $R_X$.

| $x$ | $-2$ | $-1$ | $0$ | $1$ | $2$ |
|---|---|---|---|---|---|
| $g(x)=x^2+3$ | $7$ | $4$ | $3$ | $4$ | $7$ |

Con esto, obtenemos que

$$
R_Y=\{3,4,7\}.
$$

Ahora calcularemos las probabilidades correspondientes a cada valor.

### Probabilidad de que $Y=3$

Por definición,

$$
\begin{aligned}
p_Y(3)
&=P(Y=3)\\
&=P(X^2+3=3)\\
&=P(X^2=0)\\
&=P(X=0)\\
&=\frac15\\
&=\frac6{30}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{p_Y(3)=\frac6{30}.}
$$

### Probabilidad de que $Y=4$

Tenemos que

$$
\begin{aligned}
p_Y(4)
&=P(Y=4)\\
&=P(X^2+3=4)\\
&=P(X^2=1).
\end{aligned}
$$

Observemos que

$$
X^2=1
\iff
X=-1\ \text{o}\ X=1.
$$

Continuando con el ejemplo anterior,

$$
p_Y(4)=P(X^2=1).
$$

Como los eventos $\{X=-1\}$ y $\{X=1\}$ son ajenos, tenemos que

$$
\begin{aligned}
p_Y(4)
&=P(X=-1)+P(X=1)\\
&=\frac16+\frac1{15}\\
&=\frac5{30}+\frac2{30}\\
&=\frac7{30}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{p_Y(4)=\frac7{30}.}
$$

### Probabilidad de que $Y=7$

Tenemos que

$$
\begin{aligned}
p_Y(7)
&=P(Y=7)\\
&=P(X^2+3=7)\\
&=P(X^2=4)\\
&=P(X=-2)+P(X=2)\\
&=\frac15+\frac{11}{30}\\
&=\frac6{30}+\frac{11}{30}\\
&=\frac{17}{30}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{p_Y(7)=\frac{17}{30}.}
$$

En consecuencia, la función de masa de $Y$ está dada por

$$
\boxed{
p_Y(k)=
\begin{cases}
\dfrac6{30}, & k=3,\\[6pt]
\dfrac7{30}, & k=4,\\[6pt]
\dfrac{17}{30}, & k=7,\\[6pt]
0, & \text{en otro caso}.
\end{cases}
}
$$

Verifiquemos que efectivamente es una función de masa.

$$
\begin{aligned}
\sum_{k\in R_Y}p_Y(k)
&=\frac6{30}+\frac7{30}+\frac{17}{30}\\
&=\frac{30}{30}\\
&=1.
\end{aligned}
$$

Por lo tanto, $p_Y$ es una función de masa de probabilidad.

# Transformación de una variable aleatoria continua en una discreta

Una variable aleatoria continua puede transformarse en una variable aleatoria discreta.

Para mostrarlo, consideremos el siguiente ejemplo.

## Ejemplo 3

Supongamos que

$$
X\sim U(-1,1).
$$

Entonces su función de densidad está dada por

$$
f_X(x)=
\frac12\mathbb{1}_{(-1,1)}(x).
$$

Es decir,

$$
f_X(x)=
\begin{cases}
\dfrac12, & x\in(-1,1),\\[6pt]
0, & \text{en otro caso}.
\end{cases}
$$

Definamos la función

$$
g(x)=
\begin{cases}
1, & x\geq 0,\\
0, & x<0.
\end{cases}
$$

Y consideremos la variable aleatoria

$$
Y=g(X).
$$

Como $X$ es una variable aleatoria continua y $g$ es una función medible, sabemos que $Y$ es una variable aleatoria.

Sin embargo, observemos que $Y$ solamente puede tomar dos valores:

$$
R_Y=\{0,1\}.
$$

Por lo tanto, $Y$ es una variable aleatoria discreta.

Ahora determinaremos su función de masa.

### Probabilidad de que $Y=0$

Observemos que

$$
Y=0
\iff
X<0.
$$

Por lo tanto,

$$
\begin{aligned}
p_Y(0)
&=P(Y=0)\\
&=P(X<0)\\
&=\int_{-1}^{0}\frac12\,dt\\
&=\frac12[t]_{-1}^{0}\\
&=\frac12.
\end{aligned}
$$

Es decir,

$$
\boxed{p_Y(0)=\frac12.}
$$

Continuando con el ejemplo anterior, tenemos que

$$
X\sim U(-1,1)
$$

y

$$
Y=
\begin{cases}
1, & X\geq 0,\\
0, & X<0.
\end{cases}
$$

Ya encontramos que

$$
p_Y(0)=\frac12.
$$

Por otro lado,

$$
\begin{aligned}
p_Y(1)
&=P(Y=1)\\
&=P(X\geq 0)\\
&=1-P(X<0)\\
&=1-\frac12\\
&=\frac12.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{p_Y(1)=\frac12.}
$$

Así, la función de masa de $Y$ está dada por

$$
\boxed{
p_Y(k)=
\begin{cases}
\dfrac12, & k=0,1,\\[6pt]
0, & \text{en otro caso}.
\end{cases}
}
$$

Esto nos dice que

$$
\boxed{
Y\sim\operatorname{Bernoulli}\left(\frac12\right).
}
$$

Por lo tanto, hemos transformado una variable aleatoria continua en una variable aleatoria discreta.

# Transformaciones que producen distribuciones mixtas

Una transformación también puede producir una distribución mixta.

Consideremos nuevamente

$$
X\sim U(-1,1).
$$

Ahora definimos la función

$$
g(x)=
\begin{cases}
x, & x<0,\\
1, & x\geq 0.
\end{cases}
$$

Y tomamos

$$
Y=g(X).
$$

Queremos determinar qué valores puede tomar $Y$.

### Caso 1: $X<0$

Si

$$
X<0,
$$

entonces

$$
Y=X.
$$

Por lo tanto, en este caso $Y$ puede tomar valores en el intervalo

$$
(-1,0).
$$

### Caso 2: $X\geq 0$

Si

$$
X\geq 0,
$$

entonces

$$
Y=g(X)=1.
$$

Por lo tanto, en este caso $Y$ solamente puede tomar el valor $1$.

En consecuencia, los valores posibles de $Y$ son

$$
\boxed{
R_Y=(-1,0)\cup\{1\}.
}
$$

A este tipo de distribución se le llama **distribución mixta**, pues tiene una componente continua y una componente discreta.

---

## Masa puntual de $Y$

Calculemos la masa puntual en $Y=1$.

Por definición,

$$
\begin{aligned}
P(Y=1)
&=P(X\geq 0)\\
&=\frac12.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(Y=1)=\frac12.
}
$$

Esto nos dice que $Y$ tiene una masa puntual de probabilidad igual a $\frac12$ en el valor $1$.

Ahora queremos encontrar su función de distribución acumulada.

Recordemos que

$$
\boxed{
F_Y(y)=P(Y\leq y).
}
$$

Para calcularla, consideraremos diferentes casos dependiendo de los valores de $y$.

## Cálculo de la función de distribución acumulada

Recordemos que

$$
X\sim U(-1,1)
$$

y

$$
Y=g(X)=
\begin{cases}
X, & X<0,\\
1, & X\geq 0.
\end{cases}
$$

Queremos determinar

$$
F_Y(y)=P(Y\leq y).
$$

Consideraremos cuatro casos.

---

### Caso 1: $y\leq -1$

Como los valores posibles de $Y$ pertenecen a

$$
R_Y=(-1,0)\cup\{1\},
$$

tenemos que, si $y\leq -1$,

$$
P(Y\leq y)=0.
$$

Por lo tanto,

$$
\boxed{
F_Y(y)=0,
\qquad y\leq -1.
}
$$

---

### Caso 2: $-1<y<0$

Observemos que, para $-1<y<0$, el evento

$$
Y\leq y
$$

solamente puede ocurrir cuando $X<0$.

En este caso,

$$
Y=X.
$$

Por lo tanto,

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(X\leq y)\\
&=F_X(y).
\end{aligned}
$$

Es decir,

$$
F_Y(y)=F_X(y),
\qquad y\in(-1,0).
$$

Como

$$
X\sim U(-1,1),
$$

sabemos que

$$
f_X(x)=\frac12,
\qquad x\in(-1,1).
$$

Entonces,

$$
\begin{aligned}
F_X(y)
&=\int_{-1}^{y}\frac12\,dt\\
&=\frac12[t]_{-1}^{y}\\
&=\frac12(y+1).
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
F_Y(y)=\frac{y+1}{2},
\qquad -1<y<0.
}
$$

---

### Caso 3: $0\leq y<1$

Si $X<0$, entonces

$$
Y=X<0.
$$

Por lo tanto,

$$
Y\leq y.
$$

Sin embargo, si $X\geq 0$, entonces

$$
Y=1>y.
$$

En consecuencia, para $0\leq y<1$, el evento $\{Y\leq y\}$ coincide con el evento $\{X<0\}$.

Así,

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(X<0)\\
&=\frac12.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
F_Y(y)=\frac12,
\qquad 0\leq y<1.
}
$$

---

### Caso 4: $y\geq 1$

Como el valor máximo que puede tomar $Y$ es $1$, tenemos que

$$
P(Y\leq y)=1
$$

para todo $y\geq 1$.

Por lo tanto,

$$
\boxed{
F_Y(y)=1,
\qquad y\geq 1.
}
$$

---

## Función de distribución acumulada

Reuniendo los cuatro casos, obtenemos que

$$
\boxed{
F_Y(y)=
\begin{cases}
0, & y\leq -1,\\[6pt]
\dfrac{y+1}{2}, & -1<y<0,\\[8pt]
\dfrac12, & 0\leq y<1,\\[8pt]
1, & y\geq 1.
\end{cases}
}
$$

Esta es la función de distribución acumulada de $Y$.

# Características de una distribución mixta

Observemos que no existe una función de densidad ordinaria $f_Y$ respecto de la medida de Lebesgue que describa toda la distribución de $Y$.

Esto se debe a que

$$
\boxed{
P(Y=1)=\frac12>0.
}
$$

Recordemos que, si $Y$ es una variable aleatoria absolutamente continua, entonces

$$
P(Y=y)=0,
\qquad \forall y\in\mathbb{R}.
$$

En efecto, si $Y$ tiene una función de densidad $f_Y$, entonces

$$
P(Y=y)
=
\int_{\{y\}}f_Y(t)\,dt
=
0,
$$

pues un conjunto formado por un solo punto tiene medida de Lebesgue igual a cero.

Sin embargo, en nuestro ejemplo,

$$
P(Y=1)=\frac12.
$$

Por lo tanto, $Y$ no es una variable aleatoria absolutamente continua.

Tampoco es una variable aleatoria discreta, pues toma valores en todo el intervalo $(-1,0)$ y

$$
P(Y\in(-1,0))=\frac12.
$$

Así, concluimos que

$$
\boxed{
Y\text{ tiene una distribución mixta}.
}
$$

---

## Observación adicional: Componente continua y componente discreta

La distribución de $Y$ puede separarse en dos componentes.

### Componente continua

Cuando $X<0$, tenemos que

$$
Y=X.
$$

Como

$$
X\sim U(-1,1),
$$

la componente continua está descrita por

$$
f_c(y)=\frac12\mathbb{1}_{(-1,0)}(y).
$$

Observemos que

$$
\begin{aligned}
\int_{-\infty}^{\infty}f_c(y)\,dy
&=\int_{-1}^{0}\frac12\,dy\\
&=\frac12.
\end{aligned}
$$

Es decir, la componente continua representa una probabilidad total de $\frac12$.

### Componente discreta

Por otro lado, tenemos una masa puntual en $y=1$ dada por

$$
P(Y=1)=\frac12.
$$

Así, la probabilidad total se obtiene sumando ambas componentes:

$$
\frac12+\frac12=1.
$$

Por lo tanto, la distribución de $Y$ está formada por:

- Una componente continua en $(-1,0)$, con probabilidad total $\frac12$.
- Una masa puntual en $y=1$, con probabilidad $\frac12$.

Esta es precisamente la estructura de una distribución mixta.
