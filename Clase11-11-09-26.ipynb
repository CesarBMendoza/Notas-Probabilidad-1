# Probabilidad condicional e independencia

Recordemos que, si $P(B)>0$, la probabilidad condicional de $A$ dado $B$ se define como

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}.
$$

Supongamos que conocer que ocurrió $B$ no cambia la probabilidad de que ocurra $A$.

Es decir,

$$
P(A\mid B)=P(A).
$$

Usando la definición de probabilidad condicional,

$$
P(A)
=
P(A\mid B)
=
\frac{P(A\cap B)}{P(B)}.
$$

Multiplicando por $P(B)$ obtenemos

$$
P(A\cap B)=P(A)P(B).
$$

En este caso diremos que los eventos $A$ y $B$ son **independientes**.

> **Nota.**
> Las dos condiciones
>
> $$
> P(A\mid B)=P(A)
> $$
>
> y
>
> $$
> P(A\cap B)=P(A)P(B)
> $$
>
> son equivalentes siempre que $P(B)>0$.

---

## Independencia con el complemento

Supongamos que $A$ y $B$ son independientes. Entonces

$$
P(A\cap B)=P(A)P(B).
$$

Queremos estudiar ahora la relación entre $A$ y $B^c$.

Observamos que

$$
A=(A\cap B)\cup(A\cap B^c),
$$

y además los eventos $A\cap B$ y $A\cap B^c$ son disjuntos.

Por lo tanto,

$$
P(A)=P(A\cap B)+P(A\cap B^c).
$$

Despejando,

$$
P(A\cap B^c)
=
P(A)-P(A\cap B).
$$

Como $A$ y $B$ son independientes,

$$
P(A\cap B^c)
=
P(A)-P(A)P(B).
$$

Factorizando $P(A)$,

$$
P(A\cap B^c)
=
P(A)(1-P(B)).
$$

Pero

$$
1-P(B)=P(B^c),
$$

por lo tanto

$$
P(A\cap B^c)=P(A)P(B^c).
$$

Así, $A$ y $B^c$ también son independientes.

De manera análoga, si $A$ y $B$ son independientes, entonces también son independientes los pares

$$
A^c,\ B,
$$

$$
A,\ B^c,
$$

y

$$
A^c,\ B^c.
$$

---

# Independencia de varios eventos

Sean $E_1,E_2,\ldots,E_n$ eventos.

Diremos que son **independientes a pares** si para todo $i\neq j$ se cumple

$$
P(E_i\cap E_j)=P(E_i)P(E_j).
$$

Es importante notar que independencia a pares no necesariamente implica independencia de todos los eventos.

Diremos que $E_1,\ldots,E_n$ son **mutuamente independientes** si para cualquier colección de índices distintos

$$
i_1,\ldots,i_k,
$$

se cumple

$$
P(E_{i_1}\cap\cdots\cap E_{i_k})
=
P(E_{i_1})\cdots P(E_{i_k}).
$$

Es decir, la condición debe cumplirse para todas las intersecciones de dos, tres, cuatro, etc., eventos.

> **Nota.**
> No basta comprobar únicamente que
>
> $$
> P(E_1\cap\cdots\cap E_n)
> =
> \prod_{i=1}^nP(E_i).
> $$
>
> También se deben verificar las intersecciones de todas las subcolecciones.

---

# Ejemplo: independencia a pares pero no independencia mutua

Consideremos un dado de cuatro caras con espacio muestral

$$
\Omega=\{1,2,3,4\},
$$

donde todos los resultados son igualmente probables.

Definimos los eventos

$$
A=\{1,2\},
$$

$$
B=\{1,3\},
$$

y

$$
C=\{1,4\}.
$$

Tenemos

$$
P(A)=P(B)=P(C)=\frac{2}{4}=\frac12.
$$

Ahora calculamos las intersecciones por pares.

Primero,

$$
A\cap B=\{1\},
$$

por lo que

$$
P(A\cap B)=\frac14.
$$

Por otro lado,

$$
P(A)P(B)
=
\frac12\frac12
=
\frac14.
$$

Por lo tanto,

$$
P(A\cap B)=P(A)P(B).
$$

De la misma manera,

$$
B\cap C=\{1\},
$$

de modo que

$$
P(B\cap C)
=
\frac14
=
\frac12\frac12
=
P(B)P(C).
$$

También,

$$
A\cap C=\{1\},
$$

por lo que

$$
P(A\cap C)
=
\frac14
=
\frac12\frac12
=
P(A)P(C).
$$

Así, $A$, $B$ y $C$ son **independientes a pares**.

Sin embargo,

$$
A\cap B\cap C=\{1\},
$$

de modo que

$$
P(A\cap B\cap C)=\frac14.
$$

Mientras que

$$
P(A)P(B)P(C)
=
\frac12\frac12\frac12
=
\frac18.
$$

Como

$$
\frac14\neq\frac18,
$$

tenemos

$$
P(A\cap B\cap C)
\neq
P(A)P(B)P(C).
$$

Por lo tanto, $A$, $B$ y $C$ son independientes a pares, pero **no son mutuamente independientes**.

# Ejemplo: juego con un dado y una moneda

Una amiga y yo jugamos el siguiente juego:

1. Yo tiro un dado.
2. Si sale $6$, gano inmediatamente.
3. Si no sale $6$, ella lanza una moneda.
4. Si sale águila, ella gana.
5. Si nadie gana, se repite el procedimiento hasta que alguien gane.

Queremos calcular

$$
P(\text{yo gano}).
$$

Definamos

$$
A=\{\text{yo gano el juego}\}.
$$

---

## Forma 1: sumar sobre la ronda en la que gano

Para que nadie gane durante una ronda deben ocurrir dos cosas:

- que yo no saque $6$, cuya probabilidad es

$$
\frac56,
$$

- y que ella no obtenga águila, cuya probabilidad es

$$
\frac12.
$$

Como el lanzamiento del dado y el de la moneda son independientes,

$$
P(\text{nadie gana en una ronda})
=
\frac56\frac12
=
\frac{5}{12}.
$$

### Ganar en la primera ronda

Para ganar en la primera ronda solamente necesito sacar un $6$:

$$
P(\text{gano en la ronda 1})=\frac16.
$$

### Ganar en la segunda ronda

Para ganar en la segunda ronda:

1. nadie debe ganar en la primera ronda, con probabilidad $\frac{5}{12}$;
2. después debo sacar un $6$, con probabilidad $\frac16$.

Por lo tanto,

$$
P(\text{gano en la ronda 2})
=
\frac{5}{12}\frac16.
$$

### Ganar en la tercera ronda

Necesitamos que nadie gane durante las primeras dos rondas y después obtener un $6$:

$$
P(\text{gano en la ronda 3})
=
\left(\frac{5}{12}\right)^2\frac16.
$$

En general,

$$
P(\text{gano en la ronda }i)
=
\left(\frac{5}{12}\right)^{i-1}\frac16.
$$

Por lo tanto,

$$
P(A)
=
\sum_{i=1}^{\infty}
\left(\frac{5}{12}\right)^{i-1}\frac16.
$$

Sacando $\frac16$ como factor común,

$$
P(A)
=
\frac16
\sum_{j=0}^{\infty}
\left(\frac{5}{12}\right)^j.
$$

Utilizamos la serie geométrica

$$
\sum_{j=0}^{\infty}r^j
=
\frac{1}{1-r},
\qquad |r|<1.
$$

Con $r=\frac{5}{12}$,

$$
P(A)
=
\frac16
\frac{1}{1-\frac{5}{12}}.
$$

Como

$$
1-\frac{5}{12}
=
\frac{7}{12},
$$

tenemos

$$
P(A)
=
\frac16\frac{12}{7}
=
\frac27.
$$

Así,

$$
\boxed{P(\text{yo gano})=\frac27}.
$$

---

## Forma 2: condicionar en la ronda en la que termina el juego

Definimos

$$
F_i=\{\text{el juego termina en la ronda }i\}.
$$

Primero consideremos $F_1$.

El juego puede terminar en la primera ronda de dos maneras:

- yo saco un $6$, cuya probabilidad es

$$
\frac16;
$$

- yo no saco $6$ y ella obtiene águila, cuya probabilidad es

$$
\frac56\frac12
=
\frac{5}{12}.
$$

Por lo tanto,

$$
P(F_1)
=
\frac16+\frac{5}{12}
=
\frac{2}{12}+\frac{5}{12}
=
\frac{7}{12}.
$$

Ahora,

$$
P(A\cap F_1)=\frac16,
$$

porque para que yo gane y el juego termine en la primera ronda debo sacar $6$.

Entonces

$$
P(A\mid F_1)
=
\frac{P(A\cap F_1)}{P(F_1)}
=
\frac{\frac16}{\frac{7}{12}}.
$$

Por lo tanto,

$$
P(A\mid F_1)
=
\frac16\frac{12}{7}
=
\frac27.
$$

Veamos ahora una ronda cualquiera $i$.

Para que el juego termine en la ronda $i$, nadie debe ganar durante las primeras $i-1$ rondas.

La probabilidad de esto es

$$
\left(\frac{5}{12}\right)^{i-1}.
$$

Después, en la ronda $i$, alguien debe ganar.

Por tanto,

$$
P(F_i)
=
\left(\frac{5}{12}\right)^{i-1}
\left(
\frac16+\frac{5}{12}
\right).
$$

Además, para que yo gane precisamente en la ronda $i$,

$$
P(A\cap F_i)
=
\left(\frac{5}{12}\right)^{i-1}\frac16.
$$

Entonces

$$
P(A\mid F_i)
=
\frac{
\left(\frac{5}{12}\right)^{i-1}\frac16
}{
\left(\frac{5}{12}\right)^{i-1}
\left(
\frac16+\frac{5}{12}
\right)
}.
$$

Cancelando el factor común,

$$
P(A\mid F_i)
=
\frac{\frac16}{
\frac16+\frac{5}{12}
}.
$$

Así,

$$
P(A\mid F_i)=\frac27
$$

para todo $i\in\mathbb{N}$.

Es decir, conocer la ronda en la que terminó el juego no cambia la probabilidad de que yo haya sido el ganador.

Además,

$$
P(\text{el juego continúa después de }n\text{ rondas})
=
\left(\frac{5}{12}\right)^n.
$$

Como

$$
\lim_{n\to\infty}
\left(\frac{5}{12}\right)^n
=
0,
$$

el juego termina con probabilidad $1$.

Por la fórmula de probabilidad total,

$$
P(A)
=
\sum_{i=1}^{\infty}
P(A\mid F_i)P(F_i).
$$

Como

$$
P(A\mid F_i)=\frac27
$$

para todo $i$,

$$
P(A)
=
\frac27
\sum_{i=1}^{\infty}P(F_i).
$$

Y como el juego termina con probabilidad $1$,

$$
\sum_{i=1}^{\infty}P(F_i)=1.
$$

Por lo tanto,

$$
\boxed{P(A)=\frac27}.
$$

# Ejemplo: problema de Monty Hall

Vamos a un concurso de televisión.

El presentador nos ofrece tres puertas.

- Detrás de dos de ellas hay una cabra.
- Detrás de la tercera hay un coche.

Elegimos una puerta.

Después, el presentador abre una de las otras puertas mostrando una cabra y nos ofrece la posibilidad de cambiar nuestra elección.

La pregunta es:

**¿Conviene cambiar de puerta?**

---

## Caso 1: el presentador sabe dónde está el coche

Supongamos que el presentador:

1. sabe dónde está el coche;
2. siempre abre una puerta con una cabra;
3. nunca abre la puerta que elegimos;
4. después siempre nos ofrece cambiar.

Definamos los eventos

$$
A=\{\text{elegí inicialmente la puerta que tiene el coche}\},
$$

y

$$
B=\{\text{el presentador abre una puerta con una cabra}\}.
$$

Como inicialmente las tres puertas son igualmente probables,

$$
P(A)=\frac13.
$$

Por lo tanto,

$$
P(A^c)=\frac23.
$$

Si inicialmente escogimos el coche, el presentador puede abrir alguna de las otras dos puertas y necesariamente mostrará una cabra.

Por lo tanto,

$$
P(B\mid A)=1.
$$

Si inicialmente escogimos una cabra, entre las otras dos puertas hay una cabra y un coche.

Como el presentador sabe dónde está el coche, necesariamente abre la puerta que contiene la otra cabra.

Así,

$$
P(B\mid A^c)=1.
$$

Por el teorema de Bayes,

$$
P(A\mid B)
=
\frac{P(B\mid A)P(A)}
{
P(B\mid A)P(A)
+
P(B\mid A^c)P(A^c)
}.
$$

Sustituyendo,

$$
P(A\mid B)
=
\frac{
1\left(\frac13\right)
}{
1\left(\frac13\right)
+
1\left(\frac23\right)
}.
$$

Entonces

$$
P(A\mid B)=\frac13.
$$

Esto significa que nuestra puerta inicial sigue teniendo probabilidad

$$
\frac13
$$

de contener el coche.

Como solamente quedan dos puertas cerradas y una de ellas es nuestra elección inicial, la otra puerta cerrada tiene probabilidad

$$
1-\frac13=\frac23.
$$

Por lo tanto,

$$
\boxed{
P(\text{ganar si permanecemos})=\frac13
}
$$

y

$$
\boxed{
P(\text{ganar si cambiamos})=\frac23
}.
$$

Así, bajo estas reglas, cambiar de puerta duplica la probabilidad de ganar.

---

# Caso 2: el presentador no sabe dónde está el coche

Ahora cambiemos las reglas.

Supongamos que el presentador no sabe dónde está el coche y simplemente elige al azar una de las dos puertas que nosotros no escogimos.

Puede ocurrir que accidentalmente abra la puerta que contiene el coche.

Supongamos que observamos que abrió una puerta y apareció una cabra.

Definamos nuevamente

$$
A=\{\text{nuestra elección inicial contiene el coche}\},
$$

y

$$
B=\{\text{el presentador abre una puerta con una cabra}\}.
$$

Tenemos

$$
P(A)=\frac13
$$

y

$$
P(A^c)=\frac23.
$$

Si inicialmente elegimos el coche, las otras dos puertas tienen cabras.

Por lo tanto, sin importar cuál de ellas abra el presentador,

$$
P(B\mid A)=1.
$$

En cambio, si inicialmente escogimos una cabra, entre las otras dos puertas hay

- una cabra,
- un coche.

Como el presentador elige una de esas dos puertas al azar,

$$
P(B\mid A^c)=\frac12.
$$

Aplicamos Bayes:

$$
P(A\mid B)
=
\frac{P(B\mid A)P(A)}
{
P(B\mid A)P(A)
+
P(B\mid A^c)P(A^c)
}.
$$

Sustituyendo,

$$
P(A\mid B)
=
\frac{
1\left(\frac13\right)
}{
1\left(\frac13\right)
+
\frac12\left(\frac23\right)
}.
$$

En el denominador,

$$
\frac13+\frac12\frac23
=
\frac13+\frac13
=
\frac23.
$$

Por lo tanto,

$$
P(A\mid B)
=
\frac{\frac13}{\frac23}
=
\frac12.
$$

Así, después de observar que el presentador abrió una cabra,

$$
\boxed{
P(\text{el coche está en nuestra puerta})=\frac12
}.
$$

La otra puerta que permanece cerrada también tiene probabilidad

$$
\boxed{\frac12}.
$$

Por lo tanto, bajo estas nuevas reglas,

$$
P(\text{ganar si permanecemos})
=
P(\text{ganar si cambiamos})
=
\frac12.
$$

En este caso, cambiar o no cambiar da la misma probabilidad de ganar.

---

## Conclusión del problema de Monty Hall

La estrategia depende de cómo actúa el presentador.

### El presentador sabe dónde está el coche

Si siempre evita abrir la puerta del coche,

$$
P(\text{ganar permaneciendo})=\frac13,
$$

mientras que

$$
P(\text{ganar cambiando})=\frac23.
$$

Por lo tanto, conviene cambiar.

### El presentador abre una puerta al azar

Si el presentador no sabe dónde está el coche y abre al azar una de las otras dos puertas, condicionando a que observamos una cabra obtenemos

$$
P(\text{ganar permaneciendo})=\frac12,
$$

y

$$
P(\text{ganar cambiando})=\frac12.
$$

En este caso, ambas opciones tienen la misma probabilidad.
