
# Esperanza matemática y sus propiedades

## 1. Esperanza de una función de una variable aleatoria

Recordemos que si $X$ es una variable aleatoria y $g$ es una función medible, entonces

$$
Y=g(X)
$$

también es una variable aleatoria.

Si $X$ es una variable aleatoria discreta, por ejemplo, con recorrido

$$
R_X=\{-2,-1,0,1,2\},
$$

y definimos

$$
Y=X^2,
$$

entonces

$$
R_Y=\{0,1,4\}.
$$

Observemos que podemos agrupar los elementos de $R_X$ de acuerdo con los valores que toma $g(X)$.

En particular,

$$
\begin{aligned}
\sum_{y\in R_Y}
\sum_{\substack{k\in R_X\\g(k)=y}}k
&=0+(1)+(-1)+(2)+(-2)\\
&=\sum_{k\in R_X}k.
\end{aligned}
$$

Esto nos permite reorganizar las sumas sobre el recorrido de $X$ en términos del recorrido de $Y$.

### Teorema. Esperanza de una función de una variable aleatoria

Sea $X$ una variable aleatoria discreta y sea $Y=g(X)$, donde $g$ es una función medible.

Si $g(X)$ es integrable, entonces

$$
\boxed{
\mathbb{E}[Y]
=
\mathbb{E}[g(X)]
=
\sum_{k\in R_X}g(k)P_X(k)
}
$$

donde

$$
P_X(k)=P(X=k).
$$

### Demostración

Por definición de esperanza matemática,

$$
\mathbb{E}[Y]
=
\sum_{y\in R_Y}yP_Y(y).
$$

Como

$$
P_Y(y)=P(Y=y),
$$

tenemos

$$
\mathbb{E}[Y]
=
\sum_{y\in R_Y}yP(Y=y).
$$

Ahora, como $Y=g(X)$, el evento $\{Y=y\}$ corresponde a todos los valores $k\in R_X$ tales que $g(k)=y$.

Por lo tanto,

$$
P(Y=y)
=
\sum_{\substack{k\in R_X\\g(k)=y}}P(X=k).
$$

Sustituyendo en la expresión de la esperanza,

$$
\begin{aligned}
\mathbb{E}[Y]
&=
\sum_{y\in R_Y}
y\left(
\sum_{\substack{k\in R_X\\g(k)=y}}
P(X=k)
\right)\\
&=
\sum_{y\in R_Y}
\sum_{\substack{k\in R_X\\g(k)=y}}
yP(X=k).
\end{aligned}
$$

Pero, para los valores $k$ que aparecen en la suma interior, se cumple que $g(k)=y$.

Entonces,

$$
\begin{aligned}
\mathbb{E}[Y]
&=
\sum_{y\in R_Y}
\sum_{\substack{k\in R_X\\g(k)=y}}
g(k)P(X=k)\\
&=
\sum_{k\in R_X}g(k)P(X=k).
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[g(X)]
=
\sum_{k\in R_X}g(k)P_X(k)
}
$$

$\square$


## 2. Linealidad de la esperanza matemática

Ahora veremos que la esperanza matemática es lineal.

### Propiedad. Linealidad de la esperanza

Sean $X$ y $Y$ dos variables aleatorias integrables y sean $a,b\in\mathbb{R}$.

Entonces,

$$
\boxed{
\mathbb{E}[aX+bY]
=
a\mathbb{E}[X]+b\mathbb{E}[Y]
}
$$

### Demostración

Supongamos inicialmente que $X$ y $Y$ son variables aleatorias discretas.

Por definición,

$$
\mathbb{E}[aX+bY]
=
\sum_{x\in R_X}
\sum_{y\in R_Y}
(ax+by)P(X=x,Y=y).
$$

Distribuyendo los términos de la suma,

$$
\begin{aligned}
\mathbb{E}[aX+bY]
&=
a\sum_{x\in R_X}
\sum_{y\in R_Y}
xP(X=x,Y=y)\\
&\quad+
b\sum_{x\in R_X}
\sum_{y\in R_Y}
yP(X=x,Y=y).
\end{aligned}
$$

Recordemos que, cuando las probabilidades condicionantes son positivas,

$$
P(X=x,Y=y)
=
P(X=x\mid Y=y)P(Y=y),
$$

y también

$$
P(X=x,Y=y)
=
P(Y=y\mid X=x)P(X=x).
$$

Por lo tanto,

$$
\begin{aligned}
\mathbb{E}[aX+bY]
&=
a\sum_{x\in R_X}
\sum_{y\in R_Y}
xP(X=x\mid Y=y)P(Y=y)\\
&\quad+
b\sum_{x\in R_X}
\sum_{y\in R_Y}
yP(Y=y\mid X=x)P(X=x).
\end{aligned}
$$

Reorganizando las sumas,

$$
\begin{aligned}
\mathbb{E}[aX+bY]
&=
a\sum_{x\in R_X}
x\left(
\sum_{y\in R_Y}
P(X=x\mid Y=y)P(Y=y)
\right)\\
&\quad+
b\sum_{y\in R_Y}
y\left(
\sum_{x\in R_X}
P(Y=y\mid X=x)P(X=x)
\right).
\end{aligned}
$$

Por la ley de la probabilidad total,

$$
\sum_{y\in R_Y}
P(X=x\mid Y=y)P(Y=y)
=
P(X=x),
$$

y

$$
\sum_{x\in R_X}
P(Y=y\mid X=x)P(X=x)
=
P(Y=y).
$$

En estas sumas se omiten los términos cuya probabilidad condicionante es cero, ya que no contribuyen a la suma.

Así,

$$
\begin{aligned}
\mathbb{E}[aX+bY]
&=
a\sum_{x\in R_X}xP(X=x)
+
b\sum_{y\in R_Y}yP(Y=y)\\
&=
a\mathbb{E}[X]+b\mathbb{E}[Y].
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[aX+bY]
=
a\mathbb{E}[X]+b\mathbb{E}[Y]
}
$$

$\square$

### Observación

La linealidad de la esperanza no requiere que las variables aleatorias $X$ y $Y$ sean independientes.

Es suficiente con que sus esperanzas estén bien definidas y sean finitas.

## 3. Esperanza de una función de una variable aleatoria continua

Para variables aleatorias continuas también podemos calcular la esperanza de una función de la variable aleatoria.

Sea $X$ una variable aleatoria continua con función de densidad $f_X$.

Si $g$ es una función medible y $g(X)$ es integrable, entonces

$$
\boxed{
\mathbb{E}[g(X)]
=
\int_{-\infty}^{\infty}g(x)f_X(x)\,dx
}
$$

Esta expresión es el análogo continuo de la fórmula

$$
\mathbb{E}[g(X)]
=
\sum_{k\in R_X}g(k)P_X(k)
$$

para variables aleatorias discretas.

## 4. Monotonía de la esperanza matemática

Otra propiedad importante de la esperanza matemática es que es monótona.

### Propiedad. Monotonía de la esperanza

Sean $X$ y $Y$ dos variables aleatorias integrables.

Si

$$
P(X\geq Y)=1,
$$

entonces

$$
\boxed{
\mathbb{E}[X]\geq\mathbb{E}[Y]
}
$$

### Demostración

Definamos la variable aleatoria

$$
V=X-Y.
$$

Por hipótesis,

$$
P(X\geq Y)=1.
$$

Como

$$
\{X\geq Y\}
=
\{X-Y\geq 0\}
=
\{V\geq 0\},
$$

se tiene que

$$
P(V\geq 0)=1.
$$

Esto significa que $V$ es no negativa casi seguramente.

En efecto, si

$$
P(V<0)>0,
$$

entonces

$$
P(V\geq 0)
=
1-P(V<0)<1,
$$

lo cual contradice nuestra hipótesis.

Por lo tanto,

$$
P(V<0)=0.
$$

Es decir,

$$
V\geq 0
\quad \text{casi seguramente}.
$$

Como $V$ es una variable aleatoria no negativa casi seguramente, su esperanza también es no negativa.

En particular, si $V$ es discreta,

$$
\mathbb{E}[V]
=
\sum_{v\in R_V}vP(V=v)\geq 0.
$$

La desigualdad se cumple porque los términos con $v<0$ tienen probabilidad cero y los restantes son no negativos.

Para el caso continuo se obtiene la misma conclusión mediante la integral correspondiente.

Por lo tanto,

$$
\mathbb{E}[V]\geq 0.
$$

Ahora, por la linealidad de la esperanza,

$$
\begin{aligned}
\mathbb{E}[V]
&=
\mathbb{E}[X-Y]\\
&=
\mathbb{E}[X]-\mathbb{E}[Y].
\end{aligned}
$$

Como $\mathbb{E}[V]\geq 0$, tenemos

$$
\mathbb{E}[X]-\mathbb{E}[Y]\geq 0.
$$

De donde concluimos que

$$
\boxed{
\mathbb{E}[X]\geq\mathbb{E}[Y]
}
$$

$\square$

## 5. Variable aleatoria no negativa con esperanza cero

### Teorema

Sea $X$ una variable aleatoria tal que

$$
P(X\geq 0)=1
$$

y

$$
\mathbb{E}[X]=0.
$$

Entonces,

$$
\boxed{
P(X=0)=1
}
$$

Es decir, una variable aleatoria no negativa casi seguramente cuya esperanza es cero debe ser igual a cero casi seguramente.

### Demostración

Para cada $n\in\mathbb{N}$, definamos la variable aleatoria

$$
Y_n
=
\frac{1}{n}\mathbb{1}_{\{X>1/n\}}.
$$

Recordemos que la función indicadora se define mediante

$$
\mathbb{1}_{\{X>1/n\}}
=
\begin{cases}
1, & \text{si }X>\frac{1}{n},\\
0, & \text{si }X\leq\frac{1}{n}.
\end{cases}
$$

Por lo tanto,

$$
Y_n
=
\begin{cases}
\frac{1}{n}, & \text{si }X>\frac{1}{n},\\
0, & \text{si }X\leq\frac{1}{n}.
\end{cases}
$$

Observemos que

$$
P(X\geq Y_n)
=
P\left(
\{X\geq Y_n\}\cap
\left\{X>\frac{1}{n}\right\}
\right)
+
P\left(
\{X\geq Y_n\}\cap
\left\{X\leq\frac{1}{n}\right\}
\right).
$$

En el primer evento, cuando $X>1/n$, tenemos que

$$
Y_n=\frac{1}{n}.
$$

Por lo tanto,

$$
P\left(
\{X\geq Y_n\}\cap
\left\{X>\frac{1}{n}\right\}
\right)
=
P\left(X>\frac{1}{n}\right).
$$

Por otra parte, cuando $X\leq 1/n$, tenemos que $Y_n=0$.

Así,

$$
P\left(
\{X\geq Y_n\}\cap
\left\{X\leq\frac{1}{n}\right\}
\right)
=
P\left(0\leq X\leq\frac{1}{n}\right).
$$

De esta manera,

$$
\begin{aligned}
P(X\geq Y_n)
&=
P\left(X>\frac{1}{n}\right)
+
P\left(0\leq X\leq\frac{1}{n}\right)\\
&=
P(X\geq 0)\\
&=1.
\end{aligned}
$$

Por la monotonía de la esperanza,

$$
\mathbb{E}[X]\geq\mathbb{E}[Y_n].
$$

Como por hipótesis $\mathbb{E}[X]=0$, obtenemos

$$
0\geq\mathbb{E}[Y_n].
$$

Por otra parte, como $Y_n\geq 0$, se tiene que

$$
\mathbb{E}[Y_n]\geq 0.
$$

Por lo tanto,

$$
0\geq\mathbb{E}[Y_n]\geq 0.
$$

De donde

$$
\boxed{
\mathbb{E}[Y_n]=0
}
$$

Ahora calculemos la esperanza de $Y_n$.

Como $Y_n$ únicamente toma los valores $0$ y $1/n$,

$$
\begin{aligned}
\mathbb{E}[Y_n]
&=
\sum_{y\in R_{Y_n}}yP(Y_n=y)\\
&=
0\cdot P\left(X\leq\frac{1}{n}\right)
+
\frac{1}{n}P\left(X>\frac{1}{n}\right)\\
&=
\frac{1}{n}P\left(X>\frac{1}{n}\right).
\end{aligned}
$$

Como $\mathbb{E}[Y_n]=0$, tenemos

$$
0=
\frac{1}{n}P\left(X>\frac{1}{n}\right).
$$

Puesto que $1/n>0$, concluimos que

$$
\boxed{
P\left(X>\frac{1}{n}\right)=0
}
$$

para todo $n\in\mathbb{N}$.

Observemos que los eventos

$$
A_n=\left\{X>\frac{1}{n}\right\}
$$

forman una sucesión creciente, es decir,

$$
A_n\subseteq A_{n+1}.
$$

Además,

$$
\bigcup_{n=1}^{\infty}A_n
=
\{X>0\}.
$$

Por la continuidad de la probabilidad para sucesiones crecientes de eventos,

$$
P(X>0)
=
\lim_{n\to\infty}
P\left(X>\frac{1}{n}\right).
$$

Pero sabemos que

$$
P\left(X>\frac{1}{n}\right)=0
$$

para todo $n\in\mathbb{N}$.

Por lo tanto,

$$
\boxed{
P(X>0)=0
}
$$

Finalmente, por hipótesis,

$$
P(X\geq 0)=1.
$$

Como los eventos $\{X>0\}$ y $\{X=0\}$ son ajenos y su unión es $\{X\geq 0\}$,

$$
\begin{aligned}
1
&=P(X\geq 0)\\
&=P\left(\{X>0\}\cup\{X=0\}\right)\\
&=P(X>0)+P(X=0)\\
&=0+P(X=0).
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=0)=1
}
$$

$\square$

## 6. Ejemplo: Esperanza de una variable aleatoria binomial

**Ejemplo.** Calcular $\mathbb{E}[X]$ si

$$
X\sim\operatorname{Binomial}(n,p).
$$

### Solución

Recordemos que una variable aleatoria binomial cuenta el número de éxitos en $n$ ensayos de Bernoulli independientes, cada uno con probabilidad de éxito $p$.

Por lo tanto, podemos escribir

$$
X=\sum_{i=1}^{n}X_i,
$$

donde

$$
X_i\sim\operatorname{Bernoulli}(p),
\qquad i=1,\ldots,n.
$$

Además, las variables aleatorias $X_1,\ldots,X_n$ son independientes.

Recordemos que la esperanza de una variable aleatoria Bernoulli es

$$
\begin{aligned}
\mathbb{E}[X_i]
&=0\cdot P(X_i=0)+1\cdot P(X_i=1)\\
&=0\cdot(1-p)+1\cdot p\\
&=p.
\end{aligned}
$$

Utilizando la linealidad de la esperanza matemática,

$$
\begin{aligned}
\mathbb{E}[X]
&=
\mathbb{E}\left[\sum_{i=1}^{n}X_i\right]\\
&=
\sum_{i=1}^{n}\mathbb{E}[X_i]\\
&=
\sum_{i=1}^{n}p\\
&=np.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]=np
}
$$

### Observación

La linealidad de la esperanza permite calcular la esperanza de una variable aleatoria binomial sin utilizar directamente su función de masa de probabilidad.

Además, aunque los ensayos de Bernoulli son independientes por la definición del modelo binomial, la independencia no es necesaria para aplicar la linealidad de la esperanza.
