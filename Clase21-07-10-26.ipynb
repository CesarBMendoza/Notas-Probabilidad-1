# Transformación integral de probabilidad

Recordemos que, si

$$
U\sim U[a,b],
$$

entonces su función de densidad está dada por

$$
f_U(u)=\frac{1}{b-a}\mathbb{1}_{(a,b)}(u).
$$

Es decir,

$$
f_U(u)=
\begin{cases}
\dfrac{1}{b-a}, & a<u<b,\\[6pt]
0, & \text{en otro caso}.
\end{cases}
$$

Su función de distribución acumulada es

$$
F_U(z)=P(U\leq z).
$$

Por lo tanto,

$$
\boxed{
F_U(z)=
\begin{cases}
0, & z<a,\\[6pt]
\dfrac{z-a}{b-a}, & a\leq z<b,\\[8pt]
1, & z\geq b.
\end{cases}
}
$$

En particular, si $a=0$ y $b=1$, tenemos que

$$
U\sim U[0,1].
$$

En este caso,

$$
\boxed{
F_U(z)=
\begin{cases}
0, & z<0,\\
z, & 0\leq z<1,\\
1, & z\geq 1.
\end{cases}
}
$$

---

## Teorema: Transformación integral de probabilidad

Sea $X$ una variable aleatoria continua con función de distribución acumulada $F_X$.

Entonces,

$$
\boxed{
Y=F_X(X)\sim U[0,1].
}
$$

Es decir, al evaluar una variable aleatoria continua en su propia función de distribución acumulada, obtenemos una variable aleatoria con distribución uniforme en $[0,1]$.

### Demostración

Sabemos que $F_X$ es una función continua y no decreciente.

En particular, $F_X$ es medible y, por lo tanto,

$$
Y=F_X(X)
$$

es una variable aleatoria.

Queremos determinar su función de distribución acumulada.

Por definición,

$$
F_Y(y)=P(Y\leq y).
$$

Como $Y=F_X(X)$, tenemos que

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(F_X(X)\leq y).
\end{aligned}
$$

Supongamos primero que $F_X$ es estrictamente creciente en el intervalo donde $X$ toma valores.

En este caso, $F_X$ tiene una función inversa.

Como $F_X$ es estrictamente creciente,

$$
F_X(X)\leq y
\iff
X\leq F_X^{-1}(y).
$$

Por lo tanto, para $0<y<1$,

$$
\begin{aligned}
F_Y(y)
&=P(F_X(X)\leq y)\\
&=P(X\leq F_X^{-1}(y))\\
&=F_X(F_X^{-1}(y))\\
&=y.
\end{aligned}
$$

Es decir,

$$
\boxed{
F_Y(y)=y,
\qquad 0<y<1.
}
$$

Además, sabemos que

$$
0\leq F_X(x)\leq 1
$$

para todo $x\in\mathbb{R}$.

Por lo tanto,

$$
F_Y(y)=0,
\qquad y<0,
$$

y

$$
F_Y(y)=1,
\qquad y\geq 1.
$$

En consecuencia,

$$
\boxed{
F_Y(y)=
\begin{cases}
0, & y<0,\\
y, & 0\leq y<1,\\
1, & y\geq 1.
\end{cases}
}
$$

Esta es precisamente la función de distribución acumulada de una variable aleatoria uniforme en $[0,1]$.

Por lo tanto,

$$
\boxed{
Y=F_X(X)\sim U[0,1].
}
$$

---

### Observación: ¿Es necesario que $F_X$ sea estrictamente creciente?

No. El teorema sigue siendo válido cuando $F_X$ es continua, aunque tenga intervalos donde sea constante.

La demostración anterior utiliza una inversa ordinaria, por lo que requiere estricta monotonía.

Para justificar el resultado general, consideremos $0<y<1$ y definamos

$$
a_y=\sup\{x\in\mathbb{R}:F_X(x)\leq y\}.
$$

Como $F_X$ es continua y no decreciente,

$$
F_X(a_y)=y.
$$

Además,

$$
\{F_X(X)\leq y\}
=
\{X\leq a_y\}.
$$

Por lo tanto,

$$
\begin{aligned}
F_Y(y)
&=P(F_X(X)\leq y)\\
&=P(X\leq a_y)\\
&=F_X(a_y)\\
&=y.
\end{aligned}
$$

Así, el resultado es válido para cualquier función de distribución acumulada continua.

En conclusión,

$$
\boxed{
F_X\text{ continua}
\implies
F_X(X)\sim U[0,1].
}
$$

# Método de la transformada inversa

Del teorema anterior sabemos que, si $X$ es una variable aleatoria con función de distribución acumulada continua $F_X$, entonces

$$
\boxed{
Y=F_X(X)\sim U[0,1].
}
$$

Esto nos dice que podemos transformar una variable aleatoria continua en una variable aleatoria uniforme en $[0,1]$.

Ahora queremos responder la siguiente pregunta:

**Si tenemos una variable aleatoria uniforme en $[0,1]$, ¿podemos transformarla en una variable aleatoria con una distribución determinada?**

La respuesta es afirmativa.

---

## Teorema: Método de la transformada inversa

Sea

$$
U\sim U[0,1].
$$

Supongamos que $F_X$ es una función de distribución acumulada continua y estrictamente creciente en el intervalo correspondiente, de manera que existe su inversa.

Definamos

$$
Z=F_X^{-1}(U).
$$

Entonces,

$$
\boxed{
F_Z(z)=F_X(z).
}
$$

Es decir, $Z$ tiene la misma distribución que $X$.

### Demostración

Queremos determinar la función de distribución acumulada de $Z$.

Por definición,

$$
F_Z(z)=P(Z\leq z).
$$

Como

$$
Z=F_X^{-1}(U),
$$

tenemos que

$$
\begin{aligned}
F_Z(z)
&=P(Z\leq z)\\
&=P(F_X^{-1}(U)\leq z).
\end{aligned}
$$

Como $F_X$ es estrictamente creciente, su inversa también es estrictamente creciente.

Por lo tanto,

$$
F_X^{-1}(U)\leq z
\iff
U\leq F_X(z).
$$

Así,

$$
\begin{aligned}
F_Z(z)
&=P(F_X^{-1}(U)\leq z)\\
&=P(U\leq F_X(z))\\
&=F_U(F_X(z)).
\end{aligned}
$$

Recordemos que

$$
U\sim U[0,1].
$$

Por lo tanto,

$$
F_U(u)=u,
\qquad 0\leq u\leq 1.
$$

Además, como $F_X$ es una función de distribución acumulada,

$$
0\leq F_X(z)\leq 1.
$$

En consecuencia,

$$
\begin{aligned}
F_Z(z)
&=F_U(F_X(z))\\
&=F_X(z).
\end{aligned}
$$

Es decir,

$$
\boxed{
F_Z(z)=F_X(z).
}
$$

Por lo tanto,

$$
\boxed{
Z\overset{d}{=}X.
}
$$

La notación $Z\overset{d}{=}X$ significa que ambas variables aleatorias tienen la misma distribución.

No necesariamente significa que sean iguales para cada resultado del espacio muestral.

---

## Interpretación gráfica

Podemos interpretar este procedimiento mediante las gráficas de $F_X$ y $f_X$.

### Caso continuo

Si $X$ es una variable aleatoria continua, su función de distribución acumulada $F_X$ es continua y no decreciente.

Además,

$$
\lim_{x\to-\infty}F_X(x)=0
$$

y

$$
\lim_{x\to\infty}F_X(x)=1.
$$

Si $F_X$ es estrictamente creciente, podemos utilizar su inversa para transformar valores uniformes en valores con la distribución deseada.

El procedimiento consiste en:

1. Generar un valor $u\in(0,1)$ de una variable uniforme.
2. Encontrar el valor $x$ que satisface

   $$
   F_X(x)=u.
   $$

3. Tomar

   $$
   x=F_X^{-1}(u).
   $$

Al repetir este procedimiento, obtenemos valores con la distribución $F_X$.

---

## ¿Podemos utilizar este método para variables aleatorias discretas?

Sí.

Sin embargo, cuando $X$ es una variable aleatoria discreta, su función de distribución acumulada generalmente tiene saltos.

Por lo tanto, $F_X$ no es invertible en el sentido usual.

Para resolver este problema utilizamos una función llamada **inversa generalizada**.

Esta función permite aplicar el método de la transformada inversa a distribuciones discretas, continuas y mixtas.

# Inversa generalizada de una función de distribución acumulada

## Definición

Sea $X$ una variable aleatoria con función de distribución acumulada $F_X$.

Definimos su **inversa generalizada** mediante

$$
\boxed{
F_X^{-1}(u)
=
\inf\{y\in\mathbb{R}:F_X(y)\geq u\},
\qquad 0<u<1.
}
$$

Como toda función de distribución acumulada es continua por la derecha, también podemos escribir

$$
\boxed{
F_X^{-1}(u)
=
\min\{y\in\mathbb{R}:F_X(y)\geq u\}.
}
$$

Esta definición nos permite encontrar el menor valor $y$ para el cual la probabilidad acumulada es mayor o igual que $u$.

A diferencia de la inversa ordinaria, la inversa generalizada puede utilizarse aunque $F_X$ tenga saltos o intervalos donde sea constante.

---

# Ejemplo: Variable aleatoria Bernoulli

Supongamos que

$$
X\sim\operatorname{Bernoulli}(p),
$$

donde

$$
0<p<1.
$$

Sabemos que

$$
R_X=\{0,1\}.
$$

La función de masa de $X$ está dada por

$$
\boxed{
p_X(k)=
\begin{cases}
1-p, & k=0,\\
p, & k=1,\\
0, & \text{en otro caso}.
\end{cases}
}
$$

Queremos encontrar su función de distribución acumulada y su inversa generalizada.

---

## Función de distribución acumulada

Por definición,

$$
F_X(x)=P(X\leq x).
$$

Consideremos tres casos.

### Caso 1: $x<0$

Como $X$ solamente toma los valores $0$ y $1$, tenemos que

$$
P(X\leq x)=0.
$$

Por lo tanto,

$$
F_X(x)=0,
\qquad x<0.
$$

### Caso 2: $0\leq x<1$

En este caso, el único valor posible de $X$ que satisface $X\leq x$ es $X=0$.

Por lo tanto,

$$
\begin{aligned}
F_X(x)
&=P(X\leq x)\\
&=P(X=0)\\
&=1-p.
\end{aligned}
$$

Así,

$$
F_X(x)=1-p,
\qquad 0\leq x<1.
$$

### Caso 3: $x\geq 1$

En este caso,

$$
P(X\leq x)=1.
$$

Por lo tanto,

$$
F_X(x)=1,
\qquad x\geq 1.
$$

En consecuencia,

$$
\boxed{
F_X(x)=
\begin{cases}
0, & x<0,\\
1-p, & 0\leq x<1,\\
1, & x\geq 1.
\end{cases}
}
$$

---

## Cálculo de la inversa generalizada

Recordemos que

$$
F_X^{-1}(u)
=
\min\{y\in\mathbb{R}:F_X(y)\geq u\}.
$$

Consideraremos dos casos.

### Caso 1: $0<u\leq 1-p$

Queremos encontrar el conjunto

$$
\{y\in\mathbb{R}:F_X(y)\geq u\}.
$$

Como

$$
0<u\leq 1-p,
$$

observamos que

$$
F_X(y)\geq u
$$

si y solo si

$$
y\geq 0.
$$

Por lo tanto,

$$
\{y\in\mathbb{R}:F_X(y)\geq u\}
=
[0,\infty).
$$

En consecuencia,

$$
\begin{aligned}
F_X^{-1}(u)
&=\min[0,\infty)\\
&=0.
\end{aligned}
$$

Es decir,

$$
\boxed{
F_X^{-1}(u)=0,
\qquad 0<u\leq 1-p.
}
$$

---

### Caso 2: $1-p<u<1$

Nuevamente, queremos determinar

$$
\{y\in\mathbb{R}:F_X(y)\geq u\}.
$$

Como

$$
u>1-p,
$$

los valores $y<1$ no satisfacen la condición.

En cambio, para $y\geq 1$,

$$
F_X(y)=1\geq u.
$$

Por lo tanto,

$$
\{y\in\mathbb{R}:F_X(y)\geq u\}
=
[1,\infty).
$$

En consecuencia,

$$
\begin{aligned}
F_X^{-1}(u)
&=\min[1,\infty)\\
&=1.
\end{aligned}
$$

Es decir,

$$
\boxed{
F_X^{-1}(u)=1,
\qquad 1-p<u<1.
}
$$

---

## Resultado

Reuniendo los dos casos, obtenemos

$$
\boxed{
F_X^{-1}(u)=
\begin{cases}
0, & 0<u\leq 1-p,\\
1, & 1-p<u<1.
\end{cases}
}
$$

También podemos escribir

$$
\boxed{
F_X^{-1}(u)
=
\mathbb{1}_{(1-p,1)}(u),
\qquad 0<u<1.
}
$$

Esta es la inversa generalizada de la función de distribución acumulada de una variable aleatoria Bernoulli.

Observemos que la inversa generalizada no es una función inversa ordinaria.

En particular, si $0<u\leq 1-p$, tenemos que

$$
F_X^{-1}(u)=0,
$$

mientras que, si $1-p<u<1$,

$$
F_X^{-1}(u)=1.
$$

# Aplicación de la inversa generalizada a una distribución Bernoulli

Continuando con el ejemplo anterior, supongamos que

$$
X\sim\operatorname{Bernoulli}(p).
$$

Encontramos que su inversa generalizada es

$$
F_X^{-1}(u)=
\begin{cases}
0, & 0<u\leq 1-p,\\
1, & 1-p<u<1.
\end{cases}
$$

Ahora consideremos una variable aleatoria

$$
U\sim U[0,1]
$$

y definamos

$$
\boxed{
Y=F_X^{-1}(U).
}
$$

Queremos demostrar que

$$
Y\overset{d}{=}X.
$$

Para ello, encontraremos la función de masa de $Y$.

Como la inversa generalizada solamente toma los valores $0$ y $1$, tenemos que

$$
R_Y=\{0,1\}.
$$

---

## Probabilidad de que $Y=0$

Por definición,

$$
p_Y(0)=P(Y=0).
$$

Como

$$
Y=F_X^{-1}(U),
$$

tenemos que

$$
\begin{aligned}
p_Y(0)
&=P(Y=0)\\
&=P(F_X^{-1}(U)=0).
\end{aligned}
$$

Por la definición de la inversa generalizada,

$$
F_X^{-1}(U)=0
\iff
0<U\leq 1-p,
$$

salvo por los valores extremos, que tienen probabilidad cero.

Por lo tanto,

$$
\begin{aligned}
p_Y(0)
&=P(0<U\leq 1-p).
\end{aligned}
$$

Como

$$
U\sim U[0,1],
$$

su función de densidad está dada por

$$
f_U(u)=\mathbb{1}_{(0,1)}(u).
$$

En consecuencia,

$$
\begin{aligned}
p_Y(0)
&=\int_0^{1-p}f_U(u)\,du\\
&=\int_0^{1-p}1\,du\\
&=[u]_0^{1-p}\\
&=1-p.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
p_Y(0)=1-p.
}
$$

---

## Probabilidad de que $Y=1$

Por definición,

$$
p_Y(1)=P(Y=1).
$$

Como $Y$ solamente toma los valores $0$ y $1$, tenemos que

$$
P(Y=0)+P(Y=1)=1.
$$

Por lo tanto,

$$
\begin{aligned}
p_Y(1)
&=P(Y=1)\\
&=1-P(Y=0)\\
&=1-(1-p)\\
&=p.
\end{aligned}
$$

Es decir,

$$
\boxed{
p_Y(1)=p.
}
$$

También podemos calcular esta probabilidad directamente.

Observemos que

$$
Y=1
\iff
1-p<U<1,
$$

salvo por el extremo $U=1$, que tiene probabilidad cero.

Por lo tanto,

$$
\begin{aligned}
p_Y(1)
&=P(1-p<U<1)\\
&=\int_{1-p}^{1}f_U(u)\,du\\
&=\int_{1-p}^{1}1\,du\\
&=[u]_{1-p}^{1}\\
&=1-(1-p)\\
&=p.
\end{aligned}
$$

Obtenemos el mismo resultado.

---

## Función de masa de $Y$

Por lo anterior,

$$
\boxed{
p_Y(k)=
\begin{cases}
1-p, & k=0,\\
p, & k=1,\\
0, & \text{en otro caso}.
\end{cases}
}
$$

Esta es precisamente la función de masa de una variable aleatoria Bernoulli con parámetro $p$.

Por lo tanto,

$$
\boxed{
Y\sim\operatorname{Bernoulli}(p).
}
$$

Como

$$
X\sim\operatorname{Bernoulli}(p),
$$

concluimos que

$$
\boxed{
Y\overset{d}{=}X.
}
$$

Es decir, hemos demostrado que la transformación

$$
Y=F_X^{-1}(U)
$$

produce una variable aleatoria con la misma distribución que $X$.

---

# Teorema general: Método de la transformada inversa

El procedimiento anterior no solamente funciona para una distribución Bernoulli.

Sea $F_X$ cualquier función de distribución acumulada y sea

$$
U\sim U[0,1].
$$

Definamos

$$
Y=F_X^{-1}(U),
$$

donde $F_X^{-1}$ es la inversa generalizada.

Entonces,

$$
\boxed{
F_Y(y)=F_X(y).
}
$$

### Demostración

Por definición,

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(F_X^{-1}(U)\leq y).
\end{aligned}
$$

Por las propiedades de la inversa generalizada,

$$
F_X^{-1}(u)\leq y
\iff
u\leq F_X(y),
\qquad 0<u<1.
$$

Por lo tanto,

$$
\begin{aligned}
F_Y(y)
&=P(F_X^{-1}(U)\leq y)\\
&=P(U\leq F_X(y))\\
&=F_U(F_X(y)).
\end{aligned}
$$

Como $U$ tiene distribución uniforme en $[0,1]$,

$$
F_U(u)=u,
\qquad 0\leq u\leq 1.
$$

En consecuencia,

$$
\begin{aligned}
F_Y(y)
&=F_U(F_X(y))\\
&=F_X(y).
\end{aligned}
$$

Así,

$$
\boxed{
Y\overset{d}{=}X.
}
$$

Este resultado permite generar variables aleatorias con una distribución específica utilizando una variable aleatoria uniforme en $[0,1]$.
