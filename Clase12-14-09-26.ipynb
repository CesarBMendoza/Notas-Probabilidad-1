# Variables aleatorias

Sea

$$
(\Omega,\mathcal{F},P)
$$

un espacio de probabilidad.

## Definición

Una **variable aleatoria** es una función

$$
X:\Omega\longrightarrow \mathbb{R}
$$

tal que

$$
X^{-1}(B)
=
\{\omega\in\Omega:X(\omega)\in B\}
\in\mathcal{F}
$$

para todo

$$
B\in\mathcal{B}(\mathbb{R}),
$$

donde $\mathcal{B}(\mathbb{R})$ denota la $\sigma$-álgebra de Borel de $\mathbb{R}$.

Es decir, para todo conjunto boreliano $B$, la preimagen

$$
X^{-1}(B)
$$

debe ser un evento de $\mathcal{F}$.

> **Nota.**
> La condición de que $X$ sea variable aleatoria puede verificarse equivalentemente comprobando que
>
> $$
> \{\omega\in\Omega:X(\omega)\leq x\}\in\mathcal{F}
> $$
>
> para todo $x\in\mathbb{R}$.

---

# Ejemplo: variable aleatoria constante

Consideremos la función constante

$$
X(\omega)=c,
\qquad c\in\mathbb{R},
$$

para todo $\omega\in\Omega$.

Queremos demostrar que $X$ es una variable aleatoria.

Sea

$$
B\in\mathcal{B}(\mathbb{R}).
$$

Queremos estudiar

$$
X^{-1}(B).
$$

Como

$$
X(\omega)=c
$$

para todo $\omega\in\Omega$, solamente tenemos dos posibilidades.

### Caso 1: $c\in B$

En este caso,

$$
X(\omega)=c\in B
$$

para todo $\omega\in\Omega$.

Por lo tanto,

$$
X^{-1}(B)=\Omega.
$$

Como

$$
\Omega\in\mathcal{F},
$$

la preimagen pertenece a $\mathcal{F}$.

### Caso 2: $c\notin B$

En este caso ningún $\omega\in\Omega$ satisface

$$
X(\omega)\in B.
$$

Por lo tanto,

$$
X^{-1}(B)=\varnothing.
$$

Y como

$$
\varnothing\in\mathcal{F},
$$

también en este caso la preimagen pertenece a $\mathcal{F}$.

Así,

$$
X^{-1}(B)
=
\begin{cases}
\Omega, & c\in B,\\
\varnothing, & c\notin B.
\end{cases}
$$

En ambos casos,

$$
X^{-1}(B)\in\mathcal{F}.
$$

Por lo tanto,

$$
\boxed{X(\omega)=c\text{ es una variable aleatoria}.}
$$

# Función indicadora

## Definición

Sea

$$
A\in\mathcal{F}.
$$

La **función indicadora** del conjunto $A$ se define como

$$
\mathbf{1}_A(\omega)
=
\begin{cases}
1, & \omega\in A,\\
0, & \omega\notin A.
\end{cases}
$$

También se utiliza la notación

$$
I_A(\omega).
$$

---

# Teorema

Toda función indicadora de un evento es una variable aleatoria.

Es decir, si

$$
A\in\mathcal{F},
$$

entonces

$$
\mathbf{1}_A:\Omega\longrightarrow\mathbb{R}
$$

es una variable aleatoria.

## Demostración

Sea

$$
B\in\mathcal{B}(\mathbb{R}).
$$

Como la función indicadora solamente puede tomar los valores

$$
0
\qquad\text{y}\qquad
1,
$$

la preimagen

$$
\mathbf{1}_A^{-1}(B)
$$

solamente puede ser uno de los siguientes cuatro conjuntos:

$$
\varnothing,\qquad
\Omega,\qquad
A,\qquad
A^c.
$$

Veamos por qué.

### Si $0\notin B$ y $1\notin B$

Ningún valor de la indicadora pertenece a $B$, por lo que

$$
\mathbf{1}_A^{-1}(B)=\varnothing.
$$

### Si $0\in B$ y $1\in B$

Ambos valores posibles pertenecen a $B$, por lo que

$$
\mathbf{1}_A^{-1}(B)=\Omega.
$$

### Si $1\in B$ y $0\notin B$

La indicadora pertenece a $B$ precisamente cuando

$$
\mathbf{1}_A(\omega)=1,
$$

es decir, cuando

$$
\omega\in A.
$$

Por lo tanto,

$$
\mathbf{1}_A^{-1}(B)=A.
$$

### Si $0\in B$ y $1\notin B$

La indicadora pertenece a $B$ precisamente cuando

$$
\mathbf{1}_A(\omega)=0,
$$

es decir, cuando

$$
\omega\notin A.
$$

Por lo tanto,

$$
\mathbf{1}_A^{-1}(B)=A^c.
$$

Como

$$
A\in\mathcal{F},
$$

sabemos por las propiedades de una $\sigma$-álgebra que

$$
A^c\in\mathcal{F},
$$

y además

$$
\Omega,\varnothing\in\mathcal{F}.
$$

Por lo tanto,

$$
\mathbf{1}_A^{-1}(B)\in\mathcal{F}
$$

para todo $B\in\mathcal{B}(\mathbb{R})$.

Así,

$$
\boxed{\mathbf{1}_A\text{ es una variable aleatoria}.}
$$

---

# Ejemplo

Consideremos

$$
\Omega=\{1,2,3,4\}
$$

y la $\sigma$-álgebra

$$
\mathcal{F}
=
\{
\varnothing,
\Omega,
\{1,2\},
\{3,4\}
\}.
$$

Definimos la función indicadora

$$
X=\mathbf{1}_{\{2,3\}},
$$

es decir,

$$
X(\omega)
=
\begin{cases}
1, & \omega\in\{2,3\},\\
0, & \omega\notin\{2,3\}.
\end{cases}
$$

Queremos determinar si $X$ es una variable aleatoria.

Tomemos, por ejemplo, el conjunto boreliano

$$
B=\left(\frac12,\frac32\right).
$$

Como este intervalo contiene a $1$ pero no contiene a $0$,

$$
X^{-1}(B)
=
\{\omega:X(\omega)=1\}.
$$

Por la definición de $X$,

$$
X^{-1}(B)=\{2,3\}.
$$

Pero

$$
\{2,3\}\notin\mathcal{F}.
$$

Por lo tanto,

$$
X^{-1}(B)\notin\mathcal{F}.
$$

Así,

$$
\boxed{\mathbf{1}_{\{2,3\}}\text{ no es una variable aleatoria respecto de }\mathcal{F}.}
$$

> **Observación.**
> Esto no contradice el teorema anterior, porque
>
> $$
> \{2,3\}\notin\mathcal{F}.
> $$
>
> El teorema afirma que $\mathbf{1}_A$ es variable aleatoria cuando
>
> $$
> A\in\mathcal{F}.
> $$

# Lema: distribución inducida por una variable aleatoria

Sea

$$
(\Omega,\mathcal{F},P)
$$

un espacio de probabilidad y sea

$$
X:\Omega\longrightarrow\mathbb{R}
$$

una variable aleatoria.

Definimos una función

$$
P_X:\mathcal{B}(\mathbb{R})\longrightarrow[0,1]
$$

mediante

$$
P_X(B)
=
P\left(X^{-1}(B)\right),
\qquad
B\in\mathcal{B}(\mathbb{R}).
$$

Entonces

$$
\boxed{
(\mathbb{R},\mathcal{B}(\mathbb{R}),P_X)
}
$$

es un espacio de probabilidad.

La medida $P_X$ se llama **distribución**, **ley** o **medida inducida por $X$**.

También se escribe

$$
P_X(B)=P(X\in B).
$$

Esto tiene sentido porque

$$
\{X\in B\}
=
\{\omega\in\Omega:X(\omega)\in B\}
=
X^{-1}(B).
$$
