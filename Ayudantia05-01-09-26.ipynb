# Ejemplo 1. Dados con caras de colores

Tenemos dos dados justos. Cada uno tiene sus caras pintadas de la siguiente manera:

- 2 caras rojas;
- 2 caras negras;
- 1 cara amarilla;
- 1 cara blanca.

Cuando se lanzan ambos dados, ¿cuál es la probabilidad de que los dos muestren una cara del mismo color?

---

## Probabilidad de cada color

Sea $R$ el evento de obtener rojo, $N$ el evento de obtener negro, $A$ el evento de obtener amarillo y $B$ el evento de obtener blanco.

Como cada dado tiene 6 caras equiprobables,

$\mathbb{P}(R)=\frac{2}{6}=\frac{1}{3}$.

De la misma manera,

$\mathbb{P}(N)=\frac{2}{6}=\frac{1}{3}$.

Para amarillo,

$\mathbb{P}(A)=\frac{1}{6}$.

Y para blanco,

$\mathbb{P}(B)=\frac{1}{6}$.

---

## Probabilidad de obtener el mismo color

Para que ambos dados muestren el mismo color pueden ocurrir cuatro casos:

- ambos son rojos;
- ambos son negros;
- ambos son amarillos;
- ambos son blancos.

Sea

$E=\{\text{los dos dados muestran el mismo color}\}$.

Entonces,

$E=RR\cup NN\cup AA\cup BB$.

Estos cuatro eventos son disjuntos.

Como los lanzamientos de los dos dados son independientes,

$\mathbb{P}(RR)=\mathbb{P}(R)\mathbb{P}(R)=\frac{1}{3}\frac{1}{3}=\frac{1}{9}$.

De la misma manera,

$\mathbb{P}(NN)=\frac{1}{9}$.

Para amarillo,

$\mathbb{P}(AA)=\frac{1}{6}\frac{1}{6}=\frac{1}{36}$.

Y para blanco,

$\mathbb{P}(BB)=\frac{1}{6}\frac{1}{6}=\frac{1}{36}$.

Por lo tanto,

$\mathbb{P}(E)=\frac{1}{9}+\frac{1}{9}+\frac{1}{36}+\frac{1}{36}$.

Así,

$\mathbb{P}(E)=\frac{4+4+1+1}{36}=\frac{10}{36}=\frac{5}{18}$.

Por tanto,

$\boxed{\mathbb{P}(\text{mismo color})=\frac{5}{18}}$.

---

## Otra forma: por conteo

También podemos resolverlo contando directamente los resultados de las caras de los dos dados.

Cada dado tiene 6 caras, por lo que

$|\Omega|=6\cdot 6=36$.

Contemos los casos favorables.

Para que ambos sean rojos, tenemos 2 caras rojas en el primer dado y 2 caras rojas en el segundo.

Por lo tanto, existen

$2\cdot 2=4$

resultados favorables de color rojo.

Para negro, de la misma manera, hay

$2\cdot 2=4$

resultados favorables.

Para amarillo hay

$1\cdot 1=1$

resultado favorable.

Para blanco hay

$1\cdot 1=1$

resultado favorable.

Por lo tanto,

$|E|=4+4+1+1=10$.

Así,

$\mathbb{P}(E)=\frac{|E|}{|\Omega|}=\frac{10}{36}=\frac{5}{18}$.

Llegamos al mismo resultado:

$\boxed{\mathbb{P}(E)=\frac{5}{18}}$.

# Ejemplo 2. Lanzar un dado hasta obtener un 6

Consideremos el experimento de lanzar repetidamente un dado justo hasta que aparezca por primera vez un $6$.

Queremos responder:

1. ¿Cuál es el espacio muestral de este experimento?
2. Sea $E_n$ el evento en el que se necesitan exactamente $n$ lanzamientos para terminar el experimento. ¿Qué puntos del espacio muestral están contenidos en $E_n$?
3. ¿Cuál es $\mathbb{P}(E_n)$?
4. ¿Cuál es la probabilidad de que el experimento termine eventualmente?

---

## Espacio muestral

Una realización del experimento consiste en obtener varios números distintos de $6$ y finalmente obtener un $6$.

Por ejemplo,

$(6)$

representa que obtenemos $6$ en el primer lanzamiento.

La sucesión

$(2,6)$

representa que obtenemos primero un $2$ y después un $6$.

La sucesión

$(1,3,5,6)$

representa que necesitamos cuatro lanzamientos para obtener por primera vez un $6$.

En general, podemos escribir los resultados como

$(x_1,\ldots,x_{n-1},6)$,

donde

$x_1,\ldots,x_{n-1}\in\{1,2,3,4,5\}$.

Por lo tanto, podemos escribir

$\Omega=\bigcup_{n=1}^{\infty}E_n$,

donde

$E_n=\{(x_1,\ldots,x_{n-1},6):x_1,\ldots,x_{n-1}\in\{1,2,3,4,5\}\}$.

---

## Algunos ejemplos de los eventos \(E_n\)

Si $n=1$, necesariamente obtenemos un $6$ en el primer lanzamiento:

$E_1=\{(6)\}$.

Si $n=2$, el primer lanzamiento puede ser cualquiera de $1,\ldots,5$ y el segundo debe ser $6$:

$E_2=\{(1,6),(2,6),(3,6),(4,6),(5,6)\}$.

Si $n=3$,

$E_3=\{(x_1,x_2,6):x_1,x_2\in\{1,\ldots,5\}\}$.

Por ejemplo,

$E_3=\{(1,1,6),(1,2,6),\ldots,(5,5,6)\}$.

---

## Número de elementos de \(E_n\)

Para pertenecer a $E_n$, los primeros $n-1$ lanzamientos pueden tomar cualquiera de los valores

$1,2,3,4,5$.

Cada uno tiene 5 posibilidades.

El último lanzamiento está determinado, pues debe ser un $6$.

Por lo tanto,

$|E_n|=5^{n-1}$.

---

## Probabilidad de \(E_n\)

Para necesitar exactamente $n$ lanzamientos debemos:

- no obtener un $6$ en los primeros $n-1$ lanzamientos;
- obtener un $6$ en el lanzamiento número $n$.

La probabilidad de no obtener un $6$ en un lanzamiento es

$\frac{5}{6}$.

Como los lanzamientos son independientes, la probabilidad de no obtener ningún $6$ durante los primeros $n-1$ lanzamientos es

$\left(\frac{5}{6}\right)^{n-1}$.

Finalmente, la probabilidad de obtener un $6$ es

$\frac{1}{6}$.

Por lo tanto,

$\boxed{\mathbb{P}(E_n)=\left(\frac{5}{6}\right)^{n-1}\frac{1}{6}}$.

Equivalentemente,

$\mathbb{P}(E_n)=\frac{5^{n-1}}{6^n}$.

---

## ¿El experimento termina eventualmente?

Los eventos $E_1,E_2,\ldots$ son disjuntos, pues el primer $6$ no puede aparecer simultáneamente en dos posiciones distintas.

Por lo tanto,

$\mathbb{P}\left(\bigcup_{n=1}^{\infty}E_n\right)=\sum_{n=1}^{\infty}\mathbb{P}(E_n)$.

Entonces,

$\mathbb{P}\left(\bigcup_{n=1}^{\infty}E_n\right)=\sum_{n=1}^{\infty}\left(\frac{5}{6}\right)^{n-1}\frac{1}{6}$.

Haciendo $k=n-1$,

$\mathbb{P}\left(\bigcup_{n=1}^{\infty}E_n\right)=\frac{1}{6}\sum_{k=0}^{\infty}\left(\frac{5}{6}\right)^k$.

Recordemos que

$\sum_{k=0}^{\infty}r^k=\frac{1}{1-r}$,

si $|r|<1$.

Por lo tanto,

$\mathbb{P}\left(\bigcup_{n=1}^{\infty}E_n\right)=\frac{1}{6}\frac{1}{1-\frac{5}{6}}$.

Así,

$\mathbb{P}\left(\bigcup_{n=1}^{\infty}E_n\right)=1$.

Por tanto,

$\boxed{\mathbb{P}(\text{eventualmente aparece un }6)=1}$.

---

## Otra forma de verlo

La probabilidad de no obtener ningún $6$ durante los primeros $n$ lanzamientos es

$\left(\frac{5}{6}\right)^n$.

Como

$\lim_{n\to\infty}\left(\frac{5}{6}\right)^n=0$,

tenemos

$\mathbb{P}(\text{nunca aparece un }6)=0$.

En consecuencia,

$\mathbb{P}(\text{eventualmente aparece un }6)=1$.

### Observación

Si definimos el espacio muestral usando únicamente las sucesiones finitas que terminan cuando aparece el primer $6$, entonces

$\Omega=\bigcup_{n=1}^{\infty}E_n$.

Si, en cambio, utilizamos como espacio muestral todas las sucesiones infinitas posibles de lanzamientos, entonces existen sucesiones en las que nunca aparece un $6$. En ese caso,

$\bigcup_{n=1}^{\infty}E_n\neq\Omega$,

pero

$\mathbb{P}\left(\bigcup_{n=1}^{\infty}E_n\right)=1$,

porque el conjunto de sucesiones en las que nunca aparece un $6$ tiene probabilidad cero.

# Ejemplo 3. El juego de Craps

El juego de **Craps** se juega de la siguiente manera.

Un jugador lanza dos dados justos.

En el primer lanzamiento:

- si la suma de los dados es $2$, $3$ o $12$, el jugador pierde;
- si la suma de los dados es $7$ u $11$, el jugador gana;
- si la suma es $4$, $5$, $6$, $8$, $9$ o $10$, esa suma se convierte en el **punto** del jugador.

Supongamos que el punto es $i$.

El jugador continúa lanzando los dos dados hasta que ocurra una de las siguientes cosas:

- vuelve a aparecer la suma $i$, en cuyo caso gana;
- aparece primero un $7$, en cuyo caso pierde.

¿Cuál es la probabilidad de que el jugador gane?

---

# Probabilidades de las sumas de dos dados

Sean $X_1$ y $X_2$ los resultados de los dos dados y sea

$S=X_1+X_2$.

El espacio muestral tiene

$|\Omega|=6^2=36$

resultados equiprobables.

Definimos

$p_i=\mathbb{P}(S=i)$.

Las posibles sumas y sus números de representaciones son:

| Suma | Formas |
|---|---:|
| $2$ | $1$ |
| $3$ | $2$ |
| $4$ | $3$ |
| $5$ | $4$ |
| $6$ | $5$ |
| $7$ | $6$ |
| $8$ | $5$ |
| $9$ | $4$ |
| $10$ | $3$ |
| $11$ | $2$ |
| $12$ | $1$ |

Por ejemplo, para obtener suma $4$ tenemos

$(1,3),(2,2),(3,1)$.

Por tanto,

$p_4=\frac{3}{36}=\frac{1}{12}$.

Para obtener suma $5$ tenemos

$(1,4),(2,3),(3,2),(4,1)$.

Así,

$p_5=\frac{4}{36}=\frac{1}{9}$.

En general,

$p_2=\frac{1}{36}$,

$p_3=\frac{2}{36}=\frac{1}{18}$,

$p_4=\frac{3}{36}=\frac{1}{12}$,

$p_5=\frac{4}{36}=\frac{1}{9}$,

$p_6=\frac{5}{36}$,

$p_7=\frac{6}{36}=\frac{1}{6}$,

$p_8=\frac{5}{36}$,

$p_9=\frac{4}{36}=\frac{1}{9}$,

$p_{10}=\frac{3}{36}=\frac{1}{12}$,

$p_{11}=\frac{2}{36}=\frac{1}{18}$,

$p_{12}=\frac{1}{36}$.

---

# Ganar en el primer lanzamiento

El jugador gana inmediatamente si obtiene $7$ u $11$.

Por lo tanto,

$\mathbb{P}(\text{ganar inmediatamente})=p_7+p_{11}$.

Como

$p_7=\frac{6}{36}$

y

$p_{11}=\frac{2}{36}$,

tenemos

$\mathbb{P}(\text{ganar inmediatamente})=\frac{8}{36}=\frac{2}{9}$.

---

# Ganar después de establecer un punto

Los posibles puntos son

$I=\{4,5,6,8,9,10\}$.

Fijemos $i\in I$.

Definimos

$E_{i,k}=\{\text{el punto es }i\text{ y el jugador gana exactamente en el lanzamiento }k\}$,

donde $k\geq2$.

---

## Ganar en el segundo lanzamiento

Para ganar exactamente en el segundo lanzamiento con punto $i$ necesitamos:

1. obtener suma $i$ en el primer lanzamiento;
2. volver a obtener suma $i$ en el segundo lanzamiento.

Por independencia,

$\mathbb{P}(E_{i,2})=p_i\cdot p_i=p_i^2$.

---

## Ganar en el tercer lanzamiento

Para ganar exactamente en el tercer lanzamiento necesitamos:

1. obtener $i$ en el primer lanzamiento;
2. obtener en el segundo lanzamiento algo distinto de $i$ y de $7$;
3. obtener $i$ en el tercer lanzamiento.

La probabilidad de obtener algo distinto de $i$ y $7$ es

$1-p_i-p_7$.

Por lo tanto,

$\mathbb{P}(E_{i,3})=p_i(1-p_i-p_7)p_i$.

Así,

$\mathbb{P}(E_{i,3})=p_i^2(1-p_i-p_7)$.

---

## Ganar en el cuarto lanzamiento

Para ganar exactamente en el cuarto lanzamiento necesitamos:

1. obtener $i$;
2. después obtener dos veces algo distinto de $i$ y $7$;
3. finalmente obtener nuevamente $i$.

Entonces,

$\mathbb{P}(E_{i,4})=p_i^2(1-p_i-p_7)^2$.

---

## Caso general

Para ganar exactamente en el lanzamiento $k$ necesitamos $k-2$ lanzamientos intermedios en los que no aparezca ni $i$ ni $7$.

Por lo tanto,

$\boxed{\mathbb{P}(E_{i,k})=p_i^2(1-p_i-p_7)^{k-2}}$.

---

# Probabilidad de ganar con punto \(i\)

Sea

$E_i=\{\text{el jugador gana después de haber establecido el punto }i\}$.

Tenemos

$E_i=\bigcup_{k=2}^{\infty}E_{i,k}$.

Los eventos $E_{i,k}$ son disjuntos.

Por lo tanto,

$\mathbb{P}(E_i)=\sum_{k=2}^{\infty}\mathbb{P}(E_{i,k})$.

Así,

$\mathbb{P}(E_i)=\sum_{k=2}^{\infty}p_i^2(1-p_i-p_7)^{k-2}$.

Sacando $p_i^2$,

$\mathbb{P}(E_i)=p_i^2\sum_{k=2}^{\infty}(1-p_i-p_7)^{k-2}$.

Haciendo $j=k-2$,

$\mathbb{P}(E_i)=p_i^2\sum_{j=0}^{\infty}(1-p_i-p_7)^j$.

Usando la serie geométrica,

$\sum_{j=0}^{\infty}(1-p_i-p_7)^j=\frac{1}{1-(1-p_i-p_7)}$.

Por tanto,

$\sum_{j=0}^{\infty}(1-p_i-p_7)^j=\frac{1}{p_i+p_7}$.

Así,

$\boxed{\mathbb{P}(E_i)=\frac{p_i^2}{p_i+p_7}}$.

### Interpretación

Otra manera de entender esta fórmula es la siguiente.

Una vez que el punto $i$ ya fue establecido, los únicos resultados que finalmente importan son:

- obtener nuevamente $i$, lo cual produce una victoria;
- obtener $7$, lo cual produce una derrota.

Los demás resultados simplemente hacen que el juego continúe.

Por tanto, condicionados a que uno de los dos resultados decisivos ocurra, la probabilidad de obtener primero $i$ es

$\frac{p_i}{p_i+p_7}$.

Pero para llegar a esta situación primero tuvimos que obtener $i$ en el lanzamiento inicial, lo cual tiene probabilidad $p_i$.

Así,

$p_i\frac{p_i}{p_i+p_7}=\frac{p_i^2}{p_i+p_7}$.

---

# Calculamos cada punto

## Punto 4

Tenemos

$p_4=\frac{3}{36}=\frac{1}{12}$

y

$p_7=\frac{1}{6}$.

Entonces,

$\mathbb{P}(E_4)=\frac{p_4^2}{p_4+p_7}$.

Por tanto,

$\mathbb{P}(E_4)=\frac{\left(\frac{1}{12}\right)^2}{\frac{1}{12}+\frac{1}{6}}$.

Como

$\frac{1}{12}+\frac{1}{6}=\frac{3}{12}=\frac{1}{4}$,

tenemos

$\mathbb{P}(E_4)=\frac{1/144}{1/4}=\frac{1}{36}$.

Así,

$\boxed{\mathbb{P}(E_4)=\frac{1}{36}}$.

Por simetría,

$\boxed{\mathbb{P}(E_{10})=\frac{1}{36}}$.

---

## Punto 5

Tenemos

$p_5=\frac{4}{36}=\frac{1}{9}$.

Entonces,

$\mathbb{P}(E_5)=\frac{\left(\frac{1}{9}\right)^2}{\frac{1}{9}+\frac{1}{6}}$.

Como

$\frac{1}{9}+\frac{1}{6}=\frac{5}{18}$,

obtenemos

$\mathbb{P}(E_5)=\frac{1}{81}\frac{18}{5}=\frac{2}{45}$.

Así,

$\boxed{\mathbb{P}(E_5)=\frac{2}{45}}$.

Por simetría,

$\boxed{\mathbb{P}(E_9)=\frac{2}{45}}$.

---

## Punto 6

Tenemos

$p_6=\frac{5}{36}$.

Entonces,

$\mathbb{P}(E_6)=\frac{\left(\frac{5}{36}\right)^2}{\frac{5}{36}+\frac{6}{36}}$.

Así,

$\mathbb{P}(E_6)=\frac{\frac{25}{1296}}{\frac{11}{36}}$.

Por lo tanto,

$\mathbb{P}(E_6)=\frac{25}{1296}\frac{36}{11}=\frac{25}{396}$.

Entonces,

$\boxed{\mathbb{P}(E_6)=\frac{25}{396}}$.

Por simetría,

$\boxed{\mathbb{P}(E_8)=\frac{25}{396}}$.

---

# Probabilidad total de ganar en Craps

El jugador puede ganar de las siguientes maneras:

- inmediatamente con un $7$;
- inmediatamente con un $11$;
- estableciendo punto $4$ y ganando;
- estableciendo punto $5$ y ganando;
- estableciendo punto $6$ y ganando;
- estableciendo punto $8$ y ganando;
- estableciendo punto $9$ y ganando;
- estableciendo punto $10$ y ganando.

Por lo tanto,

$\mathbb{P}(\text{ganar})=p_7+p_{11}+\mathbb{P}(E_4)+\mathbb{P}(E_5)+\mathbb{P}(E_6)+\mathbb{P}(E_8)+\mathbb{P}(E_9)+\mathbb{P}(E_{10})$.

Sustituyendo,

$\mathbb{P}(\text{ganar})=\frac{2}{9}+\frac{1}{36}+\frac{2}{45}+\frac{25}{396}+\frac{25}{396}+\frac{2}{45}+\frac{1}{36}$.

Agrupando los términos simétricos,

$\mathbb{P}(\text{ganar})=\frac{2}{9}+2\left(\frac{1}{36}\right)+2\left(\frac{2}{45}\right)+2\left(\frac{25}{396}\right)$.

Simplificando,

$\boxed{\mathbb{P}(\text{ganar})=\frac{244}{495}}$.

Numéricamente,

$\boxed{\mathbb{P}(\text{ganar})\approx0.492929}$.

Por lo tanto, la probabilidad de que el jugador gane una partida de Craps es aproximadamente

$\boxed{49.29\%}$.

En consecuencia, la probabilidad de perder es

$\mathbb{P}(\text{perder})=1-\frac{244}{495}=\frac{251}{495}$.

Numéricamente,

$\boxed{\mathbb{P}(\text{perder})\approx50.71\%}$.

Por eso el juego favorece ligeramente a la casa.
