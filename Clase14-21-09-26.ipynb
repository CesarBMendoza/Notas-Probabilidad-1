## Ejemplos de variables aleatorias discretas

### 1. Variable aleatoria de Bernoulli

Definimos

$$
X=
\begin{cases}
1, & \text{éxito},\\
0, & \text{falla}.
\end{cases}
$$

En un experimento aleatorio donde solo hay dos resultados, sea $p$ la probabilidad de éxito, por lo que

$$
p\in[0,1].
$$

El rango de $X$ es

$$
R_X=\{0,1\}.
$$

La función de probabilidad puntual está dada por

$$
p_X(0)
=
\mathbb{P}[X=0]
=
1-p,
$$

y

$$
p_X(1)
=
\mathbb{P}[X=1]
=
p.
$$

Si sabemos que una variable aleatoria $X$ tiene esta distribución, denotaremos

$$
X\sim\operatorname{Bernoulli}(p),
$$

y se lee como:

> $X$ se distribuye como una variable aleatoria Bernoulli con parámetro $p$.

**Ejemplo:** tirar una moneda se puede modelar mediante una distribución de Bernoulli si definimos, por ejemplo,

$$
X=
\begin{cases}
1, & \text{si cae águila},\\
0, & \text{si cae sol}.
\end{cases}
$$

### 2. Uniforme discreta

Sea

$$
X\sim U(B),
$$

donde $X$ toma uno de los resultados equiprobables del conjunto $B$.

Es decir, todos los elementos de $B$ tienen la misma probabilidad de ser obtenidos.

**Ejemplo:** lanzar un dado justo se comporta como una variable aleatoria uniforme discreta.

Además, como todos los resultados son equiprobables,

$$
R_X=B
$$

y

$$
p_X(k)=\frac{1}{|B|},
\qquad \forall k\in B.
$$

Por tanto, la función de distribución acumulada puede escribirse como

$$
F_X(z)
=
\mathbb{P}[X\leq z]
=
\frac{
\left|\{b\in B:b\leq z\}\right|
}{
|B|
}.
$$

Aquí, $|B|$ denota el número de elementos del conjunto $B$.

### 3. Variable aleatoria geométrica

Sea

$$
X\sim\operatorname{Geo}(p).
$$

Definimos $X$ como el número de intentos independientes necesarios hasta obtener el primer éxito.

En cada intento, el éxito ocurre con probabilidad

$$
p.
$$

**Ejemplo:** tirar un dado repetidamente hasta obtener por primera vez un $6$ se comporta como una variable aleatoria geométrica, tomando

$$
p=\frac16.
$$

El rango de $X$ es

$$
R_X=\{1,2,3,\ldots\}=\mathbb{N}.
$$

Para $k\in\mathbb{N}$,

$$
p_X(k)
=
\mathbb{P}[X=k].
$$

El evento $X=k$ significa que tenemos $k-1$ fallas seguidas y después un éxito:

$$
X=k
\quad\Longleftrightarrow\quad
\underbrace{FF\cdots F}_{k-1\text{ fallas}}E.
$$

Como los intentos son independientes,

$$
p_X(k)
=
\mathbb{P}[F]\cdots\mathbb{P}[F]\mathbb{P}[E].
$$

Hay $k-1$ fallas, cada una con probabilidad $1-p$, y un éxito con probabilidad $p$. Por tanto,

$$
p_X(k)
=
(1-p)^{k-1}p.
$$

Así,

$$
\boxed{
p_X(k)=(1-p)^{k-1}p
}
\qquad
k=1,2,\ldots
$$

Ahora calculemos la función de distribución acumulada:

$$
F_X(n)
=
\mathbb{P}[X\leq n].
$$

Como los eventos $\{X=k\}$ son disjuntos,

$$
F_X(n)
=
\sum_{k=1}^{n}\mathbb{P}[X=k].
$$

Por lo tanto,

$$
F_X(n)
=
\sum_{k=1}^{n}(1-p)^{k-1}p.
$$

Continuando con el cálculo de la función de distribución de la variable geométrica,

$$
F_X(n)
=
p\sum_{k=1}^{n}(1-p)^{k-1}.
$$

Haciendo el cambio de índice

$$
j=k-1,
$$

obtenemos

$$
F_X(n)
=
p\sum_{j=0}^{n-1}(1-p)^j.
$$

Recordemos que para una serie geométrica finita,

$$
\sum_{k=0}^{m}r^k
=
\frac{1-r^{m+1}}{1-r},
\qquad r\neq1.
$$

Tomando

$$
r=1-p,
$$

tenemos

$$
\sum_{j=0}^{n-1}(1-p)^j
=
\frac{1-(1-p)^n}{1-(1-p)}.
$$

Como

$$
1-(1-p)=p,
$$

entonces

$$
\sum_{j=0}^{n-1}(1-p)^j
=
\frac{1-(1-p)^n}{p}.
$$

Por tanto,

$$
F_X(n)
=
p
\left(
\frac{1-(1-p)^n}{p}
\right),
$$

y obtenemos

$$
\boxed{
F_X(n)=1-(1-p)^n
}.
$$

Como la función de supervivencia está definida por

$$
S_X(n)
=
\mathbb{P}[X>n]
=
1-F_X(n),
$$

tenemos

$$
\boxed{
S_X(n)=(1-p)^n
}.
$$

## Falta de memoria

La variable aleatoria geométrica tiene una propiedad importante llamada **falta de memoria**.

Queremos calcular

$$
\mathbb{P}[X>n+m\mid X>n].
$$

Por definición de probabilidad condicional,

$$
\mathbb{P}[X>n+m\mid X>n]
=
\frac{
\mathbb{P}\big(
\{X>n+m\}\cap\{X>n\}
\big)
}{
\mathbb{P}[X>n]
}.
$$

Como

$$
\{X>n+m\}\subseteq\{X>n\},
$$

entonces

$$
\{X>n+m\}\cap\{X>n\}
=
\{X>n+m\}.
$$

Por tanto,

$$
\mathbb{P}[X>n+m\mid X>n]
=
\frac{
\mathbb{P}[X>n+m]
}{
\mathbb{P}[X>n]
}.
$$

Usando la función de supervivencia,

$$
=
\frac{
S_X(n+m)
}{
S_X(n)
}.
$$

Como

$$
S_X(k)=(1-p)^k,
$$

tenemos

$$
\mathbb{P}[X>n+m\mid X>n]
=
\frac{
(1-p)^{n+m}
}{
(1-p)^n
}.
$$

Por las propiedades de los exponentes,

$$
\mathbb{P}[X>n+m\mid X>n]
=
(1-p)^m.
$$

Pero

$$
(1-p)^m
=
\mathbb{P}[X>m].
$$

Por lo tanto,

$$
\boxed{
\mathbb{P}[X>n+m\mid X>n]
=
\mathbb{P}[X>m]
}.
$$

Es decir, el hecho de que ya hayan ocurrido $n$ intentos sin éxito no modifica la distribución del número de intentos adicionales necesarios para obtener el primer éxito.

### 4. Variable aleatoria binomial

Sea

$$
X\sim\operatorname{Binomial}(n,p),
$$

donde

$$
n\in\mathbb{N}
\qquad\text{y}\qquad
p\in[0,1].
$$

La variable aleatoria $X$ representa el número de éxitos que ocurren al realizar $n$ intentos independientes, donde en cada intento la probabilidad de éxito es $p$.

El rango de $X$ es

$$
R_X=\{0,1,\ldots,n\}.
$$

Queremos calcular

$$
p_X(k)
=
\mathbb{P}[X=k].
$$

El evento $\{X=k\}$ significa obtener exactamente $k$ éxitos en los $n$ intentos.

Primero, una configuración particular con $k$ éxitos y $n-k$ fallas tiene probabilidad

$$
p^k(1-p)^{n-k}.
$$

Ahora debemos elegir cuáles de los $n$ intentos corresponden a los $k$ éxitos.

Esto puede hacerse de

$$
\binom{n}{k}
$$

formas.

Por tanto,

$$
p_X(k)
=
\binom{n}{k}
p^k(1-p)^{n-k}.
$$

Así,

$$
\boxed{
p_X(k)
=
\binom{n}{k}
p^k(1-p)^{n-k}
}
$$

para

$$
k=0,1,\ldots,n.
$$
