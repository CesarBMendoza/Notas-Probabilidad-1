# ¿Qué se necesita para una medida de probabilidad?

## 1. Espacio muestral

El **espacio muestral**, denotado por $\Omega$, es el conjunto de todos los posibles resultados de un experimento aleatorio.

### Ejemplos

Al lanzar una moneda:

$$
\Omega_1=\{\text{águila},\text{sol}\}.
$$

Al lanzar un dado:

$$
\Omega_2=\{1,\ldots,6\}.
$$

Para describir el clima:

$$
\Omega_3=\{\text{soleado},\text{lluvioso},\text{nublado}\}.
$$

Otros ejemplos pueden involucrar:

$$
\Omega_4=\{\text{colores posibles}\},
$$

$$
\Omega_5=\{\text{posibles resultados de equipos y goles}\},
$$

$$
\Omega_6=\{\text{temperaturas posibles}\}.
$$

También podemos tener espacios muestrales continuos. Por ejemplo,

$$
\Omega_7=[0,1].
$$

En este caso, $\Omega_7$ no es discreto, sino continuo.

---

## 2. Espacio de eventos

Sea

$$
\Omega=\{1,\ldots,6\}.
$$

Algunos posibles eventos son

$$
E_1=\{2,4,6\},
$$

$$
E_2=\{1,3,5\},
$$

$$
E_3=\{1,2,6\}.
$$

A partir de estos eventos podemos construir otros. Por ejemplo,

$$
E_4=E_1\cup E_2.
$$

Como

$$
E_1\cup E_2
=
\{1,2,3,4,5,6\},
$$

tenemos

$$
E_4=\Omega.
$$

También,

$$
E_5
=
\{1,2,6\}\cap\{2,4,6\}^c.
$$

Como

$$
\{2,4,6\}^c=\{1,3,5\},
$$

se sigue que

$$
E_5=\{1\}.
$$

---

Si $\Omega$ es finito, comúnmente podemos tomar como colección de eventos al conjunto potencia

$$
\mathcal{F}=\mathcal{P}(\Omega),
$$

es decir, el conjunto de todos los subconjuntos de $\Omega$.

¿Qué propiedades debe cumplir $\mathcal{F}$?

1. Si $A\in\mathcal{F}$, entonces

   $$
   A^c\in\mathcal{F}.
   $$

2. Se debe cumplir que

   $$
   \Omega\in\mathcal{F}.
   $$

Continuando con las propiedades que debe satisfacer $\mathcal{F}$:

3. Si

   $$
   A_1,A_2,A_3,\ldots\in\mathcal{F},
   $$

   entonces

   $$
   \bigcup_{j=1}^{\infty}A_j\in\mathcal{F}.
   $$

### Definición

A una familia $\mathcal{F}$ de subconjuntos de $\Omega$ que satisface las propiedades anteriores se le llama una **$\sigma$-álgebra** sobre $\Omega$.

---

### Ejemplo

Sea

$$
\Omega=\{1,2,3\}.
$$

Una posible $\sigma$-álgebra es

$$
\mathcal{F}_1=\mathcal{P}(\Omega).
$$

En este caso,

$$
\mathcal{F}_1
=
\{
\varnothing,
\{1\},
\{2\},
\{3\},
\{1,2\},
\{1,3\},
\{2,3\},
\Omega
\}.
$$

Otra $\sigma$-álgebra es

$$
\mathcal{F}_2
=
\{\varnothing,\Omega\}.
$$

También podemos considerar

$$
\mathcal{F}_3
=
\{\varnothing,\Omega,\{1\},\{2,3\}\}.
$$

---

## 3. Medida de probabilidad

Lo último que necesitamos es una **medida de probabilidad**.

Una medida de probabilidad es una función

$$
P:\mathcal{F}\longrightarrow[0,1].
$$

Debe satisfacer:

1. Para todo $A\in\mathcal{F}$,

   $$
   P(A)\geq0.
   $$

2. Se cumple que

   $$
   P(\Omega)=1.
   $$

3. Si

   $$
   A_1,A_2,\ldots\in\mathcal{F}
   $$

   son eventos disjuntos dos a dos, es decir,

   $$
   A_i\cap A_j=\varnothing
   \qquad\text{si }i\neq j,
   $$

   entonces

   $$
   P\left(
   \bigcup_{n=1}^{\infty}A_n
   \right)
   =
   \sum_{n=1}^{\infty}P(A_n).
   $$

# Formulación axiomática de la probabilidad

Los axiomas de probabilidad son los siguientes.

## 1. No negatividad

Para todo

$$
A\in\mathcal{F},
$$

se cumple

$$
P(A)\geq0.
$$

---

## 2. Normalización

Se cumple que

$$
P(\Omega)=1.
$$

---

## 3. Aditividad numerable

Si

$$
\{A_i\}_{i\in\mathbb{N}}
$$

es una colección de eventos tales que

$$
A_i\cap A_j=\varnothing
\qquad
\text{para todo }i\neq j,
$$

entonces

$$
P\left(
\bigcup_{i=1}^{\infty}A_i
\right)
=
\sum_{i=1}^{\infty}P(A_i).
$$

---

# Consecuencias de los axiomas

## 1. Probabilidad del conjunto vacío

Se cumple que

$$
P(\varnothing)=0.
$$

En efecto, como

$$
\Omega=\Omega\cup\varnothing
$$

y

$$
\Omega\cap\varnothing=\varnothing,
$$

por aditividad tenemos

$$
P(\Omega)
=
P(\Omega)+P(\varnothing).
$$

Como

$$
P(\Omega)=1,
$$

entonces

$$
1=1+P(\varnothing).
$$

Por lo tanto,

$$
\boxed{P(\varnothing)=0}.
$$

---

## 2. Aditividad finita

Sean

$$
A_1,A_2,\ldots,A_n\in\mathcal{F}
$$

eventos disjuntos dos a dos, es decir,

$$
A_i\cap A_j=\varnothing
\qquad
\text{si }i\neq j.
$$

Entonces

$$
P\left(
\bigcup_{i=1}^{n}A_i
\right)
=
\sum_{i=1}^{n}P(A_i).
$$

---

## 3. Probabilidad del complemento

Si

$$
A\in\mathcal{F},
$$

entonces

$$
P(A^c)=1-P(A).
$$

En efecto,

$$
\Omega=A\cup A^c
$$

y

$$
A\cap A^c=\varnothing.
$$

Por lo tanto,

$$
P(\Omega)
=
P(A)+P(A^c).
$$

Como

$$
P(\Omega)=1,
$$

se sigue que

$$
1=P(A)+P(A^c).
$$

Así,

$$
\boxed{P(A^c)=1-P(A)}.
$$

## 4. La probabilidad está entre $0$ y $1$

Si

$$
A\in\mathcal{F},
$$

entonces

$$
0\leq P(A)\leq1.
$$

En efecto, por el primer axioma,

$$
P(A)\geq0.
$$

Además,

$$
P(A^c)\geq0.
$$

Pero sabemos que

$$
P(A^c)=1-P(A).
$$

Por lo tanto,

$$
1-P(A)\geq0,
$$

de donde

$$
P(A)\leq1.
$$

Así,

$$
\boxed{0\leq P(A)\leq1}.
$$

---

## 5. Monotonía de la probabilidad

Si

$$
A,B\in\mathcal{F}
$$

y

$$
A\subseteq B,
$$

entonces

$$
P(A)\leq P(B).
$$

En efecto, podemos escribir

$$
B=A\cup(B\setminus A).
$$

Además,

$$
A\cap(B\setminus A)=\varnothing.
$$

Por aditividad,

$$
P(B)
=
P(A)+P(B\setminus A).
$$

Como

$$
P(B\setminus A)\geq0,
$$

obtenemos

$$
P(B)\geq P(A).
$$

Por lo tanto,

$$
\boxed{P(A)\leq P(B)}.
$$

---

## 6. Probabilidad de la unión de dos eventos

Si

$$
A,B\in\mathcal{F},
$$

entonces

$$
P(A\cup B)
=
P(A)+P(B)-P(A\cap B).
$$

En efecto,

$$
A\cup B
=
(A\setminus B)
\cup
(A\cap B)
\cup
(B\setminus A).
$$

Estos tres conjuntos son disjuntos dos a dos. Por lo tanto,

$$
P(A\cup B)
=
P(A\setminus B)
+
P(A\cap B)
+
P(B\setminus A).
$$

Por otro lado,

$$
A
=
(A\setminus B)\cup(A\cap B),
$$

por lo que

$$
P(A)
=
P(A\setminus B)+P(A\cap B).
$$

De igual manera,

$$
B
=
(B\setminus A)\cup(A\cap B),
$$

de modo que

$$
P(B)
=
P(B\setminus A)+P(A\cap B).
$$

Sumando las expresiones anteriores,

$$
P(A)+P(B)
=
P(A\setminus B)
+
P(B\setminus A)
+
2P(A\cap B).
$$

Mientras que

$$
P(A\cup B)
=
P(A\setminus B)
+
P(B\setminus A)
+
P(A\cap B).
$$

Por lo tanto,

$$
P(A\cup B)
=
P(A)+P(B)-P(A\cap B).
$$

Así,

$$
\boxed{
P(A\cup B)
=
P(A)+P(B)-P(A\cap B)
}.
$$

---

# Fórmula de inclusión-exclusión

Para eventos

$$
A_1,A_2,\ldots,A_n,
$$

se tiene

$$
P\left(
\bigcup_{i=1}^{n}A_i
\right)
=
\sum_{i=1}^{n}P(A_i)
-
\sum_{1\leq i<j\leq n}
P(A_i\cap A_j)
$$

$$
+
\sum_{1\leq i<j<k\leq n}
P(A_i\cap A_j\cap A_k)
-\cdots
+
(-1)^{n+1}
P(A_1\cap A_2\cap\cdots\cap A_n).
$$

Es decir,

$$
\boxed{
P\left(
\bigcup_{i=1}^{n}A_i
\right)
=
\sum_{i=1}^{n}P(A_i)
-
\sum_{1\leq i<j\leq n}
P(A_i\cap A_j)
+
\sum_{1\leq i<j<k\leq n}
P(A_i\cap A_j\cap A_k)
-\cdots
+
(-1)^{n+1}
P\left(
\bigcap_{i=1}^{n}A_i
\right)
}.
$$
