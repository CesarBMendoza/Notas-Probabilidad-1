# Media o esperanza de una variable aleatoria discreta

## Motivación: Promedio de calificaciones

Supongamos que tenemos $N$ exámenes cuyas calificaciones pueden tomar valores entre $0$ y $10$.

Queremos calcular el promedio de las calificaciones.

Por ejemplo, si las calificaciones obtenidas fueron

$$
9,8,5,10,\ldots,10,
$$

entonces su promedio está dado por

$$
\text{Promedio}=\frac{9+8+5+10+\cdots+10}{N},
$$

donde $N$ es el número total de exámenes.

Observemos que podemos agrupar las calificaciones de acuerdo con su valor.

Por lo tanto,

$$
\begin{aligned}
\text{Promedio}
&=\frac{(10+\cdots+10)+(9+\cdots+9)+\cdots+(0+\cdots+0)}{N}\\[6pt]
&=\frac{10n_{10}+9n_9+\cdots+0n_0}{N}\\[6pt]
&=10\frac{n_{10}}{N}+9\frac{n_9}{N}+\cdots+0\frac{n_0}{N}\\[6pt]
&=\sum_{j=0}^{10}j\frac{n_j}{N}.
\end{aligned}
$$

Donde $n_j$ representa el número de veces que aparece la calificación $j$.

Observemos que

$$
\frac{n_j}{N}\in[0,1].
$$

Además,

$$
\begin{aligned}
\sum_{j=0}^{10}\frac{n_j}{N}
&=\frac{1}{N}\sum_{j=0}^{10}n_j\\
&=\frac{N}{N}\\
&=1.
\end{aligned}
$$

Es decir,

$$
\boxed{
\sum_{j=0}^{10}\frac{n_j}{N}=1.
}
$$

La cantidad

$$
\frac{n_j}{N}
$$

representa la **frecuencia relativa** de la calificación $j$.

Podemos interpretarla como una probabilidad empírica de obtener la calificación $j$ al seleccionar un examen al azar.

De esta manera, el promedio puede escribirse como

$$
\boxed{
\text{Promedio}=
\sum_{j=0}^{10}j\frac{n_j}{N}.
}
$$

Esta expresión motiva la definición de esperanza matemática.

---

# Definición: Media o esperanza matemática

Sea $X$ una variable aleatoria discreta con función de masa

$$
p_X(k)=P(X=k)
$$

y conjunto de valores posibles $R_X$.

La **media, esperanza matemática, valor esperado o primer momento** de $X$ se define como

$$
\boxed{
\mu_X=E[X]=\langle X\rangle
=\sum_{k\in R_X}k\,p_X(k),
}
$$

siempre que la serie converja absolutamente.

Es decir,

$$
\boxed{
E[X]=\sum_{k\in R_X}kP(X=k).
}
$$

**Observación:** Si $X$ es no negativa, también podemos considerar el caso en que la esperanza sea infinita.

---

# Ejercicio 1: Esperanza de una variable Bernoulli

Calcular la media de una variable aleatoria Bernoulli.

Supongamos que

$$
X\sim\operatorname{Bernoulli}(p),
$$

donde $0\leq p\leq1$.

Sabemos que

$$
R_X=\{0,1\}
$$

y que su función de masa está dada por

$$
p_X(k)=
\begin{cases}
p, & k=1,\\
1-p, & k=0,\\
0, & \text{en otro caso}.
\end{cases}
$$

Por definición,

$$
E[X]=\sum_{k\in R_X}k\,p_X(k).
$$

Como $R_X=\{0,1\}$, tenemos que

$$
\begin{aligned}
E[X]
&=\sum_{k=0}^{1}k\,p_X(k)\\
&=0\cdot p_X(0)+1\cdot p_X(1)\\
&=0\cdot(1-p)+1\cdot p\\
&=p.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
E[X]=p.
}
$$

# Ejercicio 2: Esperanza de una variable binomial

Calcular la media de una variable aleatoria binomial.

Supongamos que

$$
X\sim\operatorname{Binomial}(n,p),
$$

donde $n\in\mathbb{N}$ y $0\leq p\leq1$.

Sabemos que

$$
R_X=\{0,1,\ldots,n\}
$$

y que su función de masa está dada por

$$
p_X(k)=
\binom{n}{k}p^k(1-p)^{n-k}.
$$

Queremos calcular

$$
E[X]=\sum_{k=0}^{n}k\,p_X(k).
$$

Sustituyendo la función de masa, tenemos que

$$
\begin{aligned}
E[X]
&=\sum_{k=0}^{n}k\binom{n}{k}p^k(1-p)^{n-k}\\[6pt]
&=\sum_{k=0}^{n}k\frac{n!}{k!(n-k)!}p^k(1-p)^{n-k}.
\end{aligned}
$$

Observemos que el término correspondiente a $k=0$ es cero.

Por lo tanto, podemos comenzar la suma en $k=1$:

$$
E[X]
=
\sum_{k=1}^{n}k\frac{n!}{k!(n-k)!}p^k(1-p)^{n-k}.
$$

Recordemos que

$$
k!=k(k-1)!
$$

y

$$
n!=n(n-1)!.
$$

Entonces,

$$
\begin{aligned}
E[X]
&=\sum_{k=1}^{n}
k\frac{n(n-1)!}{k(k-1)!(n-k)!}
p^k(1-p)^{n-k}\\[6pt]
&=\sum_{k=1}^{n}
\frac{n(n-1)!}{(k-1)!(n-k)!}
p^k(1-p)^{n-k}.
\end{aligned}
$$

Ahora utilizamos que

$$
p^k=p\,p^{k-1}.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=np\sum_{k=1}^{n}
\frac{(n-1)!}{(k-1)!(n-k)!}
p^{k-1}(1-p)^{n-k}.
\end{aligned}
$$

Realizamos el cambio de índice

$$
j=k-1.
$$

Entonces,

$$
k=j+1.
$$

Observemos que:

- Si $k=1$, entonces $j=0$.
- Si $k=n$, entonces $j=n-1$.

Además,

$$
n-k=n-(j+1)=n-1-j.
$$

En consecuencia,

$$
E[X]
=
np\sum_{j=0}^{n-1}
\frac{(n-1)!}{j!(n-1-j)!}
p^j(1-p)^{n-1-j}.
$$

Recordemos que

$$
\binom{n-1}{j}
=
\frac{(n-1)!}{j!(n-1-j)!}.
$$

Por lo tanto,

$$
E[X]
=
np\sum_{j=0}^{n-1}
\binom{n-1}{j}p^j(1-p)^{n-1-j}.
$$

Observemos que la suma corresponde a las probabilidades de una distribución binomial con parámetros $n-1$ y $p$.

Por lo tanto,

$$
\sum_{j=0}^{n-1}
\binom{n-1}{j}p^j(1-p)^{n-1-j}
=1.
$$

También podemos verificarlo mediante el teorema del binomio:

$$
\begin{aligned}
\sum_{j=0}^{n-1}
\binom{n-1}{j}p^j(1-p)^{n-1-j}
&=(p+(1-p))^{n-1}\\
&=1^{n-1}\\
&=1.
\end{aligned}
$$

En consecuencia,

$$
\begin{aligned}
E[X]
&=np\cdot1\\
&=np.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
E[X]=np.
}
$$

# Ejercicio 3: Esperanza de una variable Poisson

La distribución de Poisson se utiliza para modelar el número de veces que ocurre un evento durante un intervalo de tiempo o una región determinados, bajo ciertas condiciones.

Supongamos que

$$
X\sim\operatorname{Poisson}(\lambda),
$$

donde $\lambda>0$.

Sabemos que

$$
R_X=\mathbb{N}_0=\{0,1,2,\ldots\}
$$

y que su función de masa está dada por

$$
p_X(k)=e^{-\lambda}\frac{\lambda^k}{k!},
\qquad k=0,1,2,\ldots
$$

Queremos calcular la esperanza de $X$.

Por definición,

$$
E[X]=\sum_{k=0}^{\infty}k\,p_X(k).
$$

Sustituyendo la función de masa, tenemos que

$$
\begin{aligned}
E[X]
&=\sum_{k=0}^{\infty}
k\,e^{-\lambda}\frac{\lambda^k}{k!}\\[6pt]
&=e^{-\lambda}\sum_{k=0}^{\infty}
k\frac{\lambda^k}{k!}.
\end{aligned}
$$

Observemos que cuando $k=0$, el término correspondiente es cero.

Por lo tanto,

$$
E[X]
=
e^{-\lambda}\sum_{k=1}^{\infty}
k\frac{\lambda^k}{k!}.
$$

Recordemos que

$$
k!=k(k-1)!.
$$

Entonces,

$$
\begin{aligned}
E[X]
&=e^{-\lambda}\sum_{k=1}^{\infty}
\frac{k\lambda^k}{k(k-1)!}\\[6pt]
&=e^{-\lambda}\sum_{k=1}^{\infty}
\frac{\lambda^k}{(k-1)!}.
\end{aligned}
$$

Como

$$
\lambda^k=\lambda\lambda^{k-1},
$$

podemos factorizar $\lambda$:

$$
E[X]
=
\lambda e^{-\lambda}
\sum_{k=1}^{\infty}
\frac{\lambda^{k-1}}{(k-1)!}.
$$

Ahora realizamos el cambio de índice

$$
j=k-1.
$$

Por lo tanto,

$$
k=j+1.
$$

Observemos que:

- Si $k=1$, entonces $j=0$.
- Cuando $k\to\infty$, también $j\to\infty$.

Así,

$$
E[X]
=
\lambda e^{-\lambda}
\sum_{j=0}^{\infty}\frac{\lambda^j}{j!}.
$$

Recordemos la expansión en serie de la función exponencial:

$$
e^x=\sum_{j=0}^{\infty}\frac{x^j}{j!}.
$$

En particular,

$$
e^\lambda
=
\sum_{j=0}^{\infty}\frac{\lambda^j}{j!}.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=\lambda e^{-\lambda}e^\lambda\\
&=\lambda e^{-\lambda+\lambda}\\
&=\lambda.
\end{aligned}
$$

En consecuencia,

$$
\boxed{
E[X]=\lambda.
}
$$

# Ejercicio 4: Esperanza de variables aleatorias con soporte infinito

Sean $X$ y $Y$ variables aleatorias discretas con funciones de masa dadas por

$$
\boxed{
p_X(k)=
\begin{cases}
\dfrac{4}{k(k+1)(k+2)}, & k\in\mathbb{N},\\[8pt]
0, & \text{en otro caso},
\end{cases}
}
$$

y

$$
\boxed{
p_Y(k)=
\begin{cases}
\dfrac{1}{k(k+1)}, & k\in\mathbb{N},\\[8pt]
0, & \text{en otro caso},
\end{cases}
}
$$

donde

$$
\mathbb{N}=\{1,2,3,\ldots\}.
$$

Por lo tanto,

$$
R_X=R_Y=\mathbb{N}.
$$

Queremos calcular sus esperanzas.

---

## Parte 1: Cálculo de $E[X]$

Por definición,

$$
E[X]=\sum_{k=1}^{\infty}k\,p_X(k).
$$

Sustituyendo la función de masa, tenemos que

$$
\begin{aligned}
E[X]
&=\sum_{k=1}^{\infty}
k\frac{4}{k(k+1)(k+2)}\\[6pt]
&=\sum_{k=1}^{\infty}
\frac{4}{(k+1)(k+2)}\\[6pt]
&=4\sum_{k=1}^{\infty}
\frac{1}{(k+1)(k+2)}.
\end{aligned}
$$

Para calcular esta serie, utilizaremos una descomposición en fracciones parciales.

Observemos que

$$
\frac{1}{(k+1)(k+2)}
=
\frac{1}{k+1}-\frac{1}{k+2}.
$$

En efecto,

$$
\begin{aligned}
\frac{1}{k+1}-\frac{1}{k+2}
&=\frac{(k+2)-(k+1)}{(k+1)(k+2)}\\[6pt]
&=\frac{1}{(k+1)(k+2)}.
\end{aligned}
$$

Por lo tanto,

$$
E[X]
=
4\sum_{k=1}^{\infty}
\left(
\frac{1}{k+1}-\frac{1}{k+2}
\right).
$$

Desarrollando los primeros términos de la suma, obtenemos

$$
\begin{aligned}
E[X]
&=4\left(
\frac12-\frac13
+\frac13-\frac14
+\frac14-\frac15
+\frac15-\frac16
+\cdots
\right).
\end{aligned}
$$

Observemos que los términos se van cancelando.

A este tipo de series se les llama **series telescópicas**.

Para justificar el resultado, consideremos la suma parcial

$$
S_n=
\sum_{k=1}^{n}
\left(
\frac{1}{k+1}-\frac{1}{k+2}
\right).
$$

Entonces,

$$
\begin{aligned}
S_n
&=\left(\frac12-\frac13\right)
+\left(\frac13-\frac14\right)
+\cdots
+\left(\frac{1}{n+1}-\frac{1}{n+2}\right)\\[6pt]
&=\frac12-\frac{1}{n+2}.
\end{aligned}
$$

Tomando el límite cuando $n\to\infty$, tenemos que

$$
\begin{aligned}
\lim_{n\to\infty}S_n
&=\lim_{n\to\infty}
\left(
\frac12-\frac{1}{n+2}
\right)\\[6pt]
&=\frac12.
\end{aligned}
$$

En consecuencia,

$$
\begin{aligned}
E[X]
&=4\left(\frac12\right)\\
&=2.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
E[X]=2.
}
$$

---

## Parte 2: Cálculo de $E[Y]$

Recordemos que la función de masa de $Y$ está dada por

$$
p_Y(k)=\frac{1}{k(k+1)},
\qquad k\in\mathbb{N}.
$$

Por definición,

$$
E[Y]=\sum_{k=1}^{\infty}k\,p_Y(k).
$$

Sustituyendo la función de masa, obtenemos

$$
\begin{aligned}
E[Y]
&=\sum_{k=1}^{\infty}
k\frac{1}{k(k+1)}\\[6pt]
&=\sum_{k=1}^{\infty}
\frac{1}{k+1}.
\end{aligned}
$$

Desarrollando los primeros términos,

$$
E[Y]
=
\frac12+\frac13+\frac14+\frac15+\cdots
$$

Observemos que esta serie está relacionada con la serie armónica:

$$
\sum_{k=1}^{\infty}\frac1k
=
1+\frac12+\frac13+\frac14+\cdots
$$

Sabemos que la serie armónica es divergente.

Por lo tanto,

$$
\begin{aligned}
E[Y]
&=\sum_{k=1}^{\infty}\frac{1}{k+1}\\[6pt]
&=\left(\sum_{k=1}^{\infty}\frac1k\right)-1\\[6pt]
&=+\infty.
\end{aligned}
$$

En consecuencia,

$$
\boxed{
E[Y]=+\infty.
}
$$

Es decir, la variable aleatoria $Y$ no tiene una esperanza finita.

---

## Conclusión del ejercicio

Hemos encontrado que

$$
\boxed{
E[X]=2
}
$$

mientras que

$$
\boxed{
E[Y]=+\infty.
}
$$

Esto muestra que una variable aleatoria discreta puede tener una función de masa de probabilidad bien definida, pero no necesariamente una esperanza finita.

En particular, aunque

$$
\sum_{k\in R_Y}p_Y(k)=1,
$$

esto no garantiza que

$$
\sum_{k\in R_Y}k\,p_Y(k)
$$

sea una serie convergente.

---

## Observación adicional: Verificación de las funciones de masa

Podemos comprobar que ambas funciones son efectivamente funciones de masa de probabilidad.

### Para la variable $X$

Queremos verificar que

$$
\sum_{k=1}^{\infty}
\frac{4}{k(k+1)(k+2)}=1.
$$

Observemos que

$$
\frac{4}{k(k+1)(k+2)}
=
2\left(
\frac{1}{k(k+1)}
-
\frac{1}{(k+1)(k+2)}
\right).
$$

Por lo tanto,

$$
\begin{aligned}
\sum_{k=1}^{\infty}p_X(k)
&=
2\sum_{k=1}^{\infty}
\left(
\frac{1}{k(k+1)}
-
\frac{1}{(k+1)(k+2)}
\right)\\[6pt]
&=2\left(\frac{1}{1\cdot2}\right)\\[6pt]
&=1.
\end{aligned}
$$

### Para la variable $Y$

Queremos verificar que

$$
\sum_{k=1}^{\infty}\frac{1}{k(k+1)}=1.
$$

Como

$$
\frac{1}{k(k+1)}
=
\frac1k-\frac{1}{k+1},
$$

tenemos que

$$
\begin{aligned}
\sum_{k=1}^{\infty}p_Y(k)
&=
\sum_{k=1}^{\infty}
\left(
\frac1k-\frac{1}{k+1}
\right)\\[6pt]
&=1.
\end{aligned}
$$

Por lo tanto, ambas funciones son funciones de masa válidas.

Sin embargo, solamente $X$ tiene esperanza finita.
