
Sea $X$ una variable aleatoria con función de densidad de probabilidad

$$
f(x)=
\begin{cases}
c(1-x^2), & -1<x<1,\\
0, & \text{en otro caso}.
\end{cases}
$$

### (a) ¿Cuál es el valor de $c$?

Recordemos que para que $f$ sea una función de densidad debe satisfacer:

$$
f(x)\geq 0
$$

y

$$
\int_{-\infty}^{\infty} f(x)\,dx=1.
$$

Como $f(x)=0$ fuera del intervalo $(-1,1)$, tenemos

$$
\int_{-1}^{1}c(1-x^2)\,dx=1.
$$

Sacando la constante $c$,

$$
c\int_{-1}^{1}(1-x^2)\,dx=1.
$$

Calculamos la integral:

$$
\int(1-x^2)\,dx
=
x-\frac{x^3}{3}.
$$

Por lo tanto,

$$
c
\left[
x-\frac{x^3}{3}
\right]_{-1}^{1}
=1.
$$

Evaluando en los extremos,

$$
c
\left[
\left(1-\frac13\right)
-
\left(-1+\frac13\right)
\right]
=1.
$$

Entonces,

$$
c
\left(
\frac23+\frac23
\right)
=1.
$$

Así,

$$
c\frac43=1.
$$

Por lo tanto,

$$
\boxed{c=\frac34}.
$$

La función de densidad queda

$$
f(x)=
\begin{cases}
\dfrac34(1-x^2), & -1<x<1,\\
0, & \text{en otro caso}.
\end{cases}
$$

---

### (b) ¿Cuál es la función de distribución acumulada de $X$?

La función de distribución acumulada se define como

$$
F_X(x)=P(X\leq x)
=
\int_{-\infty}^{x}f(t)\,dt.
$$

Debemos considerar tres casos.

#### Caso 1: $x\leq -1$

Como la variable aleatoria solamente toma valores en $(-1,1)$,

$$
F_X(x)=0.
$$

#### Caso 2: $-1<x<1$

Tenemos

$$
F_X(x)
=
\int_{-1}^{x}
\frac34(1-t^2)\,dt.
$$

Entonces,

$$
F_X(x)
=
\frac34
\left[
t-\frac{t^3}{3}
\right]_{-1}^{x}.
$$

Evaluando,

$$
F_X(x)
=
\frac34
\left[
x-\frac{x^3}{3}
-
\left(-1+\frac13\right)
\right].
$$

Como

$$
-1+\frac13=-\frac23,
$$

obtenemos

$$
F_X(x)
=
\frac34
\left(
x-\frac{x^3}{3}+\frac23
\right).
$$

Distribuyendo,

$$
F_X(x)
=
\frac{3x}{4}
-\frac{x^3}{4}
+\frac12.
$$

Por lo tanto,

$$
F_X(x)
=
\frac12+\frac{3x}{4}-\frac{x^3}{4}.
$$

#### Caso 3: $x\geq 1$

Para $x\geq1$ ya se ha acumulado toda la probabilidad, por lo que

$$
F_X(x)=1.
$$

Así, la función de distribución acumulada es

$$
\boxed{
F_X(x)=
\begin{cases}
0, & x\leq -1,\\[2mm]
\dfrac12+\dfrac{3x}{4}-\dfrac{x^3}{4}, & -1<x<1,\\[2mm]
1, & x\geq1.
\end{cases}
}
$$

Como comprobación,

$$
F_X(-1)=0
$$

y

$$
F_X(1)=1.
$$

## 2)

Considere la función

$$
f(x)=
\begin{cases}
C(2x-x^3), & 0<x<\dfrac52,\\
0, & \text{en otro caso}.
\end{cases}
$$

¿Podría ser una función de densidad de probabilidad?  
Si es así, determine $C$.

Repita el problema si

$$
f(x)=
\begin{cases}
C(2x-x^2), & 0<x<\dfrac52,\\
0, & \text{en otro caso}.
\end{cases}
$$

Recordemos que una función $f$ es una densidad de probabilidad si satisface

$$
f(x)\geq 0
$$

para todo $x$, y además

$$
\int_{-\infty}^{\infty} f(x)\,dx=1.
$$

En este ejercicio conviene revisar primero la condición de no negatividad.

---

### Primera función

Tenemos

$$
f(x)=C(2x-x^3),
\qquad
0<x<\frac52.
$$

Factorizamos:

$$
2x-x^3=x(2-x^2).
$$

Como $x>0$, el signo depende de

$$
2-x^2.
$$

Tenemos

$$
2-x^2>0
$$

si

$$
x<\sqrt2,
$$

mientras que

$$
2-x^2<0
$$

si

$$
x>\sqrt2.
$$

Como

$$
\sqrt2<\frac52,
$$

la expresión cambia de signo dentro del intervalo.

Por tanto,

$$
2x-x^3>0
$$

para

$$
0<x<\sqrt2,
$$

mientras que

$$
2x-x^3<0
$$

para

$$
\sqrt2<x<\frac52.
$$

Si $C>0$, entonces $f(x)<0$ para $x>\sqrt2$.

Si $C<0$, entonces $f(x)<0$ para $0<x<\sqrt2$.

Finalmente, si $C=0$, entonces

$$
\int_{-\infty}^{\infty}f(x)\,dx=0,
$$

por lo que tampoco puede ser una densidad.

Por lo tanto,

$$
\boxed{\text{no existe ningún valor de }C\text{ que haga a }f\text{ una densidad}.}
$$

---

### Segunda función

Ahora tenemos

$$
f(x)=C(2x-x^2),
\qquad
0<x<\frac52.
$$

Factorizamos:

$$
2x-x^2=x(2-x).
$$

Como $x>0$, el signo depende de

$$
2-x.
$$

Por tanto,

$$
2x-x^2>0
$$

si

$$
0<x<2,
$$

pero

$$
2x-x^2<0
$$

si

$$
2<x<\frac52.
$$

De nuevo, la función cambia de signo dentro del intervalo.

Si $C>0$, entonces $f(x)<0$ para

$$
2<x<\frac52.
$$

Si $C<0$, entonces $f(x)<0$ para

$$
0<x<2.
$$

Y si $C=0$, la integral total de la función es cero, por lo que tampoco puede ser una densidad.

Por tanto,

$$
\boxed{\text{tampoco existe un valor de }C\text{ que haga a }f\text{ una densidad}.}
$$

---

### Observación

Para que una función sea una densidad de probabilidad debe cumplir simultáneamente

$$
f(x)\geq0
$$

y

$$
\int_{-\infty}^{\infty}f(x)\,dx=1.
$$

Por eso no basta con buscar una constante $C$ que haga que la integral sea igual a $1$.  
Primero debemos verificar que la función sea no negativa en todo su soporte.

## 3)

La función de densidad de probabilidad de $X$, el tiempo de vida de cierto tipo de dispositivo electrónico medido en horas, está dada por

$$
f(x)=
\begin{cases}
\dfrac{10}{x^2}, & x>10,\\[2mm]
0, & x\leq10.
\end{cases}
$$

**(a)** Encuentra

$$
P(X>20).
$$


Como $X$ es una variable aleatoria continua,

$$
P(X>20)
=
\int_{20}^{\infty}f(x)\,dx.
$$

Usando la densidad,

$$
P(X>20)
=
\int_{20}^{\infty}\frac{10}{x^2}\,dx.
$$

Podemos escribir

$$
\frac{10}{x^2}=10x^{-2}.
$$

Entonces,

$$
\int10x^{-2}\,dx
=
-\frac{10}{x}.
$$

Por tanto,

$$
P(X>20)
=
\left[
-\frac{10}{x}
\right]_{20}^{\infty}.
$$

Evaluamos:

$$
P(X>20)
=
\lim_{b\to\infty}
\left(
-\frac{10}{b}
+
\frac{10}{20}
\right).
$$

Como

$$
\lim_{b\to\infty}\frac{10}{b}=0,
$$

obtenemos

$$
P(X>20)=\frac{10}{20}.
$$

Por lo tanto,

$$
\boxed{P(X>20)=\frac12}.
$$

**(b)** ¿Cuál es la función de distribución acumulada de $X$?

Recordemos que

$$
F_X(x)
=
P(X\leq x)
=
\int_{-\infty}^{x}f(t)\,dt.
$$

Debemos considerar dos casos.

#### Caso 1: $x\leq10$

Como la densidad es cero para $x\leq10$,

$$
F_X(x)=0.
$$

#### Caso 2: $x>10$

Tenemos

$$
F_X(x)
=
\int_{10}^{x}\frac{10}{t^2}\,dt.
$$

Integrando,

$$
F_X(x)
=
\left[
-\frac{10}{t}
\right]_{10}^{x}.
$$

Por lo tanto,

$$
F_X(x)
=
-\frac{10}{x}
-\left(-\frac{10}{10}\right).
$$

Así,

$$
F_X(x)
=
1-\frac{10}{x}.
$$

Por tanto,

$$
\boxed{
F_X(x)=
\begin{cases}
0, & x\leq10,\\[2mm]
1-\dfrac{10}{x}, & x>10.
\end{cases}
}
$$

Como comprobación,

$$
\lim_{x\to\infty}F_X(x)=1.
$$

---
**(c)** ¿Cuál es la probabilidad de que, de $6$ dispositivos de este tipo, al menos $3$ funcionen durante al menos $15$ horas?

Primero calculamos la probabilidad de que un dispositivo funcione durante al menos $15$ horas:

$$
P(X\geq15).
$$

Como $X$ es una variable aleatoria continua,

$$
P(X\geq15)=P(X>15),
$$

ya que

$$
P(X=15)=0.
$$

Entonces,

$$
P(X>15)
=
\int_{15}^{\infty}\frac{10}{x^2}\,dx.
$$

Integrando,

$$
P(X>15)
=
\left[
-\frac{10}{x}
\right]_{15}^{\infty}.
$$

Por tanto,

$$
P(X>15)
=
\frac{10}{15}
=
\frac23.
$$

Así, la probabilidad de éxito para cada dispositivo es

$$
\boxed{p=\frac23}.
$$

Definimos

$$
Y=
\text{número de dispositivos, de los 6, que duran al menos 15 horas}.
$$

Si suponemos que los tiempos de vida de los dispositivos son independientes y que todos tienen la misma distribución, entonces

$$
Y\sim\operatorname{Binomial}\left(6,\frac23\right).
$$

Queremos calcular

$$
P(Y\geq3).
$$

Utilizamos el complemento:

$$
P(Y\geq3)
=
1-P(Y<3).
$$

Por lo tanto,

$$
P(Y\geq3)
=
1-
\left[
P(Y=0)+P(Y=1)+P(Y=2)
\right].
$$

Para una variable aleatoria binomial,

$$
P(Y=k)
=
\binom{6}{k}
\left(\frac23\right)^k
\left(\frac13\right)^{6-k}.
$$

Así,

$$
P(Y\geq3)
=
1-
\sum_{k=0}^{2}
\binom{6}{k}
\left(\frac23\right)^k
\left(\frac13\right)^{6-k}.
$$

Para $k=0$,

$$
P(Y=0)
=
\binom60
\left(\frac23\right)^0
\left(\frac13\right)^6
=
\frac{1}{729}.
$$

Para $k=1$,

$$
P(Y=1)
=
\binom61
\left(\frac23\right)
\left(\frac13\right)^5
=
\frac{12}{729}.
$$

Para $k=2$,

$$
P(Y=2)
=
\binom62
\left(\frac23\right)^2
\left(\frac13\right)^4
=
\frac{60}{729}.
$$

Entonces,

$$
P(Y<3)
=
\frac{1+12+60}{729}
=
\frac{73}{729}.
$$

Por tanto,

$$
P(Y\geq3)
=
1-\frac{73}{729}.
$$

Finalmente,

$$
\boxed{
P(Y\geq3)=\frac{656}{729}
}
$$

y aproximadamente,

$$
\boxed{
P(Y\geq3)\approx0.8999.
}
$$

Es decir, la probabilidad de que al menos $3$ de los $6$ dispositivos funcionen durante al menos $15$ horas es aproximadamente

$$
\boxed{89.99\%}.
$$

---

¿Qué supuestos estás haciendo?

### Supuestos

Para utilizar una distribución binomial estamos suponiendo que:

1. Se consideran exactamente $6$ dispositivos.
2. Cada dispositivo tiene dos resultados relevantes: dura al menos $15$ horas o no dura al menos $15$ horas.
3. Todos los dispositivos tienen la misma probabilidad de durar al menos $15$ horas:

$$
p=\frac23.
$$

4. Los tiempos de vida de los dispositivos son independientes.

## 4)

Una estación de servicio recibe gasolina una vez por semana.

Si el volumen semanal de ventas, medido en **miles de galones**, es una variable aleatoria $X$ con función de densidad

$$
f(x)=
\begin{cases}
5(1-x)^4, & 0<x<1,\\
0, & \text{en otro caso},
\end{cases}
$$

¿cuál debe ser la capacidad del tanque para que la probabilidad de que se agote la gasolina en una semana determinada sea $0.01$?

Sea

$$
c=\text{capacidad del tanque, en miles de galones}.
$$

La gasolina se agotará cuando las ventas semanales sean mayores que la cantidad disponible en el tanque.

Por lo tanto, el evento de interés es

$$
\{X>c\}.
$$

El problema nos dice que queremos que

$$
P(X>c)=0.01.
$$

---

### Paso 1. Escribir la probabilidad utilizando la densidad

Como $X$ es una variable aleatoria continua,

$$
P(X>c)
=
\int_c^\infty f(x)\,dx.
$$

Pero la densidad solamente es distinta de cero cuando

$$
0<x<1.
$$

Por lo tanto,

$$
P(X>c)
=
\int_c^1 5(1-x)^4\,dx.
$$

Queremos que esta probabilidad sea $0.01$, así que

$$
\int_c^1 5(1-x)^4\,dx=0.01.
$$

---

### Paso 2. Calcular la integral

Queremos calcular

$$
\int_c^1 5(1-x)^4\,dx.
$$

Una primitiva es

$$
-(1-x)^5,
$$

pues

$$
\frac{d}{dx}
\left[-(1-x)^5\right]
=
5(1-x)^4.
$$

Entonces,

$$
\int_c^1 5(1-x)^4\,dx
=
\left[-(1-x)^5\right]_c^1.
$$

Evaluando,

$$
=
-(1-1)^5
-
\left[-(1-c)^5\right].
$$

Como

$$
(1-1)^5=0,
$$

obtenemos

$$
P(X>c)=(1-c)^5.
$$

Por lo tanto,

$$
(1-c)^5=0.01.
$$

---

### Paso 3. Despejar $c$

Tomamos quinta raíz:

$$
1-c=(0.01)^{1/5}.
$$

Entonces,

$$
c
=
1-(0.01)^{1/5}.
$$

Como

$$
0.01=10^{-2},
$$

tenemos

$$
(0.01)^{1/5}
=
10^{-2/5}.
$$

Numéricamente,

$$
(0.01)^{1/5}
\approx0.398107.
$$

Por tanto,

$$
c
\approx
1-0.398107
=
0.601893.
$$

Así,

$$
\boxed{c\approx0.6019}
$$

miles de galones.

---

### Paso 4. Interpretar las unidades

Recordemos que $X$ está medido en miles de galones.

Entonces,

$$
0.601893(1000)
\approx601.893.
$$

Por lo tanto, la capacidad del tanque debe ser aproximadamente

$$
\boxed{602\text{ galones}}.
$$

Más precisamente, el valor obtenido por el modelo continuo es

$$
\boxed{601.9\text{ galones aproximadamente}.}
$$

---

### Forma alternativa usando la función de distribución

La condición

$$
P(X>c)=0.01
$$

también puede escribirse como

$$
P(X\leq c)=0.99.
$$

Es decir,

$$
F_X(c)=0.99.
$$

Para $0<x<1$, la función de distribución es

$$
F_X(x)
=
\int_0^x5(1-t)^4\,dt.
$$

Calculando,

$$
F_X(x)
=
1-(1-x)^5.
$$

Por lo tanto,

$$
F_X(c)=0.99
$$

implica

$$
1-(1-c)^5=0.99.
$$

Entonces,

$$
(1-c)^5=0.01.
$$

Finalmente,

$$
\boxed{
c=1-(0.01)^{1/5}\approx0.6019.
}
$$

Por lo tanto, la capacidad requerida es aproximadamente

$$
\boxed{602\text{ galones}.}
$$

## 5)

Los trenes que se dirigen al destino $A$ llegan a una estación a intervalos de $15$ minutos, comenzando a las 7:00 a.m.:

$$
7{:}00,\quad 7{:}15,\quad 7{:}30,\quad 7{:}45,\quad 8{:}00,\ldots
$$

Mientras que los trenes que se dirigen al destino $B$ llegan también cada $15$ minutos, pero comenzando a las 7:05 a.m.:

$$
7{:}05,\quad 7{:}20,\quad 7{:}35,\quad 7{:}50,\quad 8{:}05,\ldots
$$

Un pasajero llega a la estación en un instante elegido uniformemente entre las 7:00 y las 8:00 a.m. y toma el **primer tren que llegue**.

**(a)** ¿Cuál es la probabilidad de que el pasajero vaya al destino $A$?


Como el tiempo de llegada del pasajero es uniforme, la probabilidad de caer en cierto intervalo es proporcional a la longitud de ese intervalo.

Observemos primero qué ocurre entre las 7:00 y las 7:15.

Los trenes llegan así:

$$
7{:}00\;(A),
\qquad
7{:}05\;(B),
\qquad
7{:}15\;(A).
$$

Si el pasajero llega entre las 7:00 y las 7:05, el siguiente tren que llegará será el de $B$.

Por tanto,

$$
7{:}00<t<7{:}05
\quad\Longrightarrow\quad
B.
$$

Este intervalo tiene longitud

$$
5\text{ minutos}.
$$

En cambio, si llega entre las 7:05 y las 7:15, el siguiente tren será el de $A$:

$$
7{:}05<t<7{:}15
\quad\Longrightarrow\quad
A.
$$

Este intervalo tiene longitud

$$
10\text{ minutos}.
$$

Por lo tanto, en cada periodo de $15$ minutos tenemos

$$
\underbrace{5\text{ min}}_{\text{toma }B}
\qquad
\underbrace{10\text{ min}}_{\text{toma }A}.
$$

Así,

$$
P(A)
=
\frac{10}{15}.
$$

Por tanto,

$$
\boxed{P(A)=\frac23}.
$$

Es decir,

$$
\boxed{P(A)\approx0.6667}.
$$

---

### Otra forma de verlo

Entre las 7:00 y las 8:00 hay $60$ minutos.

Los intervalos en los que el pasajero tomaría un tren hacia $A$ son

$$
(7{:}05,7{:}15),
$$

$$
(7{:}20,7{:}30),
$$

$$
(7{:}35,7{:}45),
$$

y

$$
(7{:}50,8{:}00).
$$

Cada uno tiene longitud $10$ minutos.

Por tanto, el tiempo total favorable es

$$
4(10)=40\text{ minutos}.
$$

Como el intervalo total tiene $60$ minutos,

$$
P(A)
=
\frac{40}{60}
=
\frac23.
$$


**(b)** ¿Cuál sería la probabilidad si el pasajero llegara uniformemente entre las 7:10 y las 8:10 a.m.?

Ahora el pasajero llega uniformemente en

$$
[7{:}10,8{:}10].
$$

Este intervalo tiene longitud

$$
60\text{ minutos}.
$$

Los intervalos en los que el siguiente tren será hacia $A$ son:

$$
(7{:}10,7{:}15),
$$

de longitud $5$ minutos,

$$
(7{:}20,7{:}30),
$$

de longitud $10$ minutos,

$$
(7{:}35,7{:}45),
$$

de longitud $10$ minutos,

$$
(7{:}50,8{:}00),
$$

de longitud $10$ minutos,

y

$$
(8{:}05,8{:}10),
$$

de longitud $5$ minutos.

Por lo tanto, el tiempo total favorable es

$$
5+10+10+10+5=40\text{ minutos}.
$$

Como el intervalo total tiene longitud $60$ minutos,

$$
P(A)
=
\frac{40}{60}.
$$

Por tanto,

$$
\boxed{P(A)=\frac23}.
$$

---

### Conclusión

En ambos casos,

$$
\boxed{P(\text{viajar a }A)=\frac23}.
$$

La razón es que en cada ciclo de $15$ minutos hay $10$ minutos durante los cuales el siguiente tren es hacia $A$ y $5$ minutos durante los cuales el siguiente tren es hacia $B$.

Además, los instantes exactos en los que llega un tren no afectan la probabilidad, ya que para una variable aleatoria continua,

$$
P(X=a)=0.
$$
