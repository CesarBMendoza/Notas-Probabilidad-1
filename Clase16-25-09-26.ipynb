## Motivación para variables aleatorias continuas

Tenemos un blanco de radio $1$ y estamos lanzando un dardo.

Supongamos que

$$
X=\text{distancia del punto donde cae el dardo al centro}.
$$

Queremos que $X$ se comporte de manera uniforme.

Como $X$ puede tomar valores entre $0$ y $1$, para aproximar esta variable aleatoria continua mediante variables aleatorias discretas, discretizamos el radio en $n$ partes:

$$
\frac{1}{n},
\frac{2}{n},
\frac{3}{n},
\ldots,
\frac{n-1}{n},
1.
$$

Por tanto, definimos

$$
R_X
=
\left\{
\frac{1}{n},
\frac{2}{n},
\frac{3}{n},
\ldots,
\frac{n-1}{n},
1
\right\}.
$$

Suponemos que

$$
X\sim U(R_X),
$$

es decir, $X$ es uniforme sobre $R_X$.

Entonces su función de probabilidad puntual es

$$
p_X(k)
=
\mathbb{P}[X=k]
=
\frac{1}{n},
\qquad
k\in R_X.
$$

Queremos aproximar una variable aleatoria continua mediante estas variables aleatorias discretas.

Notemos que

$$
\lim_{n\to\infty}\frac{1}{n}=0.
$$

Es decir,

$$
\lim_{n\to\infty}p_X(k)=0.
$$

Ahora consideremos el intervalo

$$
A=(a,b),
\qquad
0<a<b<1.
$$

Queremos calcular

$$
\mathbb{P}[X\in A].
$$

Como cada punto de $R_X$ tiene probabilidad $\frac1n$,

$$
\mathbb{P}[X\in A]
=
\frac{1}{n}
\left|A\cap R_X\right|.
$$

Queremos entonces contar cuántos puntos de la forma

$$
\frac{j}{n}
$$

pertenecen al intervalo $(a,b)$.

Continuando con el ejemplo anterior, los puntos del conjunto $R_X$ que pertenecen al intervalo

$$
A=(a,b)
$$

son aquellos para los cuales

$$
a<\frac{j}{n}<b.
$$

Equivalentemente,

$$
an<j<bn.
$$

Por tanto, aproximadamente hay

$$
\lfloor bn\rfloor-\lfloor an\rfloor-1
$$

puntos de $R_X$ dentro de $A$.

Así,

$$
\mathbb{P}[X\in A]
=
\frac{1}{n}
\left(
\lfloor bn\rfloor-\lfloor an\rfloor-1
\right).
$$

Es decir,

$$
\mathbb{P}[X\in A]
=
\frac{\lfloor bn\rfloor}{n}
-
\frac{\lfloor an\rfloor}{n}
-
\frac{1}{n}.
$$

Ahora tomamos el límite cuando $n\to\infty$.

Como

$$
\lim_{n\to\infty}
\frac{\lfloor bn\rfloor}{n}
=
b,
$$

$$
\lim_{n\to\infty}
\frac{\lfloor an\rfloor}{n}
=
a,
$$

y

$$
\lim_{n\to\infty}\frac1n=0,
$$

obtenemos

$$
\lim_{n\to\infty}
\mathbb{P}[X\in A]
=
b-a.
$$

Esto motiva que, para variables aleatorias continuas, podamos tener

$$
\mathbb{P}[X\in A]\neq0,
$$

aunque para cada punto individual

$$
\mathbb{P}[X=k]=0.
$$

## Definición

Diremos que $X$ es una **variable aleatoria continua** si existe una función

$$
f_X:\mathbb{R}\longrightarrow[0,\infty)
$$

tal que, para todo conjunto apropiado $A\subseteq\mathbb{R}$,

$$
\mathbb{P}[X\in A]
=
\int_A f_X(t)\,dt.
$$

A la función $f_X$ se le conoce como la **función de densidad de probabilidad** de $X$.

---

## Ejemplo

Consideremos la función

$$
f_X(x)
=
10\,\mathbf{1}_{[0,1/10]}(x),
$$

donde $\mathbf{1}_{[0,1/10]}$ es la función indicadora del intervalo $[0,1/10]$.

Equivalentemente,

$$
f_X(x)
=
\begin{cases}
0, & x<0,\\[4pt]
10, & 0\leq x\leq \dfrac{1}{10},\\[6pt]
0, & x>\dfrac{1}{10}.
\end{cases}
$$

Para verificar que $f_X$ es una función de densidad, debemos comprobar que la probabilidad total es $1$:

$$
\mathbb{P}[X\in\mathbb{R}]
=
\int_{-\infty}^{\infty}
f_X(t)\,dt.
$$

Usando la definición por partes,

$$
\int_{-\infty}^{\infty}f_X(t)\,dt
=
\int_{-\infty}^{0}f_X(t)\,dt
+
\int_{0}^{1/10}f_X(t)\,dt
+
\int_{1/10}^{\infty}f_X(t)\,dt.
$$

Como $f_X(t)=0$ fuera de $[0,1/10]$,

$$
=
\int_{0}^{1/10}10\,dt.
$$

Por tanto,

$$
=
10t\Big|_{0}^{1/10}
=
1.
$$

Así,

$$
\boxed{
\int_{-\infty}^{\infty}f_X(t)\,dt=1
}.
$$

Ahora evaluemos, por ejemplo,

$$
f_X\left(\frac{1}{100}\right).
$$

Como

$$
\frac{1}{100}\in\left[0,\frac{1}{10}\right],
$$

tenemos

$$
f_X\left(\frac{1}{100}\right)=10.
$$

Esto muestra que una función de densidad puede tomar valores mayores que $1$.

Lo importante no es que

$$
f_X(x)\leq1,
$$

sino que

$$
f_X(x)\geq0
$$

y que

$$
\int_{-\infty}^{\infty}f_X(t)\,dt=1.
$$

## Variables aleatorias continuas

Para una variable aleatoria continua,

$$
\mathbb{P}[X=a]=0
$$

para cualquier $a\in\mathbb{R}$.

Por esta razón, incluir o excluir los extremos de un intervalo no cambia su probabilidad:

$$
\mathbb{P}[a<X<b]
=
\mathbb{P}[a\leq X<b]
=
\mathbb{P}[a<X\leq b]
=
\mathbb{P}[a\leq X\leq b].
$$

---

## Función de distribución acumulada

La función de distribución acumulada de $X$ es

$$
F_X(z)
=
\mathbb{P}[X\leq z].
$$

Si $X$ tiene densidad $f_X$, entonces

$$
F_X(z)
=
\int_{-\infty}^{z}f_X(t)\,dt.
$$

Además, por el Teorema Fundamental del Cálculo,

$$
\frac{d}{dz}F_X(z)
=
\frac{d}{dz}
\int_{-\infty}^{z}f_X(t)\,dt
=
f_X(z),
$$

siempre que se cumplan las condiciones necesarias para poder derivar.

Es decir,

$$
\boxed{
F_X'(z)=f_X(z)
}.
$$

Para una variable aleatoria continua,

$$
\mathbb{P}[a<X\leq b]
=
\int_a^b f_X(t)\,dt.
$$

Podemos escribir

$$
\int_a^b f_X(t)\,dt
=
\int_{-\infty}^{b}f_X(t)\,dt
-
\int_{-\infty}^{a}f_X(t)\,dt.
$$

Recordando que

$$
F_X(z)
=
\int_{-\infty}^{z}f_X(t)\,dt,
$$

obtenemos

$$
\boxed{
\mathbb{P}[a<X\leq b]
=
F_X(b)-F_X(a)
}.
$$

## Ejemplo

Sea

$$
D=\text{diámetro de un árbol, medido en cm}.
$$

Supongamos que su función de densidad tiene la forma

$$
f_D(d)
=
k\,d\,\mathbf{1}_{[0,100]}(d).
$$

Es decir,

$$
f_D(d)
=
\begin{cases}
kd, & 0\leq d\leq100,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

Queremos determinar el valor de $k$.

Como $f_D$ debe ser una función de densidad,

$$
\mathbb{P}[D\in\mathbb{R}]
=
\int_{-\infty}^{\infty}f_D(d)\,dd
=
1.
$$

Como la densidad es cero fuera de $[0,100]$,

$$
\int_{0}^{100}kd\,dd
=
1.
$$

Sacamos la constante $k$:

$$
k\int_0^{100}d\,dd
=
1.
$$

Entonces

$$
k
\left[
\frac{d^2}{2}
\right]_{0}^{100}
=
1.
$$

Por tanto,

$$
k\frac{100^2}{2}
=
1.
$$

Como

$$
\frac{100^2}{2}
=
5000,
$$

tenemos

$$
5000k=1.
$$

Así,

$$
\boxed{
k=\frac{1}{5000}
}.
$$

La función de densidad queda

$$
\boxed{
f_D(d)
=
\frac{d}{5000}
\mathbf{1}_{[0,100]}(d)
}.
$$

### ¿Cuál es la probabilidad de que el diámetro de un árbol esté entre $40$ y $60$ cm?

Queremos calcular

$$
\mathbb{P}[40\leq D\leq60].
$$

Como $D$ es continua,

$$
\mathbb{P}[40\leq D\leq60]
=
\int_{40}^{60}f_D(t)\,dt.
$$

Sustituyendo la densidad,

$$
=
\int_{40}^{60}
\frac{t}{5000}\,dt.
$$

Entonces

$$
=
\frac{1}{5000}
\left[
\frac{t^2}{2}
\right]_{40}^{60}.
$$

Por tanto,

$$
=
\frac{60^2-40^2}{10000}.
$$

Como

$$
60^2=3600
$$

y

$$
40^2=1600,
$$

obtenemos

$$
=
\frac{3600-1600}{10000}
=
\frac{2000}{10000}.
$$

Finalmente,

$$
\boxed{
\mathbb{P}[40\leq D\leq60]
=
\frac15
=
0.2
}.
$$

Por tanto, la probabilidad de que el diámetro esté entre $40$ y $60$ cm es del

$$
20\%.
$$
