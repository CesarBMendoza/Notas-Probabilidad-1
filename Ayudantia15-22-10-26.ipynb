
# Ejercicios de esperanza matemática

## Ejercicio 1. Juego con una baraja de 52 cartas

De un mazo de 52 cartas se reparte una carta a ti y dos cartas a tu oponente, sin reemplazo.

Las reglas del juego son las siguientes:

1. Si tu carta es un **as**, el oponente debe pagarte \$10.
2. Si tu carta es un **rey**, el oponente debe pagarte \$5.
3. En ambos casos anteriores, la ganancia es independiente de las cartas que tenga tu oponente.
4. Si tu carta no es un as ni un rey, pero es roja y ninguna de las dos cartas de tu oponente es roja, el oponente debe pagarte \$1.
5. En cualquier otro caso, debes pagarle \$1 a tu oponente.

**Pregunta:** ¿Cuál es la ganancia esperada de este juego?

### Solución

Definamos primero nuestra variable aleatoria.

Sea $X$ la variable aleatoria que representa la ganancia obtenida después de repartir las tres cartas.

De acuerdo con las reglas del juego,

$$
X=
\begin{cases}
10, & \text{si nuestra carta es un as},\\[4pt]
5, & \text{si nuestra carta es un rey},\\[4pt]
1, & \substack{\text{si nuestra carta es roja, no es as ni rey,}\\
\text{y las dos cartas del oponente son negras}},\\[4pt]
-1, & \text{en cualquier otro caso}.
\end{cases}
$$

Por lo tanto, el recorrido de nuestra variable aleatoria es

$$
\boxed{R_X=\{10,5,1,-1\}}
$$

Recordemos que, para una variable aleatoria discreta,

$$
\mathbb{E}[X]=\sum_{x\in R_X}xP(X=x).
$$

Por lo tanto, necesitamos calcular las probabilidades

$$
P(X=10),\quad P(X=5),\quad P(X=1),\quad P(X=-1).
$$

### Paso 1. Calcular $P(X=10)$

El evento $\{X=10\}$ ocurre cuando nuestra carta es un as.

Como existen 4 ases en una baraja de 52 cartas,

$$
\begin{aligned}
P(X=10)
&=P(\text{nuestra carta es un as})\\
&=\frac{4}{52}\\
&=\frac{1}{13}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{P(X=10)=\frac{1}{13}}
$$

### Paso 2. Calcular $P(X=5)$

El evento $\{X=5\}$ ocurre cuando nuestra carta es un rey.

Como existen 4 reyes en una baraja de 52 cartas,

$$
\begin{aligned}
P(X=5)
&=P(\text{nuestra carta es un rey})\\
&=\frac{4}{52}\\
&=\frac{1}{13}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{P(X=5)=\frac{1}{13}}
$$

### Paso 3. Calcular $P(X=1)$

El evento $\{X=1\}$ ocurre cuando se cumplen simultáneamente dos condiciones:

- Nuestra carta es roja, pero no es un as ni un rey.
- Las dos cartas de nuestro oponente son negras.

Definamos los eventos

$$
A=\{\text{nuestra carta es roja y no es as ni rey}\}
$$

y

$$
B=\{\text{las dos cartas del oponente son negras}\}.
$$

Entonces,

$$
P(X=1)=P(A\cap B).
$$

Por la fórmula de probabilidad condicionada,

$$
P(A\cap B)=P(B\mid A)P(A).
$$

**Primero, calculemos $P(A)$.**

En una baraja de 52 cartas existen 26 cartas rojas.

De estas:

- 2 son ases.
- 2 son reyes.

Por lo tanto, tenemos

$$
26-2-2=22
$$

cartas rojas que no son ases ni reyes.

Así,

$$
\boxed{
P(A)=\frac{22}{52}
}
$$

**Ahora calculemos $P(B\mid A)$.**

Sabemos que nuestra carta es roja y que ya fue retirada de la baraja.

Por lo tanto, quedan 51 cartas, de las cuales 26 son negras.

La probabilidad de que la primera carta del oponente sea negra es

$$
\frac{26}{51}.
$$

Como las cartas se reparten sin reemplazo, si la primera carta del oponente es negra, quedan 25 cartas negras entre las 50 restantes.

Por lo tanto,

$$
P(B\mid A)
=
\frac{26}{51}\frac{25}{50}.
$$

También podemos calcular esta probabilidad mediante combinaciones:

$$
P(B\mid A)
=
\frac{\binom{26}{2}}{\binom{51}{2}}.
$$

Ambas expresiones son equivalentes.

Finalmente,

$$
\begin{aligned}
P(X=1)
&=P(A\cap B)\\
&=P(B\mid A)P(A)\\
&=
\left(\frac{26}{51}\frac{25}{50}\right)
\left(\frac{22}{52}\right)\\
&=\frac{11}{102}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=1)=\frac{11}{102}
}
$$

### Paso 4. Calcular $P(X=-1)$

El evento $\{X=-1\}$ ocurre cuando no se cumple ninguna de las condiciones para ganar dinero.

Como los eventos asociados a las ganancias de \$10, \$5 y \$1 son ajenos, tenemos

$$
P(X=-1)
=
1-P(X=10)-P(X=5)-P(X=1).
$$

Sustituyendo,

$$
\begin{aligned}
P(X=-1)
&=
1-\frac{4}{52}-\frac{4}{52}-\frac{11}{102}\\
&=
1-\frac{2}{13}-\frac{11}{102}\\
&=
\frac{979}{1326}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=-1)=\frac{979}{1326}
}
$$

### Paso 5. Determinar la función de masa de probabilidad

Con los resultados anteriores, nuestra función de masa es

$$
P_X(k)=
\begin{cases}
\frac{1}{13}, & \text{si }k=10,\\[4pt]
\frac{1}{13}, & \text{si }k=5,\\[4pt]
\frac{11}{102}, & \text{si }k=1,\\[4pt]
\frac{979}{1326}, & \text{si }k=-1,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

### Paso 6. Calcular la esperanza matemática

Por definición,

$$
\mathbb{E}[X]
=
\sum_{x\in R_X}xP(X=x).
$$

Por lo tanto,

$$
\begin{aligned}
\mathbb{E}[X]
&=
10P(X=10)+5P(X=5)\\
&\quad+P(X=1)-P(X=-1).
\end{aligned}
$$

Sustituyendo,

$$
\begin{aligned}
\mathbb{E}[X]
&=
10\left(\frac{1}{13}\right)
+
5\left(\frac{1}{13}\right)\\
&\quad+
\frac{11}{102}
-
\frac{979}{1326}\\
&=
\frac{347}{663}\\
&\approx0.5234.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]=\frac{347}{663}\approx0.5234
}
$$

### Interpretación

Esto significa que, en promedio, ganaremos aproximadamente **52.34 centavos por partida**.

Aunque existe una probabilidad considerable de perder \$1, las ganancias de \$10 y \$5 compensan dichas pérdidas.

Por lo tanto, desde el punto de vista de la esperanza matemática, el juego es favorable para nosotros.

## Ejercicio 2. Juego con un dado

Un jugador participa en un juego que consiste en lanzar un dado equilibrado de 6 caras.

Dependiendo del resultado, recibe las siguientes ganancias o pérdidas:

- Si obtiene un 6, gana \$8.
- Si obtiene un 5 o un 4, gana \$3.
- Si obtiene un 1, 2 o 3, pierde \$2.

**Pregunta:** Calcular la ganancia esperada del jugador.

### Solución

Definamos $X$ como la variable aleatoria que representa la ganancia obtenida al lanzar el dado.

Entonces,

$$
X=
\begin{cases}
8, & \text{si sale }6,\\
3, & \text{si sale }4\text{ o }5,\\
-2, & \text{si sale }1,2\text{ o }3.
\end{cases}
$$

Por lo tanto,

$$
\boxed{R_X=\{8,3,-2\}}
$$

### Paso 1. Calcular la función de masa de probabilidad

Como el dado es equilibrado, todos los resultados tienen probabilidad $1/6$.

**Caso 1. Calcular $P(X=8)$.**

Esto ocurre únicamente cuando sale un 6.

Por lo tanto,

$$
\boxed{
P(X=8)=\frac{1}{6}
}
$$

**Caso 2. Calcular $P(X=3)$.**

Esto ocurre cuando sale un 4 o un 5.

Como estos eventos son ajenos,

$$
\begin{aligned}
P(X=3)
&=P(\text{sale }4\text{ o }5)\\
&=\frac{1}{6}+\frac{1}{6}\\
&=\frac{2}{6}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=3)=\frac{2}{6}
}
$$

**Caso 3. Calcular $P(X=-2)$.**

Esto ocurre cuando sale un 1, 2 o 3.

Por lo tanto,

$$
\begin{aligned}
P(X=-2)
&=P(\text{sale }1,2\text{ o }3)\\
&=\frac{1}{6}+\frac{1}{6}+\frac{1}{6}\\
&=\frac{3}{6}.
\end{aligned}
$$

Así,

$$
\boxed{
P(X=-2)=\frac{3}{6}
}
$$

La función de masa de probabilidad es

$$
P_X(k)=
\begin{cases}
\frac{1}{6}, & \text{si }k=8,\\[4pt]
\frac{2}{6}, & \text{si }k=3,\\[4pt]
\frac{3}{6}, & \text{si }k=-2,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

### Paso 2. Calcular la esperanza matemática

Por definición,

$$
\mathbb{E}[X]
=
\sum_{x\in R_X}xP(X=x).
$$

Sustituyendo,

$$
\begin{aligned}
\mathbb{E}[X]
&=
8P(X=8)+3P(X=3)-2P(X=-2)\\
&=
8\left(\frac{1}{6}\right)
+
3\left(\frac{2}{6}\right)
-
2\left(\frac{3}{6}\right)\\
&=
\frac{8}{6}+\frac{6}{6}-\frac{6}{6}\\
&=
\frac{8}{6}\\
&=
\frac{4}{3}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]=\frac{4}{3}\approx1.3333
}
$$

### Interpretación

Esto significa que, en promedio, el jugador ganará aproximadamente **\$1.33 por lanzamiento**.

Es importante observar que ganar \$1.33 no es un resultado posible en una sola partida.

Sin embargo, si el juego se repite muchas veces, la ganancia promedio por partida tenderá a aproximarse a este valor.

## Ejercicio 3. Juego con una baraja de UNO

Se propone el siguiente juego.

Tenemos una baraja tradicional de UNO de 108 cartas, bien mezclada.

Se te reparte una carta y se reparten dos cartas a tu oponente, sin reemplazo.

Las reglas del juego son las siguientes:

1. Si tu carta es un **comodín +4**, el oponente debe pagarte \$12.
2. Si tu carta es un **0**, el oponente debe pagarte \$7.
3. Si tu carta es roja, pero no es un 0, y ninguna de las dos cartas del oponente es roja, el oponente debe pagarte \$2.
4. En cualquier otro caso, debes pagarle \$5 al oponente.

**Pregunta:** ¿Cuál es la ganancia esperada de este juego?

### Solución

Primero, recordemos algunas características de una baraja tradicional de UNO de 108 cartas:

- Tiene 4 comodines +4.
- Tiene 4 cartas con el número 0, una por cada color.
- Tiene 25 cartas rojas.
- De las 25 cartas rojas, solamente una es un 0.
- Por lo tanto, existen 24 cartas rojas que no son un 0.

### Paso 1. Definir la variable aleatoria

Sea $X$ la variable aleatoria que representa la ganancia obtenida después de repartir las tres cartas.

De acuerdo con las reglas,

$$
X=
\begin{cases}
12, & \text{si nuestra carta es un comodín }+4,\\[4pt]
7, & \text{si nuestra carta es un }0,\\[4pt]
2, & \substack{\text{si nuestra carta es roja, no es }0,\\
\text{y las dos cartas del oponente no son rojas}},\\[4pt]
-5, & \text{en cualquier otro caso}.
\end{cases}
$$

Por lo tanto,

$$
\boxed{
R_X=\{12,7,2,-5\}
}
$$

### Paso 2. Calcular $P(X=12)$

El evento $\{X=12\}$ ocurre cuando nuestra carta es un comodín +4.

Como existen 4 cartas de este tipo en la baraja,

$$
\begin{aligned}
P(X=12)
&=
P(\text{nuestra carta es un comodín }+4)\\
&=
\frac{4}{108}\\
&=
\frac{1}{27}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=12)=\frac{4}{108}
}
$$

### Paso 3. Calcular $P(X=7)$

El evento $\{X=7\}$ ocurre cuando nuestra carta es un 0.

Como existen 4 cartas con el número 0,

$$
\begin{aligned}
P(X=7)
&=P(\text{nuestra carta es un }0)\\
&=\frac{4}{108}\\
&=\frac{1}{27}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=7)=\frac{4}{108}
}
$$

### Paso 4. Calcular $P(X=2)$

El evento $\{X=2\}$ ocurre cuando se cumplen simultáneamente las siguientes condiciones:

- Nuestra carta es roja y no es un 0.
- Ninguna de las dos cartas del oponente es roja.

Definamos los eventos

$$
A=\{\text{nuestra carta es roja y no es un }0\}
$$

y

$$
B=\{\text{las dos cartas del oponente no son rojas}\}.
$$

Por lo tanto,

$$
P(X=2)=P(A\cap B).
$$

Utilizando la fórmula de probabilidad condicionada,

$$
P(A\cap B)=P(B\mid A)P(A).
$$

**Primero calculemos $P(A)$.**

Sabemos que existen 25 cartas rojas, de las cuales una es un 0.

Por lo tanto, existen

$$
25-1=24
$$

cartas rojas que no son un 0.

Así,

$$
\boxed{
P(A)=\frac{24}{108}
}
$$

**Ahora calculemos $P(B\mid A)$.**

Como nuestra carta es roja y ya fue retirada de la baraja, quedan 107 cartas.

Originalmente había

$$
108-25=83
$$

cartas que no son rojas.

Por lo tanto, la probabilidad de que la primera carta del oponente no sea roja es

$$
\frac{83}{107}.
$$

Como las cartas se reparten sin reemplazo, si la primera carta del oponente no es roja, quedan 82 cartas que no son rojas entre las 106 restantes.

Así,

$$
\boxed{
P(B\mid A)
=
\frac{83}{107}\frac{82}{106}
}
$$

Por lo tanto,

$$
\begin{aligned}
P(X=2)
&=
P(A\cap B)\\
&=
P(B\mid A)P(A)\\
&=
\frac{24}{108}
\frac{83}{107}
\frac{82}{106}\\
&=
\frac{6806}{51039}\\
&\approx0.13335.
\end{aligned}
$$

Concluimos que

$$
\boxed{
P(X=2)\approx0.13335
}
$$

### Paso 5. Calcular $P(X=-5)$

El evento $\{X=-5\}$ ocurre cuando no se cumple ninguna de las condiciones para ganar \$12, \$7 o \$2.

Como estos eventos son ajenos,

$$
P(X=-5)
=
1-P(X=12)-P(X=7)-P(X=2).
$$

Sustituyendo,

$$
\begin{aligned}
P(X=-5)
&=
1-\frac{4}{108}-\frac{4}{108}\\
&\quad-
\frac{24}{108}\frac{83}{107}\frac{82}{106}\\
&\approx0.79258.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X=-5)\approx0.79258
}
$$

### Paso 6. Determinar la función de masa de probabilidad

Con los resultados anteriores, tenemos

$$
P_X(k)=
\begin{cases}
\frac{4}{108}, & \text{si }k=12,\\[4pt]
\frac{4}{108}, & \text{si }k=7,\\[4pt]
\frac{6806}{51039}, & \text{si }k=2,\\[4pt]
\frac{121357}{153117}, & \text{si }k=-5,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

### Paso 7. Calcular la esperanza matemática

Por definición,

$$
\mathbb{E}[X]
=
\sum_{x\in R_X}xP(X=x).
$$

Como

$$
R_X=\{12,7,2,-5\},
$$

tenemos

$$
\begin{aligned}
\mathbb{E}[X]
&=
12P(X=12)+7P(X=7)\\
&\quad+
2P(X=2)-5P(X=-5).
\end{aligned}
$$

Sustituyendo,

$$
\begin{aligned}
\mathbb{E}[X]
&=
12\left(\frac{4}{108}\right)
+
7\left(\frac{4}{108}\right)\\
&\quad+
2\left(\frac{6806}{51039}\right)
-
5\left(\frac{121357}{153117}\right)\\
&\approx
0.44444+0.25926+0.26670-3.96288\\
&\approx-2.99248.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]\approx-2.99
}
$$

### Interpretación

Esto significa que, en promedio, **perderemos aproximadamente \$2.99 por partida**.

Aunque podemos obtener ganancias de \$12, \$7 o \$2, la probabilidad de perder \$5 es considerablemente mayor.

En particular,

$$
P(X=-5)\approx0.79258.
$$

Es decir, tenemos aproximadamente un $79.26\%$ de probabilidad de perder \$5.

Por lo tanto, desde el punto de vista de la esperanza matemática, este juego es desfavorable para nosotros.

### Observación

Este ejercicio muestra que no basta con conocer las cantidades que podemos ganar o perder para determinar si un juego es favorable.

También necesitamos conocer las probabilidades asociadas a cada resultado.

En este ejemplo, aunque la ganancia máxima es de \$12 y la pérdida máxima es de \$5, la esperanza matemática es negativa.
