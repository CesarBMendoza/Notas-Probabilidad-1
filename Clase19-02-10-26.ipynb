## Transformaciones de variables aleatorias

Sea $X$ una variable aleatoria. Podemos construir una nueva variable aleatoria aplicando una función $g$, es decir,

$$
Y=g(X).
$$

### Ejemplo

Supongamos que $X$ representa la distancia al centro de una región circular y que $Y$ representa el número de puntos asignados dependiendo de la distancia.

Entonces podemos definir

$$
g(x)=
\begin{cases}
20, & 0\leq x\leq 2,\\
10, & 2<x\leq 5,\\
5, & 5<x\leq R,\\
0, & \text{en otro caso}.
\end{cases}
$$

Entonces,

$$
Y=g(X).
$$

Esta situación puede representarse mediante la composición

$$
\Omega
\xrightarrow{X}
\mathbb{R}
\xrightarrow{g}
\mathbb{R}.
$$

Es decir,

$$
Y=g\circ X.
$$

Para cada $\omega\in\Omega$,

$$
Y(\omega)=g(X(\omega)).
$$

Además, para cualquier conjunto $B\subseteq\mathbb{R}$, sabemos que

$$
Y^{-1}(B)=X^{-1}(g^{-1}(B)).
$$

Esto se debe a que

$$
\begin{aligned}
Y^{-1}(B)
&=\{\omega\in\Omega:Y(\omega)\in B\}\\
&=\{\omega\in\Omega:g(X(\omega))\in B\}\\
&=\{\omega\in\Omega:X(\omega)\in g^{-1}(B)\}\\
&=X^{-1}(g^{-1}(B)).
\end{aligned}
$$

### Definición: Función medible

Una función

$$
g:\mathbb{R}\longrightarrow\mathbb{R}
$$

se llama **medible** si y solo si

$$
\forall B\in\mathcal{B}(\mathbb{R}),
\qquad
g^{-1}(B)\in\mathcal{B}(\mathbb{R}).
$$

Donde $\mathcal{B}(\mathbb{R})$ representa la $\sigma$-álgebra de Borel de $\mathbb{R}$.

## Lema

Sea $X$ una variable aleatoria y sea

$$
g:\mathbb{R}\longrightarrow\mathbb{R}
$$

una función medible.

Entonces,

$$
Y=g(X)
$$

es una variable aleatoria.

### Demostración

Para $B\in\mathcal{B}(\mathbb{R})$, tenemos que

$$
\begin{aligned}
Y^{-1}(B)
&=(g\circ X)^{-1}(B)\\
&=X^{-1}(g^{-1}(B)).
\end{aligned}
$$

Como $g$ es medible,

$$
g^{-1}(B)\in\mathcal{B}(\mathbb{R}).
$$

Por otra parte, como $X$ es una variable aleatoria,

$$
X^{-1}(g^{-1}(B))\in\mathcal{F}.
$$

Es decir,

$$
Y^{-1}(B)\in\mathcal{F}.
$$

Por lo tanto,

$$
\boxed{Y=g\circ X\text{ es una variable aleatoria}.}
$$

---

## Lema: Ejemplos de funciones medibles

Sea

$$
g:\mathbb{R}\longrightarrow\mathbb{R}.
$$

Si se cumple alguna de las siguientes condiciones:

1. $g$ es continua.
2. $g$ es monótona.
3. $g=\mathbb{1}_A$, donde $A\in\mathcal{B}(\mathbb{R})$.

Entonces $g$ es medible.

## Ejemplo: Estandarización de una distribución normal

Supongamos que

$$
X\sim N(\mu,\sigma^2),
$$

donde $\sigma>0$.

Si definimos

$$
Z=\frac{X-\mu}{\sigma},
$$

entonces

$$
\boxed{Z\sim N(0,1).}
$$

En efecto, si tomamos

$$
g(x)=\frac{x-\mu}{\sigma},
$$

observamos que $g$ es una función continua.

Por lo tanto, $g$ es medible y, en consecuencia,

$$
Z=g\circ X
$$

es una variable aleatoria.

Además, si $\sigma>0$, la función

$$
g(x)=\frac{x-\mu}{\sigma}
$$

es estrictamente creciente y tiene inversa.

Es decir,

$$
\boxed{g^{-1}(z)=\mu+\sigma z.}
$$

## Transformación de densidades

Ahora queremos contestar una pregunta diferente:

**Si conocemos la densidad de $X$, ¿cómo podemos encontrar la densidad de $Y=g(X)$?**

Observemos que, para utilizar la fórmula de cambio de variable, necesitamos que $g$ sea invertible en el intervalo donde $X$ toma valores.

### Teorema: Transformación de densidades

Sea $X$ una variable aleatoria continua con función de densidad $f_X$.

Supongamos que los valores de $X$ pertenecen a un intervalo $I$.

Sea

$$
g:I\longrightarrow\mathbb{R}
$$

una función diferenciable y estrictamente monótona en $I$.

Supongamos además que su inversa $g^{-1}$ es diferenciable en el interior de $g(I)$.

Entonces, si la densidad de probabilidad de $X$ es $f_X(x)$, la densidad de

$$
Y=g(X)
$$

está dada por

$$
\boxed{
f_Y(y)=
f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|
}
$$

para $y$ en el interior de $g(I)$.

Además,

$$
f_Y(y)=0,
\qquad y\notin g(I).
$$

Por lo tanto, podemos escribir

$$
\boxed{
f_Y(y)=
f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|
\mathbb{1}_{g(I)}(y).
}
$$

La expresión se entiende en los puntos donde existe la derivada de la inversa, y los valores de la densidad en los extremos del intervalo no afectan la distribución.

---

### Observación

Una función puede no ser invertible en todo $\mathbb{R}$, pero sí puede ser invertible cuando la restringimos a un intervalo de $\mathbb{R}$.

Por ejemplo, la función

$$
g(x)=x^2
$$

no es invertible en todo $\mathbb{R}$.

Sin embargo, sí es invertible cuando se restringe a cualquiera de los intervalos

$$
(-\infty,0]
\qquad\text{o}\qquad
[0,\infty).
$$

# Demostración del teorema de transformación de densidades

Consideraremos dos casos.

## Caso 1: $g$ es estrictamente creciente

Como $g$ es estrictamente creciente, si

$$
g(X)\leq y,
$$

entonces

$$
X\leq g^{-1}(y).
$$

Por lo tanto,

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(g(X)\leq y)\\
&=P(X\leq g^{-1}(y))\\
&=F_X(g^{-1}(y)).
\end{aligned}
$$

Es decir,

$$
\boxed{
F_Y(y)=F_X(g^{-1}(y)).
}
$$

Derivando respecto de $y$, obtenemos

$$
\begin{aligned}
f_Y(y)
&=\frac{dF_Y(y)}{dy}\\
&=\frac{d}{dy}\left(F_X(g^{-1}(y))\right)\\
&=F_X'(g^{-1}(y))
\frac{d}{dy}g^{-1}(y).
\end{aligned}
$$

Como

$$
F_X'(x)=f_X(x),
$$

tenemos que

$$
f_Y(y)=
f_X(g^{-1}(y))
\frac{d}{dy}g^{-1}(y).
$$

Además, como $g$ es estrictamente creciente, su inversa $g^{-1}$ también es estrictamente creciente.

Por lo tanto, donde la derivada existe,

$$
\frac{d}{dy}g^{-1}(y)\geq 0.
$$

En consecuencia,

$$
\boxed{
f_Y(y)=
f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|.
}
$$

---

## Caso 2: $g$ es estrictamente decreciente

Como $g$ es estrictamente decreciente, si

$$
g(X)\leq y,
$$

entonces

$$
g^{-1}(y)\leq X.
$$

Por lo tanto,

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(g(X)\leq y)\\
&=P(g^{-1}(y)\leq X)\\
&=1-F_X(g^{-1}(y)).
\end{aligned}
$$

Es decir,

$$
\boxed{
F_Y(y)=1-F_X(g^{-1}(y)).
}
$$

Continuando con el caso en que $g$ es estrictamente decreciente, tenemos que

$$
F_Y(y)=1-F_X(g^{-1}(y)).
$$

Derivando respecto de $y$, obtenemos

$$
\begin{aligned}
f_Y(y)
&=\frac{dF_Y(y)}{dy}\\
&=\frac{d}{dy}
\left(1-F_X(g^{-1}(y))\right)\\
&=-\frac{d}{dy}
\left(F_X(g^{-1}(y))\right)\\
&=-F_X'(g^{-1}(y))
\frac{d}{dy}g^{-1}(y)\\
&=-f_X(g^{-1}(y))
\frac{d}{dy}g^{-1}(y).
\end{aligned}
$$

Es decir,

$$
f_Y(y)=
-f_X(g^{-1}(y))
\frac{d}{dy}g^{-1}(y).
$$

Como $g$ es estrictamente decreciente, su inversa $g^{-1}$ también es estrictamente decreciente.

Por lo tanto, donde la derivada existe,

$$
\frac{d}{dy}g^{-1}(y)\leq 0.
$$

Así,

$$
-\frac{d}{dy}g^{-1}(y)
=
\left|
\frac{d}{dy}g^{-1}(y)
\right|.
$$

Por lo tanto,

$$
\boxed{
f_Y(y)=
f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|.
}
$$

De esta manera, en ambos casos obtenemos la misma expresión.

### Fórmula general

$$
\boxed{
f_Y(y)=
f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|,
\qquad y\in g(I).
}
$$

# Ejemplo 1: Transformación de una variable uniforme

Supongamos que

$$
X\sim U[0,1]
$$

y definimos

$$
Y=X^2.
$$

Tomando

$$
g(x)=x^2,
$$

queremos encontrar la densidad de

$$
Y=g(X)=X^2.
$$

Recordemos que, si

$$
X\sim U[a,b],
$$

entonces

$$
f_X(x)=
\frac{1}{b-a}\mathbb{1}_{(a,b)}(x).
$$

En nuestro caso,

$$
f_X(x)=\mathbb{1}_{(0,1)}(x).
$$

Aunque

$$
g(x)=x^2
$$

no es invertible en todo $\mathbb{R}$, sí lo es en

$$
I=[0,1].
$$

Además,

$$
g^{-1}(y)=\sqrt{y}.
$$

De donde

$$
\boxed{
\frac{d}{dy}g^{-1}(y)=\frac{1}{2\sqrt{y}}.
}
$$

Aplicando el teorema de transformación de densidades, tenemos que

$$
f_Y(y)=
f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|.
$$

Continuando con el ejemplo anterior,

$$
X\sim U[0,1],
\qquad
Y=X^2.
$$

Tenemos que

$$
g^{-1}(y)=\sqrt{y}
$$

y

$$
\frac{d}{dy}g^{-1}(y)=\frac{1}{2\sqrt{y}}.
$$

Por lo tanto,

$$
\begin{aligned}
f_Y(y)
&=f_X(g^{-1}(y))
\left|
\frac{d}{dy}g^{-1}(y)
\right|\\
&=f_X(\sqrt{y})
\left|
\frac{1}{2\sqrt{y}}
\right|\\
&=\frac{1}{2\sqrt{y}},
\qquad y\in(0,1).
\end{aligned}
$$

Es decir,

$$
\boxed{
f_Y(y)=
\frac{1}{2\sqrt{y}}
\mathbb{1}_{(0,1)}(y).
}
$$

# Ejemplo 2: Transformación de una uniforme en $[-3,3]$

Supongamos que

$$
X\sim U[-3,3]
$$

y definimos

$$
Y=X^2.
$$

Observemos que

$$
g(x)=x^2
$$

no es invertible en todo el intervalo

$$
[-3,3].
$$

Sin embargo, sí lo es en cada uno de los intervalos

$$
[-3,0]
\qquad\text{y}\qquad
[0,3].
$$

Para encontrar la densidad de $Y$, utilizaremos su función de distribución acumulada.

Observemos que, para $y\geq 0$,

$$
X^2\leq y
$$

si y solo si

$$
-\sqrt{y}\leq X\leq\sqrt{y}.
$$

Con esto, para $0\leq y\leq 9$,

$$
\begin{aligned}
F_Y(y)
&=P(Y\leq y)\\
&=P(X^2\leq y)\\
&=P(-\sqrt{y}\leq X\leq\sqrt{y})\\
&=F_X(\sqrt{y})-F_X(-\sqrt{y}).
\end{aligned}
$$

Es decir,

$$
\boxed{
F_Y(y)=
F_X(\sqrt{y})-F_X(-\sqrt{y}),
\qquad 0\leq y\leq 9.
}
$$

Además, como

$$
X\sim U[-3,3],
$$

sabemos que

$$
f_X(x)=
\frac{1}{6}\mathbb{1}_{(-3,3)}(x).
$$

Ahora, derivando la función de distribución acumulada para $0<y<9$, tenemos que

$$
\begin{aligned}
f_Y(y)
&=\frac{dF_Y(y)}{dy}\\
&=\frac{d}{dy}
\left(
F_X(\sqrt{y})-F_X(-\sqrt{y})
\right)\\
&=\frac{d}{dy}F_X(\sqrt{y})
-\frac{d}{dy}F_X(-\sqrt{y})\\
&=F_X'(\sqrt{y})
\left(\frac{1}{2\sqrt{y}}\right)
-F_X'(-\sqrt{y})
\left(-\frac{1}{2\sqrt{y}}\right)\\
&=f_X(\sqrt{y})
\left(\frac{1}{2\sqrt{y}}\right)
+f_X(-\sqrt{y})
\left(\frac{1}{2\sqrt{y}}\right).
\end{aligned}
$$

Continuando con el ejemplo

$$
X\sim U[-3,3],
\qquad
Y=X^2,
$$

tenemos que

$$
f_Y(y)=
\frac{1}{2\sqrt{y}}
\left(
f_X(\sqrt{y})+f_X(-\sqrt{y})
\right)
$$

para $y\in(0,9)$.

Por lo tanto,

$$
f_Y(y)=
\frac{1}{2\sqrt{y}}
\left(
f_X(\sqrt{y})+f_X(-\sqrt{y})
\right)
\mathbb{1}_{(0,9)}(y).
$$

Como

$$
X\sim U[-3,3],
$$

sabemos que

$$
f_X(x)=
\frac{1}{6}\mathbb{1}_{(-3,3)}(x).
$$

Entonces, para $y\in(0,9)$,

$$
f_X(\sqrt{y})=
\frac{1}{6}
$$

y

$$
f_X(-\sqrt{y})=
\frac{1}{6}.
$$

Sustituyendo, obtenemos

$$
\begin{aligned}
f_Y(y)
&=\frac{1}{2\sqrt{y}}
\left(
\frac{1}{6}+\frac{1}{6}
\right)\\
&=\frac{1}{2\sqrt{y}}
\left(\frac{2}{6}\right)\\
&=\frac{1}{6\sqrt{y}}.
\end{aligned}
$$

Por lo tanto, la función de densidad de $Y=X^2$ es

$$
\boxed{
f_Y(y)=
\frac{1}{6\sqrt{y}}
\mathbb{1}_{(0,9)}(y).
}
$$
