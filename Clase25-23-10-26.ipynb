# Ejercicio 1

Sea $X$ una variable aleatoria continua con función de densidad

$$
f_X(x)=c(x+\sqrt{x})\mathbb{1}_{(0,1]}(x).
$$

Se pide:

a) Calcular $c$.

b) Calcular $\left\langle \frac{1}{X}\right\rangle$.

### Solución del inciso a)

Recordemos que toda función de densidad debe satisfacer

$$
1=P(X\in\mathbb{R})=\int_{-\infty}^{\infty}f_X(x)\,dx.
$$

Por lo tanto,

$$
\begin{aligned}
1
&=c\int_0^1(x+x^{1/2})\,dx\\
&=c\left[\frac{x^2}{2}+\frac{x^{3/2}}{3/2}\right]_0^1\\
&=c\left(\frac12+\frac23\right)\\
&=c\left(\frac36+\frac46\right)\\
&=c\frac76.
\end{aligned}
$$

De donde obtenemos

$$
\boxed{c=\frac67.}
$$

En consecuencia, nuestra función de densidad es

$$
\boxed{
f_X(x)=\frac67(x+\sqrt{x})\mathbb{1}_{(0,1]}(x).
}
$$

### Solución del inciso b)

Recordemos que, dada una función $g$, podemos calcular su esperanza mediante

$$
\langle g(X)\rangle=
\begin{cases}
\displaystyle\sum_{k\in X(\Omega)}g(k)p_X(k), & \text{si }X\text{ es discreta},\\[10pt]
\displaystyle\int_{-\infty}^{\infty}g(x)f_X(x)\,dx, & \text{si }X\text{ es continua},
\end{cases}
$$

siempre que la esperanza exista.

En nuestro caso,

$$
g(x)=\frac1x.
$$

Por lo tanto,

$$
\begin{aligned}
\left\langle\frac1X\right\rangle
&=\int_{-\infty}^{\infty}\frac1x f_X(x)\,dx\\
&=\frac67\int_0^1\frac1x(x+x^{1/2})\,dx\\
&=\frac67\int_0^1(1+x^{-1/2})\,dx\\
&=\frac67\left[x+2x^{1/2}\right]_0^1\\
&=\frac67(1+2)\\
&=\frac{18}{7}.
\end{aligned}
$$

Finalmente,

$$
\boxed{\left\langle\frac1X\right\rangle=\frac{18}{7}.}
$$

# Ejercicio 2

Sea $X$ una variable aleatoria continua con función de densidad

$$
f_X(x)=(a+bx)\mathbb{1}_{(0,2]}(x),
$$

tal que

$$
\langle X\rangle=\frac76.
$$

Se pide:

a) Calcular $a$ y $b$.

b) Calcular $\langle X^2\rangle$.

### Solución del inciso a)

Como $f_X$ es una función de densidad, tenemos que

$$
1=\int_{-\infty}^{\infty}f_X(x)\,dx.
$$

Sustituyendo,

$$
\begin{aligned}
1
&=\int_0^2(a+bx)\,dx\\
&=\left[ax+\frac{bx^2}{2}\right]_0^2\\
&=2a+2b.
\end{aligned}
$$

Por lo tanto,

$$
2a+2b=1.
$$

Por otra parte, sabemos que

$$
\langle X\rangle=\frac76.
$$

De modo que

$$
\begin{aligned}
\frac76
&=\int_{-\infty}^{\infty}xf_X(x)\,dx\\
&=\int_0^2x(a+bx)\,dx\\
&=\int_0^2(ax+bx^2)\,dx\\
&=\left[\frac{ax^2}{2}+\frac{bx^3}{3}\right]_0^2\\
&=2a+\frac{8b}{3}.
\end{aligned}
$$

Así, obtenemos el sistema de ecuaciones

$$
\begin{cases}
2a+2b=1,\\[4pt]
2a+\dfrac{8b}{3}=\dfrac76.
\end{cases}
$$

Restando la segunda ecuación de la primera,

$$
2b-\frac{8b}{3}=1-\frac76.
$$

Factorizando,

$$
b\left(2-\frac83\right)=1-\frac76.
$$

Entonces,

$$
b\left(-\frac23\right)=-\frac16.
$$

Por lo tanto,

$$
\boxed{b=\frac14.}
$$

Sustituyendo en la primera ecuación,

$$
2a+2\left(\frac14\right)=1.
$$

De donde

$$
2a+\frac12=1.
$$

Así,

$$
2a=\frac12.
$$

Finalmente,

$$
\boxed{a=\frac14.}
$$

Por lo tanto, la función de densidad queda dada por

$$
\boxed{
f_X(x)=\frac14(1+x)\mathbb{1}_{(0,2]}(x).
}
$$

### Solución del inciso b)

Queremos calcular

$$
\langle X^2\rangle.
$$

Por definición,

$$
\langle X^2\rangle
=
\int_{-\infty}^{\infty}x^2f_X(x)\,dx.
$$

Sustituyendo nuestra función de densidad,

$$
\begin{aligned}
\langle X^2\rangle
&=\int_0^2x^2\frac14(1+x)\,dx\\
&=\frac14\int_0^2(x^2+x^3)\,dx\\
&=\frac14\left[\frac{x^3}{3}+\frac{x^4}{4}\right]_0^2\\
&=\frac14\left(\frac83+\frac{16}{4}\right)\\
&=\frac14\left(\frac83+4\right)\\
&=\frac14\left(\frac{20}{3}\right)\\
&=\frac53.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{\langle X^2\rangle=\frac53.}
$$

# Varianza y desviación estándar

## Definición. $k$-ésimo momento de una variable aleatoria

El **$k$-ésimo momento** de una variable aleatoria $X$ se define como

$$
\boxed{\langle X^k\rangle.}
$$

Es decir,

$$
\langle X^k\rangle=\mathbb{E}[X^k].
$$

## Definición. $k$-ésimo momento central de una variable aleatoria

El **$k$-ésimo momento central** de una variable aleatoria $X$ se define como

$$
\boxed{\left\langle(X-\langle X\rangle)^k\right\rangle.}
$$

Es decir,

$$
\left\langle(X-\langle X\rangle)^k\right\rangle
=
\mathbb{E}\left[(X-\mathbb{E}[X])^k\right].
$$

## Definición. Varianza

Al segundo momento central se le llama **varianza** y se denota por

$$
\boxed{
\sigma^2=\operatorname{Var}(X)
=\left\langle(X-\langle X\rangle)^2\right\rangle.
}
$$

Equivalentemente,

$$
\operatorname{Var}(X)
=
\mathbb{E}\left[(X-\mathbb{E}[X])^2\right].
$$

## Definición. Desviación estándar

Definimos la **desviación estándar** como la raíz cuadrada de la varianza y se denota mediante

$$
\boxed{
\sigma=\sqrt{\sigma^2}=\sqrt{\operatorname{Var}(X)}.
}
$$

**Observación.** Tanto la varianza como la desviación estándar permiten estudiar la dispersión de los valores de una variable aleatoria alrededor de su esperanza.

La diferencia es que la varianza utiliza desviaciones elevadas al cuadrado, mientras que la desviación estándar se expresa en las mismas unidades que la variable aleatoria.

# Ejemplo 1. Varianza de una distribución normal

Sea

$$
X\sim N(\mu,\sigma^2).
$$

¿Cuál es la varianza de $X$?

Sabemos que

$$
\langle X\rangle=\mu.
$$

Podemos definir la función

$$
g(X)=(X-\langle X\rangle)^2=(X-\mu)^2.
$$

Por definición,

$$
\operatorname{Var}(X)
=
\left\langle(X-\langle X\rangle)^2\right\rangle.
$$

Como $\langle X\rangle=\mu$, tenemos que

$$
\operatorname{Var}(X)=\langle(X-\mu)^2\rangle.
$$

Recordemos que la función de densidad de una variable aleatoria normal es

$$
f_X(x)=
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}.
$$

Por lo tanto,

$$
\operatorname{Var}(X)
=
\int_{-\infty}^{\infty}
(x-\mu)^2
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

Como $\sigma>0$, podemos escribir

$$
\sqrt{2\pi\sigma^2}=\sigma\sqrt{2\pi}.
$$

Así,

$$
\operatorname{Var}(X)
=
\int_{-\infty}^{\infty}
(x-\mu)^2
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

Reescribiendo el integrando,

$$
\operatorname{Var}(X)
=
\int_{-\infty}^{\infty}
\frac{(x-\mu)\sigma}{\sqrt{2\pi}}
\frac{x-\mu}{\sigma^2}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

Utilizaremos integración por partes:

$$
\int u\,dv=uv-\int v\,du.
$$

Tomamos

$$
u=\frac{(x-\mu)\sigma}{\sqrt{2\pi}},
$$

y

$$
dv=
\frac{x-\mu}{\sigma^2}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

Entonces,

$$
du=\frac{\sigma}{\sqrt{2\pi}}\,dx,
$$

y

$$
v=-e^{-\frac{(x-\mu)^2}{2\sigma^2}}.
$$

Aplicando integración por partes,

$$
\begin{aligned}
\operatorname{Var}(X)
&=
-\left[
\frac{(x-\mu)\sigma}{\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\right]_{-\infty}^{\infty}\\
&\quad+
\int_{-\infty}^{\infty}
\frac{\sigma}{\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
\end{aligned}
$$

Observemos que

$$
\lim_{x\to\pm\infty}
(x-\mu)e^{-\frac{(x-\mu)^2}{2\sigma^2}}=0.
$$

Por lo tanto,

$$
-\left[
\frac{(x-\mu)\sigma}{\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\right]_{-\infty}^{\infty}=0.
$$

Así,

$$
\operatorname{Var}(X)
=
\int_{-\infty}^{\infty}
\frac{\sigma}{\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

Multiplicando y dividiendo por $\sigma$, obtenemos

$$
\begin{aligned}
\operatorname{Var}(X)
&=
\sigma^2
\int_{-\infty}^{\infty}
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx\\
&=
\sigma^2
\int_{-\infty}^{\infty}f_X(x)\,dx\\
&=\sigma^2(1)\\
&=\sigma^2.
\end{aligned}
$$

Finalmente,

$$
\boxed{\operatorname{Var}(X)=\sigma^2.}
$$

# Proposición. Segunda forma de calcular la varianza

Sea $X$ una variable aleatoria con segundo momento finito. Entonces,

$$
\boxed{
\operatorname{Var}(X)=\langle X^2\rangle-\langle X\rangle^2.
}
$$

### Demostración

Por definición,

$$
\operatorname{Var}(X)
=
\left\langle(X-\langle X\rangle)^2\right\rangle.
$$

Desarrollando el cuadrado,

$$
\begin{aligned}
\operatorname{Var}(X)
&=
\left\langle
X^2-2X\langle X\rangle+\langle X\rangle^2
\right\rangle.
\end{aligned}
$$

Utilizando la linealidad de la esperanza,

$$
\begin{aligned}
\operatorname{Var}(X)
&=
\langle X^2\rangle
-2\langle X\langle X\rangle\rangle
+\langle\langle X\rangle^2\rangle.
\end{aligned}
$$

Como $\langle X\rangle$ es una constante, tenemos que

$$
\langle X\langle X\rangle\rangle
=
\langle X\rangle\langle X\rangle
=
\langle X\rangle^2.
$$

Además,

$$
\langle\langle X\rangle^2\rangle=\langle X\rangle^2.
$$

Por lo tanto,

$$
\begin{aligned}
\operatorname{Var}(X)
&=\langle X^2\rangle-2\langle X\rangle^2+\langle X\rangle^2\\
&=\langle X^2\rangle-\langle X\rangle^2.
\end{aligned}
$$

Concluimos que

$$
\boxed{
\operatorname{Var}(X)=\langle X^2\rangle-\langle X\rangle^2.
}
$$

# Ejemplo 2. Varianza de una distribución uniforme

Sea

$$
X\sim U(a,b),
$$

con $a<b$.

Queremos calcular

$$
\operatorname{Var}(X).
$$

Recordemos que la función de densidad de $X$ está dada por

$$
f_X(x)=\frac{1}{b-a}\mathbb{1}_{(a,b]}(x).
$$

Además, sabemos que

$$
\boxed{\langle X\rangle=\frac{a+b}{2}.}
$$

Utilizaremos la segunda forma de calcular la varianza:

$$
\operatorname{Var}(X)=\langle X^2\rangle-\langle X\rangle^2.
$$

Primero, calculemos

$$
\langle X^2\rangle
=
\int_{-\infty}^{\infty}x^2f_X(x)\,dx.
$$

Sustituyendo nuestra función de densidad,

$$
\begin{aligned}
\langle X^2\rangle
&=
\int_{-\infty}^{\infty}
x^2\frac{1}{b-a}\mathbb{1}_{(a,b]}(x)\,dx\\
&=\frac{1}{b-a}\int_a^b x^2\,dx\\
&=\frac{1}{b-a}\left[\frac{x^3}{3}\right]_a^b\\
&=\frac{1}{b-a}\frac{b^3-a^3}{3}.
\end{aligned}
$$

Recordemos que

$$
b^3-a^3=(b-a)(b^2+ab+a^2).
$$

Por lo tanto,

$$
\begin{aligned}
\langle X^2\rangle
&=\frac{1}{b-a}\frac{(b-a)(b^2+ab+a^2)}{3}\\
&=\frac{b^2+ab+a^2}{3}.
\end{aligned}
$$

Así,

$$
\boxed{
\langle X^2\rangle=\frac{b^2+ab+a^2}{3}.
}
$$

Ahora calculemos la varianza:

$$
\operatorname{Var}(X)
=
\langle X^2\rangle-\langle X\rangle^2.
$$

Sustituyendo los momentos,

$$
\begin{aligned}
\operatorname{Var}(X)
&=\frac{b^2+ab+a^2}{3}
-\left(\frac{b+a}{2}\right)^2\\
&=\frac{b^2+ab+a^2}{3}
-\frac{b^2+2ab+a^2}{4}\\
&=\frac{4b^2+4ab+4a^2-3b^2-6ab-3a^2}{12}\\
&=\frac{b^2-2ab+a^2}{12}\\
&=\frac{(b-a)^2}{12}.
\end{aligned}
$$

Finalmente,

$$
\boxed{
\operatorname{Var}(X)=\frac{(b-a)^2}{12}.
}
$$
