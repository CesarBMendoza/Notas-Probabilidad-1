## Definición

La **función de distribución acumulada** $F_X$ se define como

$$
F_X(z)
=
\mathbb{P}[X\leq z]
=
\mathbb{P}
\left[
X^{-1}((-\infty,z])
\right].
$$

Veamos algunas propiedades:

### (i)

$F_X(z)$ es monótona no decreciente.

### (ii)

$$
\lim_{z\to-\infty}F_X(z)=0.
$$

### (iii)

$$
\lim_{z\to+\infty}F_X(z)=1.
$$

### (iv)

$F_X(z)$ es *càdlàg*, es decir,

$$
\lim_{h\uparrow 0}F_X(z+h)
=
F_X(z^-),
$$

y este límite existe.

Además,

$$
\lim_{h\downarrow 0}F_X(z+h)
=
F_X(z).
$$

Es decir, $F_X$ es continua por la derecha.

### Ejemplo

La función de distribución puede tener saltos, pero es continua por la derecha.

> En la figura se muestra una función de distribución con un salto:  
> el punto correspondiente al valor por la derecha pertenece a la gráfica, mientras que el valor límite por la izquierda puede ser distinto.

Ahora, notemos que

$$
\mathbb{P}[X>z]
+
\mathbb{P}[X\leq z]
=
1.
$$

Por tanto,

$$
\mathbb{P}[X>z]
=
1-F_X(z)
=
S_X(z).
$$

A $S_X$ se le conoce como la **función de supervivencia**.

Además,

$$
\mathbb{P}[a<X\leq b]
=
F_X(b)-F_X(a).
$$

### Teorema

La función $F_X(z)$ caracteriza completamente a la variable aleatoria (v.a.).

Es decir, para toda función $F_X(z)$ que cumpla las propiedades (i)-(iv), existe una variable aleatoria $X$ cuya función de distribución es $F_X$.

Si tengo $X$ y $Y$ tales que

$$
F_X(z)=F_Y(z),
$$

entonces

$$
X=Y.
$$

Si $\Omega$ es discreto, entonces $X$ es discreto.

Por tanto, diremos que $X$ es una **variable aleatoria discreta**.

Una variable aleatoria discreta es aquella que puede tomar un número finito o numerable de valores.

Además, definimos el **rango de una variable aleatoria discreta** $X$ como

$$
R_X
=
\left\{
x\in\mathbb{R}:
\mathbb{P}[X=x]>0
\right\}.
$$

Es decir, $R_X$ es el conjunto de valores posibles de $X$.

Donde

$$
[X=x]
=
\{\omega\in\Omega \mid X(\omega)=x\}.
$$

Ahora también definimos la **función de probabilidad puntual**.

La función $p_X(k)$ de una v.a. discreta $X$ se define como

$$
p_X(k)=\mathbb{P}[X=k],
\qquad k\in R_X.
$$

### Ejemplo

Tiro una moneda honesta 2 veces.

Sea $X$ el número de águilas que obtengo. ¿Cuál es $p_X(k)$?

El espacio muestral es

$$
\Omega
=
\{(A,A),(A,S),(S,A),(S,S)\}.
$$

El rango de $X$ es

$$
R_X=\{0,1,2\},
$$

donde los valores representan el número de águilas obtenidas.

Como buscamos $p_X(k)$ para $k\in R_X$, tenemos:

$$
p_X(0)=\frac14=\mathbb{P}[X=0],
$$

$$
p_X(1)=\frac12=\mathbb{P}[X=1],
$$

y

$$
p_X(2)=\frac14=\mathbb{P}[X=2].
$$

Por tanto,

$$
p_X(k)
=
\begin{cases}
\dfrac14, & k=0,2,\\[4pt]
\dfrac12, & k=1,\\[4pt]
0, & \text{a.c.}
\end{cases}
$$

Ahora, habíamos dicho que

$$
F_X(z)
=
\mathbb{P}[X\leq z]
=
\sum_{\substack{k\in R_X\\k\leq z}}p_X(k).
$$

### Ejemplo

Supongamos que el rango de nuestro dado es

$$
R_X=\{1,2,3\}.
$$

Además,

$$
p_X(1)=p_1,
\qquad
p_X(2)=p_2.
$$

Continuando con el ejemplo anterior,

$$
p_X(3)=p_3,
$$

donde

$$
p_1+p_2+p_3=1.
$$

La función de distribución acumulada tiene saltos en los puntos del rango $R_X$.

Los niveles de la función son:

$$
p_1,
$$

$$
p_1+p_2,
$$

y

$$
p_1+p_2+p_3=1.
$$

Por lo tanto, los saltos ocurren en los elementos de $R_X$.

La probabilidad puntual se puede obtener a partir del tamaño del salto de la función de distribución. Para $\varepsilon>0$ suficientemente pequeño,

$$
p_X(z)
=
F_X(z+\varepsilon)-F_X(z-\varepsilon).
$$

Si

$$
R_X\subseteq\mathbb{Z},
$$

entonces

$$
p_X(k)
=
F_X(k)-F_X(k-1).
$$

### Ejemplo

Tengo una urna con boletos numerados del $1$ al $n$.

Saco $r<n$ números al azar, **con reemplazo**.

Sea $X$ el número más grande de todos los que saqué.

Queremos encontrar $p_X(k)$.

Primero calculamos

$$
\mathbb{P}[X\leq z].
$$

El evento $X\leq z$ significa que **a lo más saqué el número $z$**, es decir, que en cada una de las $r$ extracciones salió alguno de los números

$$
1,2,\ldots,z.
$$

Por tanto,

$$
F_X(z)
=
\mathbb{P}[X\leq z]
=
\frac{z^r}{n^r}
=
\left(\frac{z}{n}\right)^r.
$$

Hay $r$ etapas y en cada etapa hay $n$ posibles resultados.

Como

$$
k=1,\ldots,n,
\qquad
k\in\mathbb{Z},
$$

tenemos

$$
p_X(k)
=
F_X(k)-F_X(k-1).
$$

Por lo tanto,

$$
p_X(k)
=
\left(\frac{k}{n}\right)^r
-
\left(\frac{k-1}{n}\right)^r.
$$
