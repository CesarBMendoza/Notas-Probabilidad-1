Una urna contiene 8 bolas blancas, 4 negras y 2 verdes. Elegimos al azar 2 bolas.

Supongamos que ganamos \$2 por cada bola negra seleccionada, perdemos \$1 por cada bola blanca seleccionada y, si sale verde, no ganamos nada.

Sea $X$ la cantidad de dinero que ganamos.

**¿Cuáles son los valores posibles de $X$ y cuáles son las probabilidades asociadas a cada valor?**

Si definimos

- $N_B =$ número de bolas blancas seleccionadas.
- $N_N =$ número de bolas negras seleccionadas.

entonces

$$
X=2N_N-N_B.
$$

Como solo escogemos 2 bolas, hay 6 posibles tipos de selección:

- $(B,B)\Rightarrow X=-1-1=-2$.
- $(B,V)\Rightarrow X=-1$.
- $(V,V)\Rightarrow X=0$.
- $(N,B)\Rightarrow X=1$.
- $(N,V)\Rightarrow X=2$.
- $(N,N)\Rightarrow X=4$.

Así, $X$ toma los valores

$$
X\in\{-2,-1,0,1,2,4\}.
$$

Para $X=-2$ (perder \$2), necesitamos seleccionar 2 bolas blancas:

$$
P(X=-2)
=
\frac{\binom{8}{2}}{\binom{14}{2}}
=
\frac{28}{91}.
$$

Para $X=-1$, necesitamos seleccionar 1 bola blanca y 1 verde:

$$
P(X=-1)
=
P(1\text{ blanca y }1\text{ verde})
=
\frac{\binom{8}{1}\binom{2}{1}}{\binom{14}{2}}
=
\frac{16}{91}.
$$

Para $X=0$, necesitamos seleccionar 2 bolas verdes:

$$
P(X=0)
=
P(2\text{ verdes})
=
\frac{\binom{2}{2}}{\binom{14}{2}}
=
\frac{1}{91}.
$$

Para $X=1$, necesitamos seleccionar 1 bola negra y 1 blanca:

$$
P(X=1)
=
P(1\text{ negra y }1\text{ blanca})
=
\frac{\binom{4}{1}\binom{8}{1}}{\binom{14}{2}}
=
\frac{32}{91}.
$$

Para $X=2$, necesitamos seleccionar 1 bola negra y 1 verde:

$$
P(X=2)
=
P(1\text{ negra y }1\text{ verde})
=
\frac{\binom{4}{1}\binom{2}{1}}{\binom{14}{2}}
=
\frac{8}{91}.
$$

Para $X=4$, necesitamos seleccionar 2 bolas negras:

$$
P(X=4)
=
P(2\text{ negras})
=
\frac{\binom{4}{2}}{\binom{14}{2}}
=
\frac{6}{91}.
$$

Con esto podemos definir su función de masa de probabilidad:

$$
p_X(x)=
\begin{cases}
\dfrac{28}{91}, & x=-2,\\[4pt]
\dfrac{16}{91}, & x=-1,\\[4pt]
\dfrac{1}{91}, & x=0,\\[4pt]
\dfrac{32}{91}, & x=1,\\[4pt]
\dfrac{8}{91}, & x=2,\\[4pt]
\dfrac{6}{91}, & x=4,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

Y

$$
F_X(x)=P(X\leq x).
$$

Sea $X$ la diferencia entre el número de caras y el número de cruces obtenidos cuando se lanza una moneda $n$ veces.

**¿Cuáles son los valores posibles de $X$?**

Si definimos

- $H =$ número de caras obtenidas.
- $T =$ número de cruces obtenidas.

entonces

$$
n=H+T.
$$

Además,

$$
X=H-T.
$$

Como

$$
T=n-H,
$$

tenemos

$$
X=H-(n-H)=2H-n.
$$

El número de caras puede tomar los valores

$$
H\in\{0,1,\ldots,n\}.
$$

Así:

- Para $H=0$, $X=-n$.
- Para $H=1$, $X=2-n$.
- Para $H=2$, $X=4-n$.

Por lo tanto,

$$
X\in\{-n,-n+2,-n+4,\ldots,n-2,n\}.
$$

Observemos que $X$ tiene la misma paridad que $n$.

Es decir, si $n$ es par, $X$ solo puede tomar valores pares; y si $n$ es impar, $X$ solo puede tomar valores impares.

Ahora, para $n=3$, calculemos los posibles valores.

Como

$$
X=2H-n=2H-3,
$$

tenemos:

| $H$ | Cruces $T$ | $X=H-T$ |
|---:|---:|---:|
| 0 | 3 | -3 |
| 1 | 2 | -1 |
| 2 | 1 | 1 |
| 3 | 0 | 3 |

Por lo tanto,

$$
X\in\{-3,-1,1,3\}.
$$

Como la moneda es justa, hay

$$
2^3=8
$$

resultados equiprobables.

En efecto,

$$
\Omega=\{(a,b,c):a,b,c\in\{H,T\}\},
$$

y

$$
|\Omega|=2^3=8.
$$

Para $X=-3$, necesitamos obtener 3 cruces:

$$
TTT.
$$

Por lo tanto,

$$
P(X=-3)=P(3\text{ cruces})=\frac{1}{8}.
$$

Para $X=-1$, necesitamos obtener 2 cruces y 1 cara:

$$
P(X=-1)
=
P(HTT)+P(THT)+P(TTH)
=
\frac{3}{8}.
$$

Para $X=1$,

$$
P(X=1)=\frac{3}{8}.
$$

Finalmente,

$$
P(X=3)=\frac{1}{8}.
$$

Otra forma es usando la distribución binomial.

Como $H$ es el número de caras y la moneda es justa,

$$
H\sim\operatorname{Binomial}\left(3,\frac12\right).
$$

Por lo tanto,

$$
P(H=k)
=
\binom{3}{k}
\left(\frac12\right)^k
\left(\frac12\right)^{3-k}
=
\frac{\binom{3}{k}}{8}.
$$

Es decir,

$$
P(H=k)=\frac{\binom{3}{k}}{8}.
$$

Como

$$
X=2H-3,
$$

tenemos:

$$
X=-3
\qquad\text{si}\qquad
H=0,
$$

$$
X=-1
\qquad\text{si}\qquad
H=1,
$$

$$
X=1
\qquad\text{si}\qquad
H=2,
$$

$$
X=3
\qquad\text{si}\qquad
H=3.
$$

Por lo tanto,

$$
P(X=-3)
=
P(H=0)
=
\frac{1}{8},
$$

$$
P(X=-1)
=
P(H=1)
=
\frac{\binom{3}{1}}{8}
=
\frac{3}{8},
$$

$$
P(X=1)
=
P(H=2)
=
\frac{\binom{3}{2}}{8}
=
\frac{3}{8},
$$

$$
P(X=3)
=
P(H=3)
=
\frac{\binom{3}{3}}{8}
=
\frac{1}{8}.
$$

## Ejercicio

5 números distintos se distribuyen aleatoriamente entre 5 jugadores numerados del 1 al 5.

Cuando dos jugadores comparan sus números, se declara ganador al que tenga el número mayor.

Inicialmente, los jugadores 1 y 2 comparan sus números. El ganador compara su número con el jugador 3 y así sucesivamente.

Sea $X$ el número de veces que el jugador 1 gana.

Entonces,

$$
X\in\{0,1,2,3,4\}.
$$

Calculemos $P(X=i)$.

### Para $X=0$

El jugador 1 no gana ningún juego contra el jugador 2.

Entonces,

$$
N_2>N_1.
$$

Como entre $N_1$ y $N_2$ cualquiera de los dos puede ser el mayor con la misma probabilidad,

$$
P(X=0)=\frac12.
$$

### Para $X=1$

El jugador 1 gana una vez y después pierde contra el jugador 3.

Entonces debe ocurrir

$$
N_2<N_1<N_3.
$$

Por lo tanto,

$$
P(X=1)=\frac{1}{3!}.
$$

### Para $X=2$

El jugador 1 le gana a los jugadores 2 y 3, pero no al jugador 4.

Es decir,

$$
N_1>N_2,
$$

$$
N_1>N_3,
$$

pero

$$
N_4>N_1.
$$

Equivalentemente,

$$
N_4>N_1>(N_2,N_3).
$$

Los jugadores 2 y 3 pueden aparecer en cualquier orden, por lo que hay $2!$ órdenes favorables entre los $4!$ órdenes posibles.

Así,

$$
P(X=2)=\frac{2!}{4!}=\frac{1}{12}.
$$

### Para $X=3$

El jugador 1 gana todos los juegos salvo el último.

Entonces,

$$
N_5>N_1>(N_2,N_3,N_4).
$$

Los números de los jugadores 2, 3 y 4 pueden ordenarse de $3!$ maneras.

Por lo tanto,

$$
P(X=3)=\frac{3!}{5!}=\frac{1}{20}.
$$

### Para $X=4$

El jugador 1 gana las cuatro rondas.

Entonces,

$$
N_1>(N_2,N_3,N_4,N_5).
$$

Los otros cuatro jugadores pueden ordenarse de $4!$ maneras.

Por lo tanto,

$$
P(X=4)=\frac{4!}{5!}=\frac15.
$$

La lotería del draft de la NBA involucra a los 11 equipos que tuvieron los peores récords.

En una urna se colocan 66 bolas de tal manera que:

- 11 tienen el nombre del equipo con el peor récord.
- 10 tienen el nombre del equipo con el segundo peor récord.
- 9 tienen el nombre del tercero.
- $\vdots$
- 1 bola tiene el nombre del equipo con el mejor récord entre esos 11.

Se extrae una bola al azar y el equipo cuyo nombre aparece recibe la primera selección del draft.

Se extrae otra bola. Si pertenece a un equipo diferente, este recibe la segunda selección.

Si corresponde al equipo que ya ganó, se descarta y se continúa extrayendo hasta obtener otro equipo.

El procedimiento continúa hasta que tres equipos reciben las primeras tres selecciones.

A los ocho equipos que no ganaron la lotería se les asignan las selecciones restantes en orden inverso a sus récords.

Sea $X$ la posición en el draft que recibe el equipo con el peor récord.

Hallemos la función de masa de probabilidad de $X$.

Llamemos $E_1$ al equipo con el peor récord, que tiene 11 bolas.

Entonces,

$$
X\in\{1,2,3,4\}.
$$

### Para $X=1$

El equipo 1 debe ganar la primera selección.

Por lo tanto,

$$
P(X=1)=\frac{11}{66}.
$$

### Para $X=2$

El equipo 1 obtiene la segunda posición.

Definamos

$$
A=\{\text{el equipo 1 obtiene la segunda posición}\},
$$

y

$$
B_j=\{\text{el primer equipo seleccionado tiene }j\text{ bolas}\}.
$$

Entonces, usando la ley de la probabilidad total,

$$
P(A)
=
\sum_{j=1}^{10}P(A\mid B_j)P(B_j).
$$

Notemos que

$$
P(B_j)=\frac{j}{66}.
$$

Además, dado que el primer equipo seleccionado tenía $j$ bolas, quedan $66-j$ bolas, de las cuales 11 pertenecen al equipo 1. Por tanto,

$$
P(A\mid B_j)=\frac{11}{66-j}.
$$

Así,

$$
P(X=2)
=
\sum_{j=1}^{10}
\frac{j}{66}
\frac{11}{66-j}
\approx 0.1556.
$$

### Para $X=3$

Para que el equipo con el peor récord obtenga la tercera selección, deben seleccionarse primero dos equipos distintos de él y distintos entre sí.

Por lo tanto,

$$
P(X=3)
=
\sum_{\substack{j,k=1\\ j\neq k}}^{10}
\frac{j}{66}
\frac{k}{66-j}
\frac{11}{66-j-k}.
$$

Aproximadamente,

$$
P(X=3)\approx 0.1434.
$$

### Para $X=4$

Podemos usar el complemento:

$$
P(X=4)
=
1-P(X=1)-P(X=2)-P(X=3).
$$

Por lo tanto,

$$
P(X=4)\approx 0.534.
$$

Así, la función de masa de probabilidad queda aproximadamente

$$
p_X(x)=
\begin{cases}
0.1667, & x=1,\\
0.1556, & x=2,\\
0.1434, & x=3,\\
0.5343, & x=4,\\
0, & \text{en otro caso}.
\end{cases}
$$
