
# Propiedades de la varianza

## 1. Caracterización de las variables aleatorias con varianza cero

### Proposición

Sea $X$ una variable aleatoria con segundo momento finito. Entonces,

$$
\boxed{
\operatorname{Var}(X)=0
\iff
P(X=\mathbb{E}[X])=1
}
$$

### Demostración

**$(\Leftarrow)$** Supongamos que

$$
P(X=\mathbb{E}[X])=1.
$$

Es decir, $X$ toma el valor $\mathbb{E}[X]$ con probabilidad uno.

Recordemos que

$$
\operatorname{Var}(X)
=
\mathbb{E}\left[(X-\mathbb{E}[X])^2\right].
$$

Como $X=\mathbb{E}[X]$ casi seguramente, tenemos que

$$
(X-\mathbb{E}[X])^2=0
$$

casi seguramente.

Por lo tanto,

$$
\begin{aligned}
\operatorname{Var}(X)
&=
\mathbb{E}\left[(X-\mathbb{E}[X])^2\right]\\
&=\mathbb{E}[0]\\
&=0.
\end{aligned}
$$

Así,

$$
\boxed{\operatorname{Var}(X)=0}.
$$

**$(\Rightarrow)$** Supongamos ahora que

$$
\operatorname{Var}(X)=0.
$$

Por definición,

$$
\operatorname{Var}(X)
=
\mathbb{E}\left[(X-\mathbb{E}[X])^2\right].
$$

Definamos la variable aleatoria

$$
Y=(X-\mathbb{E}[X])^2.
$$

Observemos que

$$
Y\geq 0
$$

para cualquier valor que tome $X$.

Además, por nuestra hipótesis,

$$
\mathbb{E}[Y]
=
\operatorname{Var}(X)
=
0.
$$

Recordemos el siguiente resultado:

> Si $Y$ es una variable aleatoria no negativa y $\mathbb{E}[Y]=0$, entonces $P(Y=0)=1$.

Aplicando este resultado, tenemos que

$$
P(Y=0)=1.
$$

Es decir,

$$
P\left((X-\mathbb{E}[X])^2=0\right)=1.
$$

Pero

$$
(X-\mathbb{E}[X])^2=0
\iff
X=\mathbb{E}[X].
$$

Por lo tanto,

$$
\boxed{
P(X=\mathbb{E}[X])=1
}
$$

Concluimos que

$$
\boxed{
\operatorname{Var}(X)=0
\iff
P(X=\mathbb{E}[X])=1
}
$$

$\square$

### Observación

Una variable aleatoria tiene varianza cero si y solo si es constante casi seguramente.

En particular, si $X=c$ con probabilidad uno, entonces

$$
\mathbb{E}[X]=c
$$

y

$$
\operatorname{Var}(X)=0.
$$

## 2. Invariancia de la varianza ante traslaciones

### Proposición

Sea $X$ una variable aleatoria con segundo momento finito y sea $b\in\mathbb{R}$.

Entonces,

$$
\boxed{
\operatorname{Var}(X+b)=\operatorname{Var}(X)
}
$$

### Demostración

Por definición de varianza,

$$
\operatorname{Var}(X+b)
=
\mathbb{E}\left[
\left((X+b)-\mathbb{E}[X+b]\right)^2
\right].
$$

Por la linealidad de la esperanza,

$$
\mathbb{E}[X+b]=\mathbb{E}[X]+b.
$$

Sustituyendo,

$$
\begin{aligned}
\operatorname{Var}(X+b)
&=
\mathbb{E}\left[
\left(X+b-\mathbb{E}[X]-b\right)^2
\right]\\
&=
\mathbb{E}\left[
(X-\mathbb{E}[X])^2
\right]\\
&=
\operatorname{Var}(X).
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\operatorname{Var}(X+b)=\operatorname{Var}(X)
}
$$

$\square$

### Observación

Sumar una constante a una variable aleatoria modifica su esperanza, pero no modifica su varianza.

Esto se debe a que la varianza mide la dispersión de los valores de $X$ respecto de su esperanza.

Al sumar una constante $b$, tanto los valores de $X$ como su esperanza se desplazan en la misma cantidad.

## 3. Varianza de una variable aleatoria multiplicada por una constante

### Proposición

Sea $X$ una variable aleatoria con segundo momento finito y sea $a\in\mathbb{R}$.

Entonces,

$$
\boxed{
\operatorname{Var}(aX)=a^2\operatorname{Var}(X)
}
$$

### Demostración

Por definición,

$$
\operatorname{Var}(aX)
=
\mathbb{E}\left[(aX-\mathbb{E}[aX])^2\right].
$$

Por la linealidad de la esperanza,

$$
\mathbb{E}[aX]=a\mathbb{E}[X].
$$

Por lo tanto,

$$
\begin{aligned}
\operatorname{Var}(aX)
&=
\mathbb{E}\left[
(aX-a\mathbb{E}[X])^2
\right]\\
&=
\mathbb{E}\left[
a^2(X-\mathbb{E}[X])^2
\right]\\
&=
a^2\mathbb{E}\left[
(X-\mathbb{E}[X])^2
\right]\\
&=
a^2\operatorname{Var}(X).
\end{aligned}
$$

Concluimos que

$$
\boxed{
\operatorname{Var}(aX)=a^2\operatorname{Var}(X)
}
$$

$\square$

### Observación. La varianza no es lineal

De las propiedades anteriores observamos que la varianza no es una operación lineal.

En efecto, si consideramos

$$
Y=X+X=2X,
$$

entonces, por la propiedad que acabamos de demostrar,

$$
\operatorname{Var}(Y)
=
\operatorname{Var}(2X)
=
4\operatorname{Var}(X).
$$

Sin embargo, si la varianza fuera lineal, tendríamos que

$$
\operatorname{Var}(X+X)
=
\operatorname{Var}(X)+\operatorname{Var}(X)
=
2\operatorname{Var}(X).
$$

En general,

$$
4\operatorname{Var}(X)\neq 2\operatorname{Var}(X).
$$

Por lo tanto, concluimos que la varianza no es lineal.

**Observación adicional:** Para constantes $a,b\in\mathbb{R}$, las propiedades anteriores permiten obtener

$$
\boxed{
\operatorname{Var}(aX+b)
=
a^2\operatorname{Var}(X)
}
$$

# Desigualdades probabilísticas

## 4. Desigualdad fundamental

### Teorema

Sea $X$ una variable aleatoria y sea $h:\mathbb{R}\to[0,\infty)$ una función medible.

Si $a>0$, entonces

$$
\boxed{
P(h(X)\geq a)
\leq
\frac{\mathbb{E}[h(X)]}{a}
}
$$

Esta desigualdad se conoce como una forma general de la **desigualdad de Markov**.

### Demostración

Definamos la función indicadora del evento $\{h(X)\geq a\}$:

$$
\mathbb{1}_{\{h(X)\geq a\}}
=
\begin{cases}
1, & \text{si }h(X)\geq a,\\
0, & \text{si }h(X)<a.
\end{cases}
$$

Observemos que

$$
h(X)\geq
a\mathbb{1}_{\{h(X)\geq a\}}.
$$

Para justificarlo, consideremos dos casos.

**Caso 1.** Si $h(X)\geq a$, entonces

$$
\mathbb{1}_{\{h(X)\geq a\}}=1.
$$

Por lo tanto,

$$
h(X)\geq a
=
a\mathbb{1}_{\{h(X)\geq a\}}.
$$

**Caso 2.** Si $h(X)<a$, entonces

$$
\mathbb{1}_{\{h(X)\geq a\}}=0.
$$

Como $h(X)\geq 0$, tenemos

$$
h(X)\geq 0
=
a\mathbb{1}_{\{h(X)\geq a\}}.
$$

Por lo tanto, en ambos casos se cumple que

$$
\boxed{
h(X)\geq a\mathbb{1}_{\{h(X)\geq a\}}
}
$$

Aplicando la monotonía de la esperanza,

$$
\mathbb{E}[h(X)]
\geq
a\mathbb{E}\left[
\mathbb{1}_{\{h(X)\geq a\}}
\right].
$$

Recordemos que la esperanza de una función indicadora es la probabilidad del evento correspondiente:

$$
\mathbb{E}[\mathbb{1}_A]=P(A).
$$

Por lo tanto,

$$
\begin{aligned}
\mathbb{E}[h(X)]
&\geq
a\mathbb{E}\left[
\mathbb{1}_{\{h(X)\geq a\}}
\right]\\
&=
aP(h(X)\geq a).
\end{aligned}
$$

Como $a>0$, podemos dividir entre $a$ y obtenemos

$$
\boxed{
P(h(X)\geq a)
\leq
\frac{\mathbb{E}[h(X)]}{a}
}
$$

$\square$

## 5. Desigualdad de Markov

### Teorema

Sea $X$ una variable aleatoria no negativa, es decir,

$$
P(X\geq 0)=1.
$$

Entonces, para cualquier $a>0$,

$$
\boxed{
P(X\geq a)
\leq
\frac{\mathbb{E}[X]}{a}
}
$$

### Demostración

Utilizando la desigualdad fundamental, tomemos

$$
h(x)=x
$$

para $x\geq 0$.

Como $X\geq 0$ casi seguramente, podemos aplicar la desigualdad anterior y obtenemos

$$
P(X\geq a)
\leq
\frac{\mathbb{E}[X]}{a}.
$$

Por lo tanto,

$$
\boxed{
P(X\geq a)
\leq
\frac{\mathbb{E}[X]}{a}
}
$$

$\square$

### Observación

Si $X$ no es necesariamente no negativa, podemos tomar

$$
h(x)=|x|.
$$

Como $|X|\geq 0$, obtenemos

$$
\boxed{
P(|X|\geq a)
\leq
\frac{\mathbb{E}[|X|]}{a}
}
$$

siempre que $a>0$.

## 6. Desigualdad de Chebyshev

### Teorema

Sea $X$ una variable aleatoria con segundo momento finito.

Entonces, para cualquier $a>0$,

$$
\boxed{
P(|X-\mathbb{E}[X]|\geq a)
\leq
\frac{\operatorname{Var}(X)}{a^2}
}
$$

### Demostración

Primero, tomemos

$$
h(x)=x^2.
$$

Como

$$
h(x)=x^2\geq 0,
$$

podemos aplicar la desigualdad fundamental con el umbral $a^2>0$.

De esta manera,

$$
P(X^2\geq a^2)
\leq
\frac{\mathbb{E}[X^2]}{a^2}.
$$

Observemos que

$$
\{X^2\geq a^2\}
=
\{|X|\geq a\}.
$$

Por lo tanto,

$$
\boxed{
P(|X|\geq a)
\leq
\frac{\mathbb{E}[X^2]}{a^2}
}
$$

Ahora consideremos la función

$$
h(x)=(x-\mathbb{E}[X])^2.
$$

Esta función es no negativa para cualquier $x\in\mathbb{R}$.

Aplicando nuevamente la desigualdad fundamental,

$$
P\left(
(X-\mathbb{E}[X])^2\geq a^2
\right)
\leq
\frac{
\mathbb{E}\left[
(X-\mathbb{E}[X])^2
\right]
}{a^2}.
$$

Por definición de varianza,

$$
\mathbb{E}\left[
(X-\mathbb{E}[X])^2
\right]
=
\operatorname{Var}(X).
$$

Además,

$$
\left\{
(X-\mathbb{E}[X])^2\geq a^2
\right\}
=
\left\{
|X-\mathbb{E}[X]|\geq a
\right\}.
$$

Por lo tanto,

$$
\boxed{
P(|X-\mathbb{E}[X]|\geq a)
\leq
\frac{\operatorname{Var}(X)}{a^2}
}
$$

$\square$

### Observación

La desigualdad de Chebyshev nos proporciona una cota superior para la probabilidad de que una variable aleatoria se aleje de su esperanza una distancia de al menos $a$.

Esta cota depende únicamente de su varianza y no requiere conocer su distribución de probabilidad.

Por otra parte, tomando

$$
a=k\sigma,
$$

donde

$$
\sigma=\sqrt{\operatorname{Var}(X)}
$$

es la desviación estándar y $\sigma>0$, obtenemos

$$
P(|X-\mathbb{E}[X]|\geq k\sigma)
\leq
\frac{\sigma^2}{k^2\sigma^2}.
$$

Por lo tanto,

$$
\boxed{
P(|X-\mathbb{E}[X]|\geq k\sigma)
\leq
\frac{1}{k^2}
}
$$

para cualquier $k>0$.

## 7. Desigualdad de Chernoff

La desigualdad de Chernoff permite obtener cotas para las probabilidades de que una variable aleatoria tome valores grandes.

La idea consiste en aplicar la desigualdad de Markov a una transformación exponencial de la variable aleatoria.

### Teorema

Sea $X$ una variable aleatoria y supongamos que

$$
\mathbb{E}[e^{sX}]<\infty
$$

para los valores de $s>0$ que consideraremos.

Entonces, para cualquier $a\in\mathbb{R}$,

$$
\boxed{
P(X\geq a)
\leq
\inf_{s>0}
\left\{
e^{-sa}\mathbb{E}[e^{sX}]
\right\}
}
$$

### Demostración

Sea $s>0$.

Como la función exponencial es estrictamente creciente,

$$
X\geq a
\iff
sX\geq sa
\iff
e^{sX}\geq e^{sa}.
$$

Por lo tanto,

$$
P(X\geq a)
=
P(e^{sX}\geq e^{sa}).
$$

Como

$$
e^{sX}>0,
$$

podemos aplicar la desigualdad de Markov:

$$
P(e^{sX}\geq e^{sa})
\leq
\frac{\mathbb{E}[e^{sX}]}{e^{sa}}.
$$

De esta manera,

$$
P(X\geq a)
\leq
e^{-sa}\mathbb{E}[e^{sX}].
$$

Esta desigualdad se cumple para cualquier $s>0$ para el cual exista la esperanza exponencial.

Por lo tanto, podemos tomar el ínfimo sobre todos esos valores de $s$:

$$
\boxed{
P(X\geq a)
\leq
\inf_{s>0}
e^{-sa}\mathbb{E}[e^{sX}]
}
$$

$\square$

### Observación

La función

$$
M_X(s)=\mathbb{E}[e^{sX}]
$$

se conoce como la función generadora de momentos de $X$, cuando existe en un entorno de cero.

En consecuencia, la desigualdad de Chernoff puede escribirse como

$$
\boxed{
P(X\geq a)
\leq
\inf_{s>0}e^{-sa}M_X(s)
}
$$

## 8. Ejemplo: Desigualdad de Chernoff para una variable aleatoria binomial

Sea

$$
X\sim\operatorname{Binomial}(n,p),
$$

donde $0<p<1$.

Queremos obtener una cota superior para

$$
P(X\geq bn),
$$

donde $p<b<1$.

### 8.1. Aplicando la desigualdad de Markov

Recordemos que si

$$
X\sim\operatorname{Binomial}(n,p),
$$

entonces

$$
\mathbb{E}[X]=np.
$$

Como $X\geq 0$, podemos aplicar la desigualdad de Markov:

$$
P(X\geq bn)
\leq
\frac{\mathbb{E}[X]}{bn}.
$$

Sustituyendo la esperanza,

$$
\begin{aligned}
P(X\geq bn)
&\leq
\frac{np}{bn}\\
&=
\frac{p}{b}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X\geq bn)
\leq
\frac{p}{b}
}
$$

Esta cota depende de $p$ y $b$, pero no mejora al aumentar $n$.

Ahora utilizaremos la desigualdad de Chernoff para obtener una cota más precisa.

### 8.2. Calculando la esperanza exponencial

Por definición,

$$
\mathbb{E}[e^{sX}]
=
\sum_{k\in R_X}e^{sk}P_X(k).
$$

Como $X$ es una variable aleatoria binomial,

$$
R_X=\{0,1,\ldots,n\}
$$

y

$$
P_X(k)
=
\binom{n}{k}p^k(1-p)^{n-k}.
$$

Por lo tanto,

$$
\mathbb{E}[e^{sX}]
=
\sum_{k=0}^{n}
e^{sk}
\binom{n}{k}
p^k(1-p)^{n-k}.
$$

Observemos que

$$
e^{sk}p^k=(pe^s)^k.
$$

Entonces,

$$
\mathbb{E}[e^{sX}]
=
\sum_{k=0}^{n}
\binom{n}{k}
(pe^s)^k(1-p)^{n-k}.
$$

Recordemos el teorema del binomio:

$$
(a+b)^n
=
\sum_{k=0}^{n}
\binom{n}{k}a^kb^{n-k}.
$$

Tomando

$$
a=pe^s,
\qquad
b=1-p,
$$

obtenemos

$$
\boxed{
\mathbb{E}[e^{sX}]
=
(pe^s+1-p)^n
}
$$

### 8.3. Aplicando la desigualdad de Chernoff

Por la desigualdad de Chernoff,

$$
P(X\geq bn)
\leq
\inf_{s>0}
e^{-sbn}\mathbb{E}[e^{sX}].
$$

Sustituyendo la expresión que acabamos de obtener,

$$
\boxed{
P(X\geq bn)
\leq
\inf_{s>0}
e^{-sbn}(pe^s+1-p)^n
}
$$

Por lo tanto, nuestro problema se reduce a encontrar el valor de $s>0$ que minimiza la función

$$
f(s)=e^{-sbn}(pe^s+1-p)^n.
$$

## 8.4. Encontrando el valor óptimo de $s$

Queremos minimizar la función

$$
f(s)=e^{-sbn}(pe^s+1-p)^n,
$$

para $s>0$.

Para encontrar sus puntos críticos, calculamos su derivada.

Utilizando la regla del producto y la regla de la cadena,

$$
\begin{aligned}
f'(s)
&=
-bn e^{-sbn}(pe^s+1-p)^n\\
&\quad+
n e^{-sbn}(pe^s+1-p)^{n-1}pe^s.
\end{aligned}
$$

Factorizando,

$$
f'(s)
=
n e^{-sbn}(pe^s+1-p)^{n-1}
\left[
-b(pe^s+1-p)+pe^s
\right].
$$

Buscamos los valores de $s$ tales que

$$
f'(s)=0.
$$

Como

$$
n e^{-sbn}(pe^s+1-p)^{n-1}>0,
$$

para $n\geq 1$, podemos reducir la ecuación a

$$
-b(pe^s+1-p)+pe^s=0.
$$

Desarrollando los términos,

$$
-bpe^s-b(1-p)+pe^s=0.
$$

Agrupando los términos que contienen $e^s$,

$$
pe^s-bpe^s=b(1-p).
$$

Factorizando,

$$
e^s(p-bp)=b(1-p).
$$

Observemos que

$$
p-bp=p(1-b).
$$

Por lo tanto,

$$
e^sp(1-b)=b(1-p).
$$

Despejando,

$$
e^s=\frac{b(1-p)}{p(1-b)}.
$$

Aplicando el logaritmo natural,

$$
\boxed{
s^*
=
\ln\left(
\frac{b(1-p)}{p(1-b)}
\right)
}
$$

Este es el punto crítico que buscamos.

### 8.5. Verificando que el punto crítico es un mínimo

En lugar de analizar directamente la segunda derivada de $f$, podemos considerar su logaritmo.

Definamos

$$
\varphi(s)=\ln f(s).
$$

Entonces,

$$
\varphi(s)
=
-sbn+n\ln(pe^s+1-p).
$$

Derivando,

$$
\varphi'(s)
=
-bn+
\frac{npe^s}{pe^s+1-p}.
$$

Y su segunda derivada es

$$
\varphi''(s)
=
\frac{npe^s(1-p)}{(pe^s+1-p)^2}.
$$

Como $0<p<1$, tenemos que

$$
\varphi''(s)>0.
$$

Por lo tanto, $\varphi$ es estrictamente convexa.

En consecuencia, el punto crítico $s^*$ es el único mínimo global de $\varphi$ y, por lo tanto, también de $f$.

Además, como estamos suponiendo que $b>p$, tenemos

$$
\frac{b(1-p)}{p(1-b)}>1.
$$

Por lo tanto,

$$
s^*>0.
$$

Esto garantiza que el valor encontrado pertenece al dominio de optimización.

### 8.6. Sustituyendo el valor óptimo

Recordemos que

$$
P(X\geq bn)
\leq
\inf_{s>0}
e^{-sbn}(pe^s+1-p)^n.
$$

Como el mínimo se alcanza en $s=s^*$, obtenemos

$$
P(X\geq bn)
\leq
e^{-s^*bn}(pe^{s^*}+1-p)^n.
$$

Sabemos que

$$
e^{s^*}
=
\frac{b(1-p)}{p(1-b)}.
$$

Por lo tanto,

$$
pe^{s^*}+1-p
=
\frac{b(1-p)}{1-b}+1-p.
$$

Factorizando $1-p$,

$$
\begin{aligned}
pe^{s^*}+1-p
&=
(1-p)\left(\frac{b}{1-b}+1\right)\\
&=
(1-p)\left(\frac{b+1-b}{1-b}\right)\\
&=
\frac{1-p}{1-b}.
\end{aligned}
$$

Por otra parte,

$$
e^{-s^*bn}
=
\left(
\frac{p(1-b)}{b(1-p)}
\right)^{bn}.
$$

Sustituyendo ambas expresiones,

$$
P(X\geq bn)
\leq
\left(
\frac{p(1-b)}{b(1-p)}
\right)^{bn}
\left(
\frac{1-p}{1-b}
\right)^n.
$$

Reorganizando los factores,

$$
\begin{aligned}
P(X\geq bn)
&\leq
\left(\frac{p}{b}\right)^{bn}
\left(\frac{1-b}{1-p}\right)^{bn}
\left(\frac{1-p}{1-b}\right)^n\\
&=
\left(\frac{p}{b}\right)^{bn}
\left(\frac{1-p}{1-b}\right)^{n(1-b)}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
P(X\geq bn)
\leq
\left(\frac{p}{b}\right)^{bn}
\left(\frac{1-p}{1-b}\right)^{n(1-b)}
}
$$

### 8.7. Expresión mediante la divergencia de Kullback-Leibler

Definamos

$$
D(b\|p)
=
b\ln\left(\frac{b}{p}\right)
+
(1-b)\ln\left(\frac{1-b}{1-p}\right).
$$

Esta expresión corresponde a la divergencia de Kullback-Leibler entre dos distribuciones Bernoulli con parámetros $b$ y $p$.

Observemos que

$$
\begin{aligned}
e^{-nD(b\|p)}
&=
\exp\left(
-nb\ln\left(\frac{b}{p}\right)
-n(1-b)\ln\left(\frac{1-b}{1-p}\right)
\right)\\
&=
\left(\frac{p}{b}\right)^{bn}
\left(\frac{1-p}{1-b}\right)^{n(1-b)}.
\end{aligned}
$$

Por lo tanto, la desigualdad de Chernoff para una variable aleatoria binomial puede expresarse como

$$
\boxed{
P(X\geq bn)
\leq
e^{-nD(b\|p)}
}
$$

para $0<p<b<1$.

### Observación

Como $b>p$, se cumple que

$$
D(b\|p)>0.
$$

En consecuencia,

$$
e^{-nD(b\|p)}\longrightarrow 0
$$

cuando $n\to\infty$.

Esto nos muestra que la probabilidad de que una variable aleatoria binomial supere una proporción $b$ mayor que su probabilidad de éxito $p$ decrece, al menos, exponencialmente con $n$.

A diferencia de la cota de Markov,

$$
P(X\geq bn)\leq\frac{p}{b},
$$

la cota de Chernoff incorpora explícitamente el número de ensayos $n$.
