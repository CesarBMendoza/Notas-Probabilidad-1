
# Ejercicios de esperanza matemática

## Ejercicio 1. Juego de canicas

Una caja contiene **5 canicas rojas y 5 canicas azules**. Se extraen aleatoriamente dos canicas, sin reemplazo.

Las reglas del juego son las siguientes:

- Si ambas canicas son del mismo color, ganamos \$1.10.
- Si ambas canicas son de diferente color, perdemos \$1.00.

**Pregunta:** ¿Cuál es el valor esperado del dinero ganado?

### Solución

Primero, definamos nuestra variable aleatoria.

Sea $X$ la variable aleatoria que representa la cantidad de dinero que ganamos al extraer dos canicas.

Entonces,

$$
X=
\begin{cases}
1.10, & \text{si ambas canicas son del mismo color},\\
-1.00, & \text{si las canicas son de diferente color}.
\end{cases}
$$

Por lo tanto, el recorrido de nuestra variable aleatoria es

$$
R_X=\{-1.00,1.10\}.
$$

### Paso 1. Calcular la función de masa de probabilidad

Comencemos calculando la probabilidad de ganar \$1.10.

$$
P_X(1.10)=P(X=1.10).
$$

Observemos que el evento $\{X=1.10\}$ ocurre cuando ambas canicas son del mismo color.

Es decir, podemos obtener:

- Dos canicas rojas.
- Dos canicas azules.

Como tenemos 5 canicas de cada color, el número de formas de seleccionar dos canicas del mismo color es

$$
\binom{5}{2}+\binom{5}{2}.
$$

Por otra parte, el número total de formas de seleccionar dos canicas de las 10 disponibles es

$$
\binom{10}{2}.
$$

Por lo tanto,

$$
\begin{aligned}
P_X(1.10)
&=
\frac{\binom{5}{2}+\binom{5}{2}}{\binom{10}{2}}\\
&=
\frac{10+10}{45}\\
&=
\frac{20}{45}\\
&=
\frac{4}{9}.
\end{aligned}
$$

Así,

$$
\boxed{P_X(1.10)=\frac{4}{9}}
$$

Ahora, como únicamente tenemos dos posibles resultados,

$$
P_X(-1.00)=1-P_X(1.10).
$$

Por lo tanto,

$$
\begin{aligned}
P_X(-1.00)
&=1-\frac{4}{9}\\
&=\frac{5}{9}.
\end{aligned}
$$

De donde

$$
\boxed{P_X(-1.00)=\frac{5}{9}}
$$

### Paso 2. Determinar la función de masa

Con los resultados anteriores, podemos escribir la función de masa de probabilidad de $X$:

$$
P_X(k)=
\begin{cases}
\frac{4}{9}, & \text{si }k=1.10,\\[4pt]
\frac{5}{9}, & \text{si }k=-1.00,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

### Paso 3. Calcular la esperanza matemática

Recordemos que, para una variable aleatoria discreta,

$$
\mathbb{E}[X]
=
\sum_{x\in R_X}xP(X=x).
$$

En nuestro caso,

$$
\begin{aligned}
\mathbb{E}[X]
&=
(1.10)P(X=1.10)+(-1.00)P(X=-1.00)\\
&=
(1.10)\left(\frac{4}{9}\right)
+(-1.00)\left(\frac{5}{9}\right)\\
&=
\frac{22}{45}-\frac{25}{45}\\
&=
-\frac{3}{45}\\
&=
-\frac{1}{15}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]=-\frac{1}{15}\approx -0.0667
}
$$

### Interpretación

Esto significa que, en promedio, **perderemos aproximadamente 6.67 centavos por partida**.

Es importante observar que la esperanza matemática no necesariamente es un valor que la variable aleatoria pueda tomar.

En este caso, únicamente podemos ganar \$1.10 o perder \$1.00, pero nuestra ganancia promedio por partida es aproximadamente $-\$0.0667$.

## Ejercicio 2. Lanzamiento de dos monedas

Se lanzan dos monedas.

La primera cae en cara con probabilidad $0.6$, mientras que la segunda cae en cara con probabilidad $0.7$.

Supongamos que los resultados de los lanzamientos son independientes.

Sea $X$ la variable aleatoria que representa el número total de caras obtenidas.

Se pide:

**(a)** Calcular $P(X=1)$.

**(b)** Determinar $\mathbb{E}[X]$.

### Solución

Como $X$ representa el número total de caras obtenidas al lanzar dos monedas, tenemos que

$$
R_X=\{0,1,2\}.
$$

Donde:

- $X=0$: ninguna moneda cae en cara.
- $X=1$: exactamente una moneda cae en cara.
- $X=2$: ambas monedas caen en cara.

Para resolver el ejercicio, primero determinaremos la función de masa de probabilidad de $X$.

### Paso 1. Definir los eventos

Llamemos:

- $C^{(1)}$: evento de que la primera moneda cae en cara.
- $T^{(1)}$: evento de que la primera moneda cae en cruz.
- $C^{(2)}$: evento de que la segunda moneda cae en cara.
- $T^{(2)}$: evento de que la segunda moneda cae en cruz.

Sabemos que

$$
P(C^{(1)})=0.6
$$

y

$$
P(T^{(1)})=0.4.
$$

Análogamente, para la segunda moneda,

$$
P(C^{(2)})=0.7
$$

y

$$
P(T^{(2)})=0.3.
$$

Como los lanzamientos son independientes, podemos utilizar la propiedad

$$
P(A\cap B)=P(A)P(B)
$$

cuando $A$ y $B$ corresponden a resultados de monedas distintas.

### Paso 2. Calcular la función de masa de probabilidad

**Caso 1. Calcular $P(X=0)$.**

El evento $\{X=0\}$ ocurre cuando ambas monedas caen en cruz.

Por lo tanto,

$$
\begin{aligned}
P_X(0)
&=P(X=0)\\
&=P(T^{(1)}\cap T^{(2)})\\
&=P(T^{(1)})P(T^{(2)})\\
&=(0.4)(0.3)\\
&=0.12.
\end{aligned}
$$

Así,

$$
\boxed{P(X=0)=0.12}
$$

**Caso 2. Calcular $P(X=1)$.**

El evento $\{X=1\}$ ocurre cuando exactamente una moneda cae en cara.

Tenemos dos posibilidades:

- La primera moneda cae en cara y la segunda en cruz.
- La primera moneda cae en cruz y la segunda en cara.

Estos eventos son ajenos, por lo que podemos sumar sus probabilidades.

Entonces,

$$
\begin{aligned}
P_X(1)
&=P(X=1)\\
&=P(C^{(1)}\cap T^{(2)})
+P(T^{(1)}\cap C^{(2)})\\
&=P(C^{(1)})P(T^{(2)})
+P(T^{(1)})P(C^{(2)})\\
&=(0.6)(0.3)+(0.4)(0.7)\\
&=0.18+0.28\\
&=0.46.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{P(X=1)=0.46}
$$

**Caso 3. Calcular $P(X=2)$.**

El evento $\{X=2\}$ ocurre cuando ambas monedas caen en cara.

Por lo tanto,

$$
\begin{aligned}
P_X(2)
&=P(X=2)\\
&=P(C^{(1)}\cap C^{(2)})\\
&=P(C^{(1)})P(C^{(2)})\\
&=(0.6)(0.7)\\
&=0.42.
\end{aligned}
$$

Así,

$$
\boxed{P(X=2)=0.42}
$$

Observemos que

$$
0.12+0.46+0.42=1.
$$

Por lo tanto, las probabilidades obtenidas son consistentes.

### Paso 3. Calcular la esperanza matemática

Con los resultados anteriores, nuestra función de masa de probabilidad es

$$
P_X(k)=
\begin{cases}
0.12, & \text{si }k=0,\\
0.46, & \text{si }k=1,\\
0.42, & \text{si }k=2,\\
0, & \text{en otro caso}.
\end{cases}
$$

Ahora, por definición,

$$
\mathbb{E}[X]
=
\sum_{x\in R_X}xP(X=x).
$$

Sustituyendo los valores,

$$
\begin{aligned}
\mathbb{E}[X]
&=
0P(X=0)+1P(X=1)+2P(X=2)\\
&=
0(0.12)+1(0.46)+2(0.42)\\
&=
0.46+0.84\\
&=1.30.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{\mathbb{E}[X]=1.30}
$$

### Interpretación

Esto significa que, si repetimos muchas veces el experimento de lanzar ambas monedas, obtendremos en promedio $1.30$ caras por experimento.

### Observación

También podemos calcular la esperanza utilizando la linealidad.

Definamos

$$
X=X_1+X_2,
$$

donde

$$
X_i=
\begin{cases}
1, & \text{si la moneda }i\text{ cae en cara},\\
0, & \text{si la moneda }i\text{ cae en cruz}.
\end{cases}
$$

Entonces,

$$
X_1\sim\operatorname{Bernoulli}(0.6)
$$

y

$$
X_2\sim\operatorname{Bernoulli}(0.7).
$$

Por la linealidad de la esperanza,

$$
\begin{aligned}
\mathbb{E}[X]
&=\mathbb{E}[X_1+X_2]\\
&=\mathbb{E}[X_1]+\mathbb{E}[X_2]\\
&=0.6+0.7\\
&=1.30.
\end{aligned}
$$

Obtenemos el mismo resultado.

## Ejercicio 3. Selección de estudiantes y conductores

Cuatro autobuses que transportan a **148 estudiantes** de una misma escuela llegan a un estadio de fútbol.

Los autobuses transportan, respectivamente,

$$
40,\quad 33,\quad 25,\quad 50
$$

estudiantes.

Se selecciona aleatoriamente a uno de los 148 estudiantes.

Sea $X$ la variable aleatoria que representa el número de estudiantes que viajaban en el mismo autobús que el estudiante seleccionado.

También se selecciona aleatoriamente a uno de los cuatro conductores.

Sea $Y$ la variable aleatoria que representa el número de estudiantes que viajaban en el autobús del conductor seleccionado.

**Pregunta:**

¿Cuál de las dos esperanzas, $\mathbb{E}[X]$ o $\mathbb{E}[Y]$, crees que es mayor? ¿Por qué?

### Solución

Observemos primero que ambas variables aleatorias pueden tomar los mismos valores:

$$
R_X=R_Y=\{25,33,40,50\}.
$$

Sin embargo, sus distribuciones de probabilidad son distintas.

### Paso 1. Calcular la función de masa de $X$

Recordemos que $X$ representa el número de estudiantes que viajaban en el mismo autobús que el estudiante seleccionado.

Como seleccionamos aleatoriamente a uno de los 148 estudiantes, todos tienen la misma probabilidad de ser elegidos.

Por lo tanto, la probabilidad de seleccionar a un estudiante de un autobús depende de cuántos estudiantes viajen en él.

**Autobús con 40 estudiantes:**

$$
P(X=40)=\frac{40}{148}.
$$

**Autobús con 50 estudiantes:**

$$
P(X=50)=\frac{50}{148}.
$$

**Autobús con 25 estudiantes:**

$$
P(X=25)=\frac{25}{148}.
$$

**Autobús con 33 estudiantes:**

$$
P(X=33)=\frac{33}{148}.
$$

Por lo tanto, la función de masa de probabilidad de $X$ es

$$
P_X(k)=
\begin{cases}
\frac{25}{148}, & \text{si }k=25,\\[4pt]
\frac{33}{148}, & \text{si }k=33,\\[4pt]
\frac{40}{148}, & \text{si }k=40,\\[4pt]
\frac{50}{148}, & \text{si }k=50,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

Observemos que

$$
\frac{25+33+40+50}{148}=1.
$$

### Paso 2. Calcular la función de masa de $Y$

Recordemos que $Y$ representa el número de estudiantes que viajaban en el autobús del conductor seleccionado.

Como tenemos cuatro conductores y todos tienen la misma probabilidad de ser seleccionados,

$$
P(Y=25)=P(Y=33)=P(Y=40)=P(Y=50)=\frac{1}{4}.
$$

Por lo tanto,

$$
P_Y(k)=
\begin{cases}
\frac{1}{4}, & \text{si }k\in\{25,33,40,50\},\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

### Paso 3. Comparar las distribuciones

Observemos que

$$
P(X=50)=\frac{50}{148}\approx 0.338,
$$

mientras que

$$
P(Y=50)=\frac{1}{4}=0.25.
$$

Por otra parte,

$$
P(X=25)=\frac{25}{148}\approx 0.169,
$$

mientras que

$$
P(Y=25)=\frac{1}{4}=0.25.
$$

Esto significa que, al seleccionar aleatoriamente a un estudiante, tenemos mayor probabilidad de elegir a alguien que viaje en un autobús grande.

En cambio, al seleccionar aleatoriamente a un conductor, todos los autobuses tienen la misma probabilidad de ser seleccionados.

Por esta razón, podemos esperar que

$$
\boxed{\mathbb{E}[X]>\mathbb{E}[Y]}.
$$

Ahora comprobemos esta afirmación mediante el cálculo de ambas esperanzas.

### Paso 4. Calcular $\mathbb{E}[X]$

Por definición,

$$
\mathbb{E}[X]
=
\sum_{x\in R_X}xP(X=x).
$$

Como

$$
R_X=\{25,33,40,50\},
$$

tenemos

$$
\begin{aligned}
\mathbb{E}[X]
&=
25P(X=25)+33P(X=33)\\
&\quad+
40P(X=40)+50P(X=50).
\end{aligned}
$$

Sustituyendo las probabilidades,

$$
\begin{aligned}
\mathbb{E}[X]
&=
25\left(\frac{25}{148}\right)
+
33\left(\frac{33}{148}\right)\\
&\quad+
40\left(\frac{40}{148}\right)
+
50\left(\frac{50}{148}\right)\\
&=
\frac{625+1089+1600+2500}{148}\\
&=
\frac{5814}{148}\\
&=
\frac{2907}{74}\\
&\approx39.28.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]=\frac{2907}{74}\approx39.28
}
$$

### Interpretación

Esto significa que, al seleccionar aleatoriamente a un estudiante, el número esperado de estudiantes que viajaban en su mismo autobús es aproximadamente $39.28$.

---

### Paso 5. Calcular $\mathbb{E}[Y]$

Recordemos que

$$
P(Y=25)=P(Y=33)=P(Y=40)=P(Y=50)=\frac{1}{4}.
$$

Entonces,

$$
\mathbb{E}[Y]
=
\sum_{y\in R_Y}yP(Y=y).
$$

Sustituyendo,

$$
\begin{aligned}
\mathbb{E}[Y]
&=
25\left(\frac{1}{4}\right)
+
33\left(\frac{1}{4}\right)\\
&\quad+
40\left(\frac{1}{4}\right)
+
50\left(\frac{1}{4}\right)\\
&=
\frac{25+33+40+50}{4}\\
&=
\frac{148}{4}\\
&=37.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{\mathbb{E}[Y]=37}
$$

### Interpretación

Esto significa que, al seleccionar aleatoriamente a uno de los cuatro conductores, el número esperado de estudiantes transportados por su autobús es de $37$.

Finalmente,

$$
\boxed{
\mathbb{E}[X]\approx39.28>37=\mathbb{E}[Y]
}
$$

Por lo tanto, nuestra predicción era correcta.

### Conclusión del ejercicio de autobuses

Aunque $X$ y $Y$ toman exactamente los mismos valores,

$$
R_X=R_Y=\{25,33,40,50\},
$$

sus distribuciones de probabilidad son distintas.

La diferencia está en la forma en que realizamos la selección aleatoria.

**Cuando seleccionamos a un estudiante:**

Los autobuses con mayor cantidad de estudiantes tienen mayor probabilidad de ser seleccionados.

Por ejemplo,

$$
P(X=50)=\frac{50}{148},
$$

mientras que

$$
P(X=25)=\frac{25}{148}.
$$

Por lo tanto, los autobuses grandes tienen mayor peso en el cálculo de $\mathbb{E}[X]$.

**Cuando seleccionamos a un conductor:**

Cada autobús tiene la misma probabilidad de ser seleccionado, independientemente de la cantidad de estudiantes que transporte.

Es decir,

$$
P(Y=k)=\frac{1}{4},
\qquad k\in\{25,33,40,50\}.
$$

Esto nos muestra que la esperanza matemática depende tanto de los valores que toma una variable aleatoria como de las probabilidades asociadas a dichos valores.

En particular,

$$
\boxed{
\mathbb{E}[X]\approx39.28>37=\mathbb{E}[Y]
}
$$

## Ejercicio 4. Tiempo de vida de un tubo electrónico

El tiempo de vida, medido en horas, de un tubo electrónico es una variable aleatoria continua $X$ que tiene la siguiente función de densidad de probabilidad:

$$
f_X(x)=xe^{-x}\mathbb{1}_{[0,\infty)}(x).
$$

Es decir,

$$
f_X(x)=
\begin{cases}
xe^{-x}, & \text{si }x\geq0,\\
0, & \text{si }x<0.
\end{cases}
$$

**Pregunta:** ¿Cuál es el tiempo de vida esperado de dicho tubo electrónico?

### Solución

Recordemos que, si $X$ es una variable aleatoria continua con función de densidad $f_X$, entonces

$$
\mathbb{E}[X]
=
\int_{-\infty}^{\infty}xf_X(x)\,dx.
$$

Como nuestra función de densidad únicamente es distinta de cero para $x\geq0$,

$$
\mathbb{E}[X]
=
\int_0^\infty x(xe^{-x})\,dx.
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X]
=
\int_0^\infty x^2e^{-x}\,dx
}
$$

Ahora resolveremos esta integral utilizando integración por partes.

### Cálculo de la integral

Queremos calcular

$$
\mathbb{E}[X]
=
\int_0^\infty x^2e^{-x}\,dx.
$$

Utilizaremos integración por partes.

Recordemos que

$$
\int u\,dv=uv-\int v\,du.
$$

Tomemos

$$
u=x^2,
\qquad
dv=e^{-x}\,dx.
$$

Entonces,

$$
du=2x\,dx,
\qquad
v=-e^{-x}.
$$

Por lo tanto,

$$
\begin{aligned}
\mathbb{E}[X]
&=
\int_0^\infty x^2e^{-x}\,dx\\
&=
\left[-x^2e^{-x}\right]_0^\infty
+
2\int_0^\infty xe^{-x}\,dx.
\end{aligned}
$$

Observemos que

$$
\lim_{x\to\infty}x^2e^{-x}=0.
$$

Por lo tanto,

$$
\left[-x^2e^{-x}\right]_0^\infty=0.
$$

Así,

$$
\mathbb{E}[X]
=
2\int_0^\infty xe^{-x}\,dx.
$$

Aplicando nuevamente integración por partes,

$$
u=x,
\qquad
dv=e^{-x}\,dx.
$$

Entonces,

$$
du=dx,
\qquad
v=-e^{-x}.
$$

De donde

$$
\begin{aligned}
\mathbb{E}[X]
&=
2\left(
\left[-xe^{-x}\right]_0^\infty
+
\int_0^\infty e^{-x}\,dx
\right)\\
&=
2\int_0^\infty e^{-x}\,dx\\
&=
2\left[-e^{-x}\right]_0^\infty\\
&=
2(0-(-1))\\
&=2.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{\mathbb{E}[X]=2}
$$

### Interpretación

El tiempo de vida esperado de dicho tubo electrónico es de **2 horas**.

## Ejercicio 5. Esperanza de una potencia de una variable uniforme

Sea $X$ una variable aleatoria tal que

$$
X\sim U(0,1).
$$

Calcular

$$
\mathbb{E}[X^n],
$$

donde $n\in\mathbb{N}$ y $n\geq1$.

### Solución

Definamos una nueva variable aleatoria

$$
Y=X^n.
$$

Nuestro objetivo es calcular

$$
\mathbb{E}[Y].
$$

Como $Y$ es una variable aleatoria continua, podemos utilizar

$$
\mathbb{E}[Y]
=
\int_{-\infty}^{\infty}yf_Y(y)\,dy.
$$

Sin embargo, para aplicar esta fórmula necesitamos conocer la función de densidad de probabilidad de $Y$.

Por lo tanto, primero determinaremos la distribución de $Y$.

### Paso 1. Determinar el recorrido de $Y$

Como

$$
X\sim U(0,1),
$$

sabemos que

$$
0<X<1
$$

casi seguramente.

Por lo tanto, al elevar a la potencia $n\geq1$,

$$
0<X^n<1.
$$

Como $Y=X^n$, obtenemos

$$
0<Y<1.
$$

Así, podemos tomar como recorrido de $Y$ el intervalo

$$
R_Y=(0,1).
$$

### Paso 2. Calcular la función de distribución acumulada de $Y$

Recordemos que, si

$$
X\sim U(0,1),
$$

entonces su función de distribución acumulada es

$$
F_X(x)=
\begin{cases}
0, & \text{si }x\leq0,\\
x, & \text{si }0<x<1,\\
1, & \text{si }x\geq1.
\end{cases}
$$

Por definición, la función de distribución acumulada de $Y$ es

$$
F_Y(y)=P(Y\leq y).
$$

Como

$$
Y=X^n,
$$

tenemos

$$
F_Y(y)=P(X^n\leq y).
$$

Analicemos tres casos.

**Caso 1. Si $y\leq0$.**

Como $Y>0$ casi seguramente,

$$
F_Y(y)=P(Y\leq y)=0.
$$

**Caso 2. Si $0<y<1$.**

Como la función

$$
g(x)=x^n
$$

es estrictamente creciente en $(0,1)$ para $n\geq1$, podemos utilizar su inversa.

Entonces,

$$
\begin{aligned}
F_Y(y)
&=P(X^n\leq y)\\
&=P(X\leq y^{1/n})\\
&=F_X(y^{1/n}).
\end{aligned}
$$

Como $0<y^{1/n}<1$, tenemos

$$
F_X(y^{1/n})=y^{1/n}.
$$

Por lo tanto,

$$
\boxed{
F_Y(y)=y^{1/n}
}
$$

para $0<y<1$.

**Caso 3. Si $y\geq1$.**

Como $Y<1$ casi seguramente,

$$
F_Y(y)=P(Y\leq y)=1.
$$

Por lo tanto, la función de distribución acumulada de $Y$ es

$$
\boxed{
F_Y(y)=
\begin{cases}
0, & \text{si }y\leq0,\\[4pt]
y^{1/n}, & \text{si }0<y<1,\\[4pt]
1, & \text{si }y\geq1.
\end{cases}
}
$$

### Paso 3. Calcular la función de densidad de $Y$

Recordemos que, para una variable aleatoria absolutamente continua,

$$
f_Y(y)=\frac{d}{dy}F_Y(y)
$$

en los puntos donde existe la derivada.

Por lo tanto, para $0<y<1$,

$$
\begin{aligned}
f_Y(y)
&=
\frac{d}{dy}y^{1/n}\\
&=
\frac{1}{n}y^{\frac{1}{n}-1}.
\end{aligned}
$$

Así,

$$
\boxed{
f_Y(y)=
\frac{1}{n}y^{\frac{1}{n}-1}
\mathbb{1}_{(0,1)}(y)
}
$$

Es decir,

$$
f_Y(y)=
\begin{cases}
\frac{1}{n}y^{\frac{1}{n}-1}, & \text{si }0<y<1,\\[4pt]
0, & \text{en otro caso}.
\end{cases}
$$

Ahora podemos calcular la esperanza de $Y$.

### Paso 4. Calcular la esperanza de $Y$

Recordemos que

$$
\mathbb{E}[Y]
=
\int_{-\infty}^{\infty}yf_Y(y)\,dy.
$$

Como

$$
f_Y(y)
=
\frac{1}{n}y^{\frac{1}{n}-1}
\mathbb{1}_{(0,1)}(y),
$$

tenemos

$$
\begin{aligned}
\mathbb{E}[Y]
&=
\int_0^1
y\left(
\frac{1}{n}y^{\frac{1}{n}-1}
\right)\,dy\\
&=
\frac{1}{n}
\int_0^1 y^{\frac{1}{n}}\,dy\\
&=
\frac{1}{n}
\left[
\frac{y^{\frac{1}{n}+1}}
{\frac{1}{n}+1}
\right]_0^1\\
&=
\frac{1}{n}
\left(
\frac{1}{\frac{1+n}{n}}
\right)\\
&=
\frac{1}{n}\frac{n}{n+1}\\
&=
\frac{1}{n+1}.
\end{aligned}
$$

Como $Y=X^n$, concluimos que

$$
\boxed{
\mathbb{E}[X^n]=\frac{1}{n+1}
}
$$

para $X\sim U(0,1)$ y $n\geq1$.

### Observación. Una forma alternativa

También podemos calcular esta esperanza directamente, sin necesidad de determinar la distribución de $Y$.

Recordemos que, si $X$ es una variable aleatoria continua y $g$ es una función medible e integrable respecto de la distribución de $X$, entonces

$$
\mathbb{E}[g(X)]
=
\int_{-\infty}^{\infty}g(x)f_X(x)\,dx.
$$

Tomando

$$
g(x)=x^n,
$$

y como $X\sim U(0,1)$,

$$
f_X(x)=\mathbb{1}_{(0,1)}(x),
$$

obtenemos

$$
\begin{aligned}
\mathbb{E}[X^n]
&=
\int_0^1x^n\,dx\\
&=
\left[\frac{x^{n+1}}{n+1}\right]_0^1\\
&=
\frac{1}{n+1}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
\mathbb{E}[X^n]=\frac{1}{n+1}
}
$$

Esta segunda demostración es más corta, pero la primera nos permite practicar el cálculo de la distribución de una transformación de una variable aleatoria.

## Ejercicio 6. Esperanza de variables aleatorias continuas

Sea $X$ una variable aleatoria continua cuya función de densidad de probabilidad está dada por alguna de las siguientes expresiones.

Calcular la esperanza matemática de $X$ en cada caso.

**(a)**

$$
f_X(x)=
\frac{1}{4}xe^{-x/2}
\mathbb{1}_{(0,\infty)}(x).
$$

**(b)**

$$
f_X(x)=
c(1-x^2)\mathbb{1}_{(-1,1)}(x).
$$

**(c)**

$$
f_X(x)=
\frac{5}{x^2}\mathbb{1}_{(5,\infty)}(x).
$$

### Inciso (a)

Tenemos la función de densidad

$$
f_X(x)=
\frac{1}{4}xe^{-x/2}
\mathbb{1}_{(0,\infty)}(x).
$$

Por definición,

$$
\mathbb{E}[X]
=
\int_{-\infty}^{\infty}xf_X(x)\,dx.
$$

Sustituyendo,

$$
\begin{aligned}
\mathbb{E}[X]
&=
\int_0^\infty
x\left(\frac{1}{4}xe^{-x/2}\right)\,dx\\
&=
\frac{1}{4}\int_0^\infty x^2e^{-x/2}\,dx.
\end{aligned}
$$

Resolveremos la integral mediante integración por partes.

Tomemos

$$
u=x^2,
\qquad
dv=e^{-x/2}\,dx.
$$

Entonces,

$$
du=2x\,dx,
\qquad
v=-2e^{-x/2}.
$$

Por lo tanto,

$$
\begin{aligned}
\mathbb{E}[X]
&=
\frac{1}{4}
\left(
\left[-2x^2e^{-x/2}\right]_0^\infty
+
4\int_0^\infty xe^{-x/2}\,dx
\right).
\end{aligned}
$$

Como

$$
\lim_{x\to\infty}x^2e^{-x/2}=0,
$$

obtenemos

$$
\mathbb{E}[X]
=
\int_0^\infty xe^{-x/2}\,dx.
$$

Aplicando nuevamente integración por partes,

$$
u=x,
\qquad
dv=e^{-x/2}\,dx.
$$

Entonces,

$$
du=dx,
\qquad
v=-2e^{-x/2}.
$$

De donde

$$
\begin{aligned}
\mathbb{E}[X]
&=
\left[-2xe^{-x/2}\right]_0^\infty
+
2\int_0^\infty e^{-x/2}\,dx\\
&=
2\int_0^\infty e^{-x/2}\,dx\\
&=
2\left[-2e^{-x/2}\right]_0^\infty\\
&=
2(0-(-2))\\
&=4.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{\mathbb{E}[X]=4}
$$

---

### Inciso (b)

Tenemos la función de densidad

$$
f_X(x)
=
c(1-x^2)\mathbb{1}_{(-1,1)}(x).
$$

Para calcular la esperanza, primero necesitamos determinar la constante $c$.

Recordemos que una función de densidad debe satisfacer

$$
\int_{-\infty}^{\infty}f_X(x)\,dx=1.
$$

Por lo tanto,

$$
\begin{aligned}
1
&=
\int_{-1}^{1}c(1-x^2)\,dx\\
&=
c\int_{-1}^{1}(1-x^2)\,dx\\
&=
c\left[x-\frac{x^3}{3}\right]_{-1}^{1}\\
&=
c\left(
1-\frac{1}{3}
-\left(-1+\frac{1}{3}\right)
\right)\\
&=
c\left(\frac{4}{3}\right).
\end{aligned}
$$

Despejando,

$$
\boxed{c=\frac{3}{4}}
$$

Así, nuestra función de densidad es

$$
f_X(x)
=
\frac{3}{4}(1-x^2)
\mathbb{1}_{(-1,1)}(x).
$$

Ahora calculamos la esperanza:

$$
\begin{aligned}
\mathbb{E}[X]
&=
\int_{-\infty}^{\infty}xf_X(x)\,dx\\
&=
\int_{-1}^{1}
x\left(\frac{3}{4}(1-x^2)\right)\,dx\\
&=
\frac{3}{4}
\int_{-1}^{1}(x-x^3)\,dx\\
&=
\frac{3}{4}
\left[
\frac{x^2}{2}-\frac{x^4}{4}
\right]_{-1}^{1}\\
&=
0.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{\mathbb{E}[X]=0}
$$

### Observación

Podemos obtener este resultado sin calcular explícitamente la integral.

Observemos que

$$
g(x)=x-x^3
$$

es una función impar, pues

$$
g(-x)=-g(x).
$$

Como estamos integrando sobre un intervalo simétrico respecto del origen,

$$
\int_{-1}^{1}g(x)\,dx=0.
$$

Por lo tanto,

$$
\mathbb{E}[X]=0.
$$

---

### Inciso (c)

Tenemos la función de densidad

$$
f_X(x)
=
\frac{5}{x^2}\mathbb{1}_{(5,\infty)}(x).
$$

Por definición,

$$
\mathbb{E}[X]
=
\int_{-\infty}^{\infty}xf_X(x)\,dx.
$$

Sustituyendo,

$$
\begin{aligned}
\mathbb{E}[X]
&=
\int_5^\infty
x\left(\frac{5}{x^2}\right)\,dx\\
&=
\int_5^\infty\frac{5}{x}\,dx\\
&=
5\int_5^\infty\frac{1}{x}\,dx.
\end{aligned}
$$

Recordemos que

$$
\int\frac{1}{x}\,dx=\ln|x|+C.
$$

Por lo tanto,

$$
\begin{aligned}
\mathbb{E}[X]
&=
5\left[\ln x\right]_5^\infty\\
&=
5\lim_{b\to\infty}
\left(\ln b-\ln5\right)\\
&=+\infty.
\end{aligned}
$$

Así,

$$
\boxed{\mathbb{E}[X]=+\infty}
$$

### Interpretación

En este caso, la variable aleatoria $X$ tiene una función de densidad válida, pero su esperanza no es finita.

Esto ocurre porque la función de densidad decrece suficientemente lento como para que la integral

$$
\int_5^\infty xf_X(x)\,dx
$$

diverja.

Por lo tanto, aunque $X$ es una variable aleatoria continua con distribución de probabilidad bien definida, no tiene una esperanza finita.

### Observación final

La existencia de una función de densidad de probabilidad no garantiza que la esperanza matemática sea finita.

Para que una variable aleatoria continua tenga esperanza finita, es suficiente y necesario que

$$
\int_{-\infty}^{\infty}|x|f_X(x)\,dx<\infty.
$$

Es decir, que la variable aleatoria sea integrable.
