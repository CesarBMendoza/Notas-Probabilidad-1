# Ejemplos de probabilidad y conteo

## Ejemplo 1. Reparto de ases entre 4 personas

Se tienen 4 personas y se reparten las 52 cartas de una baraja, de manera que cada persona recibe 13 cartas.

¿Cuál es la probabilidad de que **a cada persona le toque exactamente un as**?

Recordemos que en una baraja hay 4 ases y 48 cartas que no son ases.

Sea

$E=\{\text{cada una de las 4 personas recibe exactamente un as}\}$.

### Forma 1: contando todos los repartos

Primero calculemos el tamaño del espacio muestral.

Para la primera persona podemos escoger 13 de las 52 cartas:

$\binom{52}{13}$.

Una vez hecho esto quedan 39 cartas, de las cuales escogemos 13 para la segunda persona:

$\binom{39}{13}$.

Después quedan 26 cartas, de las cuales escogemos 13 para la tercera persona:

$\binom{26}{13}$.

Finalmente, las últimas 13 cartas corresponden a la cuarta persona:

$\binom{13}{13}=1$.

Por lo tanto,

$|\Omega| = \binom{52}{13} \binom{39}{13} \binom{26}{13} \binom{13}{13}$.

Usando la definición del coeficiente binomial,

$|\Omega| = \frac{52!}{13!39!} \frac{39!}{13!26!} \frac{26!}{13!13!}$.

Cancelando términos,

$|\Omega| = \frac{52!}{(13!)^4}$.

Ahora contemos los elementos del evento favorable $E$.

Queremos que cada persona reciba exactamente uno de los cuatro ases.

Primero asignamos los 4 ases a las 4 personas.

Como los ases son distintos y las personas también son distintas, hay

$4!$

formas de hacer esta asignación.

Una vez repartidos los ases, cada persona tiene 1 carta y necesita recibir otras 12 cartas.

Quedan 48 cartas, ninguna de las cuales es un as.

Para la primera persona escogemos 12 de estas 48 cartas:

$\binom{48}{12}$.

Después quedan 36 cartas, de las cuales escogemos 12 para la segunda persona:

$\binom{36}{12}$.

Después escogemos 12 de las 24 restantes para la tercera persona:

$\binom{24}{12}$.

Finalmente, las últimas 12 corresponden a la cuarta persona:

$\binom{12}{12}=1$.

Por lo tanto,

$|E| = 4! \binom{48}{12} \binom{36}{12} \binom{24}{12} \binom{12}{12}$.

Equivalentemente,

$|E| = 4!\frac{48!}{(12!)^4}$.

Así,

$\mathbb{P}(E) = \frac{|E|}{|\Omega|}$.

Entonces,

$\mathbb{P}(E) = \frac{ 4!\dfrac{48!}{(12!)^4} }{ \dfrac{52!}{(13!)^4} }$.

Reordenando,

$\mathbb{P}(E) = \frac{4!48!(13!)^4}{52!(12!)^4}$.

Como

$\frac{13!}{12!}=13$,

tenemos

$\frac{(13!)^4}{(12!)^4}=13^4$.

Por lo tanto,

$\mathbb{P}(E) = \frac{4!48!13^4}{52!}$.

Pero

$\binom{52}{4} = \frac{52!}{4!48!}$,

de donde

$\frac{4!48!}{52!} = \frac{1}{\binom{52}{4}}$.

Así,

$\boxed{ \mathbb{P}(E) = \frac{13^4}{\binom{52}{4}} }$.

Numéricamente,

$\mathbb{P}(E)\approx 0.1055$.

Por lo tanto, la probabilidad de que cada una de las cuatro personas reciba exactamente un as es aproximadamente

$\boxed{10.55\%}$.

### Forma 2: observando solamente las posiciones de los ases

Existe una manera más sencilla de hacer el conteo.

Imaginemos las 52 cartas repartidas en 52 posiciones:

- las primeras 13 posiciones corresponden a la primera persona;
- las siguientes 13 a la segunda;
- las siguientes 13 a la tercera;
- las últimas 13 a la cuarta.

Solo nos interesa saber en qué posiciones aparecen los 4 ases.

Podemos escoger las 4 posiciones de los ases entre las 52 posiciones de

$\binom{52}{4}$

formas.

Por lo tanto,

$|\Omega|=\binom{52}{4}$.

Ahora queremos que haya exactamente un as en el grupo de 13 posiciones correspondiente a cada persona.

Para la primera persona podemos escoger cualquiera de sus 13 posiciones.

Para la segunda persona también tenemos 13 posibilidades.

Lo mismo para la tercera y la cuarta.

Por lo tanto,

$|E| = 13\cdot13\cdot13\cdot13 = 13^4$.

Así,

$\mathbb{P}(E) = \frac{13^4}{\binom{52}{4}}$,

que coincide con el resultado anterior.

# Ejemplo 2. Ana, Beto y Claudia tiran un dado

Ana, Beto y Claudia tiran un dado siguiendo inicialmente el orden

$A,B,C,A,B,C,\ldots$

Cuando a una persona le sale un $6$, esa persona sale del juego.

El juego continúa con las personas que todavía no han obtenido un $6$.

Por ejemplo, si Beto sale primero, el orden continúa como

$A,C,A,C,\ldots$

hasta que salga otra persona.

Finalmente queda una sola persona, quien continúa tirando hasta obtener un $6$.

## El evento de que el juego termine

Definamos

$D_n = \{\text{la tercera y última persona sale del juego en su }n\text{-ésimo tiro}\}$.

Entonces,

$\bigcup_{n=1}^{\infty}D_n$

representa el evento

$\{\text{el juego termina eventualmente}\}$.

Es importante observar que, si nuestro espacio muestral $\Omega$ contiene todas las sucesiones infinitas posibles de resultados de los dados, entonces no es correcto escribir literalmente

$\Omega=\bigcup_{n=1}^{\infty}D_n$.

La razón es que existen sucesiones en las que alguna persona nunca obtiene un $6$.

Por ejemplo, matemáticamente es posible considerar una sucesión en la que Ana obtiene siempre números distintos de $6$.

Sin embargo, la probabilidad de que esto ocurra es cero.

Para verlo, la probabilidad de que Ana no obtenga un $6$ en sus primeros $n$ tiros es

$\left(\frac{5}{6}\right)^n$.

Por lo tanto,

$\mathbb{P}(\text{Ana nunca obtiene un }6) = \lim_{n\to\infty} \left(\frac{5}{6}\right)^n = 0$.

Lo mismo ocurre con Beto y Claudia.

Así,

$\mathbb{P} \left( \bigcup_{n=1}^{\infty}D_n \right) = 1$.

Es decir,

$\boxed{ \mathbb{P}(\text{el juego termina})=1 }$.

El complemento

$\left( \bigcup_{n=1}^{\infty}D_n \right)^c$

representa el evento

$\{\text{el juego nunca termina}\}$.

Por tanto,

$\mathbb{P} \left( \left( \bigcup_{n=1}^{\infty}D_n \right)^c \right) = 0$.

En otras palabras, aunque existen sucesiones infinitas en las que el juego nunca termina, dichas sucesiones forman un evento de probabilidad cero.

# Probabilidad de que Ana sea la segunda persona en salir

Ahora queremos calcular

$\mathbb{P}(E)$,

donde

$E = \{\text{Ana es la segunda persona en salir del juego}\}$.

Para hacer el conteo, vamos a separar los casos según el número de tiros que Ana necesita para obtener su primer $6$.

Supongamos que Ana obtiene su primer $6$ en su tiro número

$r+1$.

Esto significa que Ana no obtuvo ningún $6$ durante sus primeros $r$ tiros.

Para que Ana sea la **segunda persona en salir**, durante esos primeros $r$ tiros debe ocurrir además que:

1. una de las otras dos personas ya haya obtenido al menos un $6$;
2. la otra persona todavía no haya obtenido ningún $6$.

## Construcción mediante tiradas potenciales

Una manera conveniente de hacer el conteo consiste en imaginar que cada persona tiene previamente determinada una sucesión infinita de resultados del dado.

Aunque una persona abandone el juego, podemos imaginar que sus resultados futuros siguen existiendo, aunque ya no se utilicen.

Esto nos permite comparar los primeros $r$ resultados potenciales de Ana, Beto y Claudia.

Para un $r$ fijo consideramos:

- los primeros $r$ resultados de Ana;
- los primeros $r$ resultados de Beto;
- los primeros $r$ resultados de Claudia;
- el resultado número $r+1$ de Ana.

En total tenemos

$3r+1$

tiradas de dado.

Cada tirada tiene 6 resultados posibles.

Por lo tanto,

$|\Omega_r| = 6^{3r+1}$.

## Conteo de los casos favorables

Definamos

$E_r = \{\text{Ana es la segunda en salir y obtiene su primer }6 \text{ en su tiro }r+1\}$.

Para que ocurra $E_r$ necesitamos varias condiciones.

### 5. Ana obtiene un 6 en su siguiente tiro

Finalmente, en su tiro número $r+1$, Ana debe obtener exactamente un $6$.

Esta tirada tiene un único resultado favorable.

Por lo tanto, aporta un factor

$1$.

Juntando todos los factores obtenemos

$|E_r| = 2(5^r)(6^r-5^r)(5^r)$.

Así,

$\boxed{ |E_r| = 2(5^r)^2(6^r-5^r) }$.

Por lo tanto,

$\mathbb{P}(E_r) = \frac{|E_r|}{|\Omega_r|}$.

Entonces,

$\boxed{ \mathbb{P}(E_r) = \frac{ 2(5^r)^2(6^r-5^r) }{ 6^{3r+1} } }$.

# Sumamos sobre todos los valores posibles de \(r\)

Ana puede obtener su primer $6$ en su segundo tiro, tercero, cuarto, etc.

Por eso,

$E = \bigcup_{r=1}^{\infty}E_r$.

Además, los eventos $E_r$ son disjuntos, pues Ana no puede obtener su primer $6$ simultáneamente en dos tiros diferentes.

Por lo tanto,

$\mathbb{P}(E) = \sum_{r=1}^{\infty}\mathbb{P}(E_r)$.

Así,

$\mathbb{P}(E) = \sum_{r=1}^{\infty} \frac{ 2(5^r)^2(6^r-5^r) }{ 6^{3r+1} }$.

Ahora simplificamos la expresión.

Tenemos

$\mathbb{P}(E) = \sum_{r=1}^{\infty} \frac{ 2\cdot25^r(6^r-5^r) }{ 6^{3r+1} }$.

Distribuyendo,

$\mathbb{P}(E) = \sum_{r=1}^{\infty} \left[ \frac{2\cdot25^r6^r}{6^{3r+1}} - \frac{2\cdot25^r5^r}{6^{3r+1}} \right]$.

Como

$\frac{2}{6}=\frac{1}{3}$,

obtenemos

$\mathbb{P}(E) = \frac{1}{3} \sum_{r=1}^{\infty} \left[ \frac{25^r6^r}{6^{3r}} - \frac{125^r}{6^{3r}} \right]$.

Para el primer término,

$\frac{25^r6^r}{6^{3r}} = \frac{25^r}{6^{2r}} = \left(\frac{25}{36}\right)^r$.

Para el segundo término,

$\frac{125^r}{6^{3r}} = \left(\frac{125}{216}\right)^r$.

Por lo tanto,

$\mathbb{P}(E) = \frac{1}{3} \sum_{r=1}^{\infty} \left[ \left(\frac{25}{36}\right)^r - \left(\frac{125}{216}\right)^r \right]$.

Separando las sumas,

$\mathbb{P}(E) = \frac{1}{3} \left[ \sum_{r=1}^{\infty} \left(\frac{25}{36}\right)^r - \sum_{r=1}^{\infty} \left(\frac{125}{216}\right)^r \right]$.

Recordemos que, para $|x|<1$,

$\sum_{r=1}^{\infty}x^r = \frac{x}{1-x}$.

Entonces,

$\sum_{r=1}^{\infty} \left(\frac{25}{36}\right)^r = \frac{ \frac{25}{36} }{ 1-\frac{25}{36} } = \frac{25}{11}$.

Por otro lado,

$\sum_{r=1}^{\infty} \left(\frac{125}{216}\right)^r = \frac{ \frac{125}{216} }{ 1-\frac{125}{216} } = \frac{125}{91}$.

Por lo tanto,

$\mathbb{P}(E) = \frac{1}{3} \left( \frac{25}{11} - \frac{125}{91} \right)$.

Tomando denominador común,

$\frac{25}{11}-\frac{125}{91} = \frac{900}{1001}$.

Así,

$\mathbb{P}(E) = \frac{1}{3} \frac{900}{1001}$.

Finalmente,

$\boxed{ \mathbb{P}(E) = \frac{300}{1001} }$.

Numéricamente,

$\boxed{ \mathbb{P}(E) \approx 0.2997 }$.

Por lo tanto, la probabilidad de que **Ana sea la segunda persona en salir del juego** es aproximadamente

$\boxed{29.97\%}$.
