# Esperanza matemática de variables aleatorias continuas

Recordemos que, si $X$ es una variable aleatoria discreta con función de masa $p_X$, entonces su esperanza está dada por

$$
\boxed{
E[X]=\sum_{k\in R_X}k\,p_X(k).
}
$$

Donde $R_X$ representa el conjunto de valores que puede tomar la variable aleatoria $X$.

Sin embargo, surge una pregunta:

**¿Cómo podemos calcular la esperanza de una variable aleatoria continua?**

## Definición: Esperanza de una variable aleatoria continua

Sea $X$ una variable aleatoria continua con función de densidad $f_X$.

La esperanza matemática de $X$ se define como

$$
\boxed{
E[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx,
}
$$

siempre que la integral sea absolutamente convergente, es decir, que

$$
\int_{-\infty}^{\infty}|x|f_X(x)\,dx<\infty.
$$

Si $X$ es no negativa, también podemos considerar el caso en que

$$
E[X]=+\infty.
$$

---

# Ejemplos

## Ejemplo 1: Esperanza de una distribución uniforme

Supongamos que

$$
X\sim U[a,b],
$$

donde $a<b$.

Sabemos que su función de densidad está dada por

$$
f_X(x)=\frac{1}{b-a}\mathbb{1}_{[a,b]}(x).
$$

Es decir,

$$
f_X(x)=
\begin{cases}
\dfrac{1}{b-a}, & a\leq x\leq b,\\[6pt]
0, & \text{en otro caso}.
\end{cases}
$$

Queremos calcular la esperanza de $X$.

Por definición,

$$
E[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx.
$$

Sustituyendo la función de densidad, obtenemos

$$
E[X]=
\int_{-\infty}^{\infty}
x\frac{1}{b-a}\mathbb{1}_{[a,b]}(x)\,dx.
$$

Podemos separar la integral en tres intervalos:

$$
\begin{aligned}
E[X]
&=
\int_{-\infty}^{a}
x\frac{1}{b-a}\mathbb{1}_{[a,b]}(x)\,dx\\[6pt]
&\quad+
\int_a^b
x\frac{1}{b-a}\mathbb{1}_{[a,b]}(x)\,dx\\[6pt]
&\quad+
\int_b^\infty
x\frac{1}{b-a}\mathbb{1}_{[a,b]}(x)\,dx.
\end{aligned}
$$

Observemos que

$$
\mathbb{1}_{[a,b]}(x)=0
$$

cuando $x\notin[a,b]$.

Por lo tanto, la primera y tercera integrales son cero.

Así,

$$
E[X]=\int_a^b x\frac{1}{b-a}\,dx.
$$

Como $\frac{1}{b-a}$ es constante, podemos sacarla de la integral:

$$
\begin{aligned}
E[X]
&=\frac{1}{b-a}\int_a^b x\,dx\\[6pt]
&=\frac{1}{b-a}
\left[\frac{x^2}{2}\right]_a^b\\[6pt]
&=\frac{1}{b-a}
\left(\frac{b^2}{2}-\frac{a^2}{2}\right)\\[6pt]
&=\frac{b^2-a^2}{2(b-a)}.
\end{aligned}
$$

Recordemos la identidad

$$
b^2-a^2=(b-a)(b+a).
$$

Entonces,

$$
\begin{aligned}
E[X]
&=\frac{(b-a)(b+a)}{2(b-a)}\\[6pt]
&=\frac{a+b}{2}.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
E[X]=\frac{a+b}{2}.
}
$$

### Interpretación

La esperanza de una variable aleatoria uniforme en $[a,b]$ corresponde al punto medio del intervalo.

Esto resulta intuitivo, pues todos los subintervalos de la misma longitud tienen la misma probabilidad.

Por ejemplo, si

$$
X\sim U[0,10],
$$

entonces

$$
E[X]=\frac{0+10}{2}=5.
$$

# Ejemplo 2: Esperanza de una distribución exponencial

Supongamos que

$$
X\sim\operatorname{Exp}(\lambda),
$$

donde $\lambda>0$.

Sabemos que su función de densidad está dada por

$$
f_X(x)=\lambda e^{-\lambda x}\mathbb{1}_{[0,\infty)}(x).
$$

Es decir,

$$
f_X(x)=
\begin{cases}
\lambda e^{-\lambda x}, & x\geq 0,\\[6pt]
0, & x<0.
\end{cases}
$$

Queremos calcular la esperanza de $X$.

Por definición,

$$
E[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx.
$$

Como la función de densidad es cero para $x<0$, tenemos que

$$
\begin{aligned}
E[X]
&=\int_0^\infty x\lambda e^{-\lambda x}\,dx.
\end{aligned}
$$

Para resolver esta integral realizaremos un cambio de variable.

### Paso 1. Cambio de variable

Tomemos

$$
s=\lambda x.
$$

Entonces,

$$
ds=\lambda\,dx.
$$

Por lo tanto,

$$
dx=\frac{ds}{\lambda}.
$$

Además,

$$
x=\frac{s}{\lambda}.
$$

Como $\lambda>0$, los límites de integración se transforman de la siguiente manera:

- Si $x=0$, entonces $s=0$.
- Si $x\to\infty$, entonces $s\to\infty$.

Sustituyendo en la integral,

$$
\begin{aligned}
E[X]
&=\int_0^\infty x\lambda e^{-\lambda x}\,dx\\[6pt]
&=\int_0^\infty
\frac{s}{\lambda}\lambda e^{-s}
\frac{ds}{\lambda}\\[6pt]
&=\frac{1}{\lambda}
\int_0^\infty s e^{-s}\,ds.
\end{aligned}
$$

Ahora debemos calcular

$$
\int_0^\infty s e^{-s}\,ds.
$$

### Paso 2. Integración por partes

Recordemos la fórmula de integración por partes:

$$
\int u\,dv=uv-\int v\,du.
$$

Tomemos

$$
u=s,
\qquad
dv=e^{-s}\,ds.
$$

Entonces,

$$
du=ds,
\qquad
v=-e^{-s}.
$$

Aplicando integración por partes,

$$
\begin{aligned}
\int_0^\infty s e^{-s}\,ds
&=\left[-se^{-s}\right]_0^\infty
+\int_0^\infty e^{-s}\,ds.
\end{aligned}
$$

Observemos que

$$
\lim_{s\to\infty}se^{-s}
=
\lim_{s\to\infty}\frac{s}{e^s}
=0.
$$

Este límite puede calcularse utilizando la regla de L'Hôpital:

$$
\begin{aligned}
\lim_{s\to\infty}\frac{s}{e^s}
&=\lim_{s\to\infty}\frac{1}{e^s}\\
&=0.
\end{aligned}
$$

Por lo tanto,

$$
\left[-se^{-s}\right]_0^\infty=0.
$$

Además,

$$
\begin{aligned}
\int_0^\infty e^{-s}\,ds
&=\left[-e^{-s}\right]_0^\infty\\[6pt]
&=0-(-1)\\[6pt]
&=1.
\end{aligned}
$$

En consecuencia,

$$
\int_0^\infty s e^{-s}\,ds=1.
$$

### Paso 3. Sustituir el resultado

Recordemos que

$$
E[X]=\frac{1}{\lambda}
\int_0^\infty s e^{-s}\,ds.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=\frac{1}{\lambda}\cdot1\\[6pt]
&=\frac{1}{\lambda}.
\end{aligned}
$$

Así, concluimos que

$$
\boxed{
E[X]=\frac{1}{\lambda}.
}
$$

---

# Ejemplo 3: Esperanza de una distribución normal

Supongamos que

$$
X\sim N(\mu,\sigma^2),
$$

donde $\mu\in\mathbb{R}$ y $\sigma>0$.

Sabemos que la función de densidad de $X$ está dada por

$$
f_X(x)=
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}.
$$

Queremos calcular la esperanza de $X$.

Por definición,

$$
E[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx.
$$

Sustituyendo la función de densidad,

$$
E[X]
=
\int_{-\infty}^{\infty}
x\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

Es decir,

$$
\boxed{
E[X]
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
x e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
}
$$

Para resolver esta integral realizaremos un cambio de variable.

# Continuación del ejemplo 3: Esperanza de una normal

Recordemos que

$$
X\sim N(\mu,\sigma^2)
$$

y que queremos calcular

$$
E[X]
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
x e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\,dx.
$$

## Paso 1. Cambio de variable

Tomemos

$$
s=x-\mu.
$$

Entonces,

$$
ds=dx
$$

y

$$
x=s+\mu.
$$

Observemos que los límites de integración no cambian:

$$
x\to-\infty
\implies
s\to-\infty,
$$

y

$$
x\to\infty
\implies
s\to\infty.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
(s+\mu)e^{-\frac{s^2}{2\sigma^2}}
\,ds.
\end{aligned}
$$

Distribuyendo el integrando, obtenemos

$$
\begin{aligned}
E[X]
&=
\frac{1}{\sqrt{2\pi\sigma^2}}
\left(
\int_{-\infty}^{\infty}
s e^{-\frac{s^2}{2\sigma^2}}
\,ds
+
\mu
\int_{-\infty}^{\infty}
e^{-\frac{s^2}{2\sigma^2}}
\,ds
\right).
\end{aligned}
$$

Separando los términos,

$$
\begin{aligned}
E[X]
&=
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
s e^{-\frac{s^2}{2\sigma^2}}
\,ds\\[6pt]
&\quad+
\mu
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
e^{-\frac{s^2}{2\sigma^2}}
\,ds.
\end{aligned}
$$

Ahora estudiaremos cada integral por separado.

---

## Paso 2. Cálculo de la segunda integral

Observemos que

$$
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{s^2}{2\sigma^2}}
$$

es la función de densidad de una variable aleatoria

$$
Z\sim N(0,\sigma^2).
$$

Recordemos que toda función de densidad satisface

$$
\int_{-\infty}^{\infty}f_Z(s)\,ds=1.
$$

Por lo tanto,

$$
\boxed{
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
e^{-\frac{s^2}{2\sigma^2}}
\,ds=1.
}
$$

En consecuencia, el segundo término de la esperanza es igual a $\mu$.

Así,

$$
E[X]
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
s e^{-\frac{s^2}{2\sigma^2}}
\,ds
+\mu.
$$

---

## Paso 3. Cálculo de la primera integral

Queremos calcular

$$
\int_{-\infty}^{\infty}
s e^{-\frac{s^2}{2\sigma^2}}
\,ds.
$$

Para ello, recordemos las definiciones de funciones pares e impares.

### Función par

Una función $f$ es par si

$$
f(-x)=f(x)
$$

para todo $x$ de su dominio.

### Función impar

Una función $f$ es impar si

$$
f(-x)=-f(x)
$$

para todo $x$ de su dominio.

Observemos que la función

$$
f(s)=e^{-\frac{s^2}{2\sigma^2}}
$$

es par.

En efecto,

$$
\begin{aligned}
f(-s)
&=e^{-\frac{(-s)^2}{2\sigma^2}}\\[6pt]
&=e^{-\frac{s^2}{2\sigma^2}}\\[6pt]
&=f(s).
\end{aligned}
$$

Por otro lado, la función

$$
g(s)=s
$$

es impar, pues

$$
g(-s)=-s=-g(s).
$$

Recordemos que el producto de una función par y una función impar es una función impar.

Por lo tanto,

$$
h(s)=s e^{-\frac{s^2}{2\sigma^2}}
$$

es una función impar.

Podemos verificarlo directamente:

$$
\begin{aligned}
h(-s)
&=(-s)e^{-\frac{(-s)^2}{2\sigma^2}}\\[6pt]
&=-s e^{-\frac{s^2}{2\sigma^2}}\\[6pt]
&=-h(s).
\end{aligned}
$$

Sabemos que, si $h$ es una función impar e integrable en un intervalo simétrico respecto del origen, entonces

$$
\int_{-a}^{a}h(s)\,ds=0.
$$

En particular, si $h$ es integrable en toda la recta real,

$$
\int_{-\infty}^{\infty}h(s)\,ds=0.
$$

Como

$$
h(s)=s e^{-\frac{s^2}{2\sigma^2}}
$$

es impar y absolutamente integrable, obtenemos

$$
\boxed{
\int_{-\infty}^{\infty}
s e^{-\frac{s^2}{2\sigma^2}}
\,ds=0.
}
$$

---

## Paso 4. Conclusión

Recordemos que habíamos obtenido

$$
E[X]
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\int_{-\infty}^{\infty}
s e^{-\frac{s^2}{2\sigma^2}}
\,ds
+\mu.
$$

Como la integral es cero,

$$
\begin{aligned}
E[X]
&=\frac{1}{\sqrt{2\pi\sigma^2}}\cdot0+\mu\\[6pt]
&=\mu.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
E[X]=\mu.
}
$$

### Interpretación

La esperanza de una variable aleatoria normal coincide con el parámetro $\mu$.

Además, la función de densidad de una normal es simétrica respecto de la recta vertical

$$
x=\mu.
$$

Por esta razón, la media se encuentra en el centro de simetría de la distribución.

# Ejemplo 4: Esperanza de una distribución de Pareto

Supongamos que

$$
X\sim\operatorname{Pareto}(\alpha),
$$

con la parametrización cuya función de densidad está dada por

$$
\boxed{
f_X(x)=
(\alpha-1)x^{-\alpha}
\mathbb{1}_{[1,\infty)}(x),
}
$$

donde $\alpha>1$.

Es decir,

$$
f_X(x)=
\begin{cases}
(\alpha-1)x^{-\alpha}, & x\geq1,\\[6pt]
0, & x<1.
\end{cases}
$$

Queremos calcular la esperanza de $X$ y determinar para qué valores de $\alpha$ es finita.

---

## Paso 1. Planteamiento de la integral

Por definición,

$$
E[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx.
$$

Como la función de densidad es cero para $x<1$, tenemos que

$$
E[X]
=
\int_1^\infty
x(\alpha-1)x^{-\alpha}\,dx.
$$

Utilizando las propiedades de las potencias,

$$
x\cdot x^{-\alpha}=x^{1-\alpha}.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=\int_1^\infty
(\alpha-1)x^{1-\alpha}\,dx\\[6pt]
&=(\alpha-1)
\int_1^\infty x^{1-\alpha}\,dx.
\end{aligned}
$$

Así,

$$
\boxed{
E[X]
=
(\alpha-1)
\int_1^\infty x^{1-\alpha}\,dx.
}
$$

Debemos analizar la convergencia de esta integral.

---

## Paso 2. Integración para $\alpha\neq2$

Recordemos que

$$
\int x^r\,dx
=
\frac{x^{r+1}}{r+1}+C,
\qquad r\neq-1.
$$

En nuestro caso,

$$
r=1-\alpha.
$$

Por lo tanto,

$$
r+1=2-\alpha.
$$

Si $\alpha\neq2$, tenemos que

$$
\int x^{1-\alpha}\,dx
=
\frac{x^{2-\alpha}}{2-\alpha}+C.
$$

Como la integral es impropia, utilizamos la definición

$$
E[X]
=
(\alpha-1)
\lim_{m\to\infty}
\int_1^m x^{1-\alpha}\,dx.
$$

Integrando,

$$
\begin{aligned}
E[X]
&=(\alpha-1)
\lim_{m\to\infty}
\left[
\frac{x^{2-\alpha}}{2-\alpha}
\right]_1^m\\[6pt]
&=(\alpha-1)
\lim_{m\to\infty}
\left(
\frac{m^{2-\alpha}}{2-\alpha}
-\frac{1}{2-\alpha}
\right)\\[6pt]
&=\frac{\alpha-1}{2-\alpha}
\lim_{m\to\infty}
\left(m^{2-\alpha}-1\right).
\end{aligned}
$$

El comportamiento de esta expresión depende del valor de $\alpha$.

---

## Caso 1: $\alpha>2$

Si

$$
\alpha>2,
$$

entonces

$$
2-\alpha<0.
$$

En consecuencia,

$$
m^{2-\alpha}
=
\frac{1}{m^{\alpha-2}}.
$$

Como $\alpha-2>0$, tenemos que

$$
\lim_{m\to\infty}m^{2-\alpha}=0.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=\frac{\alpha-1}{2-\alpha}
\left(0-1\right)\\[6pt]
&=-\frac{\alpha-1}{2-\alpha}\\[6pt]
&=\frac{\alpha-1}{\alpha-2}.
\end{aligned}
$$

Así,

$$
\boxed{
E[X]=\frac{\alpha-1}{\alpha-2},
\qquad \alpha>2.
}
$$

---

## Caso 2: $1<\alpha<2$

Si

$$
1<\alpha<2,
$$

entonces

$$
2-\alpha>0.
$$

Por lo tanto,

$$
\lim_{m\to\infty}m^{2-\alpha}=+\infty.
$$

Como

$$
\frac{\alpha-1}{2-\alpha}>0,
$$

obtenemos que

$$
\begin{aligned}
E[X]
&=\frac{\alpha-1}{2-\alpha}
\lim_{m\to\infty}
\left(m^{2-\alpha}-1\right)\\[6pt]
&=+\infty.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{
E[X]=+\infty,
\qquad 1<\alpha<2.
}
$$

Esto significa que la variable aleatoria no tiene esperanza finita para estos valores del parámetro.

---

## Caso 3: $\alpha=2$

Si

$$
\alpha=2,
$$

entonces la integral de la esperanza es

$$
E[X]
=
\int_1^\infty
x(2-1)x^{-2}\,dx.
$$

Simplificando,

$$
\begin{aligned}
E[X]
&=\int_1^\infty x^{-1}\,dx\\[6pt]
&=\int_1^\infty\frac1x\,dx.
\end{aligned}
$$

Recordemos que

$$
\int\frac1x\,dx=\ln|x|+C.
$$

Por lo tanto,

$$
\begin{aligned}
E[X]
&=\lim_{m\to\infty}
\int_1^m\frac1x\,dx\\[6pt]
&=\lim_{m\to\infty}
\left[\ln x\right]_1^m\\[6pt]
&=\lim_{m\to\infty}
(\ln m-\ln1)\\[6pt]
&=\lim_{m\to\infty}\ln m\\[6pt]
&=+\infty.
\end{aligned}
$$

En consecuencia,

$$
\boxed{
E[X]=+\infty,
\qquad \alpha=2.
}
$$

---

## Conclusión

Reuniendo los tres casos, obtenemos que

$$
\boxed{
E[X]=
\begin{cases}
\dfrac{\alpha-1}{\alpha-2}, & \alpha>2,\\[10pt]
+\infty, & 1<\alpha\leq2.
\end{cases}
}
$$

Por lo tanto, la esperanza de la distribución de Pareto considerada es finita si y solo si

$$
\boxed{
\alpha>2.
}
$$

---

## Observación: ¿Por qué necesitamos $\alpha>1$?

Recordemos que la función de densidad propuesta es

$$
f_X(x)=
(\alpha-1)x^{-\alpha}
\mathbb{1}_{[1,\infty)}(x).
$$

Para que sea una función de densidad, necesitamos verificar que

$$
\int_{-\infty}^{\infty}f_X(x)\,dx=1.
$$

Como la función es cero para $x<1$, tenemos que

$$
\begin{aligned}
\int_{-\infty}^{\infty}f_X(x)\,dx
&=(\alpha-1)\int_1^\infty x^{-\alpha}\,dx.
\end{aligned}
$$

Para $\alpha>1$, sabemos que

$$
\int x^{-\alpha}\,dx
=
\frac{x^{1-\alpha}}{1-\alpha}+C.
$$

Por lo tanto,

$$
\begin{aligned}
\int_{-\infty}^{\infty}f_X(x)\,dx
&=(\alpha-1)
\lim_{m\to\infty}
\left[
\frac{x^{1-\alpha}}{1-\alpha}
\right]_1^m\\[6pt]
&=(\alpha-1)
\lim_{m\to\infty}
\frac{m^{1-\alpha}-1}{1-\alpha}.
\end{aligned}
$$

Como $\alpha>1$, tenemos que

$$
1-\alpha<0
$$

y, por lo tanto,

$$
\lim_{m\to\infty}m^{1-\alpha}=0.
$$

En consecuencia,

$$
\begin{aligned}
\int_{-\infty}^{\infty}f_X(x)\,dx
&=(\alpha-1)
\frac{-1}{1-\alpha}\\[6pt]
&=1.
\end{aligned}
$$

Así, hemos verificado que la función propuesta es una densidad de probabilidad válida cuando $\alpha>1$.
