## 1. Personas sentadas en una fila

Tenemos $3$ niños y $3$ niñas, todos distintos.

### a) ¿De cuántas maneras pueden sentarse los $3$ niños y las $3$ niñas en una fila?

En total tenemos $6$ personas distintas.

Para el primer lugar tenemos $6$ posibilidades; después de sentar a una persona, para el segundo lugar quedan $5$ posibilidades; después quedan $4$, y así sucesivamente.

Por lo tanto,

$$
6\cdot5\cdot4\cdot3\cdot2\cdot1
=
6!.
$$

Aquí:

- $6$ representa las posibilidades para el primer asiento;
- $5$ las posibilidades para el segundo;
- $4$ para el tercero;
- y así sucesivamente.

Entonces,

$$
6!=720.
$$

Por lo tanto, hay

$$
\boxed{720}
$$

formas de sentarlos.

---

### b) ¿De cuántas maneras pueden sentarse si los niños deben estar juntos y las niñas también deben estar juntas?

Podemos pensar en dos bloques:

$$
[\text{niños}]
\qquad
[\text{niñas}].
$$

Los dos bloques pueden ordenarse de

$$
2!
$$

formas:

$$
[\text{niños}][\text{niñas}]
$$

o

$$
[\text{niñas}][\text{niños}].
$$

Dentro del bloque de los niños, los $3$ niños pueden acomodarse de

$$
3!
$$

formas.

De igual manera, las $3$ niñas pueden acomodarse dentro de su bloque de

$$
3!
$$

formas.

Por el principio multiplicativo,

$$
2!\,3!\,3!.
$$

Aquí:

- $2!$ representa las formas de ordenar los dos bloques;
- el primer $3!$ representa las formas de ordenar a los niños dentro de su bloque;
- el segundo $3!$ representa las formas de ordenar a las niñas dentro de su bloque.

Así,

$$
2!\,3!\,3!
=
2(6)(6)
=
72.
$$

Por lo tanto,

$$
\boxed{72}
$$

formas.

---

### c) ¿Y si únicamente los niños deben sentarse juntos?

Consideramos a los $3$ niños como un solo bloque:

$$
[\text{niños}].
$$

Además tenemos a las $3$ niñas por separado.

Por lo tanto, debemos ordenar $4$ objetos:

$$
[\text{niños}],\quad N_1,\quad N_2,\quad N_3.
$$

Estos $4$ objetos pueden ordenarse de

$$
4!
$$

formas.

Dentro del bloque, los $3$ niños pueden ordenarse de

$$
3!
$$

formas.

Entonces,

$$
4!\,3!.
$$

Aquí:

- $4!$ representa las formas de acomodar el bloque de niños junto con las $3$ niñas;
- $3!$ representa las formas de ordenar a los niños dentro de su bloque.

Así,

$$
4!\,3!
=
24(6)
=
144.
$$

Por lo tanto,

$$
\boxed{144}
$$

formas.

---

### d) ¿Y si no pueden sentarse juntas dos personas del mismo sexo?

Para que nunca aparezcan juntas dos personas del mismo sexo, los niños y las niñas deben alternarse.

Hay solamente dos patrones posibles:

$$
N\,M\,N\,M\,N\,M
$$

o

$$
M\,N\,M\,N\,M\,N.
$$

En cualquiera de los dos patrones, los $3$ niños pueden acomodarse en sus lugares de

$$
3!
$$

formas y las $3$ niñas de

$$
3!
$$

formas.

Por lo tanto,

$$
2(3!)(3!).
$$

Aquí:

- $2$ representa los dos patrones posibles de alternancia;
- un $3!$ representa cómo acomodar a los niños;
- el otro $3!$ representa cómo acomodar a las niñas.

Así,

$$
2(3!)(3!)
=
2(6)(6)
=
72.
$$

Por lo tanto,

$$
\boxed{72}
$$

formas.

# Arreglos de letras

¿Cuántos arreglos diferentes pueden formarse con las letras de cada palabra?

## a) FLUKE

La palabra

$$
\text{FLUKE}
$$

tiene $5$ letras distintas.

Por lo tanto, podemos ordenar las $5$ letras de

$$
5!
$$

formas.

Así,

$$
5!
=
120.
$$

Por lo tanto,

$$
\boxed{120}
$$

arreglos diferentes.

---

## b) PROPOSE

La palabra

$$
\text{PROPOSE}
$$

tiene $7$ letras.

Si inicialmente consideráramos todas las letras distintas, tendríamos

$$
7!
$$

ordenamientos.

Sin embargo:

- la letra $P$ aparece $2$ veces;
- la letra $O$ aparece $2$ veces.

Intercambiar entre sí las dos letras $P$ no genera un nuevo arreglo. Por ello debemos dividir entre

$$
2!.
$$

De igual manera, debemos dividir entre $2!$ por las dos letras $O$.

Así,

$$
\frac{7!}{2!2!}.
$$

Aquí:

- $7!$ cuenta los ordenamientos si todas las letras fueran distintas;
- el primer $2!$ elimina las repeticiones producidas por intercambiar las dos $P$;
- el segundo $2!$ elimina las repeticiones producidas por intercambiar las dos $O$.

Por lo tanto,

$$
\frac{7!}{2!2!}
=
1260.
$$

Así, hay

$$
\boxed{1260}
$$

arreglos diferentes.

---

## c) MISSISSIPPI

La palabra

$$
\text{MISSISSIPPI}
$$

tiene $11$ letras.

Las cantidades de cada letra son:

| Letra | Cantidad |
|---|---:|
| $M$ | $1$ |
| $I$ | $4$ |
| $S$ | $4$ |
| $P$ | $2$ |

Si las $11$ letras fueran distintas, tendríamos

$$
11!
$$

ordenamientos.

Pero debemos dividir por las permutaciones de las letras repetidas:

$$
\frac{11!}{4!4!2!}.
$$

Aquí:

- $11!$ representa todos los ordenamientos si las letras fueran distinguibles;
- un $4!$ elimina los órdenes repetidos producidos por las cuatro $I$;
- el otro $4!$ elimina los producidos por las cuatro $S$;
- $2!$ elimina los producidos por las dos $P$.

Por lo tanto,

$$
\frac{11!}{4!4!2!}
=
34650.
$$

Así,

$$
\boxed{34650}
$$

arreglos diferentes.

---

## d) ARRANGE

La palabra

$$
\text{ARRANGE}
$$

tiene $7$ letras.

Las letras $A$ y $R$ aparecen dos veces cada una.

Por lo tanto,

$$
\frac{7!}{2!2!}.
$$

Aquí:

- $7!$ cuenta los ordenamientos suponiendo que todas las letras fueran distintas;
- un $2!$ corrige la repetición de las dos $A$;
- el otro $2!$ corrige la repetición de las dos $R$.

Entonces,

$$
\frac{7!}{2!2!}
=
1260.
$$

Por lo tanto,

$$
\boxed{1260}.
$$

# Coeficiente multinomial

## Definición

Sean

$$
n_1,\ldots,n_r
$$

enteros no negativos tales que

$$
n_1+\cdots+n_r=n.
$$

Definimos el **coeficiente multinomial** como

$$
\binom{n}{n_1,n_2,\ldots,n_r}
=
\frac{n!}{n_1!n_2!\cdots n_r!}.
$$

Este número representa las formas de dividir $n$ objetos distintos en $r$ grupos distinguibles, donde:

- el primer grupo contiene $n_1$ objetos;
- el segundo grupo contiene $n_2$ objetos;
- $\vdots$
- el $r$-ésimo grupo contiene $n_r$ objetos.

---

## ¿Por qué aparece esta fórmula?

Para formar el primer grupo debemos escoger $n_1$ objetos de los $n$ disponibles.

Esto puede hacerse de

$$
\binom{n}{n_1}
$$

formas.

El número

$$
\binom{n}{n_1}
$$

representa las formas de decidir cuáles $n_1$ de los $n$ objetos pertenecerán al primer grupo.

Después quedan

$$
n-n_1
$$

objetos.

Para formar el segundo grupo debemos escoger $n_2$ de esos objetos, lo cual puede hacerse de

$$
\binom{n-n_1}{n_2}
$$

formas.

Después quedan

$$
n-n_1-n_2
$$

objetos.

Para el tercer grupo tenemos

$$
\binom{n-n_1-n_2}{n_3}
$$

posibilidades.

Continuando de esta manera obtenemos

$$
\binom{n}{n_1}
\binom{n-n_1}{n_2}
\binom{n-n_1-n_2}{n_3}
\cdots
\binom{n-n_1-\cdots-n_{r-1}}{n_r}.
$$

Al escribir cada coeficiente binomial mediante factoriales,

$$
\binom{n}{n_1}
=
\frac{n!}{n_1!(n-n_1)!},
$$

$$
\binom{n-n_1}{n_2}
=
\frac{(n-n_1)!}
{n_2!(n-n_1-n_2)!},
$$

y así sucesivamente.

Los factoriales intermedios se cancelan y obtenemos

$$
\boxed{
\binom{n}{n_1,n_2,\ldots,n_r}
=
\frac{n!}{n_1!n_2!\cdots n_r!}
}.
$$

# Invitaciones a una fiesta

Una persona tiene $8$ amigos, de los cuales invitará a $5$ a una fiesta.

## a) Dos de los amigos están peleados y no asistirán juntos

Llamemos $A$ y $B$ a los dos amigos peleados.

Sin ninguna restricción, podemos escoger $5$ de los $8$ amigos de

$$
\binom{8}{5}
$$

formas.

El número

$$
\binom{8}{5}
$$

representa las formas de escoger cuáles $5$ de los $8$ amigos serán invitados.

Ahora debemos quitar los grupos en los que aparecen juntos $A$ y $B$.

Si $A$ y $B$ ya están seleccionados, necesitamos escoger otros $3$ amigos entre los $6$ restantes.

Esto puede hacerse de

$$
\binom{6}{3}
$$

formas.

Por lo tanto,

$$
\binom{8}{5}-\binom{6}{3}.
$$

Así,

$$
56-20=36.
$$

Por lo tanto,

$$
\boxed{36}
$$

elecciones posibles.

---

## b) Dos amigos asistirán únicamente si van juntos

Supongamos nuevamente que los amigos son $A$ y $B$.

Esto significa que solamente hay dos posibilidades:

1. van los dos;
2. no va ninguno.

### Caso 1: van los dos

Si $A$ y $B$ ya están invitados, debemos elegir otros $3$ amigos entre los $6$ restantes.

Hay

$$
\binom{6}{3}
$$

formas.

### Caso 2: no va ninguno

Entonces debemos escoger los $5$ invitados entre los otros $6$ amigos.

Hay

$$
\binom{6}{5}
$$

formas.

Por el principio aditivo,

$$
\binom{6}{3}+\binom{6}{5}.
$$

Así,

$$
20+6=26.
$$

Por lo tanto,

$$
\boxed{26}
$$

elecciones posibles.

# Subconjuntos con al menos un elemento especial

Sea

$$
\{1,2,\ldots,20\}.
$$

¿Cuántos subconjuntos de tamaño $4$ contienen al menos uno de los elementos

$$
1,2,3,4,5?
$$

Primero contamos todos los subconjuntos de tamaño $4$.

Hay

$$
\binom{20}{4}
$$

formas.

El número

$$
\binom{20}{4}
$$

representa las formas de escoger $4$ elementos entre los $20$ disponibles.

Ahora preguntamos:

> ¿Cuáles son los subconjuntos que no queremos?

Son aquellos que no contienen ninguno de

$$
1,2,3,4,5.
$$

Si eliminamos esos $5$ elementos, quedan

$$
20-5=15
$$

elementos disponibles.

Por lo tanto, hay

$$
\binom{15}{4}
$$

subconjuntos que no contienen ninguno de los cinco elementos especiales.

Entonces,

$$
\binom{20}{4}-\binom{15}{4}.
$$

Así,

$$
4845-1365
=
3480.
$$

Por lo tanto,

$$
\boxed{3480}.
$$

Esta es una aplicación del **conteo por complemento**:

$$
\text{al menos uno}
=
\text{todos}
-
\text{ninguno}.
$$

# Caminos en una cuadrícula

Queremos ir del punto $A$ al punto $B$ siguiendo únicamente caminos mínimos en una cuadrícula de $4$ unidades horizontales y $3$ unidades verticales.

Para llegar de $A$ a $B$ necesariamente debemos realizar:

- $4$ movimientos hacia la derecha;
- $3$ movimientos hacia arriba.

En total necesitamos

$$
4+3=7
$$

movimientos.

Por ejemplo, un camino podría representarse como

$$
D\,A\,D\,D\,A\,D\,A,
$$

donde:

- $D$ significa un movimiento hacia la derecha;
- $A$ significa un movimiento hacia arriba.

Todo camino queda determinado al escoger en cuáles $4$ de los $7$ lugares aparecen los movimientos hacia la derecha.

Por lo tanto,

$$
\binom{7}{4}.
$$

El número

$$
\binom{7}{4}
$$

representa las formas de escoger cuáles $4$ de los $7$ movimientos serán hacia la derecha.

Así,

$$
\binom{7}{4}=35.
$$

Equivalentemente, podríamos escoger cuáles $3$ posiciones serán movimientos hacia arriba:

$$
\binom{7}{3}=35.
$$

Por lo tanto,

$$
\boxed{
\binom{7}{4}
=
\binom{7}{3}
=
35
}.
$$

# Lanzamientos de un dado

Tenemos un dado y hacemos $4$ tiradas.

¿Cuántas secuencias tienen exactamente un número que aparece dos veces, mientras que los otros dos resultados son distintos entre sí y distintos del número repetido?

Por ejemplo,

$$
2,2,5,4
$$

es una secuencia de este tipo.

Primero escogemos cuál será el número que aparece dos veces.

Como el dado tiene $6$ caras, tenemos

$$
6
$$

posibilidades.

Luego escogemos en cuáles $2$ de las $4$ posiciones aparecerá ese número.

Esto puede hacerse de

$$
\binom{4}{2}
$$

formas.

Después quedan dos posiciones.

Para la primera de ellas podemos escoger cualquiera de los otros $5$ números.

Por lo tanto, tenemos

$$
5
$$

posibilidades.

Para la última posición necesitamos un número diferente tanto del repetido como del resultado que acabamos de utilizar.

Por eso quedan

$$
4
$$

posibilidades.

Por el principio multiplicativo,

$$
6\binom{4}{2}(5)(4).
$$

Aquí:

- $6$ representa la elección del número que se repetirá;
- $\binom{4}{2}$ representa la elección de las dos posiciones donde aparecerá;
- $5$ representa la elección del primero de los otros dos resultados;
- $4$ representa la elección del último resultado.

Entonces,

$$
6\binom{4}{2}(5)(4)
=
6(6)(5)(4)
=
720.
$$

Por lo tanto,

$$
\boxed{720}
$$

secuencias.

# Ejercicio de póker

Tenemos una baraja de $52$ cartas.

¿Cuál es la probabilidad de obtener una mano de $5$ cartas que contenga:

- exactamente $2$ ases;
- exactamente $1$ corazón?

El número total de manos de $5$ cartas es

$$
|\Omega|
=
\binom{52}{5}.
$$

El número

$$
\binom{52}{5}
$$

representa las formas de escoger $5$ cartas de las $52$, sin importar el orden.

Para contar las manos favorables debemos separar en dos casos.

---

## Caso 1: el as de corazones está entre los dos ases

El as de corazones ya está seleccionado.

Debemos escoger exactamente otro as entre los otros $3$ ases.

Esto puede hacerse de

$$
\binom{3}{1}
$$

formas.

Ahora necesitamos otras $3$ cartas que no sean:

- corazones, porque ya tenemos exactamente un corazón;
- ases, porque ya tenemos exactamente dos ases.

Hay $36$ cartas que no son ni ases ni corazones.

Por lo tanto, podemos escoger las otras $3$ cartas de

$$
\binom{36}{3}
$$

formas.

Así, el número de manos de este caso es

$$
\binom{3}{1}\binom{36}{3}.
$$

Aquí:

- $\binom{3}{1}$ escoge cuál de los otros $3$ ases acompañará al as de corazones;
- $\binom{36}{3}$ escoge las otras $3$ cartas que no son ni ases ni corazones.

---

## Caso 2: ninguno de los dos ases es el as de corazones

Debemos escoger los $2$ ases entre los $3$ ases que no son corazones.

Esto puede hacerse de

$$
\binom{3}{2}
$$

formas.

Todavía necesitamos exactamente un corazón.

Como el as de corazones no puede aparecer, porque entonces tendríamos $3$ ases, debemos escoger el corazón entre los otros $12$ corazones.

Esto puede hacerse de

$$
\binom{12}{1}
$$

formas.

Después necesitamos otras $2$ cartas que no sean ni ases ni corazones.

Hay $36$ cartas de este tipo.

Por lo tanto, tenemos

$$
\binom{36}{2}
$$

posibilidades.

Así, el número de manos favorables en este caso es

$$
\binom{3}{2}
\binom{12}{1}
\binom{36}{2}.
$$

Aquí:

- $\binom{3}{2}$ escoge los $2$ ases que no son de corazones;
- $\binom{12}{1}$ escoge el único corazón, que no puede ser el as de corazones;
- $\binom{36}{2}$ escoge las otras $2$ cartas, que no son ni ases ni corazones.

---

## Probabilidad total

Los dos casos son mutuamente excluyentes, por lo que sumamos:

$$
|E|
=
\binom{3}{1}\binom{36}{3}
+
\binom{3}{2}\binom{12}{1}\binom{36}{2}.
$$

Por lo tanto,

$$
P(E)
=
\frac{
\binom{3}{1}\binom{36}{3}
+
\binom{3}{2}\binom{12}{1}\binom{36}{2}
}{
\binom{52}{5}
}.
$$

Numéricamente,

$$
P(E)
\approx
0.0169683.
$$

Es decir,

$$
\boxed{P(E)\approx0.01697}.
$$

En porcentaje,

$$
P(E)\approx1.697\%.
$$
