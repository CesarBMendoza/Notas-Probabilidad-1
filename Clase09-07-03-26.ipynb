# Probabilidad condicional

## Ejemplo 1. Bolas de diferentes colores y texturas

Tengo una bolsa con $11$ bolas:

- $7$ bolas peluditas:
  - $3$ rojas,
  - $4$ azules.
- $4$ bolas lisas:
  - $2$ rojas,
  - $2$ azules.

En total hay

$3+2=5$

bolas rojas y

$4+2=6$

bolas azules.

---

### (a) ¿Cuál es la probabilidad de sacar una bola roja?

Como hay $5$ bolas rojas de un total de $11$ bolas,

$P(\text{roja})=\frac{5}{11}$.

> **¿De dónde sale $\frac{5}{11}$?**
>
> Estamos suponiendo que todas las bolas tienen la misma probabilidad de ser elegidas.
>
> Hay $5$ resultados favorables, correspondientes a las $5$ bolas rojas, y $11$ resultados posibles en total.
>
> Por ello,
>
> $P(\text{roja})=\frac{\text{número de bolas rojas}}{\text{número total de bolas}}=\frac{5}{11}$.

---

### (b) Antes de sacar la bola noto que es peludita. ¿Cuál es la probabilidad de que sea roja?

Ahora tenemos información adicional: sabemos que la bola pertenece al grupo de las $7$ bolas peluditas.

Entre esas $7$ bolas peluditas hay $3$ rojas.

Por lo tanto,

$P(\text{roja}\mid\text{peludita})=\frac{3}{7}$.

> **Interpretación.**
>
> Al saber que la bola es peludita, nuestro espacio de posibilidades se reduce.
>
> Ya no estamos considerando las $11$ bolas, sino solamente las $7$ bolas peluditas.
>
> De esas $7$, exactamente $3$ son rojas.

Podemos obtener el mismo resultado usando probabilidades.

Tenemos

$P(\text{roja y peludita})=\frac{3}{11}$,

pues hay $3$ bolas que son simultáneamente rojas y peluditas.

Además,

$P(\text{peludita})=\frac{7}{11}$.

Entonces,

$P(\text{roja}\mid\text{peludita})=\frac{P(\text{roja}\cap\text{peludita})}{P(\text{peludita})}$.

Sustituyendo,

$P(\text{roja}\mid\text{peludita})=\frac{\frac{3}{11}}{\frac{7}{11}}$.

Por lo tanto,

$P(\text{roja}\mid\text{peludita})=\frac{3}{7}$.

---

## Definición de probabilidad condicional

Sean $A$ y $B$ dos eventos con $P(B)>0$.

La **probabilidad condicional de $A$ dado $B$** se define como

$P(A\mid B)=\frac{P(A\cap B)}{P(B)}$.

> **Interpretación.**
>
> Cuando sabemos que ocurrió $B$, nuestro nuevo espacio de posibilidades es $B$.
>
> Dentro de ese nuevo espacio, los casos favorables para $A$ son precisamente aquellos que pertenecen simultáneamente a $A$ y a $B$, es decir, $A\cap B$.

# Ejemplo 2. Lanzamiento de una moneda

Tengo los ojos vendados y lanzo una moneda dos veces.

Me dicen que salió **al menos un águila**.

¿Cuál es la probabilidad de haber obtenido **dos águilas**?

Definamos

$A=\{\text{salieron dos águilas}\}$,

y

$B=\{\text{salió al menos un águila}\}$.

---

## Espacio muestral

Si representamos águila por $A$ y sol por $S$, los posibles resultados de lanzar dos veces una moneda son

$\Omega=\{AA,AS,SA,SS\}$.

Suponiendo que la moneda es justa, los cuatro resultados son igualmente probables.

El evento de obtener dos águilas es

$A=\{AA\}$.

Por lo tanto,

$P(A)=\frac14$.

El evento de obtener al menos un águila es

$B=\{AA,AS,SA\}$.

Por lo tanto,

$P(B)=\frac34$.

---

## Relación entre los eventos

Si obtuvimos dos águilas, entonces necesariamente obtuvimos al menos un águila.

Por ello,

$A\subseteq B$.

En consecuencia,

$A\cap B=A$.

Por lo tanto,

$P(A\cap B)=P(A)=\frac14$.

Aplicamos la definición de probabilidad condicional:

$P(A\mid B)=\frac{P(A\cap B)}{P(B)}$.

Sustituyendo,

$P(A\mid B)=\frac{\frac14}{\frac34}$.

Así,

$P(A\mid B)=\frac13$.

Por lo tanto,

$\boxed{P(\text{dos águilas}\mid\text{al menos un águila})=\frac13}$.

> **¿Por qué no es $\frac12$?**
>
> Una vez que sabemos que hubo al menos un águila, los posibles resultados ya no son los cuatro elementos de $\Omega$.
>
> El resultado $SS$ queda descartado.
>
> Los únicos resultados compatibles con la información son
>
> $\{AA,AS,SA\}$.
>
> De estos tres resultados igualmente probables, solamente uno corresponde a dos águilas. Por eso la probabilidad es $\frac13$.

# Regla de multiplicación para probabilidades

Partimos de la definición de probabilidad condicional:

$P(A\mid B)=\frac{P(A\cap B)}{P(B)}$.

Multiplicando ambos lados por $P(B)$ obtenemos

$P(A\mid B)P(B)=P(A\cap B)$.

Por lo tanto,

$\boxed{P(A\cap B)=P(A\mid B)P(B)}$.

Esta igualdad se conoce como la **regla de multiplicación**.

---

## Regla de multiplicación para tres eventos

Sean ahora $A$, $B$ y $C$ tres eventos.

Aplicamos la definición de probabilidad condicional a $C$ condicionado a que hayan ocurrido $A$ y $B$:

$P(C\mid A\cap B)=\frac{P(A\cap B\cap C)}{P(A\cap B)}$.

Multiplicando por $P(A\cap B)$,

$P(A\cap B\cap C)=P(C\mid A\cap B)P(A\cap B)$.

Pero por la regla de multiplicación,

$P(A\cap B)=P(A\mid B)P(B)$.

Sustituyendo,

$\boxed{P(A\cap B\cap C)=P(C\mid A\cap B)P(A\mid B)P(B)}$.

> **¿Qué significa esta fórmula?**
>
> Para que ocurran simultáneamente $B$, $A$ y $C$, podemos pensar el proceso en etapas:
>
> 1. Primero ocurre $B$.
> 2. Después ocurre $A$, sabiendo que ocurrió $B$.
> 3. Finalmente ocurre $C$, sabiendo que ocurrieron $A$ y $B$.
>
> La probabilidad de que ocurra todo el proceso se obtiene multiplicando las probabilidades correspondientes.

---

## Generalización

Para eventos $A_1,A_2,\ldots,A_n$, la regla de multiplicación puede escribirse como

$P(A_1\cap A_2\cap\cdots\cap A_n)=P(A_1)P(A_2\mid A_1)P(A_3\mid A_1\cap A_2)\cdots P(A_n\mid A_1\cap\cdots\cap A_{n-1})$.

Esta fórmula será especialmente útil para experimentos realizados **sin reemplazo**.

# Ejemplo 3. Tres bolas negras sin reemplazo

Tengo una urna con

- $10$ bolas blancas,
- $10$ bolas negras.

En total hay

$20$

bolas.

¿Cuál es la probabilidad de sacar **3 bolas negras sin reemplazo**?

Podemos resolver el problema de dos maneras.

---

# Método 1. Conteo combinatorio

Sea

$E=\{\text{las tres bolas seleccionadas son negras}\}$.

Como vamos a elegir $3$ bolas de un total de $20$, sin importar el orden en el que fueron elegidas, el número total de posibles selecciones es

$|\Omega|=\binom{20}{3}$.

> **¿Por qué $\binom{20}{3}$?**
>
> Estamos escogiendo un subconjunto de $3$ bolas entre las $20$ disponibles.
>
> El orden no es importante: seleccionar las bolas $1,2,3$ produce el mismo grupo que seleccionar $3,1,2$.
>
> Por eso utilizamos combinaciones.

Para que ocurra $E$, las tres bolas deben escogerse entre las $10$ bolas negras.

Por lo tanto,

$|E|=\binom{10}{3}$.

Así,

$P(E)=\frac{|E|}{|\Omega|}$.

Entonces,

$P(E)=\frac{\binom{10}{3}}{\binom{20}{3}}$.

Utilizando

$\binom{n}{k}=\frac{n!}{k!(n-k)!}$,

obtenemos

$P(E)=\frac{\frac{10!}{3!7!}}{\frac{20!}{3!17!}}$.

Dividir entre una fracción equivale a multiplicar por su recíproco, así que

$P(E)=\frac{10!}{3!7!}\frac{3!17!}{20!}$.

Cancelando $3!$,

$P(E)=\frac{10!17!}{7!20!}$.

Como

$\frac{10!}{7!}=8\cdot9\cdot10$

y

$\frac{20!}{17!}=18\cdot19\cdot20$,

obtenemos

$P(E)=\frac{8\cdot9\cdot10}{18\cdot19\cdot20}$.

Simplificando,

$P(E)=\frac{2}{19}$.

Por lo tanto,

$\boxed{P(E)=\frac{2}{19}\approx0.1053}$.

Es decir, la probabilidad es aproximadamente de un $10.53\%$.

# Método 2. Probabilidad condicional

Vamos a considerar las extracciones una por una.

Definamos

$B=\{\text{la primera bola es negra}\}$,

$A=\{\text{la segunda bola es negra}\}$,

y

$C=\{\text{la tercera bola es negra}\}$.

Queremos calcular

$P(A\cap B\cap C)$.

---

## Paso 1. Sacar la primera bola negra

Inicialmente tenemos $10$ bolas negras entre $20$ bolas.

Por lo tanto,

$P(B)=\frac{10}{20}$.

Así,

$P(B)=\frac12$.

---

## Paso 2. Sacar una segunda bola negra

Suponiendo que la primera bola fue negra, quedan

$19$

bolas en la urna.

Como retiramos una bola negra, ahora quedan

$9$

bolas negras.

Por lo tanto,

$P(A\mid B)=\frac{9}{19}$.

> **¿Por qué cambia la probabilidad?**
>
> El experimento es **sin reemplazo**.
>
> Después de extraer la primera bola, no la regresamos a la urna. Por eso ahora tenemos $19$ bolas en lugar de $20$, y solamente $9$ bolas negras en lugar de $10$.

---

## Paso 3. Sacar una tercera bola negra

Suponiendo que las dos primeras bolas fueron negras, quedan

$18$

bolas en la urna.

De ellas,

$8$

son negras.

Por lo tanto,

$P(C\mid A\cap B)=\frac{8}{18}$.

Simplificando,

$P(C\mid A\cap B)=\frac49$.

---

## Usamos la regla de multiplicación

Tenemos

$P(A\cap B\cap C)=P(C\mid A\cap B)P(A\mid B)P(B)$.

Sustituyendo,

$P(A\cap B\cap C)=\frac{8}{18}\frac{9}{19}\frac{10}{20}$.

Equivalentemente,

$P(A\cap B\cap C)=\frac49\frac9{19}\frac12$.

Cancelando el factor $9$,

$P(A\cap B\cap C)=\frac{4}{19}\frac12$.

Finalmente,

$\boxed{P(A\cap B\cap C)=\frac2{19}}$.

Observamos que obtenemos exactamente el mismo resultado que mediante el conteo combinatorio:

$\frac{\binom{10}{3}}{\binom{20}{3}}=\frac{10}{20}\frac9{19}\frac8{18}=\frac2{19}$.

> **Idea importante.**
>
> Este ejemplo muestra dos maneras diferentes de resolver un mismo problema:
>
> - mediante **conteo combinatorio**;
> - mediante **probabilidades condicionales y la regla de multiplicación**.
>
> La igualdad de ambos resultados no es casual: los dos métodos están contando el mismo fenómeno desde perspectivas diferentes.

# Fórmula de la probabilidad total

Sea

$F_1,F_2,\ldots,F_n$

una colección de eventos tales que

$F_i\cap F_j=\varnothing\quad\text{si }i\neq j$,

y

$\bigcup_{i=1}^{n}F_i=\Omega$.

Además, suponemos que

$P(F_i)>0$

para todo $i$.

Estas condiciones significan que los eventos

$F_1,F_2,\ldots,F_n$

forman una **partición del espacio muestral $\Omega$**.

Entonces, para cualquier evento $A$,

$\boxed{P(A)=\sum_{i=1}^{n}P(F_i)P(A\mid F_i)}$.

Esta expresión se conoce como la **fórmula o ley de la probabilidad total**.

---

## Demostración

Como

$\bigcup_{i=1}^{n}F_i=\Omega$,

tenemos

$P(A)=P(A\cap\Omega)$.

Sustituyendo $\Omega$ por la unión de los $F_i$,

$P(A)=P\left(A\cap\left(\bigcup_{i=1}^{n}F_i\right)\right)$.

La intersección se distribuye sobre la unión, así que

$A\cap\left(\bigcup_{i=1}^{n}F_i\right)=\bigcup_{i=1}^{n}(A\cap F_i)$.

Por lo tanto,

$P(A)=P\left(\bigcup_{i=1}^{n}(A\cap F_i)\right)$.

Como los eventos $F_i$ son mutuamente excluyentes, también los eventos

$A\cap F_1,A\cap F_2,\ldots,A\cap F_n$

son mutuamente excluyentes.

Entonces podemos sumar sus probabilidades:

$P(A)=\sum_{i=1}^{n}P(A\cap F_i)$.

Por la regla de multiplicación,

$P(A\cap F_i)=P(A\mid F_i)P(F_i)$.

Sustituyendo,

$P(A)=\sum_{i=1}^{n}P(F_i)P(A\mid F_i)$.

Por lo tanto,

$\boxed{P(A)=\sum_{i=1}^{n}P(F_i)P(A\mid F_i)}$.

---

## Interpretación

Los eventos

$F_1,F_2,\ldots,F_n$

representan todas las distintas maneras mutuamente excluyentes en las que puede ocurrir el experimento.

Para que ocurra $A$, necesariamente debe ocurrir a través de alguno de los eventos $F_i$.

Para cada $i$, la cantidad

$P(F_i)P(A\mid F_i)$

representa la probabilidad de que primero estemos en el caso $F_i$ y después ocurra $A$.

Al sumar sobre todos los posibles casos obtenemos la probabilidad total de $A$.

# Fórmula de Bayes

La fórmula anterior es la **fórmula de la probabilidad total**.

A partir de ella podemos obtener la **fórmula de Bayes**.

Sean $F_1,\ldots,F_n$ una partición de $\Omega$ y sea $A$ un evento con $P(A)>0$.

Por definición de probabilidad condicional,

$P(F_j\mid A)=\frac{P(F_j\cap A)}{P(A)}$.

Por la regla de multiplicación,

$P(F_j\cap A)=P(A\mid F_j)P(F_j)$.

Además, por la fórmula de la probabilidad total,

$P(A)=\sum_{i=1}^{n}P(A\mid F_i)P(F_i)$.

Sustituyendo ambas expresiones,

$\boxed{P(F_j\mid A)=\frac{P(A\mid F_j)P(F_j)}{\displaystyle\sum_{i=1}^{n}P(A\mid F_i)P(F_i)}}$.

Esta es la **fórmula de Bayes**.

> **Idea central.**
>
> La probabilidad total nos permite calcular $P(A)$ considerando todas las maneras en que $A$ puede ocurrir.
>
> Bayes nos permite invertir una probabilidad condicional: a partir de información sobre $P(A\mid F_j)$ podemos calcular $P(F_j\mid A)$.
