## Fórmula de Bayes

Sean

$$
F_1,\ldots,F_n
$$

eventos mutuamente excluyentes, es decir,

$$
F_i\cap F_j=\varnothing
\qquad
\text{si }i\neq j,
$$

y además supongamos que

$$
\Omega
=
\bigcup_{i=1}^{n}F_i.
$$

Entonces, para cualquier evento $A$,

$$
P(A)
=
\sum_{i=1}^{n}
P(F_i)P(A\mid F_i).
$$


## Ejemplo

Tiro una moneda.

- Si sale **águila**, tiro un dado.
- Si sale **sol**, tiro 2 dados.

**¿Cuál es la probabilidad de sacar exactamente un 6?**

Definimos

$$
A=\{\text{sacar exactamente un 6}\}.
$$

Además,

$$
F_1=\{\text{sale águila}\},
\qquad
F_2=\{\text{sale sol}\}.
$$

Como la moneda es justa,

$$
P(F_1)=\frac12=P(F_2).
$$

### Caso 1: sale águila

En este caso tiramos solamente un dado, por lo que

$$
P(A\mid F_1)=\frac16.
$$

### Caso 2: sale sol

En este caso tiramos dos dados.

Para obtener exactamente un 6, puede ocurrir:

- El primer dado es 6 y el segundo no es 6.
- El primer dado no es 6 y el segundo es 6.

Por lo tanto,

$$
P(A\mid F_2)
=
\frac{10}{36}.
$$

Usando la fórmula de probabilidad total,

$$
P(A)
=
P(F_1)P(A\mid F_1)
+
P(F_2)P(A\mid F_2).
$$

Entonces,

$$
P(A)
=
\frac12\left(\frac16\right)
+
\frac12\left(\frac{10}{36}\right).
$$

Así,

$$
P(A)
=
\frac{2}{9}
\approx 0.22.
$$

## Ejemplo

Tengo 2 monedas idénticas en todo, excepto que una está cargada y al tirarla siempre sale águila.

Escojo una moneda al azar, la tiro 2 veces y salen dos águilas, $AA$.

**¿Cuál es la probabilidad de que sea la moneda honesta?**

Definimos

$$
A=\{\text{salen 2 águilas}\}.
$$

Además,

$$
F_1=\{\text{la moneda es honesta}\},
$$

$$
F_2=\{\text{la moneda está cargada}\}.
$$

Queremos calcular

$$
P(F_1\mid A).
$$

Por la fórmula de Bayes,

$$
P(F_1\mid A)
=
\frac{P(F_1)P(A\mid F_1)}
{P(A)}.
$$

Como la moneda se escoge al azar,

$$
P(F_1)=P(F_2)=\frac12.
$$

Si la moneda es honesta,

$$
P(A\mid F_1)
=
\frac12\cdot\frac12
=
\frac14.
$$

Si la moneda está cargada,

$$
P(A\mid F_2)=1.
$$

Por la fórmula de probabilidad total,

$$
P(A)
=
P(F_1)P(A\mid F_1)
+
P(F_2)P(A\mid F_2).
$$

Así,

$$
P(A)
=
\frac12\left(\frac14\right)
+
\frac12(1)
=
\frac58.
$$

Por lo tanto,

$$
P(F_1\mid A)
=
\frac{\frac12\left(\frac14\right)}
{\frac58}
=
\frac15
=
0.2.
$$

## Ejemplo

Tiro un dado de 6 lados.

Tomo tantas cartas como indique el dado.

**¿Cuál es la probabilidad de tener al menos un as?**

Definimos

$$
A=\{\text{sacar al menos un as}\}.
$$

También definimos

$$
F_i=\{\text{el dado muestra }i\},
\qquad i\in\{1,\ldots,6\}.
$$

Como el dado es justo,

$$
P(F_i)=\frac16,
\qquad
\forall i\in\{1,\ldots,6\}.
$$

Condicionado a que el dado haya mostrado $i$, tomamos $i$ cartas.

La probabilidad de obtener al menos un as es

$$
P(A\mid F_i)
=
1-P(A^c\mid F_i).
$$

Como existen 48 cartas que no son ases,

$$
P(A^c\mid F_i)
=
\frac{\binom{48}{i}}
{\binom{52}{i}}.
$$

Por lo tanto,

$$
P(A\mid F_i)
=
1-
\frac{\binom{48}{i}}
{\binom{52}{i}}.
$$

Usando la fórmula de probabilidad total,

$$
P(A)
=
\sum_{i=1}^{6}
P(F_i)P(A\mid F_i).
$$

Entonces,

$$
P(A)
=
\sum_{i=1}^{6}
\left(
1-\frac{\binom{48}{i}}
{\binom{52}{i}}
\right)
\frac16
\approx 0.25.
$$


## Segunda fórmula de Bayes

Sabemos que

$$
P(F_i\mid A)
=
\frac{P(F_i\cap A)}
{P(A)}.
$$

Podemos escribir

$$
P(F_i\mid A)
=
\frac{P(A\cap F_i)}
{P(F_i)}
\frac{P(F_i)}
{P(A)}.
$$

Pero

$$
\frac{P(A\cap F_i)}
{P(F_i)}
=
P(A\mid F_i).
$$

Por lo tanto,

$$
P(F_i\mid A)
=
\frac{P(F_i)}{P(A)}
P(A\mid F_i).
$$

Equivalentemente,

$$
\boxed{
P(F_i\mid A)
=
\frac{P(F_i)P(A\mid F_i)}
{P(A)}
}
$$

donde:

- $P(F_i\mid A)$ es la **probabilidad a posteriori**.
- $P(F_i)$ es la **probabilidad a priori**.
- $P(A\mid F_i)$ es la **verosimilitud**.
- $P(A)$ es la **evidencia**.

## Ejemplo

Supongamos que el $10\%$ de la población tiene COVID.  
Hay una prueba de COVID con precisión del $80\%$.

$$
P(\text{positiva}\mid\text{enferma})=0.8
$$

**Verdadero positivo.**

$$
P(\text{positiva}\mid\text{sana})=0.2
$$

**Falso positivo.**

Una persona se hace la prueba y sale positiva.

**¿Cuál es la probabilidad de que sí esté enferma?**

Definimos los eventos:

- $F_1=\{\text{la persona está enferma}\}$.
- $F_2=\{\text{la persona está sana}\}$.
- $A=\{\text{la prueba sale positiva}\}$.

Queremos calcular

$$
P(F_1\mid A).
$$

Sabemos que

$$
P(F_1)=0.1,
\qquad
P(F_2)=0.9.
$$

Además,

$$
P(A\mid F_1)=0.8,
\qquad
P(A\mid F_2)=0.2.
$$

Por la fórmula de probabilidad total,

$$
P(A)
=
P(F_1)P(A\mid F_1)
+
P(F_2)P(A\mid F_2).
$$

Por lo tanto,

$$
P(A)
=
(0.1)(0.8)+(0.9)(0.2).
$$

Usando la fórmula de Bayes,

$$
P(F_1\mid A)
=
\frac{P(F_1)P(A\mid F_1)}
{P(F_1)P(A\mid F_1)+P(F_2)P(A\mid F_2)}.
$$

Entonces,

$$
P(F_1\mid A)
=
\frac{(0.1)(0.8)}
{(0.1)(0.8)+(0.9)(0.2)}
\approx 0.31.
$$
