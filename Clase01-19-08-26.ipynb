El **espacio muestral** es el conjunto de todos los posibles resultados de un experimento  
(lo denotaremos comúnmente como $\Omega$).

Un evento $E \subseteq \Omega$ es un subconjunto del espacio muestral, representando un posible resultado del experimento.

Si todos los resultados posibles del experimento son equiprobables:

$$
P(E) = \frac{|E|}{|\Omega|}
$$

> No puede haber un conjunto con infinitos resultados equiprobables en un experimento.

## Ejemplo

Una gallina pone **4 huevos**.

Suponiendo que la probabilidad de que nazca un **gallo** o una **gallina** de cada huevo es equiprobable:

**¿Cuál es la probabilidad de que en 4 huevos salgan 2 gallos y 2 gallinas?**

### Solución

Denotaremos por:

- $N =$ Gallina
- $Y =$ Gallo

Así:

$$
\Omega = \{
XXXX,\; YXXX,\; XYXX,\; XXYX,\; XXXY,
$$

$$
YYXX,\; YXYX,\; YXXY,\; XYYX,
$$

$$
XYXY,\; XXYY,\; YXYY,\; YYXY,
$$

$$
YYYX,\; XYYY,\; YYYY
\}
$$

El evento:

$$
E = \{
YYXX,\; YXYX,\; YXXY,\; XYYX,\; XYXY,\; XXYY
\}
$$

Entonces:

$$
P(E) = \frac{|E|}{|\Omega|}
= \frac{6}{16}
= \frac{3}{8}
$$

El evento más probable es que haya **3 de un sexo y 1 del otro**, es decir:

- 3 gallos y 1 gallina, o
- 3 gallinas y 1 gallo.

## Principio de conteo

Si mi experimento tiene $r$ etapas independientes, con $n_i$ posibles resultados para la etapa $i$-ésima, el número de resultados posibles es:

$$
\prod_{i=1}^{r} n_i
$$

## Ejemplo

### Primera, segunda y tercera etapa

- Primera etapa: $3$ posibilidades
- Segunda etapa: $2$ posibilidades
- Tercera etapa: $3$ posibilidades

Por el principio de conteo:

$$
3 \times 2 \times 3 = 18
$$

## Ejemplo

Tiro un dado 4 veces.

**¿Cuál es la probabilidad de que salgan números diferentes en cada tiro?**

El espacio muestral es:

$$
\Omega = \{(a,b,c,d) : a,b,c,d \in \{1,\ldots,6\}\}
$$

Hay 4 etapas y cada etapa tiene 6 posibilidades, entonces:

$$
|\Omega| = 6^4
$$

Definimos el evento:

$$
E = \{(a,b,c,d) \mid a,b,c,d \in \{1,\ldots,6\},\;
a\neq b,\;
a\neq c,\;
a\neq d,\;
b\neq c,\;
b\neq d,\;
c\neq d
\}
$$

Hay 6 posibilidades para la primera etapa.

Elegimos un número.  
(Pongamos que sale 1).

Entonces, para la segunda etapa solo quedan 5 posibilidades:

$$
2,\;3,\;4,\;5,\;6
$$

Volvemos a elegir. (Pongamos que sale 3).

Nos quedan 4 posibilidades para la **3ra etapa**, que son:

$$
2,\;4,\;5,\;6
$$

Volvemos a escoger y nos quedan 3 posibilidades para la **4ta etapa**, es decir:

| Etapa 1 | Etapa 2 | Etapa 3 | Etapa 4 |
|---|---|---|---|
| 6 | 5 | 4 | 3 |

Entonces:

$$
P(E)
=
\frac{|E|}{|\Omega|}
=
\frac{6\cdot5\cdot4\cdot3}{|\Omega|}
=
\frac{5\cdot4\cdot3}{6^3}
\approx 0.28
$$

## Permutaciones

Si tengo $n$ elementos, el número de formas de ordenar todos es $n!$, ya que:

$$
(a_1, a_2, a_3, a_4, \ldots, a_n)
$$

Para el primer elemento hay $n$ posibilidades, para el segundo $n-1$ posibilidades, para el tercero $n-2$ posibilidades, y así sucesivamente hasta llegar a $1$.

Entonces:

$$
n(n-1)(n-2)\cdots 1 = n!
$$

Si queremos ordenar un subconjunto con $k \leq n$ elementos de todos los $n$ elementos:

Si $k=1$, hay $n$ formas de escoger.

Si $k=2$, hay $n$ posibilidades para la primera etapa y $n-1$ para $a_2$.

Por lo tanto:

$$
n(n-1)
$$

Si $k=3$, hay:

- $n$ posibilidades para $a_1$,
- $n-1$ posibilidades para $a_2$,
- $n-2$ posibilidades para $a_3$.

Entonces hay:

$$
n(n-1)(n-2)
$$

En general, el número de formas de ordenar $k$ elementos es:

$$
n(n-1)\cdots(n-k+1)
=
\frac{n!}{(n-k)!}
$$

## Ejemplo: Juego de cartas

Hay 4 palos con 13 cartas cada palo. Es decir, hay **52 cartas**.

**¿Cuál es la probabilidad de que salga 1 as después de barajear?**

Sabemos que:

$$
|\Omega| = 52!
$$

pues:

$$
\Omega = \{\text{todos los posibles ordenamientos de las 52 cartas después de barajear}\}
$$

Definimos el evento:

$$
E = \{\text{un as hasta arriba}\}
$$

En la primera etapa tenemos los 4 ases, y entonces hay 4 posibilidades.

Para la segunda etapa hay:

$$
51!
$$

posibilidades.

Por lo tanto:

$$
|E| = 4\cdot 51!
$$

Entonces:

$$
P(E)
=
\frac{4\cdot 51!}{52!}
=
\frac{1}{13}
$$

**¿Cuál es la probabilidad de que todos los corazones estén juntos?**

Hay 13 corazones.

Si juntamos los 13 corazones, hay:

$$
13!
$$

formas de ordenar los corazones.

Y las posiciones como bloque son:

$$
40!
$$

Por lo tanto:

$$
|E| = 13! \cdot 40!
$$

Entonces:

$$
P(E)
=
\frac{13! \cdot 40!}{52!}
$$
