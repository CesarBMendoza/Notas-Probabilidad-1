A la terna $(\Omega,\mathcal{F},\mathbb{P})$ lo llamaremos el espacio de probabilidad.

$\rightarrow$ Cuando $\Omega$ es discreto diremos que:

$(\Omega,\mathcal{F},\mathbb{P})$ es un E. probabilidad discreto.

$\rightarrow$ Cuando $\Omega$ es continuo diremos que:

$(\Omega,\mathcal{F},\mathbb{P})$ es un E. probabilidad continuo.

### 1) Espacios discretos

$\Omega=\{\omega_1,\omega_2,\omega_3,\ldots\}$ y $\mathcal{F}=\mathcal{P}(\Omega)$.

Si $i\neq j$ y para $\omega_i,\omega_j\in\Omega$,

$\{\omega_i\}\cap\{\omega_j\}=\varnothing$.

Entonces:

Si $A\in\mathcal{F}$,

$A=\{\omega_j,\omega_k,\omega_\ell,\ldots\}$

$=\{\omega_j\}\cup\{\omega_k\}\cup\{\omega_\ell\}\cup\cdots$

Así:

$\mathbb{P}(A)=\mathbb{P}(\{\omega_j\})\cup\cdots\cup\mathbb{P}(\{\omega_k\})\cup\mathbb{P}(\{\omega_\ell\})\cup\cdots$

Y como es engorroso estar escribiendo, denotaremos como

$p_j=\mathbb{P}(\{\omega_j\})$,

la función de probabilidad puntual (pmf).

**Ejemplo:** Tiramos 1 dado justo.

$\Omega=\{1,\ldots,6\}$. Entonces:

$p_j=\mathbb{P}(\{j\})=\frac{1}{6}$, en general,

$p_j=\frac{1}{6},\qquad \forall j\in\{1,\ldots,6\}$.

¿Cuál es la proba de que salga número par?

$\mathbb{P}(E)=p_2+p_4+p_6
=\frac{1}{6}+\frac{1}{6}+\frac{1}{6}
=\frac{1}{2}$.

---

**Ejemplo:**

Tirar una moneda hasta que caiga sol.

$\Omega=\{S,AS,AAS,AAAS,\ldots\}$.

Notemos que para $S$ salieron $0$ águilas.

$AS\longrightarrow$ salió $1$ águila.

$AAS\longrightarrow$ salieron $2$ águilas.

$\vdots$

Entonces, mejor consideremos

$\Omega=\{1,\ldots\}$.

Supongamos que la moneda es justa.

$p_n=\mathbb{P}(\{n\})
=\mathbb{P}(\{AA\ldots AS\})
=\frac{1}{2^n}$,

donde aparecen $n-1$ veces $A$.

¿Y si es?

Notemos que:

$\mathbb{P}(\Omega)
=\sum_{k=1}^{\infty}\mathbb{P}(\{k\})
=\sum_{k=1}^{\infty}p_k
=\sum_{k=1}^{\infty}\frac{1}{2^k}
=1$.

Pues

$\sum_{k=0}^{\infty}\frac{1}{2^k}$

es una serie geométrica y es tal que

$\sum_{k=0}^{\infty}\frac{1}{2^k}
=\frac{1}{1-\frac{1}{2}}
=2$.

¿Cuál es la probabilidad de que necesite tirar la moneda un número par de veces?

$E=\{2,4,6,\ldots\}=2\mathbb{N}$.

$\mathbb{P}(E)
=\sum_{j=1}^{\infty}p_{2j}
=\frac{1}{2^2}+\frac{1}{2^4}+\cdots
=\sum_{j=1}^{\infty}\frac{1}{2^{2j}}$

$=\sum_{j=1}^{\infty}\frac{1}{4^j}
=\frac{1}{1-\frac{1}{4}}-1
=\frac{4}{3}-1
=\frac{1}{3}$.

$\boxed{\mathbb{P}(E)=\frac{1}{3}}$

## Espacios de probabilidad continuos

Casi siempre supondremos que $\Omega\subseteq\mathbb{R}$.

Pero...

¿$\mathcal{F}=\mathcal{P}(\mathbb{R})$? **NO**, pues este conjunto es tan grande que tiene conjuntos que no se pueden medir.

(Los conjuntos de Vitali's).

Consideraremos a

$A=\{\text{uniones finitas de subintervalos de }\mathbb{R}\}$.

$\sigma(A)$ es el conjunto donde a $A$ le agregamos todos los conjuntos necesarios para medir eventos

[uniones numerables de los intervalos].

Entonces

$\sigma(A)=\mathcal{B}_{\mathbb{R}}=\mathcal{B}$

($\sigma$-álgebra de Borel).

---

En espacios de proba continua,

si $\omega\in\Omega$

$\Rightarrow\mathbb{P}(\omega)=0$.

**Ejemplo:** ¿Cuál es la proba que un insecto en el suelo mida

$3.14162041\ldots$?

Si $A\subseteq\Omega$,

$\mathbb{P}(A)=\int_A f(x)\,dx$.

Donde a $f(x)$ se le llamará la función de densidad de probabilidad.

$\Rightarrow$

$\mathbb{P}(E)=\int_E f(x)\,dx$.

Si $f(x)$ puede tener esta forma, ¿sí?

No, pues hay una región Donde $f$ toma valores negativos. Así, para una región $E$ en dichos valores negativos,

$\Rightarrow$

$\int_E f(x)\,dx<0$,

por lo que $f$ tiene que ser no negativa.

$\mathbb{P}(\Omega)
=\mathbb{P}(\mathbb{R})
=\int_{-\infty}^{\infty}f(x)\,dx
=1$.

Por lo tanto, $f$ **no** es una probabilidad.
