# Ejemplo: parejas casadas alrededor de una mesa redonda

Calcule la probabilidad de que, al sentar al azar a $10$ parejas casadas alrededor de una mesa redonda, ninguna esposa quede sentada junto a su esposo.

Tenemos $20$ personas.

Como se trata de una mesa redonda, fijamos a una persona para eliminar las rotaciones equivalentes. Por lo tanto, el número total de formas de sentar a las $20$ personas es

$$
(20-1)!=19!.
$$

Definimos los eventos

$$
A_i=\{\text{la pareja }i\text{ queda sentada junta}\},
\qquad i\in\{1,\ldots,10\}.
$$

Entonces

$$
\bigcap_{i=1}^{10}A_i
$$

representa el evento en el que **todas las parejas quedan juntas**, mientras que

$$
\bigcup_{i=1}^{10}A_i
$$

representa el evento en el que **al menos una pareja queda junta**.

Lo que queremos calcular es

$$
P\left(\bigcap_{i=1}^{10}A_i^c\right).
$$

Por las leyes de De Morgan,

$$
\bigcap_{i=1}^{10}A_i^c
=
\left(\bigcup_{i=1}^{10}A_i\right)^c,
$$

por lo que

$$
P\left(\bigcap_{i=1}^{10}A_i^c\right)
=
1-
P\left(\bigcup_{i=1}^{10}A_i\right).
$$

Por lo tanto, primero calcularemos la probabilidad de que **al menos una pareja quede junta** usando inclusión-exclusión.

---

## Probabilidad de que una pareja específica quede junta

Consideremos $A_i$.

Queremos que el esposo y la esposa de la pareja $i$ queden juntos.

Podemos considerar a esta pareja como un solo bloque.

Así, en lugar de tener $20$ objetos alrededor de la mesa, tenemos

$$
19
$$

objetos:

- el bloque formado por la pareja $i$;
- las otras $18$ personas.

Como están sentados alrededor de una mesa redonda, estos $19$ objetos pueden acomodarse de

$$
(19-1)!=18!
$$

formas.

Además, dentro del bloque, los miembros de la pareja pueden sentarse en dos órdenes:

$$
XY
\qquad\text{o}\qquad
YX.
$$

Por lo tanto, el número de acomodos favorables es

$$
18!\,2.
$$

Así,

$$
P(A_i)
=
\frac{18!\,2}{19!}.
$$

---

## Dos parejas específicas juntas

Consideremos ahora

$$
A_i\cap A_j.
$$

Esto significa que las parejas $i$ y $j$ deben quedar juntas.

Consideramos cada pareja como un bloque.

Entonces tenemos

$$
20-2=18
$$

objetos alrededor de la mesa:

- $2$ bloques de parejas;
- las otras $16$ personas.

El número de acomodos circulares es entonces

$$
(18-1)!=17!.
$$

Cada una de las dos parejas puede acomodarse internamente de $2$ maneras, por lo que aparecen

$$
2^2
$$

ordenamientos internos.

Por lo tanto,

$$
P(A_i\cap A_j)
=
\frac{17!\,2^2}{19!}.
$$

---

## Tres parejas específicas juntas

Para

$$
A_i\cap A_j\cap A_k,
$$

consideramos a las tres parejas como bloques.

Tenemos entonces

$$
20-3=17
$$

objetos alrededor de la mesa.

El número de acomodos circulares es

$$
(17-1)!=16!.
$$

Además, cada pareja tiene $2$ posibles órdenes internos, por lo que tenemos

$$
2^3
$$

posibilidades.

Así,

$$
P(A_i\cap A_j\cap A_k)
=
\frac{16!\,2^3}{19!}.
$$

---

## Caso general

Si escogemos $k$ parejas específicas y exigimos que cada una de ellas quede junta, tratamos esas $k$ parejas como bloques.

Entonces quedan

$$
20-k
$$

objetos alrededor de la mesa.

Por lo tanto, existen

$$
(20-k-1)!=(19-k)!
$$

acomodos circulares.

Además, cada una de las $k$ parejas puede intercambiar sus lugares internamente, lo que da

$$
2^k
$$

posibilidades.

Así,

$$
P(A_{i_1}\cap\cdots\cap A_{i_k})
=
\frac{(19-k)!\,2^k}{19!}.
$$

---

## Aplicamos inclusión-exclusión

Hay

$$
\binom{10}{1}
$$

formas de escoger una pareja,

$$
\binom{10}{2}
$$

formas de escoger dos parejas,

$$
\binom{10}{3}
$$

formas de escoger tres parejas,

y, en general,

$$
\binom{10}{k}
$$

formas de escoger $k$ parejas.

Por el principio de inclusión-exclusión,

$$
P\left(\bigcup_{i=1}^{10}A_i\right)
=
\sum_{k=1}^{10}
(-1)^{k+1}
\binom{10}{k}
\frac{(19-k)!\,2^k}{19!}.
$$

Es decir,

$$
P\left(\bigcup_{i=1}^{10}A_i\right)
=
\binom{10}{1}\frac{18!\,2}{19!}
-
\binom{10}{2}\frac{17!\,2^2}{19!}
+
\binom{10}{3}\frac{16!\,2^3}{19!}
-\cdots
-
\binom{10}{10}\frac{9!\,2^{10}}{19!}.
$$

Numéricamente,

$$
P\left(\bigcup_{i=1}^{10}A_i\right)
\approx 0.66046277.
$$

Esta es la probabilidad de que **al menos una pareja quede junta**.

Pero nosotros queremos que **ninguna pareja quede junta**. Por lo tanto,

$$
P\left(\bigcap_{i=1}^{10}A_i^c\right)
=
1-
P\left(\bigcup_{i=1}^{10}A_i\right).
$$

Entonces

$$
P\left(\bigcap_{i=1}^{10}A_i^c\right)
\approx
1-0.66046277.
$$

Finalmente,

$$
\boxed{
P(\text{ninguna pareja queda junta})
\approx 0.33953723
}.
$$

Es decir, aproximadamente

$$
\boxed{33.95\%}.
$$

# Ejemplo: placas de automóvil de la CDMX

Las placas de un automóvil de la CDMX tienen el formato

$$
\text{Letra}-\text{Número}-\text{Número}-\text{Letra}-\text{Letra}.
$$

Supongamos que:

- cada letra puede ser cualquier letra del alfabeto excepto la $\tilde{N}$;
- tenemos, por lo tanto, $26$ letras disponibles;
- cada número puede ser cualquiera de los $10$ dígitos $0,1,\ldots,9$;
- todas las placas posibles son igualmente probables.

Queremos calcular la probabilidad de obtener una placa en la que **todos los símbolos del mismo tipo sean distintos**, es decir:

- las tres letras sean distintas entre sí;
- los dos números sean distintos entre sí.

---

## Número total de placas

Para cada una de las tres posiciones correspondientes a letras tenemos $26$ opciones.

Para cada una de las dos posiciones numéricas tenemos $10$ opciones.

Por la regla de multiplicación,

$$
|\Omega|
=
26\cdot10\cdot10\cdot26\cdot26.
$$

Por lo tanto,

$$
|\Omega|
=
26^3\,10^2.
$$

---

## Placas favorables

Sea

$$
E=\{\text{las tres letras son distintas y los dos números son distintos}\}.
$$

Para las letras:

- la primera letra puede escogerse de $26$ maneras;
- la segunda debe ser distinta de la primera, así que tenemos $25$ opciones;
- la tercera debe ser distinta de las dos anteriores, así que quedan $24$ opciones.

Por lo tanto,

$$
26\cdot25\cdot24
$$

formas de elegir las letras.

Para los números:

- el primer número tiene $10$ posibilidades;
- el segundo debe ser diferente, así que tiene $9$ posibilidades.

Por lo tanto,

$$
10\cdot9
$$

formas de escoger los números.

Por la regla de multiplicación,

$$
|E|
=
26\cdot25\cdot24\cdot10\cdot9.
$$

También podemos escribir

$$
26\cdot25\cdot24
=
\frac{26!}{23!},
$$

y

$$
10\cdot9
=
\frac{10!}{8!}.
$$

Así,

$$
|E|
=
\frac{26!}{23!}
\frac{10!}{8!}.
$$

La probabilidad buscada es

$$
P(E)
=
\frac{|E|}{|\Omega|}.
$$

Por lo tanto,

$$
\boxed{
P(E)
=
\frac{
26\cdot25\cdot24\cdot10\cdot9
}{
26^3\,10^2
}
}.
$$

Podemos simplificar:

$$
P(E)
=
\frac{25}{26}
\frac{24}{26}
\frac{9}{10}.
$$

# Método de estrellas y barras

Queremos contar el número de vectores de enteros positivos

$$
(x_1,\ldots,x_r)
$$

que satisfacen

$$
x_1+\cdots+x_r=n,
$$

con

$$
x_i>0,
\qquad
i=1,\ldots,r.
$$

El número de soluciones es

$$
\boxed{
\binom{n-1}{r-1}
}.
$$

---

## Ejemplo

Consideremos

$$
x_1+x_2+x_3=7,
$$

con

$$
x_1,x_2,x_3>0.
$$

Representamos los $7$ objetos mediante estrellas:

$$
*\,*\,*\,*\,*\,*\,*.
$$

Queremos dividir estas $7$ estrellas en $3$ grupos no vacíos.

Como cada grupo debe contener al menos una estrella, los separadores solamente pueden colocarse **entre** dos estrellas consecutivas.

Entre $7$ estrellas existen

$$
7-1=6
$$

espacios:

$$
*\ \square\ *\ \square\ *\ \square\ *\ \square\ *\ \square\ *\ \square\ *.
$$

Como queremos formar $3$ grupos, necesitamos

$$
3-1=2
$$

separadores.

Por ejemplo,

$$
**\mid *\mid ****
$$

representa la solución

$$
(x_1,x_2,x_3)=(2,1,4).
$$

Por lo tanto, solamente debemos escoger $2$ de los $6$ espacios disponibles.

Así, el número de soluciones es

$$
\binom{6}{2}.
$$

En general, para

$$
x_1+\cdots+x_r=n,
\qquad x_i>0,
$$

tenemos $n-1$ espacios entre las estrellas y debemos escoger $r-1$ de ellos.

Por lo tanto,

$$
\boxed{
\#\{(x_1,\ldots,x_r)\in\mathbb{N}^r:
x_1+\cdots+x_r=n\}
=
\binom{n-1}{r-1}
}.
$$

---

# Ejemplo: rachas de victorias y derrotas

Considere un equipo que acaba de terminar su temporada con un récord final de

$$
n \text{ victorias}
$$

y

$$
m \text{ derrotas}.
$$

Queremos estudiar la secuencia de victorias y derrotas para determinar si hubo periodos en los que parecía más probable ganar que en otros momentos.

Denotaremos por

$$
G=\text{ganar}
$$

y

$$
P=\text{perder}.
$$

Por ejemplo, una temporada podría tener una secuencia como

$$
GGGPPPGPP.
$$

Como el equipo tuvo $n$ victorias y $m$ derrotas, hubo en total

$$
n+m
$$

partidos.

---

## Número de secuencias posibles

Si únicamente conocemos que hubo $n$ victorias y $m$ derrotas, cada secuencia queda determinada por las posiciones en las que colocamos las $n$ victorias.

Tenemos $n+m$ posiciones y debemos escoger $n$ para colocar las victorias.

Por lo tanto, existen

$$
\binom{n+m}{n}
=
\frac{(n+m)!}{n!\,m!}
$$

secuencias posibles.

Equivalentemente podríamos escoger las posiciones de las $m$ derrotas:

$$
\binom{n+m}{m}
=
\binom{n+m}{n}.
$$

Bajo la hipótesis de que la probabilidad de ganar permanece constante e independiente de los resultados anteriores, y condicionando a haber observado exactamente $n$ victorias y $m$ derrotas, todas estas secuencias son igualmente probables.

---

# Número de secuencias con exactamente $r$ rachas de victorias

Queremos determinar la probabilidad de que haya exactamente

$$
r
$$

rachas de victorias.

Una **racha de victorias** es un bloque maximal de victorias consecutivas.

Por ejemplo,

$$
GGGPPGGPGGG
$$

tiene tres rachas de victorias:

$$
GGG,\qquad GG,\qquad GGG.
$$

---

## Paso 1: distribuir las victorias entre las $r$ rachas

Sean

$$
x_1,\ldots,x_r
$$

las longitudes de las $r$ rachas.

Entonces

$$
x_1+\cdots+x_r=n.
$$

Como cada racha debe contener al menos una victoria,

$$
x_i>0.
$$

Por el método de estrellas y barras, el número de soluciones es

$$
\binom{n-1}{r-1}.
$$

Esto representa las distintas formas de dividir las $n$ victorias en $r$ rachas no vacías.

---

## Paso 2: colocar las derrotas

Sean

$$
y_1,\ldots,y_{r+1}
$$

los números de derrotas:

- $y_1$: antes de la primera racha de victorias;
- $y_2$: entre la primera y segunda racha;
- $\cdots$
- $y_r$: entre la racha $r-1$ y la racha $r$;
- $y_{r+1}$: después de la última racha.

Entonces

$$
y_1+y_2+\cdots+y_r+y_{r+1}=m.
$$

Para que las $r$ rachas permanezcan separadas necesitamos al menos una derrota entre cada par de rachas.

Por lo tanto,

$$
y_2,\ldots,y_r>0,
$$

mientras que en los extremos podemos tener

$$
y_1\geq 0,
\qquad
y_{r+1}\geq0.
$$

Definimos

$$
\overline y_1=y_1,
$$

$$
\overline y_i=y_i-1,
\qquad i=2,\ldots,r,
$$

y

$$
\overline y_{r+1}=y_{r+1}.
$$

Entonces todas las nuevas variables son no negativas.

Como hemos retirado una derrota de cada uno de los $r-1$ espacios interiores,

$$
\overline y_1+\cdots+\overline y_{r+1}
=
m-(r-1).
$$

Ahora buscamos soluciones no negativas de esta ecuación.

Por estrellas y barras, el número es

$$
\binom{
[m-(r-1)]+(r+1)-1
}{
(r+1)-1
}.
$$

Simplificando,

$$
\binom{m+1}{r}.
$$

Por lo tanto, para cada forma de dividir las victorias en $r$ rachas existen

$$
\binom{m+1}{r}
$$

formas de distribuir las derrotas.

---

## Número total de secuencias con $r$ rachas

Tenemos

$$
\binom{n-1}{r-1}
$$

formas de formar las rachas de victorias y

$$
\binom{m+1}{r}
$$

formas de colocar las derrotas.

Por la regla de multiplicación,

$$
\boxed{
N_r
=
\binom{n-1}{r-1}
\binom{m+1}{r}
}.
$$

Como en total existen

$$
\binom{n+m}{n}
$$

secuencias posibles, obtenemos

$$
\boxed{
P(R=r)
=
\frac{
\binom{n-1}{r-1}
\binom{m+1}{r}
}{
\binom{n+m}{n}
}
}.
$$

Esta fórmula es válida siempre que

$$
1\leq r\leq \min(n,m+1).
$$

---

# Ejemplo numérico

Supongamos que

$$
n=8,
\qquad
m=6,
$$

y queremos calcular la probabilidad de observar exactamente

$$
r=7
$$

rachas de victorias.

Utilizando la fórmula,

$$
P(R=7)
=
\frac{
\binom{8-1}{7-1}
\binom{6+1}{7}
}{
\binom{8+6}{8}
}.
$$

Por lo tanto,

$$
P(R=7)
=
\frac{
\binom{7}{6}
\binom{7}{7}
}{
\binom{14}{8}
}.
$$

Como

$$
\binom{7}{6}=7,
\qquad
\binom{7}{7}=1,
\qquad
\binom{14}{8}=3003,
$$

obtenemos

$$
P(R=7)
=
\frac{7}{3003}
=
\frac{1}{429}.
$$

Numéricamente,

$$
\boxed{
P(R=7)\approx0.002331
}.
$$

Es decir, aproximadamente

$$
\boxed{0.2331\%}.
$$

Por lo tanto, bajo el modelo de una probabilidad constante de ganar, observar $7$ rachas de victorias es un evento poco frecuente.

---

## ¿Qué sucede si solamente hay una racha de victorias?

Ahora tomemos

$$
r=1.
$$

Entonces

$$
P(R=1)
=
\frac{
\binom{7}{0}
\binom{7}{1}
}{
\binom{14}{8}
}.
$$

Como

$$
\binom{7}{0}=1
$$

y

$$
\binom{7}{1}=7,
$$

tenemos

$$
P(R=1)
=
\frac{7}{3003}
=
\frac{1}{429}.
$$

Por lo tanto,

$$
\boxed{
P(R=1)=P(R=7)=\frac{1}{429}
}.
$$

Es decir, tanto una temporada con muy pocas rachas como una temporada con muchas rachas pueden ser poco frecuentes bajo este modelo.

---

## Nota importante sobre la interpretación

No debemos concluir simplemente que una **secuencia específica** es evidencia contra el modelo porque tenga una probabilidad pequeña.

Condicionados a que hubo exactamente $8$ victorias y $6$ derrotas, cada secuencia específica tiene probabilidad

$$
\frac{1}{\binom{14}{8}}
=
\frac{1}{3003}.
$$

Lo que estamos estudiando no es la probabilidad de una secuencia particular, sino la probabilidad de una **característica de la secuencia**, en este caso el número de rachas.

Por ejemplo,

$$
P(R=7)=\frac{1}{429}
$$

es la probabilidad de **todas las secuencias que poseen exactamente siete rachas de victorias**.

De manera similar,

$$
P(R=1)=\frac{1}{429}
$$

es la probabilidad de **todas las secuencias con exactamente una racha de victorias**.

Por ello, para usar el número de rachas como evidencia acerca de si la probabilidad de ganar permanece constante, debemos estudiar qué tan extremo es el número de rachas observado bajo la distribución de $R$, y no únicamente la probabilidad de una secuencia individual.
