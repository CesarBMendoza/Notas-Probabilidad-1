### Ejemplo

¿Cuál es la probabilidad de que todas las cartas de cada palo aparezcan juntas?

El número total de formas de ordenar las $52$ cartas es

$$
|\Omega|=52!.
$$

Para el evento $E$, podemos ordenar las $13$ cartas de cada palo de $13!$ formas, y además podemos ordenar los $4$ palos de $4!$ formas. Por lo tanto,

$$
|E|=(13!)^4(4!).
$$

Así,

$$
P(E)=\frac{(13!)^4(4!)}{52!}.
$$

- $13!$: formas de ordenar las cartas de cada palo.
- $4!$: formas de ordenar los $4$ palos.

## Combinaciones

Si queremos ordenar $k\leq n$ elementos de un conjunto de $n$ elementos, tenemos

$$
\frac{n!}{(n-k)!}
$$

posibilidades.

Ahora preguntamos:

**¿Cuántos subconjuntos diferentes de $k$ elementos puedo obtener de un conjunto de $n$ elementos?**

Por ejemplo, sea

$$
\{1,2,\ldots,n\}
$$

y supongamos que $k=3$.

Al contar arreglos ordenados obtenemos, por ejemplo,

$$
(1,2,3),\quad (2,1,3),\quad (1,3,2),\quad (1,5,4),\ldots
$$

Estos arreglos tienen un orden diferente. Sin embargo, si lo que queremos contar son subconjuntos, entonces

$$
\{1,2,3\}=\{2,1,3\}=\{1,3,2\}.
$$

Por lo tanto, estamos contando cada subconjunto de $k$ elementos exactamente $k!$ veces.

Así, el número de subconjuntos diferentes de $k$ elementos es

$$
\frac{n!}{(n-k)!k!}
=
\binom{n}{k}.
$$

El número

$$
\binom{n}{k}
$$

es llamado **coeficiente binomial**, y se lee también como

> "$n$ sobre $k$"

o

> "de $n$, escoger $k$".

### Ejercicio

¿Cuál es la probabilidad de que, tirando $10$ veces una moneda justa, salgan exactamente $5$ águilas y $5$ soles?

Como en cada lanzamiento tenemos dos resultados posibles,

$$
|\Omega|=2^{10}.
$$

Para obtener exactamente $5$ águilas, basta escoger en cuáles $5$ de las $10$ tiradas aparecen. Por lo tanto,

$$
|E|=\binom{10}{5}.
$$

Así,

$$
P(E)
=
\frac{\binom{10}{5}}{2^{10}}
=
\frac{252}{1024}
\approx 0.2461.
$$

### Ejercicio

Tengo una urna con $100$ pelotas:

- $50$ rojas,
- $50$ azules.

Se sacan $5$ pelotas.

¿Cuál es la probabilidad de sacar $3$ azules y $2$ rojas?

Como hay $100$ pelotas y queremos seleccionar $5$ sin importar el orden,

$$
|\Omega|=\binom{100}{5}.
$$

Para obtener $3$ pelotas azules podemos escogerlas de

$$
\binom{50}{3}
$$

formas.

De manera similar, las $2$ pelotas rojas se pueden escoger de

$$
\binom{50}{2}
$$

formas.

Por lo tanto,

$$
|E|
=
\binom{50}{3}\binom{50}{2}.
$$

Así,

$$
P(E)
=
\frac{
\binom{50}{3}\binom{50}{2}
}{
\binom{100}{5}
}.
$$

A veces hacemos los experimentos **con reemplazo** o **sin reemplazo**.

### Con reemplazo

Decimos que un experimento es **con reemplazo** cuando, después de realizar una elección, el objeto seleccionado se devuelve. Por lo tanto, en cada extracción volvemos a tener disponibles todas las opciones originales.

### Sin reemplazo

Decimos que un experimento es **sin reemplazo** cuando el objeto seleccionado no se devuelve. Por lo tanto, después de cada extracción tenemos menos opciones disponibles.

### Objetos indistinguibles y distinguibles

- **Indistinguibles:** no podemos diferenciar un objeto de otro del mismo tipo.
- **Distinguibles:** podemos diferenciar cada objeto de los demás.

### Ejemplo

Tengo $6$ papelitos con las letras

$$
P,\;P,\;P,\;E,\;E,\;R.
$$

¿Cuál es la probabilidad de sacar, en ese orden, la palabra

$$
\text{PEPPER}?
$$

Consideremos los dos casos:

1. Con reemplazo.
2. Sin reemplazo.

### 1. Con reemplazo

Como después de cada extracción regresamos el papelito, en cada una de las $6$ extracciones tenemos nuevamente $6$ papelitos disponibles.

Por lo tanto,

$$
|\Omega|=6^6.
$$

Queremos obtener

$$
P\,E\,P\,P\,E\,R.
$$

Para cada posición tenemos:

- primera $P$: $3$ posibilidades;
- $E$: $2$ posibilidades;
- segunda $P$: $3$ posibilidades;
- tercera $P$: $3$ posibilidades;
- segunda $E$: $2$ posibilidades;
- $R$: $1$ posibilidad.

Así,

$$
|E|
=
3\cdot2\cdot3\cdot3\cdot2\cdot1
=
3^3\,2^2.
$$

Por lo tanto,

$$
P(E)
=
\frac{3^3\,2^2}{6^6}
=
\frac{1}{432}
\approx 2.315\times10^{-3}.
$$

---

### 2. Sin reemplazo

Como no reemplazamos los papelitos, únicamente estamos reordenando los $6$ papelitos.

Si consideramos los papelitos físicamente distinguibles,

$$
|\Omega|=6!.
$$

Para formar la palabra

$$
P\,E\,P\,P\,E\,R
$$

podemos intercambiar entre sí los $3$ papelitos que contienen $P$, lo cual puede hacerse de

$$
3!
$$

formas.

De igual manera, los $2$ papelitos que contienen $E$ pueden intercambiarse de

$$
2!
$$

formas.

Por lo tanto,

$$
|E|=3!2!=12.
$$

Así,

$$
P(E)
=
\frac{3!2!}{6!}
=
\frac{12}{720}
=
\frac{1}{60}
\approx 0.01667.
$$

Por lo tanto,

$$
\frac{1}{60}>\frac{1}{432}.
$$

En este caso, es más probable obtener la palabra **PEPPER** sin reemplazo que con reemplazo.

Hasta ahora, cuando $\Omega$ es finito y todos los resultados elementales son equiprobables, hemos utilizado

$$
P(E)=\frac{|E|}{|\Omega|}.
$$

Pero, ¿qué ocurre si ahora $\Omega$ es infinito?

La probabilidad es una **medida**.

Pero, ¿qué es una medida?

Intuitivamente podemos pensar primero en la longitud.

Si $a<b$, entonces la longitud del intervalo $(a,b)$ es

$$
\ell((a,b))=b-a.
$$

Si los intervalos $(a,b)$ y $(c,d)$ son disjuntos, entonces

$$
\ell\big((a,b)\cup(c,d)\big)
=
\ell((a,b))+\ell((c,d)).
$$

Si los intervalos se traslapan, debemos restar la longitud de la intersección. Por lo tanto,

$$
\ell\big((a,b)\cup(c,d)\big)
=
\ell((a,b))
+
\ell((c,d))
-
\ell\big((a,b)\cap(c,d)\big).
$$

La probabilidad comparte algunas propiedades con la longitud, pues también es una medida.

Sin embargo, existe una diferencia fundamental.

Una medida de probabilidad satisface

$$
P(\Omega)=1,
$$

mientras que la longitud de toda la recta real es

$$
\ell(\mathbb{R})=+\infty.
$$

Por ejemplo, si nuestro espacio muestral es la altura posible de una persona y tomamos

$$
\Omega=\mathbb{R},
$$

entonces necesariamente

$$
P(\mathbb{R})=1,
$$

aunque

$$
\ell(\mathbb{R})=+\infty.
$$

Por lo tanto, una probabilidad no es simplemente una longitud: es una medida normalizada de modo que todo el espacio muestral tenga probabilidad $1$.
