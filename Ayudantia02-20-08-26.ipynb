## Ejercicios

Tenemos $2$ dados justos.

### Pregunta 1

¿Cuántos resultados tiene el espacio muestral si distinguimos el primer dado del segundo?

Un primer intento podría ser tomar

$$
\Omega=\{2,3,4,\ldots,12\},
$$

pues estos son los posibles valores de la suma.

Sin embargo, este espacio muestral no es adecuado si queremos que todos los resultados elementales sean equiprobables.

Por ejemplo, la suma $2$ solamente puede obtenerse como

$$
(1,1),
$$

mientras que la suma $7$ puede obtenerse de varias maneras.

Por ello debemos distinguir el resultado del primer dado del resultado del segundo.

Así,

$$
\Omega
=
\{1,\ldots,6\}\times\{1,\ldots,6\}.
$$

Como para el primer dado tenemos $6$ posibilidades y, para cada una de ellas, el segundo dado también tiene $6$ posibilidades,

$$
|\Omega|=6\cdot6=36.
$$

Es decir,

- el primer factor $6$ representa las posibles caras del primer dado;
- el segundo factor $6$ representa las posibles caras del segundo dado.

---

### Pregunta 2

¿Cuál es la probabilidad de obtener suma $7$?

Definimos el evento

$$
E=\{\text{la suma de los dos dados es }7\}.
$$

Los resultados favorables son

$$
E
=
\{
(1,6),
(2,5),
(3,4),
(4,3),
(5,2),
(6,1)
\}.
$$

Por lo tanto,

$$
|E|=6.
$$

Hay $6$ resultados favorables porque existen exactamente $6$ pares ordenados cuya suma es $7$.

Como los $36$ resultados del espacio muestral son equiprobables,

$$
P(E)
=
\frac{|E|}{|\Omega|}
=
\frac{6}{36}
=
\frac16.
$$

## Ejemplo: contraseñas

Una contraseña consta de $2$ letras seguidas de $3$ dígitos.

Se permiten repeticiones.

¿Cuántas contraseñas existen?

Tenemos cinco posiciones:

$$
\boxed{\phantom{A}}
\boxed{\phantom{A}}
\boxed{\phantom{0}}
\boxed{\phantom{0}}
\boxed{\phantom{0}}
$$

Para la primera letra tenemos $26$ posibilidades.

Para la segunda letra también tenemos $26$ posibilidades, pues se permiten repeticiones.

Para cada uno de los tres dígitos tenemos $10$ posibilidades:

$$
0,1,\ldots,9.
$$

Por el principio multiplicativo,

$$
|\Omega|
=
26\cdot26\cdot10\cdot10\cdot10.
$$

Cada factor representa:

- $26$: elección de la primera letra;
- $26$: elección de la segunda letra;
- $10$: elección del primer dígito;
- $10$: elección del segundo dígito;
- $10$: elección del tercer dígito.

Por lo tanto,

$$
|\Omega|
=
26^2\,10^3.
$$

## Contraseñas con restricciones

Continuemos con contraseñas formadas por $2$ letras seguidas de $3$ dígitos.

### 1. Las dos letras deben ser distintas

¿Cuántas contraseñas existen si las $2$ letras tienen que ser diferentes?

Para la primera letra tenemos

$$
26
$$

posibilidades.

Una vez elegida la primera letra, para la segunda solamente quedan

$$
25
$$

letras disponibles, pues no podemos repetir la anterior.

Como los dígitos sí pueden repetirse, tenemos $10$ posibilidades en cada una de sus posiciones.

Por lo tanto,

$$
26\cdot25\cdot10\cdot10\cdot10.
$$

Es decir,

$$
26\cdot25\cdot10^3.
$$

Los factores representan:

- $26$: elección de la primera letra;
- $25$: elección de la segunda letra, distinta de la primera;
- $10$: elección del primer dígito;
- $10$: elección del segundo dígito;
- $10$: elección del tercer dígito.

---

### 2. Las letras y los dígitos deben ser distintos

¿Cuántas contraseñas existen si, además, los $3$ dígitos deben ser distintos?

Para las letras seguimos teniendo

$$
26\cdot25.
$$

Para el primer dígito tenemos

$$
10
$$

posibilidades.

Después de elegirlo, para el segundo dígito quedan

$$
9
$$

posibilidades.

Finalmente, para el tercer dígito quedan

$$
8
$$

posibilidades.

Así,

$$
26\cdot25\cdot10\cdot9\cdot8.
$$

Los factores representan:

- $26$: primera letra;
- $25$: segunda letra, distinta de la primera;
- $10$: primer dígito;
- $9$: segundo dígito, distinto del primero;
- $8$: tercer dígito, distinto de los dos anteriores.

---

# Continuando con el ejemplo de los dados

Se lanza un dado justo $4$ veces.

## ¿Cuál es la probabilidad de que los $4$ resultados sean distintos?

En cada lanzamiento tenemos $6$ posibles resultados.

Por lo tanto,

$$
|\Omega|=6^4.
$$

El exponente $4$ aparece porque hacemos $4$ lanzamientos y en cada uno tenemos $6$ posibilidades:

$$
6\cdot6\cdot6\cdot6=6^4.
$$

Ahora contemos los resultados favorables.

Para el primer lanzamiento podemos obtener cualquiera de las $6$ caras.

Por lo tanto, tenemos

$$
6
$$

posibilidades.

Para que el segundo resultado sea distinto del primero, solamente quedan

$$
5
$$

posibilidades.

Para el tercero quedan

$$
4
$$

posibilidades.

Para el cuarto quedan

$$
3
$$

posibilidades.

Así,

$$
|E|
=
6\cdot5\cdot4\cdot3.
$$

Por lo tanto,

$$
P(E)
=
\frac{6\cdot5\cdot4\cdot3}{6^4}.
$$

Simplificando,

$$
P(E)
=
\frac{5\cdot4\cdot3}{6^3}.
$$

---

## ¿Cuál es la probabilidad de que aparezca exactamente un $6$ y que los otros tres resultados sean distintos?

Primero elegimos en cuál de los $4$ lanzamientos aparecerá el $6$.

Tenemos

$$
4
$$

posibilidades.

Una vez fijada la posición del $6$, los otros tres lanzamientos deben ser distintos entre sí y distintos de $6$.

Para el primero de estos tres lugares tenemos $5$ posibilidades:

$$
1,2,3,4,5.
$$

Para el siguiente tenemos $4$ posibilidades.

Para el último tenemos $3$ posibilidades.

Por lo tanto,

$$
|E|
=
4(5\cdot4\cdot3).
$$

Aquí:

- $4$ representa la elección de la posición en la que aparece el $6$;
- $5$ representa el valor del primero de los otros resultados;
- $4$ representa el valor del siguiente, distinto de los anteriores;
- $3$ representa el valor del último.

Como

$$
|\Omega|=6^4,
$$

tenemos

$$
P(E)
=
\frac{4(5\cdot4\cdot3)}{6^4}.
$$

Así,

$$
P(E)
=
\frac{240}{1296}
=
\frac{5}{27}.
$$

## Ejemplo 4

En una clase hay $10$ alumnos.

Se escogen $3$ alumnos para pasar al pizarrón.

¿Cuántos grupos distintos de $3$ estudiantes podemos escoger?

Si comenzamos contando de manera ordenada, para escoger al primer estudiante tenemos

$$
10
$$

posibilidades.

Después de escogerlo, para el segundo tenemos

$$
9
$$

posibilidades.

Finalmente, para el tercero tenemos

$$
8
$$

posibilidades.

Así obtenemos

$$
10\cdot9\cdot8.
$$

Pero este conteo distingue el orden.

Por ejemplo, si los alumnos escogidos son

$$
A,\ B,\ C,
$$

entonces estamos contando como diferentes las selecciones

$$
(A,B,C),
$$

$$
(A,C,B),
$$

$$
(B,A,C),
$$

$$
(B,C,A),
$$

$$
(C,A,B),
$$

$$
(C,B,A).
$$

Sin embargo, todas ellas representan el mismo grupo

$$
\{A,B,C\}.
$$

Como cada grupo de $3$ estudiantes fue contado

$$
3!=6
$$

veces, debemos dividir entre $3!$.

Por lo tanto,

$$
\frac{10\cdot9\cdot8}{3!}.
$$

Equivalentemente,

$$
\frac{10!}{3!7!}.
$$

Así,

$$
\boxed{
\binom{10}{3}
=
\frac{10!}{3!7!}
}.
$$

El coeficiente

$$
\binom{10}{3}
$$

representa exactamente:

> las formas de escoger $3$ estudiantes de un grupo de $10$, sin importar el orden en que son escogidos.

# Ejercicio: distinguir cuándo importa el orden

Tenemos una baraja con $52$ cartas.

Se reparten $5$ cartas.

¿Cuál es la probabilidad de obtener exactamente $2$ ases?

Primero debemos decidir cómo describir nuestro espacio muestral.

No necesitamos conocer el orden completo de las $52$ cartas.

Tampoco necesitamos distinguir el orden en que aparecen las $5$ cartas de nuestra mano.

Solamente nos importa **qué conjunto de $5$ cartas recibimos**.

Por lo tanto,

$$
|\Omega|
=
\binom{52}{5}.
$$

El número

$$
\binom{52}{5}
$$

representa:

> las formas de escoger las $5$ cartas que forman nuestra mano entre las $52$ cartas de la baraja, sin importar el orden.

Ahora contemos las manos que contienen exactamente $2$ ases.

Continuando con el ejercicio anterior:

Queremos recibir exactamente $2$ ases en una mano de $5$ cartas.

Tenemos $4$ ases en toda la baraja.

Para escoger cuáles $2$ de esos $4$ ases recibiremos tenemos

$$
\binom{4}{2}
$$

posibilidades.

Es decir,

$$
\binom{4}{2}
$$

representa:

> las formas de escoger $2$ de los $4$ ases de la baraja.

Una vez escogidos los $2$ ases, todavía necesitamos completar nuestra mano con $3$ cartas más.

Como no queremos recibir más ases, esas $3$ cartas deben elegirse entre las

$$
52-4=48
$$

cartas que no son ases.

Por lo tanto, tenemos

$$
\binom{48}{3}
$$

formas de elegir esas cartas.

El número

$$
\binom{48}{3}
$$

representa:

> las formas de escoger $3$ cartas entre las $48$ cartas que no son ases.

Por el principio multiplicativo, el número de manos favorables es

$$
|E|
=
\binom{4}{2}
\binom{48}{3}.
$$

¿Por qué multiplicamos?

Porque para **cada** manera de escoger los $2$ ases podemos escoger las otras $3$ cartas de

$$
\binom{48}{3}
$$

formas.

Por lo tanto,

$$
P(E)
=
\frac{|E|}{|\Omega|}
=
\frac{
\binom{4}{2}\binom{48}{3}
}{
\binom{52}{5}
}.
$$

Así,

$$
\boxed{
P(\text{exactamente $2$ ases})
=
\frac{
\binom{4}{2}\binom{48}{3}
}{
\binom{52}{5}
}
}.
$$

## Una idea importante para problemas de conteo

Muchas veces, contar los elementos de un evento consiste en **descomponer el problema en varias elecciones sucesivas**.

Por ejemplo, en el problema anterior hicimos dos elecciones:

1. Escoger cuáles $2$ ases recibimos:

   $$
   \binom{4}{2}.
   $$

2. Escoger cuáles $3$ cartas no-ases completan la mano:

   $$
   \binom{48}{3}.
   $$

Como cada elección del primer tipo puede combinarse con cada elección del segundo tipo, utilizamos el principio multiplicativo:

$$
\binom{4}{2}\binom{48}{3}.
$$

## Ejercicio 1

¿Cuántos números de $4$ cifras hay?

Un número de $4$ cifras tiene la forma

$$
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}.
$$

Para la primera cifra tenemos

$$
9
$$

posibilidades, pues puede ser cualquiera de

$$
1,2,\ldots,9.
$$

No podemos utilizar $0$ en la primera posición, ya que, por ejemplo, $0123$ no es un número de $4$ cifras.

Para cada una de las otras tres posiciones tenemos

$$
10
$$

posibilidades:

$$
0,1,\ldots,9.
$$

Por el principio multiplicativo,

$$
9\cdot10\cdot10\cdot10
=
9\cdot10^3.
$$

Aquí:

- $9$ representa las posibilidades para la primera cifra;
- el primer $10$ representa las posibilidades para la segunda cifra;
- el segundo $10$ representa las posibilidades para la tercera cifra;
- el tercer $10$ representa las posibilidades para la cuarta cifra.

Por lo tanto,

$$
\boxed{9\cdot10^3=9000}.
$$

Así, existen

$$
\boxed{9000}
$$

números de $4$ cifras.

---

## Ejercicio 2

¿Cuántos números de $4$ cifras tienen todos sus dígitos distintos?

Nuevamente tenemos cuatro posiciones:

$$
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}.
$$

Para la primera cifra tenemos

$$
9
$$

posibilidades, pues puede ser cualquiera de

$$
1,2,\ldots,9.
$$

Después de elegir la primera cifra, para la segunda tenemos

$$
9
$$

posibilidades.

¿Por qué nuevamente $9$?

Porque originalmente hay $10$ dígitos,

$$
0,1,\ldots,9,
$$

pero uno de ellos ya fue utilizado en la primera posición.

Además, ahora sí podemos utilizar el $0$.

Para la tercera cifra quedan

$$
8
$$

posibilidades, pues ya hemos utilizado dos dígitos.

Finalmente, para la cuarta cifra quedan

$$
7
$$

posibilidades.

Por lo tanto,

$$
9\cdot9\cdot8\cdot7.
$$

Cada factor representa:

- $9$: elección de la primera cifra;
- $9$: elección de la segunda cifra, distinta de la primera;
- $8$: elección de la tercera cifra, distinta de las dos anteriores;
- $7$: elección de la cuarta cifra, distinta de las tres anteriores.

Así,

$$
9\cdot9\cdot8\cdot7
=
4536.
$$

Por lo tanto, existen

$$
\boxed{4536}
$$

números de $4$ cifras con dígitos distintos.

## Ejercicio 3

¿Cuántos números pares de $5$ cifras hay sin repetir dígitos?

Un número de $5$ cifras tiene la forma

$$
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}.
$$

Para que el número sea par, su última cifra debe pertenecer a

$$
\{0,2,4,6,8\}.
$$

Sin embargo, el $0$ requiere un tratamiento especial porque no puede aparecer en la primera posición.

Por ello, conviene separar el problema en dos casos:

1. números pares que terminan en $0$;
2. números pares que terminan en $2,4,6$ u $8$.

---

### Caso 1: el número termina en $0$

Tenemos

$$
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{\phantom{0}}\;
\boxed{0}.
$$

Como el último dígito ya es $0$, para la primera cifra podemos utilizar cualquiera de

$$
1,2,\ldots,9.
$$

Por lo tanto, tenemos

$$
9
$$

posibilidades para la primera cifra.

Después de escogerla, ya hemos utilizado dos dígitos:

- el $0$;
- la primera cifra.

Por lo tanto, para la segunda posición quedan

$$
8
$$

posibilidades.

Para la tercera posición quedan

$$
7
$$

posibilidades.

Para la cuarta posición quedan

$$
6
$$

posibilidades.

Así, el número de posibilidades es

$$
9\cdot8\cdot7\cdot6.
$$

Aquí:

- $9$: elección de la primera cifra;
- $8$: elección de la segunda cifra;
- $7$: elección de la tercera cifra;
- $6$: elección de la cuarta cifra;
- la última cifra ya está fijada como $0$.

Por lo tanto,

$$
9\cdot8\cdot7\cdot6
=
3024.
$$

---

### Caso 2: el número no termina en $0$

En este caso, la última cifra debe ser una de

$$
2,4,6,8.
$$

Por lo tanto, tenemos

$$
4
$$

posibilidades para la última cifra.

Ahora debemos escoger la primera cifra.

La primera cifra no puede ser $0$, y tampoco puede ser el dígito par que acabamos de colocar al final.

De los $9$ dígitos

$$
1,2,\ldots,9,
$$

uno ya fue utilizado en la última posición.

Por lo tanto, quedan

$$
8
$$

posibilidades para la primera cifra.

Después de escoger la primera y la última cifra, hemos utilizado $2$ de los $10$ dígitos.

Así, para la segunda posición quedan

$$
8
$$

posibilidades.

Para la tercera posición quedan

$$
7
$$

posibilidades.

Para la cuarta posición quedan

$$
6
$$

posibilidades.

Por lo tanto,

$$
4\cdot8\cdot8\cdot7\cdot6.
$$

Cada factor representa:

- $4$: elección del último dígito entre $2,4,6,8$;
- $8$: elección de la primera cifra, distinta de $0$ y de la cifra final;
- $8$: elección de la segunda cifra;
- $7$: elección de la tercera cifra;
- $6$: elección de la cuarta cifra.

Así,

$$
4\cdot8\cdot8\cdot7\cdot6
=
10752.
$$

---

### Sumamos ambos casos

Los dos casos son mutuamente excluyentes:

- un número termina en $0$;
- o termina en $2,4,6$ u $8$.

Por el principio aditivo,

$$
3024+10752=13776.
$$

Por lo tanto, existen

$$
\boxed{13776}
$$

números pares de $5$ cifras sin repetir dígitos.
