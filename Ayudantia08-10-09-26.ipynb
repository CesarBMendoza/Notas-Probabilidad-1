# Ejemplo: tarjetas American Express y VISA

Un establecimiento minorista acepta tarjetas **American Express** o **VISA**.

Un total del $24\%$ de sus clientes lleva una tarjeta American Express, el $61\%$ lleva una tarjeta VISA y el $11\%$ lleva ambas.

¿Cuál es el porcentaje de clientes que lleva una tarjeta de crédito que el establecimiento acepta?

Definimos los eventos

$$
A=\{\text{el cliente lleva una tarjeta American Express}\},
$$

$$
B=\{\text{el cliente lleva una tarjeta VISA}\}.
$$

Sabemos que

$$
P(A)=0.24,
$$

$$
P(B)=0.61,
$$

y

$$
P(A\cap B)=0.11.
$$

Queremos calcular la probabilidad de que un cliente lleve **al menos una** de las dos tarjetas, es decir,

$$
P(A\cup B).
$$

Por la fórmula de inclusión-exclusión para dos eventos,

$$
P(A\cup B)
=
P(A)+P(B)-P(A\cap B).
$$

Sustituyendo,

$$
P(A\cup B)
=
0.24+0.61-0.11.
$$

Por lo tanto,

$$
\boxed{
P(A\cup B)=0.74
}.
$$

Así, el

$$
\boxed{74\%}
$$

de los clientes lleva al menos una tarjeta aceptada por el establecimiento.

> **¿Por qué restamos $P(A\cap B)$?**
>
> Al sumar $P(A)+P(B)$, los clientes que poseen ambas tarjetas se cuentan dos veces.
> Por eso debemos restarlos una vez.

# Ejemplo: estudiantes con anillos y collares

En una escuela, el $60\%$ de los estudiantes no lleva anillos ni collares.

Además,

- el $20\%$ lleva un anillo;
- el $30\%$ lleva un collar.

Si se elige un estudiante al azar, calcule la probabilidad de que el estudiante:

1. lleve un anillo o un collar;
2. lleve un anillo y un collar.

Definimos los eventos

$$
A=\{\text{el estudiante lleva un anillo}\},
$$

$$
B=\{\text{el estudiante lleva un collar}\}.
$$

Sabemos que

$$
P(A)=0.20,
$$

$$
P(B)=0.30.
$$

Además, sabemos que el $60\%$ no lleva ninguno de los dos.

El evento

$$
A^c\cap B^c
$$

representa que el estudiante no lleva ni anillo ni collar.

Por lo tanto,

$$
P(A^c\cap B^c)=0.60.
$$

Por las leyes de De Morgan,

$$
A^c\cap B^c=(A\cup B)^c.
$$

Entonces

$$
P((A\cup B)^c)=0.60.
$$

Como

$$
P(A\cup B)+P((A\cup B)^c)=1,
$$

tenemos

$$
P(A\cup B)
=
1-0.60.
$$

Así,

$$
\boxed{
P(A\cup B)=0.40
}.
$$

Por lo tanto, la probabilidad de que un estudiante lleve un anillo o un collar es

$$
\boxed{0.40}.
$$

---

## (b) Probabilidad de que lleve un anillo y un collar

Utilizamos nuevamente

$$
P(A\cup B)
=
P(A)+P(B)-P(A\cap B).
$$

Despejando,

$$
P(A\cap B)
=
P(A)+P(B)-P(A\cup B).
$$

Por lo tanto,

$$
P(A\cap B)
=
0.20+0.30-0.40.
$$

Así,

$$
\boxed{
P(A\cap B)=0.10
}.
$$

Es decir, el

$$
\boxed{10\%}
$$

de los estudiantes lleva tanto un anillo como un collar.

# Ejemplo: clases de idiomas

En una escuela primaria hay $100$ niños.

La escuela ofrece tres idiomas:

- inglés;
- francés;
- alemán.

Sabemos que:

- $28$ estudiantes toman inglés;
- $26$ estudiantes toman francés;
- $16$ estudiantes toman alemán;
- $12$ toman inglés y francés;
- $4$ toman inglés y alemán;
- $6$ toman francés y alemán;
- $2$ toman los tres idiomas.

Calcule:

1. la probabilidad de que un estudiante elegido al azar no esté en ninguna clase de idiomas;
2. la probabilidad de que esté exactamente en una clase de idiomas;
3. si se eligen $2$ estudiantes al azar, la probabilidad de que al menos uno de ellos esté tomando una clase de idiomas.

---

## Definición de eventos

Definimos

$$
I=\{\text{el estudiante toma inglés}\},
$$

$$
F=\{\text{el estudiante toma francés}\},
$$

$$
A=\{\text{el estudiante toma alemán}\}.
$$

Tenemos

$$
|I|=28,
\qquad
|F|=26,
\qquad
|A|=16,
$$

y

$$
|I\cap F|=12,
$$

$$
|I\cap A|=4,
$$

$$
|F\cap A|=6,
$$

$$
|I\cap F\cap A|=2.
$$

---

# (a) Probabilidad de que no tome ningún idioma

Primero contamos cuántos estudiantes toman **al menos un idioma**.

Por inclusión-exclusión,

$$
|I\cup F\cup A|
=
|I|+|F|+|A|
-|I\cap F|
-|I\cap A|
-|F\cap A|
+|I\cap F\cap A|.
$$

Sustituyendo,

$$
|I\cup F\cup A|
=
28+26+16-12-4-6+2.
$$

Por lo tanto,

$$
|I\cup F\cup A|=50.
$$

Es decir, $50$ estudiantes toman al menos un idioma.

Como hay $100$ estudiantes en total,

$$
100-50=50
$$

no toman ningún idioma.

Así,

$$
P(\text{ningún idioma})
=
\frac{50}{100}.
$$

Por lo tanto,

$$
\boxed{
P(\text{ningún idioma})=\frac12
}.
$$

---

# (b) Probabilidad de que tome exactamente un idioma

Tenemos que contar los estudiantes que toman solamente inglés, solamente francés o solamente alemán.

## Solamente inglés

De los $28$ estudiantes que toman inglés, restamos los que también toman francés y los que también toman alemán.

Sin embargo, los estudiantes que toman los tres idiomas fueron restados dos veces, así que debemos sumarlos una vez.

Por lo tanto,

$$
|I\text{ solamente}|
=
|I|-|I\cap F|-|I\cap A|+|I\cap F\cap A|.
$$

Entonces

$$
|I\text{ solamente}|
=
28-12-4+2.
$$

Así,

$$
|I\text{ solamente}|=14.
$$

---

## Solamente francés

Análogamente,

$$
|F\text{ solamente}|
=
|F|-|F\cap I|-|F\cap A|+|I\cap F\cap A|.
$$

Por lo tanto,

$$
|F\text{ solamente}|
=
26-12-6+2.
$$

Así,

$$
|F\text{ solamente}|=10.
$$

---

## Solamente alemán

De la misma forma,

$$
|A\text{ solamente}|
=
|A|-|A\cap I|-|A\cap F|+|I\cap F\cap A|.
$$

Entonces

$$
|A\text{ solamente}|
=
16-4-6+2.
$$

Por lo tanto,

$$
|A\text{ solamente}|=8.
$$

---

Sumamos los tres casos:

$$
14+10+8=32.
$$

Por lo tanto,

$$
P(\text{exactamente un idioma})
=
\frac{32}{100}.
$$

Así,

$$
\boxed{
P(\text{exactamente un idioma})=0.32
}.
$$

---

# (c) Al elegir dos estudiantes, al menos uno toma un idioma

Definimos

$$
E=\{\text{al menos uno de los dos estudiantes toma un idioma}\}.
$$

Su complemento es

$$
E^c
=
\{\text{ninguno de los dos estudiantes toma un idioma}\}.
$$

Sabemos que hay $50$ estudiantes que no toman ningún idioma.

El número total de maneras de escoger $2$ estudiantes entre los $100$ es

$$
\binom{100}{2}.
$$

El número de maneras de escoger $2$ estudiantes entre los $50$ que no toman ningún idioma es

$$
\binom{50}{2}.
$$

Por lo tanto,

$$
P(E^c)
=
\frac{\binom{50}{2}}{\binom{100}{2}}.
$$

Así,

$$
P(E)
=
1-
\frac{\binom{50}{2}}{\binom{100}{2}}.
$$

Calculando,

$$
\binom{50}{2}=1225,
$$

y

$$
\binom{100}{2}=4950.
$$

Entonces

$$
P(E)
=
1-\frac{1225}{4950}.
$$

Por lo tanto,

$$
\boxed{
P(E)
=
\frac{149}{198}
\approx0.7525
}.
$$

Es decir, la probabilidad es aproximadamente

$$
\boxed{75.25\%}.
$$

# Ejemplo: lectores de periódicos

Una ciudad con una población de $100\,000$ habitantes tiene tres periódicos:

$$
I,\qquad II,\qquad III.
$$

La proporción de habitantes que lee cada periódico o combinación de periódicos es:

| Evento | Porcentaje |
|---|---:|
| Lee $I$ | $10\%$ |
| Lee $II$ | $30\%$ |
| Lee $III$ | $5\%$ |
| Lee $I$ y $II$ | $8\%$ |
| Lee $I$ y $III$ | $2\%$ |
| Lee $II$ y $III$ | $4\%$ |
| Lee $I$, $II$ y $III$ | $1\%$ |

Por ejemplo,

$$
100\,000(0.08)=8\,000
$$

personas leen los periódicos $I$ y $II$.

Definimos los eventos

$$
I=\{\text{lee el periódico }I\},
$$

$$
II=\{\text{lee el periódico }II\},
$$

$$
III=\{\text{lee el periódico }III\}.
$$

---

# (a) ¿Cuántas personas leen solamente un periódico?

Primero calculamos el porcentaje que lee **solamente $I$**:

$$
P(I\text{ solamente})
=
P(I)-P(I\cap II)-P(I\cap III)
+P(I\cap II\cap III).
$$

Sustituyendo,

$$
P(I\text{ solamente})
=
0.10-0.08-0.02+0.01.
$$

Entonces

$$
P(I\text{ solamente})=0.01.
$$

Es decir,

$$
1\%
$$

lee solamente el periódico $I$.

---

## Solamente $II$

Análogamente,

$$
P(II\text{ solamente})
=
P(II)-P(I\cap II)-P(II\cap III)
+P(I\cap II\cap III).
$$

Entonces

$$
P(II\text{ solamente})
=
0.30-0.08-0.04+0.01.
$$

Por lo tanto,

$$
P(II\text{ solamente})=0.19.
$$

Es decir,

$$
19\%
$$

lee solamente el periódico $II$.

---

## Solamente $III$

Tenemos

$$
P(III\text{ solamente})
=
P(III)-P(I\cap III)-P(II\cap III)
+P(I\cap II\cap III).
$$

Así,

$$
P(III\text{ solamente})
=
0.05-0.02-0.04+0.01.
$$

Por lo tanto,

$$
P(III\text{ solamente})=0.
$$

Es decir, según los datos del problema, ninguna persona lee **únicamente** el periódico $III$.

---

Sumando,

$$
1\%+19\%+0\%=20\%.
$$

Como hay $100\,000$ habitantes,

$$
0.20(100\,000)=20\,000.
$$

Por lo tanto,

$$
\boxed{20\,000}
$$

personas leen solamente un periódico.

---

# (b) ¿Cuántas personas leen al menos dos periódicos?

Primero calculamos quienes leen exactamente dos.

## Solamente $I$ y $II$

La intersección $I\cap II$ también contiene a quienes leen los tres periódicos.

Por lo tanto,

$$
P(I\cap II\text{ solamente})
=
P(I\cap II)-P(I\cap II\cap III).
$$

Entonces

$$
P(I\cap II\text{ solamente})
=
0.08-0.01
=
0.07.
$$

---

## Solamente $I$ y $III$

$$
P(I\cap III\text{ solamente})
=
0.02-0.01
=
0.01.
$$

---

## Solamente $II$ y $III$

$$
P(II\cap III\text{ solamente})
=
0.04-0.01
=
0.03.
$$

---

Además, el

$$
1\%
$$

lee los tres periódicos.

Por lo tanto,

$$
P(\text{al menos dos})
=
0.07+0.01+0.03+0.01.
$$

Así,

$$
P(\text{al menos dos})=0.12.
$$

Como la población es de $100\,000$ personas,

$$
0.12(100\,000)=12\,000.
$$

Por lo tanto,

$$
\boxed{12\,000}
$$

personas leen al menos dos periódicos.

---

# (c) Periódicos matutinos y vespertinos

Supongamos que los periódicos $I$ y $III$ son matutinos y el periódico $II$ es vespertino.

Queremos calcular cuántas personas leen al menos un periódico matutino y un periódico vespertino.

El evento es

$$
II\cap(I\cup III).
$$

Por distributividad,

$$
II\cap(I\cup III)
=
(I\cap II)\cup(II\cap III).
$$

Entonces, por inclusión-exclusión,

$$
P(II\cap(I\cup III))
=
P(I\cap II)
+
P(II\cap III)
-
P(I\cap II\cap III).
$$

Sustituyendo,

$$
P(II\cap(I\cup III))
=
0.08+0.04-0.01.
$$

Así,

$$
P(II\cap(I\cup III))=0.11.
$$

Por lo tanto,

$$
0.11(100\,000)=11\,000.
$$

Así,

$$
\boxed{11\,000}
$$

personas leen al menos un periódico matutino y uno vespertino.

---

# (d) ¿Cuántas personas no leen ningún periódico?

Primero calculamos

$$
P(I\cup II\cup III).
$$

Por inclusión-exclusión,

$$
P(I\cup II\cup III)
=
P(I)+P(II)+P(III)
-P(I\cap II)
-P(I\cap III)
-P(II\cap III)
+P(I\cap II\cap III).
$$

Sustituyendo,

$$
P(I\cup II\cup III)
=
0.10+0.30+0.05
-0.08-0.02-0.04
+0.01.
$$

Por lo tanto,

$$
P(I\cup II\cup III)=0.32.
$$

El evento de no leer ningún periódico es

$$
I^c\cap II^c\cap III^c
=
(I\cup II\cup III)^c.
$$

Así,

$$
P(I^c\cap II^c\cap III^c)
=
1-0.32.
$$

Por lo tanto,

$$
P(\text{ningún periódico})=0.68.
$$

Entonces

$$
0.68(100\,000)=68\,000.
$$

Así,

$$
\boxed{68\,000}
$$

personas no leen ningún periódico.

---

# (e) ¿Cuántas personas leen solamente un periódico matutino y uno vespertino?

Recordemos que:

- $I$ y $III$ son matutinos;
- $II$ es vespertino.

Queremos personas que lean $II$ y **exactamente uno** de los periódicos matutinos.

Tenemos dos casos disjuntos:

$$
I\cap II\cap III^c,
$$

o

$$
I^c\cap II\cap III.
$$

---

## Lee $I$ y $II$, pero no $III$

Tenemos

$$
P(I\cap II\cap III^c)
=
P(I\cap II)-P(I\cap II\cap III).
$$

Por lo tanto,

$$
P(I\cap II\cap III^c)
=
0.08-0.01
=
0.07.
$$

---

## Lee $II$ y $III$, pero no $I$

Análogamente,

$$
P(I^c\cap II\cap III)
=
P(II\cap III)-P(I\cap II\cap III).
$$

Por lo tanto,

$$
P(I^c\cap II\cap III)
=
0.04-0.01
=
0.03.
$$

Como los dos casos son disjuntos,

$$
P(\text{exactamente un matutino y un vespertino})
=
0.07+0.03.
$$

Así,

$$
P(\text{exactamente un matutino y un vespertino})
=
0.10.
$$

Por lo tanto,

$$
0.10(100\,000)=10\,000.
$$

Así,

$$
\boxed{10\,000}
$$

personas leen exactamente un periódico matutino y el periódico vespertino.

# Ejemplo: perros sentados en una fila

En una fiesta de perros se sientan en una fila

- $3$ perros chicos;
- $4$ perros grandes.

Suponemos que los $7$ perros son distintos y que todos los ordenamientos son igualmente probables.

Queremos calcular:

> ¿Cuál es la probabilidad de que no todos los perros chicos estén juntos y tampoco todos los perros grandes estén juntos?

Definimos

$$
A=\{\text{los $3$ perros chicos están juntos}\},
$$

$$
B=\{\text{los $4$ perros grandes están juntos}\}.
$$

Queremos que no ocurra $A$ y tampoco ocurra $B$, es decir,

$$
A^c\cap B^c.
$$

Por las leyes de De Morgan,

$$
A^c\cap B^c
=
(A\cup B)^c.
$$

Entonces

$$
P(A^c\cap B^c)
=
1-P(A\cup B).
$$

Utilizando inclusión-exclusión,

$$
P(A\cup B)
=
P(A)+P(B)-P(A\cap B).
$$

Por lo tanto,

$$
P(A^c\cap B^c)
=
1-P(A)-P(B)+P(A\cap B).
$$

---

## Cálculo de $P(A)$

Queremos que los $3$ perros chicos estén juntos.

Tratamos a los $3$ perros chicos como un solo bloque.

Entonces tenemos:

- $1$ bloque de perros chicos;
- $4$ perros grandes.

En total tenemos

$$
5
$$

objetos que podemos ordenar de

$$
5!
$$

formas.

Dentro del bloque, los $3$ perros chicos pueden ordenarse de

$$
3!
$$

formas.

Por lo tanto, el número de ordenamientos favorables es

$$
5!3!.
$$

El número total de formas de ordenar a los $7$ perros es

$$
7!.
$$

Así,

$$
P(A)
=
\frac{5!3!}{7!}.
$$

---

## Cálculo de $P(B)$

Ahora queremos que los $4$ perros grandes estén juntos.

Tratamos a los $4$ perros grandes como un bloque.

Tenemos entonces:

- $1$ bloque de perros grandes;
- $3$ perros chicos.

En total hay

$$
4
$$

objetos.

Estos pueden ordenarse de

$$
4!
$$

formas.

Además, dentro del bloque, los $4$ perros grandes pueden ordenarse de

$$
4!
$$

formas.

Por lo tanto,

$$
P(B)
=
\frac{4!4!}{7!}.
$$

---

## Cálculo de $P(A\cap B)$

Ahora queremos que los perros chicos estén juntos **y** que los perros grandes estén juntos.

Tenemos dos bloques:

$$
[\text{chicos}]
\qquad
[\text{grandes}].
$$

Estos dos bloques pueden colocarse de

$$
2!
$$

formas:

$$
[\text{chicos}][\text{grandes}]
$$

o

$$
[\text{grandes}][\text{chicos}].
$$

Dentro del bloque de perros chicos tenemos

$$
3!
$$

ordenamientos.

Dentro del bloque de perros grandes tenemos

$$
4!
$$

ordenamientos.

Por lo tanto,

$$
P(A\cap B)
=
\frac{2!3!4!}{7!}.
$$

---

# Sustitución

Entonces

$$
P(A^c\cap B^c)
=
1
-
\frac{5!3!}{7!}
-
\frac{4!4!}{7!}
+
\frac{2!3!4!}{7!}.
$$

Como $2!=2$,

$$
P(A^c\cap B^c)
=
1-
\frac{5!3!+4!4!-2(3!4!)}{7!}.
$$

Calculamos

$$
5!3!=720,
$$

$$
4!4!=576,
$$

y

$$
2(3!4!)=288.
$$

Así,

$$
P(A^c\cap B^c)
=
1-\frac{720+576-288}{5040}.
$$

Entonces

$$
P(A^c\cap B^c)
=
1-\frac{1008}{5040}.
$$

Como

$$
\frac{1008}{5040}
=
\frac15,
$$

obtenemos

$$
\boxed{
P(A^c\cap B^c)
=
1-\frac15
=
\frac45
}.
$$

Por lo tanto,

$$
\boxed{
P(\text{ni todos los chicos juntos ni todos los grandes juntos})
=
\frac45
}.
$$
