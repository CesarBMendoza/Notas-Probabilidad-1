## V.a. Gamma

Sea

$$
X \sim \Gamma(\alpha,\beta),
$$

con $\alpha,\beta>0$.

Su función de densidad está dada por

$$
f_X(x)
=
\frac{\beta^\alpha}{\Gamma(\alpha)}
x^{\alpha-1}e^{-\beta x}
\mathbb{1}_{[0,\infty)}(x).
$$

> **Nota:** $\Gamma(\alpha)$ es tal que, para $\alpha\in\mathbb{N}$,

$$
\Gamma(\alpha)=(\alpha-1)!.
$$

Para $\alpha=1$,

$$
f_X(x)
=
\beta e^{-\beta x}
\mathbb{1}_{[0,\infty)}(x),
$$

que es la distribución exponencial.

Por lo tanto, la v.a. Gamma es una generalización de la distribución exponencial.

Su interpretación sería:

> $X$ = tiempo necesario para que pasen $\alpha$ cosas cuando, en promedio, pasa una cada $\frac{1}{\beta}$ unidades de tiempo.

Si

$$
X\sim\Gamma(\alpha,\beta),
\qquad
\alpha\in\mathbb{N},
$$

entonces $X$ es la suma de $\alpha$ variables aleatorias exponenciales con parámetro $\beta$.

## V.a. Normal (Gaussiana)

Sea

$$
X\sim\mathcal{N}(\mu,\sigma),
$$

con $\mu\in\mathbb{R}$ y $\sigma>0$.

$X$ es la v.a. con máxima incertidumbre si sólo sabemos que su promedio es $\mu$ y su varianza es $\sigma^2$.

Su función de densidad está dada por

$$
f_X(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

### V.a. Normal estándar

Para $\mu=0$ y $\sigma=1$,

$$
f_X(x)
=
\frac{1}{\sqrt{2\pi}}
\exp\left(
-\frac{x^2}{2}
\right).
$$

### Teorema

Si

$$
X\sim\mathcal{N}(\mu,\sigma),
$$

entonces

$$
Z=\frac{X-\mu}{\sigma}
\sim\mathcal{N}(0,1).
$$

Además,

$$
F_Z(z)
=
P(Z\leq z)
=
P\left(
\frac{X-\mu}{\sigma}\leq z
\right).
$$

Como $\sigma>0$,

$$
P\left(
\frac{X-\mu}{\sigma}\leq z
\right)
=
P(X\leq z\sigma+\mu).
$$

Por lo tanto,

$$
F_Z(z)
=
F_X(z\sigma+\mu).
$$

Además,

$$
\frac{d}{dz}F_X(\sigma z+\mu)
=
f_X(\sigma z+\mu)\sigma.
$$

Es decir,

$$
f_Z(z)
=
F_Z'(z)
=
\frac{d}{dz}F_X(\sigma z+\mu).
$$

Continuando con el resultado anterior,

$$
f_Z(z)
=
\frac{d}{dz}F_X(\sigma z+\mu)
=
F_X'(\sigma z+\mu)\sigma.
$$

Por lo tanto,

$$
f_Z(z)
=
f_X(\sigma z+\mu)\sigma.
$$

Sustituyendo la densidad de $X$,

$$
f_Z(z)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{(\sigma z+\mu-\mu)^2}{2\sigma^2}
\right)\sigma.
$$

Entonces,

$$
f_Z(z)
=
\frac{1}{\sqrt{2\pi}}
\exp\left(
-\frac{z^2}{2}
\right).
$$

Así,

$$
f_Z(z)
=
\frac{1}{\sqrt{2\pi}}
\exp\left(
-\frac{z^2}{2}
\right).
$$

### Teorema

Si

$$
Z\sim\mathcal{N}(0,1),
$$

entonces

$$
X=\sigma Z+\mu
\sim\mathcal{N}(\mu,\sigma).
$$

Ya que

$$
F_X(x)
=
P(X\leq x)
=
P(\sigma Z+\mu\leq x).
$$

Como $\sigma>0$,

$$
F_X(x)
=
P\left(
Z\leq\frac{x-\mu}{\sigma}
\right).
$$

Por lo tanto,

$$
F_X(x)
=
F_Z\left(
\frac{x-\mu}{\sigma}
\right).
$$

Además,

$$
f_X(x)
=
\frac{dF_X(x)}{dx}
=
\frac{d}{dx}
F_Z\left(
\frac{x-\mu}{\sigma}
\right).
$$

Aplicando la regla de la cadena,

$$
f_X(x)
=
F_Z'\left(
\frac{x-\mu}{\sigma}
\right)
\frac{1}{\sigma}.
$$

Por lo tanto,

$$
f_X(x)
=
\frac{1}{\sigma}
f_Z\left(
\frac{x-\mu}{\sigma}
\right).
$$

## V.a. Weibull

Sea

$$
X\sim\operatorname{Weibull}(\nu,\alpha,\beta),
$$

con $\nu,\alpha,\beta>0$.

$X$ representa la vida de un objeto que tiene muchas partes y falla cuando una de sus partes falla.

Su función de densidad está dada por

$$
f_X(x)
=
\frac{\beta}{\alpha}
\left(
\frac{x-\nu}{\alpha}
\right)^{\beta-1}
\exp\left[
-\left(
\frac{x-\nu}{\alpha}
\right)^\beta
\right]
\mathbb{1}_{[\nu,\infty)}(x).
$$

Y su función de distribución es

$$
F_X(x)
=
\begin{cases}
0,
& x<\nu,\\[6pt]
1-
\exp\left[
-\left(
\frac{x-\nu}{\alpha}
\right)^\beta
\right],
& x\geq\nu.
\end{cases}
$$

## V.a. Cauchy (Lorentziana)

Sea

$$
X\sim\operatorname{Cauchy}(x_0,\delta),
$$

con $x_0\in\mathbb{R}$ y $\delta>0$.

Su función de densidad está dada por

$$
f_X(x)
=
\frac{1}
{\pi\delta
\left[
1+
\left(
\frac{x-x_0}{\delta}
\right)^2
\right]}.
$$

Además,

$$
F_X(x)
=
\frac{1}{\pi}
\tan^{-1}\left(
\frac{x-x_0}{\delta}
\right)
+
\frac{1}{2}.
$$

## V.a. Beta

Sea

$$
X\sim\operatorname{Beta}(\alpha,\beta),
$$

con $\alpha,\beta>0$.

$X$ es una probabilidad aleatoria.

En general, modela fenómenos en intervalos.

Su función de densidad está dada por

$$
f_X(x)
=
\frac{\Gamma(\alpha+\beta)}
{\Gamma(\alpha)\Gamma(\beta)}
x^{\alpha-1}
(1-x)^{\beta-1}
\mathbb{1}_{[0,1]}(x).
$$
