# Regla de multiplicación

Sean $E_1,\ldots,E_n$ eventos.

La regla de multiplicación nos dice que

$$
P\left(\bigcap_{i=1}^n E_i\right)
=
P(E_1)
P(E_2\mid E_1)
P(E_3\mid E_1\cap E_2)
\cdots
P\left(E_n\mid \bigcap_{i=1}^{n-1}E_i\right).
$$

Es decir,

$$
P(E_1\cap\cdots\cap E_n)
=
P(E_1)
P(E_2\mid E_1)
\cdots
P(E_n\mid E_1\cap\cdots\cap E_{n-1}).
$$

Esta fórmula nos permite calcular la probabilidad de que ocurran varios eventos consecutivamente, tomando en cuenta la información proporcionada por los eventos anteriores.

---

# Ejemplo: el problema de los sombreros

Supongamos que hay $N$ personas con sombrero en una fiesta.

Las $N$ personas lanzan sus sombreros al centro de la habitación y, posteriormente, cada persona selecciona al azar uno de los sombreros restantes.

Queremos calcular:

> ¿Cuál es la probabilidad de que ninguna persona seleccione su propio sombrero?

Definimos, para cada

$$
i\in\{1,\ldots,N\},
$$

el evento

$$
E_i
=
\{\text{la persona }i\text{ selecciona su propio sombrero}\}.
$$

Entonces

$$
\bigcup_{i=1}^N E_i
$$

es el evento

$$
\{\text{al menos una persona selecciona su propio sombrero}\}.
$$

Por lo tanto, el evento que nos interesa es

$$
\bigcap_{i=1}^N E_i^c,
$$

es decir,

$$
\{\text{ninguna persona selecciona su propio sombrero}\}.
$$

Por las leyes de De Morgan,

$$
\bigcap_{i=1}^N E_i^c
=
\left(
\bigcup_{i=1}^N E_i
\right)^c.
$$

Así,

$$
P\left(\bigcap_{i=1}^N E_i^c\right)
=
1-
P\left(\bigcup_{i=1}^N E_i\right).
$$

Por lo tanto, comenzaremos calculando la probabilidad de que al menos una persona obtenga su propio sombrero.

---

## Representación mediante una permutación

Podemos representar el resultado del experimento mediante un vector de $N$ números.

La entrada $i$ indica el número del sombrero escogido por la persona $i$.

Por ejemplo,

$$
(1,2,3,\ldots,N)
$$

representa el resultado en el que **todas las personas obtienen su propio sombrero**.

Como cada sombrero solamente puede ser elegido por una persona, el resultado es una permutación de

$$
1,2,\ldots,N.
$$

Hay

$$
N!
$$

permutaciones posibles y suponemos que todas son igualmente probables.

---

# Aplicación del principio de inclusión-exclusión

Queremos calcular

$$
P\left(\bigcup_{i=1}^N E_i\right).
$$

Por inclusión-exclusión,

$$
P\left(\bigcup_{i=1}^N E_i\right)
=
\sum_i P(E_i)
-
\sum_{i<j}P(E_i\cap E_j)
+
\sum_{i<j<k}P(E_i\cap E_j\cap E_k)
-\cdots.
$$

Veamos cada término.

---

## Una persona obtiene su propio sombrero

Para una persona específica $i$,

$$
P(E_i)=\frac{1}{N}.
$$

Esto se debe a que la persona $i$ puede recibir cualquiera de los $N$ sombreros y solamente uno de ellos es el suyo.

Hay

$$
\binom{N}{1}=N
$$

formas de escoger qué persona obtiene su propio sombrero.

Por lo tanto, la primera suma de inclusión-exclusión es

$$
\binom{N}{1}\frac1N.
$$

---

## Dos personas específicas obtienen su propio sombrero

Sean $i\neq j$.

Queremos calcular

$$
P(E_i\cap E_j).
$$

La persona $i$ obtiene su sombrero con probabilidad

$$
\frac1N.
$$

Una vez que esto ocurrió, quedan $N-1$ sombreros para las demás personas.

Entonces la probabilidad de que la persona $j$ obtenga también su sombrero es

$$
\frac1{N-1}.
$$

Por la regla de multiplicación,

$$
P(E_i\cap E_j)
=
\frac1N\frac1{N-1}.
$$

Hay

$$
\binom{N}{2}
$$

formas de escoger las dos personas.

Por lo tanto, el segundo término de inclusión-exclusión es

$$
\binom{N}{2}
\frac1N
\frac1{N-1}.
$$

---

## Tres personas específicas obtienen su propio sombrero

Sean $i,j,k$ tres personas distintas.

Por la regla de multiplicación,

$$
P(E_i\cap E_j\cap E_k)
=
\frac1N
\frac1{N-1}
\frac1{N-2}.
$$

En efecto:

- la primera persona obtiene su sombrero con probabilidad $\frac1N$;
- quedan $N-1$ sombreros y la segunda obtiene el suyo con probabilidad $\frac1{N-1}$;
- quedan $N-2$ sombreros y la tercera obtiene el suyo con probabilidad $\frac1{N-2}$.

Hay

$$
\binom{N}{3}
$$

formas de elegir a estas tres personas.

---

## Caso general

Si seleccionamos $k$ personas específicas, la probabilidad de que todas ellas obtengan su propio sombrero es

$$
\frac1N
\frac1{N-1}
\cdots
\frac1{N-k+1}.
$$

Esta expresión puede escribirse como

$$
\frac{(N-k)!}{N!}.
$$

Hay

$$
\binom Nk
$$

formas de escoger las $k$ personas.

Por lo tanto, por inclusión-exclusión,

$$
P\left(\bigcup_{i=1}^N E_i\right)
=
\sum_{k=1}^N
(-1)^{k+1}
\binom Nk
\frac{(N-k)!}{N!}.
$$

Ahora usamos que

$$
\binom Nk
=
\frac{N!}{k!(N-k)!}.
$$

Entonces

$$
\binom Nk
\frac{(N-k)!}{N!}
=
\frac1{k!}.
$$

Así,

$$
P\left(\bigcup_{i=1}^N E_i\right)
=
1-\frac1{2!}+\frac1{3!}-\frac1{4!}
+\cdots+
\frac{(-1)^{N+1}}{N!}.
$$

---

# Probabilidad de que nadie obtenga su sombrero

Recordemos que

$$
P\left(\bigcap_{i=1}^N E_i^c\right)
=
1-
P\left(\bigcup_{i=1}^N E_i\right).
$$

Por lo tanto,

$$
P\left(\bigcap_{i=1}^N E_i^c\right)
=
1-
\left(
1-\frac1{2!}+\frac1{3!}-\cdots+\frac{(-1)^{N+1}}{N!}
\right).
$$

Así,

$$
\boxed{
P\left(\bigcap_{i=1}^N E_i^c\right)
=
\sum_{j=0}^N\frac{(-1)^j}{j!}
}.
$$

Denotemos esta probabilidad por

$$
p_N
=
\sum_{j=0}^N\frac{(-1)^j}{j!}.
$$

Recordemos que la expansión de $e^{-1}$ es

$$
e^{-1}
=
\sum_{j=0}^{\infty}
\frac{(-1)^j}{j!}.
$$

Por lo tanto,

$$
\boxed{
p_N\longrightarrow e^{-1}
}
$$

cuando

$$
N\to\infty.
$$

Es decir, para $N$ grande,

$$
\boxed{
P(\text{nadie obtiene su sombrero})
\approx
\frac1e
\approx0.367879
}.
$$

Este problema se conoce como el **problema de los desarreglos**.

---

# Probabilidad de que exactamente $k$ personas obtengan su sombrero

Ahora queremos calcular la probabilidad de que exactamente

$$
k
$$

de las $N$ personas obtengan su propio sombrero.

Primero fijemos cuáles son esas $k$ personas.

Supongamos que son

$$
i_1,\ldots,i_k.
$$

Definimos

$$
E
=
E_{i_1}\cap\cdots\cap E_{i_k}.
$$

Así,

$$
E
=
\{\text{las personas }i_1,\ldots,i_k
\text{ obtienen su propio sombrero}\}.
$$

Además, definimos

$$
G
=
\{\text{ninguna de las otras }N-k
\text{ personas obtiene su propio sombrero}\}.
$$

Entonces

$$
E\cap G
$$

es el evento en el que **exactamente esas $k$ personas** obtienen su sombrero.

Por la regla de multiplicación,

$$
P(E\cap G)
=
P(E)P(G\mid E).
$$

---

## Cálculo de $P(E)$

Por la regla de multiplicación,

$$
P(E)
=
\frac1N
\frac1{N-1}
\frac1{N-2}
\cdots
\frac1{N-k+1}.
$$

Por lo tanto,

$$
P(E)
=
\frac{(N-k)!}{N!}.
$$

---

## Cálculo de $P(G\mid E)$

Una vez que esas $k$ personas obtuvieron sus propios sombreros, quedan

$$
N-k
$$

personas y

$$
N-k
$$

sombreros.

Queremos que ninguna de esas personas obtenga su propio sombrero.

Esta es exactamente la misma situación que el problema anterior, pero ahora con $N-k$ personas.

Por lo tanto,

$$
P(G\mid E)
=
p_{N-k},
$$

donde

$$
p_{N-k}
=
\sum_{j=0}^{N-k}
\frac{(-1)^j}{j!}.
$$

Así,

$$
P(E\cap G)
=
\frac{(N-k)!}{N!}p_{N-k}.
$$

---

## Escoger cuáles son las $k$ personas

Hasta ahora habíamos fijado cuáles eran las $k$ personas que obtenían su sombrero.

Pero podemos escogerlas de

$$
\binom Nk
$$

maneras.

Por lo tanto,

$$
P(\text{exactamente }k\text{ personas obtienen su sombrero})
=
\binom Nk
\frac{(N-k)!}{N!}
p_{N-k}.
$$

Como

$$
\binom Nk
=
\frac{N!}{k!(N-k)!},
$$

tenemos

$$
\binom Nk
\frac{(N-k)!}{N!}
=
\frac1{k!}.
$$

Por lo tanto,

$$
\boxed{
P(\text{exactamente }k)
=
\frac{p_{N-k}}{k!}
}.
$$

Equivalentemente,

$$
\boxed{
P(\text{exactamente }k)
=
\frac1{k!}
\sum_{j=0}^{N-k}
\frac{(-1)^j}{j!}
}.
$$

Cuando $N-k$ es grande,

$$
p_{N-k}\approx e^{-1},
$$

por lo que

$$
\boxed{
P(\text{exactamente }k)
\approx
\frac{e^{-1}}{k!}
}.
$$

> **Nota.**
> La expresión $\frac{e^{-1}}{k!}$ es una aproximación para $N$ grande.
> La fórmula exacta es
>
> $$
> \frac{p_{N-k}}{k!}.
> $$

# Ejemplo: repartir una baraja en cuatro montones

Una baraja ordinaria de $52$ cartas se divide aleatoriamente en $4$ montones de $13$ cartas cada uno.

Queremos calcular:

> ¿Cuál es la probabilidad de que cada montón tenga exactamente un as?

Como hay cuatro ases y cuatro montones, esto equivale a pedir que los cuatro ases estén en montones distintos.

---

## Primera forma: colocar los ases uno por uno

Definimos los eventos:

$$
E_1
=
\{\text{el as de espadas puede estar en cualquier montón}\},
$$

$$
E_2
=
\{\text{los ases de espadas y corazones están en montones distintos}\},
$$

$$
E_3
=
\{\text{los tres primeros ases están en montones distintos}\},
$$

y

$$
E_4
=
\{\text{los cuatro ases están en montones distintos}\}.
$$

Queremos calcular

$$
P(E_4).
$$

Por la regla de multiplicación,

$$
P(E_4)
=
P(E_1)
P(E_2\mid E_1)
P(E_3\mid E_1\cap E_2)
P(E_4\mid E_1\cap E_2\cap E_3).
$$

---

## Primer as

El primer as puede estar en cualquier posición, por lo que

$$
P(E_1)=1.
$$

---

## Segundo as

Una vez colocado el primer as, quedan

$$
51
$$

posiciones disponibles.

En el mismo montón donde quedó el primer as quedan

$$
12
$$

posiciones.

Por lo tanto, en los otros tres montones hay

$$
51-12=39
$$

posiciones.

Para que el segundo as quede en un montón diferente al primero, debe ocupar una de estas $39$ posiciones.

Así,

$$
P(E_2\mid E_1)
=
\frac{39}{51}.
$$

---

## Tercer as

Supongamos que los primeros dos ases ya están en montones diferentes.

Quedan

$$
50
$$

posiciones disponibles.

Hay dos montones que todavía no contienen ningún as.

Cada uno tiene $13$ posiciones, así que hay

$$
13+13=26
$$

posiciones favorables.

Por lo tanto,

$$
P(E_3\mid E_1\cap E_2)
=
\frac{26}{50}.
$$

---

## Cuarto as

Supongamos ahora que los primeros tres ases están en tres montones distintos.

Quedan

$$
49
$$

posiciones disponibles.

Existe solamente un montón sin as y ese montón tiene

$$
13
$$

posiciones.

Por lo tanto,

$$
P(E_4\mid E_1\cap E_2\cap E_3)
=
\frac{13}{49}.
$$

---

## Probabilidad final

Por la regla de multiplicación,

$$
P(E_4)
=
1
\left(\frac{39}{51}\right)
\left(\frac{26}{50}\right)
\left(\frac{13}{49}\right).
$$

Por lo tanto,

$$
\boxed{
P(E_4)
=
\frac{39}{51}
\frac{26}{50}
\frac{13}{49}
}.
$$

Numéricamente,

$$
\boxed{
P(E_4)
\approx0.105498
}.
$$

Es decir, aproximadamente

$$
\boxed{10.55\%}.
$$

---

# Segunda forma: construir cada montón

Otra forma de resolver el problema consiste en ir formando los montones de $13$ cartas.

Sea

$$
E_i
=
\{\text{el montón }i\text{ contiene exactamente un as}\}.
$$

Queremos calcular

$$
P(E_1\cap E_2\cap E_3\cap E_4).
$$

Por la regla de multiplicación,

$$
P(E_1\cap E_2\cap E_3\cap E_4)
=
P(E_1)
P(E_2\mid E_1)
P(E_3\mid E_1\cap E_2)
P(E_4\mid E_1\cap E_2\cap E_3).
$$

---

## Primer montón

Queremos seleccionar $13$ cartas de las $52$ de manera que haya exactamente un as.

Hay

$$
\binom{52}{13}
$$

formas totales de formar el primer montón.

Para que tenga exactamente un as:

- elegimos $1$ de los $4$ ases;
- elegimos las otras $12$ cartas entre las $48$ cartas que no son ases.

Por lo tanto,

$$
P(E_1)
=
\frac{
\binom41\binom{48}{12}
}{
\binom{52}{13}
}.
$$

---

## Segundo montón

Si el primer montón ya contiene exactamente un as, quedan

$$
39
$$

cartas.

Entre ellas quedan:

- $3$ ases;
- $36$ cartas que no son ases.

Por lo tanto,

$$
P(E_2\mid E_1)
=
\frac{
\binom31\binom{36}{12}
}{
\binom{39}{13}
}.
$$

---

## Tercer montón

Después de formar correctamente los primeros dos montones, quedan

$$
26
$$

cartas.

Entre ellas hay:

- $2$ ases;
- $24$ cartas que no son ases.

Por lo tanto,

$$
P(E_3\mid E_1\cap E_2)
=
\frac{
\binom21\binom{24}{12}
}{
\binom{26}{13}
}.
$$

---

## Cuarto montón

Si los primeros tres montones tienen exactamente un as cada uno, queda:

- un as;
- $12$ cartas que no son ases.

Es decir, el último montón necesariamente contiene exactamente un as.

Por lo tanto,

$$
P(E_4\mid E_1\cap E_2\cap E_3)=1.
$$

---

## Probabilidad final

Así,

$$
P(E_1\cap E_2\cap E_3\cap E_4)
=
\frac{
\binom41\binom{48}{12}
}{
\binom{52}{13}
}
\frac{
\binom31\binom{36}{12}
}{
\binom{39}{13}
}
\frac{
\binom21\binom{24}{12}
}{
\binom{26}{13}
}.
$$

Al simplificar obtenemos

$$
P(E_1\cap E_2\cap E_3\cap E_4)
=
\frac{6\cdot13^3}{51\cdot50\cdot49}.
$$

Equivalentemente,

$$
P(E_1\cap E_2\cap E_3\cap E_4)
=
\frac{2197}{20825}.
$$

Por lo tanto,

$$
\boxed{
P(E_1\cap E_2\cap E_3\cap E_4)
\approx0.105498
}.
$$

Como era de esperarse, obtenemos exactamente el mismo resultado que con la primera forma.
