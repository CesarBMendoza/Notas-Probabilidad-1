# Caminos que pasan por un punto fijo

Consideremos la cuadrícula de la figura y queremos contar los caminos mínimos que van de $A$ a $B$ y que pasan por el punto $O$.

La idea es dividir el recorrido en dos partes:

1. ir de $A$ a $O$;
2. ir de $O$ a $B$.

---

## Caminos de $A$ a $O$

Para llegar de $A$ a $O$ necesitamos realizar $4$ movimientos:

- $2$ hacia la derecha;
- $2$ hacia arriba.

Por ejemplo, uno de los caminos podría representarse mediante

$$
D\,A\,D\,A,
$$

donde $D$ significa derecha y $A$ significa arriba.

Todo camino queda determinado al escoger en cuáles $2$ de los $4$ movimientos avanzaremos hacia la derecha.

Por lo tanto, hay

$$
\binom{4}{2}
$$

caminos distintos de $A$ a $O$.

El número

$$
\binom{4}{2}
$$

representa:

> las formas de escoger cuáles $2$ de los $4$ movimientos serán hacia la derecha.

Así,

$$
\binom{4}{2}=6.
$$

---

## Caminos de $O$ a $B$

Para ir de $O$ a $B$ necesitamos $3$ movimientos:

- $2$ hacia la derecha;
- $1$ hacia arriba.

Podemos escoger en cuáles $2$ de los $3$ movimientos avanzaremos hacia la derecha.

Por lo tanto,

$$
\binom{3}{2}=3.
$$

El número

$$
\binom{3}{2}
$$

representa:

> las formas de escoger cuáles $2$ de los $3$ movimientos serán hacia la derecha.

---

## Caminos de $A$ a $B$ que pasan por $O$

Cada camino de $A$ a $O$ puede combinarse con cualquiera de los caminos de $O$ a $B$.

Por el principio multiplicativo,

$$
\binom{4}{2}\binom{3}{2}.
$$

Así,

$$
\binom{4}{2}\binom{3}{2}
=
6\cdot3
=
18.
$$

Por lo tanto, existen

$$
\boxed{18}
$$

caminos mínimos distintos de $A$ a $B$ que pasan por $O$.

# Ayudantía 4

Se lanzan $2$ dados justos.

Definimos los siguientes eventos:

- $E$: la suma de los dos dados es impar.
- $F$: al menos uno de los dados muestra un $1$.
- $G$: la suma de los dos dados es $5$.

El espacio muestral es

$$
\Omega
=
\{1,2,\ldots,6\}\times\{1,2,\ldots,6\}.
$$

Como cada dado tiene $6$ posibles resultados,

$$
|\Omega|=6\cdot6=36.
$$

---

## Evento $E$

La suma es impar cuando un dado muestra un número impar y el otro un número par.

Por lo tanto,

$$
E=
\{
(1,2),(1,4),(1,6),
$$

$$
(2,1),(2,3),(2,5),
$$

$$
(3,2),(3,4),(3,6),
$$

$$
(4,1),(4,3),(4,5),
$$

$$
(5,2),(5,4),(5,6),
$$

$$
(6,1),(6,3),(6,5)
\}.
$$

Así,

$$
|E|=18.
$$

---

## Evento $F$

El evento $F$ ocurre cuando al menos uno de los dados muestra un $1$:

$$
F=
\{
(1,1),(1,2),(1,3),(1,4),(1,5),(1,6),
$$

$$
(2,1),(3,1),(4,1),(5,1),(6,1)
\}.
$$

Por lo tanto,

$$
|F|=11.
$$

---

## Evento $G$

La suma es $5$ en los resultados

$$
G=
\{
(1,4),(2,3),(3,2),(4,1)
\}.
$$

Por lo tanto,

$$
|G|=4.
$$

---

# Descripción de eventos

## 1. $E\cap F$

El evento

$$
E\cap F
$$

significa:

> la suma es impar y al menos uno de los dados muestra un $1$.

Tenemos

$$
E\cap F
=
\{
(1,2),(1,4),(1,6),
(2,1),(4,1),(6,1)
\}.
$$

Por lo tanto,

$$
|E\cap F|=6
$$

y

$$
P(E\cap F)
=
\frac{6}{36}
=
\frac16.
$$

---

## 2. $E\cup F$

El evento

$$
E\cup F
$$

significa:

> la suma es impar o al menos uno de los dados muestra un $1$.

Utilizando inclusión-exclusión,

$$
|E\cup F|
=
|E|+|F|-|E\cap F|.
$$

Sustituyendo,

$$
|E\cup F|
=
18+11-6
=
23.
$$

Por lo tanto,

$$
P(E\cup F)
=
\frac{23}{36}.
$$

---

## 3. $F\cap G$

Significa:

> al menos uno de los dados muestra un $1$ y la suma es $5$.

Los únicos resultados posibles son

$$
F\cap G
=
\{(1,4),(4,1)\}.
$$

Por lo tanto,

$$
P(F\cap G)
=
\frac{2}{36}
=
\frac{1}{18}.
$$

---

## 4. $E\cap F^c$

El evento $F^c$ significa que ninguno de los dos dados muestra un $1$.

Por lo tanto,

$$
E\cap F^c
$$

significa:

> la suma es impar y ninguno de los dados muestra un $1$.

Como

$$
|E|=18
$$

y $6$ de esos resultados pertenecen a $F$, quedan

$$
18-6=12
$$

resultados.

Así,

$$
P(E\cap F^c)
=
\frac{12}{36}
=
\frac13.
$$

---

## 5. $E\cap F\cap G$

El evento significa:

> la suma es impar, al menos uno de los dados muestra un $1$ y la suma es $5$.

Como toda suma igual a $5$ es impar, basta buscar los elementos de $F\cap G$:

$$
E\cap F\cap G
=
\{(1,4),(4,1)\}.
$$

Por lo tanto,

$$
P(E\cap F\cap G)
=
\frac{2}{36}
=
\frac{1}{18}.
$$

# Juego de monedas

$A$, $B$ y $C$ se turnan para lanzar una moneda.

El orden de los turnos es

$$
A,B,C,A,B,C,\ldots
$$

El primero en obtener **cara** gana.

Representaremos:

- $1=$ cara;
- $0=$ sol.

El espacio muestral puede escribirse como

$$
\Omega
=
\{1,01,001,0001,00001,\ldots\}.
$$

Equivalentemente,

$$
\Omega
=
\{0^n1:n=0,1,2,\ldots\}.
$$

Cada resultado termina cuando aparece la primera cara.

---

## ¿Cómo se interpreta el espacio muestral?

### Resultado $1$

$$
1
$$

significa que $A$ obtiene cara inmediatamente.

Por lo tanto, gana $A$.

---

### Resultado $01$

$$
01
$$

significa:

- $A$ obtiene sol;
- $B$ obtiene cara.

Por lo tanto, gana $B$.

---

### Resultado $001$

$$
001
$$

significa:

- $A$ obtiene sol;
- $B$ obtiene sol;
- $C$ obtiene cara.

Por lo tanto, gana $C$.

---

### Resultado $0001$

Tenemos:

- $A$: sol;
- $B$: sol;
- $C$: sol;
- $A$: cara.

Por lo tanto, gana $A$ nuevamente.

El patrón continúa periódicamente.

---

# Eventos en términos del espacio muestral

## Evento: $A$ gana

$A$ gana cuando aparecen

$$
0,3,6,9,\ldots
$$

soles antes de la primera cara.

Por lo tanto,

$$
A
=
\{
1,0001,0000001,\ldots
\}.
$$

En notación general,

$$
\boxed{
A
=
\{0^{3k}1:k=0,1,2,\ldots\}.
}
$$

---

## Evento: $B$ gana

$B$ gana cuando aparecen

$$
1,4,7,10,\ldots
$$

soles antes de la primera cara.

Por lo tanto,

$$
B
=
\{
01,00001,00000001,\ldots
\}.
$$

En general,

$$
\boxed{
B
=
\{0^{3k+1}1:k=0,1,2,\ldots\}.
}
$$

---

## Evento: $C$ gana

$C$ gana cuando aparecen

$$
2,5,8,11,\ldots
$$

soles antes de la primera cara.

Así,

$$
C
=
\{
001,000001,000000001,\ldots
\}.
$$

En general,

$$
\boxed{
C
=
\{0^{3k+2}1:k=0,1,2,\ldots\}.
}
$$

---

## Complemento de $B$

El evento

$$
B^c
$$

significa que $B$ no gana.

Como necesariamente uno de los tres jugadores gana dentro del espacio muestral $\Omega$, tenemos

$$
\boxed{
B^c=A\cup C.
}
$$

Además,

$$
A\cap B
=
A\cap C
=
B\cap C
=
\varnothing
$$

y

$$
A\cup B\cup C=\Omega.
$$

---

## El evento $A\cup B^c$

Como

$$
B^c=A\cup C,
$$

entonces

$$
A\cup B^c
=
A\cup(A\cup C).
$$

Por lo tanto,

$$
\boxed{
A\cup B^c
=
A\cup C
=
B^c.
}
$$

# Ejemplo: distribución de cumpleaños

Consideremos $20$ personas.

¿Cuál es la probabilidad de que, entre los $12$ meses del año:

- haya exactamente $4$ meses que contengan $2$ cumpleaños;
- haya exactamente $4$ meses que contengan $3$ cumpleaños?

Supongamos que:

1. el mes de nacimiento de cada persona es independiente del de las demás;
2. cada uno de los $12$ meses es igualmente probable.

---

## Espacio muestral

Cada una de las $20$ personas puede haber nacido en cualquiera de los $12$ meses.

Por lo tanto,

$$
|\Omega|
=
12^{20}.
$$

Aquí:

- el primer factor $12$ representa el mes de nacimiento de la primera persona;
- el segundo factor $12$ representa el mes de nacimiento de la segunda;
- y así sucesivamente hasta la persona $20$.

Es decir,

$$
12^{20}
=
\underbrace{12\cdot12\cdots12}_{20\text{ factores}}.
$$

---

## Verificación del número de personas

Queremos:

- $4$ meses con $3$ cumpleaños;
- $4$ meses con $2$ cumpleaños.

Por lo tanto,

$$
4(3)+4(2)
=
12+8
=
20.
$$

Así quedan distribuidas las $20$ personas.

---

# Conteo de los casos favorables

## Paso 1: escoger los meses con $3$ cumpleaños

Debemos elegir cuáles $4$ de los $12$ meses tendrán exactamente $3$ cumpleaños.

Esto puede hacerse de

$$
\binom{12}{4}
$$

formas.

El número

$$
\binom{12}{4}
$$

representa:

> las formas de escoger cuáles $4$ de los $12$ meses serán los meses con exactamente $3$ cumpleaños.

---

## Paso 2: escoger los meses con $2$ cumpleaños

Después de elegir los $4$ meses anteriores, quedan

$$
12-4=8
$$

meses.

De estos $8$ debemos escoger $4$ para que tengan exactamente $2$ cumpleaños.

Esto puede hacerse de

$$
\binom{8}{4}
$$

formas.

El número

$$
\binom{8}{4}
$$

representa:

> las formas de escoger cuáles $4$ de los $8$ meses restantes tendrán exactamente $2$ cumpleaños.

Los otros $4$ meses tendrán $0$ cumpleaños.

---

## Paso 3: distribuir a las $20$ personas

Ya hemos decidido cuáles meses tendrán:

$$
3,3,3,3,2,2,2,2
$$

cumpleaños.

Como las $20$ personas son distintas, debemos distribuirlas entre estos $8$ meses.

El número de maneras es el coeficiente multinomial

$$
\binom{20}
{3,3,3,3,2,2,2,2}.
$$

Por definición,

$$
\binom{20}
{3,3,3,3,2,2,2,2}
=
\frac{20!}
{(3!)^4(2!)^4}.
$$

Aquí:

- $20!$ cuenta inicialmente las formas de ordenar a las $20$ personas;
- cada $3!$ corrige el orden de las $3$ personas que pertenecen a uno de los meses con $3$ cumpleaños;
- hay $4$ meses de este tipo, por eso aparece $(3!)^4$;
- cada $2!$ corrige el orden de las $2$ personas de uno de los meses con $2$ cumpleaños;
- hay $4$ meses de este tipo, por eso aparece $(2!)^4$.

---

## Número de casos favorables

Por el principio multiplicativo,

$$
|E|
=
\binom{12}{4}
\binom{8}{4}
\frac{20!}{(3!)^4(2!)^4}.
$$

Cada factor tiene una función distinta:

$$
\underbrace{\binom{12}{4}}_
{\substack{\text{escoger los 4 meses}\\\text{con 3 cumpleaños}}}
\;
\underbrace{\binom{8}{4}}_
{\substack{\text{escoger los 4 meses}\\\text{con 2 cumpleaños}}}
\;
\underbrace{\frac{20!}{(3!)^4(2!)^4}}_
{\substack{\text{distribuir a las 20 personas}\\
\text{entre esos meses}}}.
$$

---

# Probabilidad

Como todos los resultados elementales son equiprobables,

$$
P(E)
=
\frac{|E|}{|\Omega|}.
$$

Por lo tanto,

$$
\boxed{
P(E)
=
\frac{
\binom{12}{4}
\binom{8}{4}
\dfrac{20!}{(3!)^4(2!)^4}
}{
12^{20}
}.
}
$$

Numéricamente,

$$
P(E)
\approx
0.00106042.
$$

Por lo tanto,

$$
\boxed{
P(E)\approx0.00106.
}
$$

En porcentaje,

$$
P(E)
\approx
0.106\%.
$$
